**الفهرسة المنطقية** بعمود واحد هي أيضًا مريحة للغاية. بناء الجملة هو `df[P(df['Name'])]`، حيث `P` هو شرط منطقي يتم التحقق منه لكل عنصر في العمود `Name`. نتيجة هذه الفهرسة هي `DataFrame` التي تتكون فقط من صفوف تفي بشرط `P` في العمود `Name`. 
لنستخدمها للإجابة على السؤال:
**ما هو متوسط قيم الميزات الرقمية للمستخدمين المختلين؟**

````python
df.select_dtypes(include=np.number)[df["Churn"] == 1].mean()
````

### تطبيق الوظائف على الخلايا والأعمدة والصفوف
**لتطبيق الوظائف على كل عمود، استخدم `apply()`:**


````python
df.apply('max')
````


يمكن أيضًا استخدام الأسلوب `apply` لتطبيق دالة على كل صف. للقيام بذلك، حدد `axis=1`. تعتبر وظائف Lambda ملائمة جدًا في مثل هذه السيناريوهات. على سبيل المثال، إذا أردنا تحديد جميع الحالات التي تبدأ بـ "W"، فيمكننا القيام بذلك على النحو التالي:


````python
df[df["State"].apply(lambda state: state[0] == "W")].head()
````

````python
d = {"No": False, "Yes": True}
df["International plan"] = df["International plan"].map(d)
df.head()
````

a_series = pd.Series(['a', 'b', 'c'])


### التجمع
بشكل عام، تجميع البيانات في Pandas يعمل على النحو التالي:



```python
df.groupby(by=grouping_columns)[columns_to_show].function()
```



1. أولاً، تقوم الطريقة `groupby` بتقسيم `grouping_columns` على قيمها. تصبح فهرسًا جديدًا في إطار البيانات الناتج.
2. ثم يتم تحديد الأعمدة محل الاهتمام (`columns_to_show`). إذا لم يتم تضمين `columns_to_show`، فسيتم تضمين كافة البنود غير المجمعة.
3. وأخيرًا، يتم تطبيق وظيفة واحدة أو عدة وظائف على المجموعات التي تم الحصول عليها لكل أعمدة محددة.
فيما يلي مثال حيث نقوم بتجميع البيانات وفقًا لقيم المتغير `Churn` ونعرض إحصائيات ثلاثة أعمدة في كل مجموعة:


````python
columns_to_show = ["Total day minutes", "Total eve minutes", "Total night minutes"]

df.groupby(["Churn"])[columns_to_show].describe(percentiles=[])
````


لنفعل نفس الشيء، ولكن بشكل مختلف قليلاً عن طريق تمرير قائمة الوظائف إلى `agg()`:


````python
columns_to_show = ["Total day minutes", "Total eve minutes", "Total night minutes"]

df.groupby(["Churn"])[columns_to_show].agg(['mean', 'std', 'min', 'max'])
````


### جداول ملخصةلنفترض أننا نريد أن نرى كيف يتم توزيع الملاحظات في مجموعة البيانات الخاصة بنا في سياق متغيرين - `Churn` و`International plan`. للقيام بذلك، يمكننا إنشاء **جدول طوارئ** باستخدام الطريقة `crosstab`:


````python
pd.crosstab(df["Churn"], df["International plan"])
````

````python
pd.crosstab(df["Churn"], df["Voice mail plan"], normalize=True)
````


يمكننا أن نرى أن معظم المستخدمين مخلصون ولا يستخدمون خدمات إضافية (الخطة الدولية/البريد الصوتي).
سيشبه هذا **الجداول المحورية** بالنسبة لأولئك الذين هم على دراية ببرنامج Excel. وبالطبع، يتم تنفيذ الجداول المحورية في Pandas: تأخذ الطريقة `pivot_table` المعلمات التالية:
* `values` - قائمة المتغيرات لحساب الإحصائيات،
* `index` - قائمة المتغيرات لتجميع البيانات حسب،
* `aggfunc` – ما هي الإحصائيات التي نحتاج لحسابها للمجموعات، على سبيل المثال. المجموع أو المتوسط ​​أو الحد الأقصى أو الحد الأدنى أو أي شيء آخر.
دعونا نلقي نظرة على متوسط عدد المكالمات النهارية والمسائية والليلية حسب رمز المنطقة:


````python
df.pivot_table(
    ["Total day calls", "Total eve calls", "Total night calls"],
    ["Area code"],
    aggfunc="mean",
)
````


### تحويلات DataFrame
مثل العديد من الأشياء الأخرى في Pandas، يمكن إضافة أعمدة إلى DataFrame بعدة طرق.
على سبيل المثال، إذا أردنا حساب العدد الإجمالي للمكالمات لجميع المستخدمين، فلنقم بإنشاء سلسلة `total_calls` ولصقها في DataFrame:


````python
total_calls = (
    df["Total day calls"]
    + df["Total eve calls"]
    + df["Total night calls"]
    + df["Total intl calls"]
)
df.insert(loc=len(df.columns), column="Total calls", value=total_calls)
# loc parameter is the number of columns after which to insert the Series object
# we set it to len(df.columns) to paste it at the very end of the dataframe
df.head()
````


من الممكن إضافة عمود بسهولة أكبر دون إنشاء مثيل سلسلة وسيطة:


````python
df["Total charge"] = (
    df["Total day charge"]
    + df["Total eve charge"]
    + df["Total night charge"]
    + df["Total intl charge"]
)
df.head()
````


لحذف الأعمدة أو الصفوف، استخدم الأسلوب `drop`، وقم بتمرير الفهارس المطلوبة والمعلمة `axis` (`1` إذا قمت بحذف الأعمدة، ولا شيء أو `0` إذا قمت بحذف الصفوف). تخبر الوسيطة `inplace` ما إذا كان سيتم تغيير DataFrame الأصلي. باستخدام `inplace=False`، لا تقوم الطريقة `drop` بتغيير DataFrame الموجود وتقوم بإرجاع إطار جديد يحتوي على صفوف أو أعمدة تم إسقاطها. باستخدام `inplace=True`، فإنه يغير DataFrame.


````python
# get rid of just created columns
df.drop(["Total charge", "Total calls"], axis=1, inplace=True)
# and here’s how you can delete rows
df.drop([1, 2]).head()
````


## 2. المحاولة الأولى للتنبؤ بتقلبات قطاع الاتصالاتدعونا نرى مدى ارتباط معدل الإلغاء بميزة *الخطة الدولية*. سنفعل ذلك باستخدام جدول الطوارئ `crosstab` وأيضًا من خلال التحليل المرئي باستخدام `Seaborn` (ومع ذلك، سيتم تغطية التحليل البصري بشكل أكثر شمولاً في المقالة التالية).


````python
pd.crosstab(df["Churn"], df["International plan"], margins=True)
````

````python
# some imports to set up plotting
import matplotlib.pyplot as plt
# !pip install seaborn
import seaborn as sns

# import some nice vis settings
sns.set()
# Graphics in the Retina format are more sharp and legible
%config InlineBackend.figure_format = 'retina'
````

````python
sns.countplot(x="International plan", hue="Churn", data=df);
````


نرى أنه مع *الخطة الدولية*، يكون معدل الإلغاء أعلى بكثير، وهي ملاحظة مثيرة للاهتمام! ربما تكون النفقات الكبيرة والتي لا يمكن التحكم فيها بشكل جيد فيما يتعلق بالمكالمات الدولية عرضة للصراع وتؤدي إلى عدم الرضا بين عملاء مشغل الاتصالات.
بعد ذلك، دعونا نلقي نظرة على ميزة أخرى مهمة – *مكالمات خدمة العملاء*. لنقم أيضًا بإنشاء جدول ملخص وصورة.


````python
pd.crosstab(df["Churn"], df["Customer service calls"], margins=True)
````

````python
sns.countplot(x="Customer service calls", hue="Churn", data=df);
````


على الرغم من أنه ليس واضحًا جدًا من الجدول الملخص، فمن السهل أن نرى من المخطط أعلاه أن معدل الإلغاء يزيد بشكل حاد من 4 مكالمات لخدمة العملاء وما فوق.
الآن دعونا نضيف ميزة ثنائية إلى DataFrame الخاص بنا – `Customer service calls > 3`. ومرة أخرى، دعونا نرى مدى ارتباطه بالزبدة. 


````python
df["Many_service_calls"] = (df["Customer service calls"] > 3).astype("int")

pd.crosstab(df["Many_service_calls"], df["Churn"], margins=True)
````

````python
sns.countplot(x="Many_service_calls", hue="Churn", data=df);
````


لنقم بإنشاء جدول طوارئ آخر يربط *Churn* بكل من *الخطة الدولية* و*Many_service_calls* التي تم إنشاؤها حديثًا.


````python
pd.crosstab(df["Many_service_calls"] & df["International plan"], df["Churn"])
````


لذلك، توقع أن العميل ليس مخلصًا (*Churn*=1) في حالة ما إذا كان عدد المكالمات إلى مركز الخدمة أكبر من 3 وتم إضافة *الخطة الدولية* (والتوقع *Churn*=0 بخلاف ذلك)، قد نتوقع دقة تبلغ 85.8% (نحن مخطئون فقط 464 + 9 مرات). هذا الرقم، 85.8%، الذي حصلنا عليه من خلال هذا المنطق البسيط للغاية، يعد بمثابة نقطة بداية جيدة (*خط الأساس*) لنماذج التعلم الآلي الإضافية التي سنقوم ببنائها.بينما نمضي قدمًا في هذه الدورة التدريبية، تذكر أنه قبل ظهور التعلم الآلي، كانت عملية تحليل البيانات تبدو على هذا النحو. دعونا نلخص ما قمنا بتغطيته:
    
- حصة العملاء المخلصين في مجموعة البيانات 85.5%. النموذج الأكثر سذاجة والذي يتنبأ دائمًا بـ "العميل المخلص" على مثل هذه البيانات سوف يخمن بشكل صحيح في حوالي 85.5٪ من جميع الحالات. أي أن نسبة الإجابات الصحيحة (*الدقة*) للنماذج اللاحقة يجب ألا تقل عن هذا العدد، ونأمل أن تكون أعلى بكثير؛
- بمساعدة تنبؤ بسيط يمكن التعبير عنه بالمعادلة التالية: `International plan = True & Customer Service calls > 3 => Churn = 1, else Churn = 0`، يمكننا أن نتوقع معدل تخمين يبلغ 85.8%، وهو ما يزيد قليلاً عن 85.5%. بعد ذلك، سنتحدث عن أشجار القرار ونتعرف على كيفية العثور على هذه القواعد **تلقائيًا** بناءً على البيانات المدخلة فقط؛
- لقد حصلنا على هذين الخطين الأساسيين دون تطبيق التعلم الآلي، وسيكونان بمثابة نقطة البداية لنماذجنا اللاحقة. إذا اتضح أنه بجهد هائل، فإننا نزيد الدقة بنسبة 0.5٪ فقط، على سبيل المثال، فمن المحتمل أننا نفعل شيئًا خاطئًا، ويكفي أن نقتصر على نموذج بسيط "إذا كان الأمر كذلك" بشرطين؛
- قبل تدريب النماذج المعقدة، يوصى بمناقشة البيانات قليلاً، وعمل بعض المخططات، والتحقق من الافتراضات البسيطة. علاوة على ذلك، في تطبيقات الأعمال التجارية للتعلم الآلي، عادةً ما يبدأون بحلول بسيطة ثم يقومون بتجربة حلول أكثر تعقيدًا.
