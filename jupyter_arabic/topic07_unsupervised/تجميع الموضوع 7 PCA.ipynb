## 1. مقدمةالسمة الرئيسية لخوارزميات التعلم غير الخاضعة للرقابة، عند مقارنتها بطرق التصنيف والانحدار، هي أن بيانات الإدخال غير مصنفة (أي لا توجد تسميات أو فئات معينة) وأن الخوارزمية تتعلم بنية البيانات دون أي مساعدة. وهذا يخلق اختلافين رئيسيين. أولاً، يسمح لنا بمعالجة كميات كبيرة من البيانات لأن البيانات لا تحتاج إلى تصنيفها يدويًا. ثانيًا، من الصعب تقييم جودة الخوارزمية غير الخاضعة للإشراف بسبب عدم وجود مقياس جودة واضح كما هو مستخدم في التعلم الخاضع للإشراف.  
إحدى المهام الأكثر شيوعًا في التعلم غير الخاضع للرقابة هي تقليل الأبعاد. من ناحية، قد يساعد تقليل الأبعاد في تصور البيانات (على سبيل المثال طريقة t-SNE)، بينما من ناحية أخرى، قد يساعد في التعامل مع العلاقة الخطية المتعددة لبياناتك وإعداد البيانات لطريقة التعلم الخاضعة للإشراف (مثل أشجار القرار).
## 2. تحليل المكونات الرئيسية (PCA)
### الحدس والنظريات وقضايا التطبيق
يعد تحليل المكون الرئيسي واحدًا من أسهل الطرق وأكثرها بديهية والأكثر استخدامًا لتقليل الأبعاد، وإسقاط البيانات على الفضاء الفرعي للميزة المتعامدة.
<img align="right" src="@@KEEP_00012@@ width="400">بشكل أكثر عمومية، يمكن اعتبار جميع الملاحظات بمثابة شكل إهليلجي في فضاء فرعي من مساحة الميزة الأولية، ويتم محاذاة الأساس الجديد المحدد في هذا الفضاء الفرعي مع محاور الإهليلج. يتيح لنا هذا الافتراض إزالة الميزات شديدة الارتباط نظرًا لأن متجهات المجموعة الأساسية متعامدة. 
في الحالة العامة، تتطابق أبعاد الشكل الإهليلجي الناتج مع أبعاد الفضاء الأولية، لكن افتراض أن بياناتنا تقع في فضاء فرعي ذي بُعد أصغر يسمح لنا بقطع الفضاء "المفرط" بالإسقاط الجديد (الفضاء الفرعي). نحن ننجز ذلك بطريقة "جشعة"، حيث نقوم باختيار كل محور من محاور الإهليلج بشكل تسلسلي من خلال تحديد مكان التشتت الأقصى.
 
> "للتعامل مع المستويات المفرطة في مساحة ذات 14 بُعدًا، تصور مساحة ثلاثية الأبعاد وقل "أربعة عشر" بصوت عالٍ للغاية. الجميع يفعل ذلك." - جيفري هينتون
دعونا نلقي نظرة على الصيغة الرياضية لهذه العملية:
من أجل تقليل أبعاد بياناتنا من $n$ إلى $k$ باستخدام $k \leq n$، نقوم بفرز قائمة المحاور لدينا من أجل تقليل التشتت ونأخذ الجزء العلوي $k$ منها.
نبدأ بحساب التشتت والتباين في الميزات الأولية. يتم ذلك عادةً باستخدام مصفوفة التغاير. وفقًا لتعريف التباين المشترك، يتم حساب التباين المشترك للميزتين كما يلي: $$cov(X_i, X_j) = E[(X_i - \mu_i) (X_j - \mu_j)] = E[X_i X_j] - \mu_i \mu_j,$$ حيث $\mu_i$ هي القيمة المتوقعة للميزة $i$. ومن الجدير بالذكر أن التغاير متماثل، وأن تباين المتجه مع نفسه يساوي تشتته.لذلك تكون مصفوفة التغاير متماثلة مع تشتت الميزات المقابلة على القطر. القيم غير القطرية هي التغايرات المشتركة لزوج المعالم المقابل. فيما يتعلق بالمصفوفات حيث $\mathbf{X}$ هي مصفوفة الملاحظات، تكون مصفوفة التغاير كما يلي:
$$\Sigma = E[(\mathbf{X} - E[\mathbf{X}]) (\mathbf{X} - E[\mathbf{X}])^{T}]$$
خلاصة سريعة: المصفوفات، باعتبارها عوامل خطية، لها قيم ذاتية ومتجهات ذاتية. إنها مريحة للغاية لأنها تصف أجزاء من مساحتنا التي لا تدور وتمتد فقط عندما نطبق عليها عوامل خطية؛ تظل المتجهات الذاتية في نفس الاتجاه ولكنها ممتدة بقيمة ذاتية مقابلة. رسميًا، مصفوفة $M$ مع المتجه الذاتي $w_i$ والقيمة الذاتية $\lambda_i$ تلبي هذه المعادلة: $M w_i = \lambda_i w_i$.
يمكن كتابة مصفوفة التغاير لعينة $\mathbf{X}$ كمنتج لـ $\mathbf{X}^{T} \mathbf{X}$. وفقًا لـ [حاصل رايلي](https://en.wikipedia.org/wiki/Rayleigh_quotient)، فإن الحد الأقصى للتباين في عينتنا يقع على طول المتجه الذاتي لهذه المصفوفة ويتوافق مع القيمة الذاتية القصوى. ولذلك، فإن المكونات الرئيسية التي نهدف إلى الاحتفاظ بها من البيانات هي مجرد المتجهات الذاتية المقابلة للقيم الذاتية الأعلى $k$ الأكبر للمصفوفة.
الخطوات التالية أسهل في الهضم. نقوم بضرب مصفوفة بياناتنا $X$ بهذه المكونات للحصول على إسقاط بياناتنا على الأساس المتعامد للمكونات المختارة. إذا كان عدد المكونات أقل من البعد المكاني الأولي، تذكر أننا سنفقد بعض المعلومات عند تطبيق هذا التحويل.



## أمثلة
### مجموعة بيانات فيشر لقزحية العين
لنبدأ بتحميل جميع الوحدات الأساسية ونجرب مثال القزحية من وثائق `scikit-learn`. 


```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

sns.set(style="white")
%matplotlib inline
%config InlineBackend.figure_format = 'retina'
from mpl_toolkits.mplot3d import Axes3D
from sklearn import datasets, decomposition

# Loading the dataset
iris = datasets.load_iris()
X = iris.data
y = iris.target

# Let's create a beautiful 3d-plot
fig = plt.figure(1, figsize=(6, 5))
plt.clf()
ax = Axes3D(fig, rect=[0, 0, 0.95, 1], elev=48, azim=134)

plt.cla()

for name, label in [("Setosa", 0), ("Versicolour", 1), ("Virginica", 2)]:
    ax.text3D(
        X[y == label, 0].mean(),
        X[y == label, 1].mean() + 1.5,
        X[y == label, 2].mean(),
        name,
        horizontalalignment="center",
        bbox=dict(alpha=0.5, edgecolor="w", facecolor="w"),
    )
# Change the order of labels, so that they match
y_clr = np.choose(y, [1, 2, 0]).astype(np.float)
ax.scatter(X[:, 0], X[:, 1], X[:, 2], c=y_clr, cmap=plt.cm.nipy_spectral)

ax.w_xaxis.set_ticklabels([])
ax.w_yaxis.set_ticklabels([])
ax.w_zaxis.set_ticklabels([]);
```

الآن دعونا نرى كيف سيعمل PCA على تحسين نتائج نموذج بسيط غير قادر على ملاءمة جميع بيانات التدريب بشكل صحيح:


```python
from sklearn.metrics import accuracy_score, roc_auc_score
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

# Train, test splits
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, stratify=y, random_state=42
)

# Decision trees with depth = 2
clf = DecisionTreeClassifier(max_depth=2, random_state=42)
clf.fit(X_train, y_train)
preds = clf.predict_proba(X_test)
print("Accuracy: {:.5f}".format(accuracy_score(y_test, preds.argmax(axis=1))))
```


دعونا نحاول ذلك مرة أخرى، ولكن هذه المرة، دعونا نقلل الأبعاد إلى بعدين:


```python
# Using PCA from sklearn PCA
pca = decomposition.PCA(n_components=2)
X_centered = X - X.mean(axis=0)
pca.fit(X_centered)
X_pca = pca.transform(X_centered)

# Plotting the results of PCA
plt.plot(X_pca[y == 0, 0], X_pca[y == 0, 1], "bo", label="Setosa")
plt.plot(X_pca[y == 1, 0], X_pca[y == 1, 1], "go", label="Versicolour")
plt.plot(X_pca[y == 2, 0], X_pca[y == 2, 1], "ro", label="Virginica")
plt.legend(loc=0);
```

```python
# Test-train split and apply PCA
X_train, X_test, y_train, y_test = train_test_split(
    X_pca, y, test_size=0.3, stratify=y, random_state=42
)

clf = DecisionTreeClassifier(max_depth=2, random_state=42)
clf.fit(X_train, y_train)
preds = clf.predict_proba(X_test)
print("Accuracy: {:.5f}".format(accuracy_score(y_test, preds.argmax(axis=1))))
```


لم تزد الدقة بشكل ملحوظ في هذه الحالة، ولكن مع مجموعات البيانات الأخرى التي تحتوي على عدد كبير من الأبعاد، يمكن لـ PCA تحسين دقة أشجار القرار وطرق التجميع الأخرى بشكل كبير.



الآن دعونا نتحقق من نسبة التباين التي يمكن تفسيرها بواسطة كل مكون من المكونات المحددة.


```python
for i, component in enumerate(pca.components_):
    print(
        "{} component: {}% of initial variance".format(
            i + 1, round(100 * pca.explained_variance_ratio_[i], 2)
        )
    )
    print(
        " + ".join(
            "%.3f x %s" % (value, name)
            for value, name in zip(component, iris.feature_names)
        )
    )
```


### مجموعة بيانات الأرقام المكتوبة بخط اليد 
دعونا نلقي نظرة على مجموعة بيانات الأرقام المكتوبة بخط اليد والتي استخدمناها من قبل في [الدرس الثالث](https://habrahabr.ru/company/ods/blog/322534/#derevya-resheniy-i-metod-blizhayshih-sosedey-v-zadache-raspoznavaniya-rukopisnyh-cifr-mnist).


```python
digits = datasets.load_digits()
X = digits.data
y = digits.target
```


لنبدأ بتصور بياناتنا. جلب أول 10 أرقام. يتم تمثيل الأرقام بمصفوفات 8 × 8 مع كثافة اللون لكل بكسل. يتم تسوية كل مصفوفة في متجه مكون من 64 رقمًا، لذلك نحصل على الإصدار المميز للبيانات.


```python
# f, axes = plt.subplots(5, 2, sharey=True, figsize=(16,6))
plt.figure(figsize=(16, 6))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(X[i, :].reshape([8, 8]), cmap="gray");
```


تحتوي بياناتنا على 64 بُعدًا، ولكننا سنقوم بتقليلها إلى بُعدين فقط ونرى أنه حتى مع وجود بُعدين فقط، يمكننا أن نرى بوضوح أن الأرقام تنقسم إلى مجموعات.


```python
pca = decomposition.PCA(n_components=2)
X_reduced = pca.fit_transform(X)

print("Projecting %d-dimensional data to 2D" % X.shape[1])

plt.figure(figsize=(12, 10))
plt.scatter(
    X_reduced[:, 0],
    X_reduced[:, 1],
    c=y,
    edgecolor="none",
    alpha=0.7,
    s=40,
    cmap=plt.cm.get_cmap("nipy_spectral", 10),
)
plt.colorbar()
plt.title("MNIST. PCA projection");
```


في الواقع، مع t-SNE، تبدو الصورة أفضل نظرًا لأن PCA له قيد خطي بينما لا يوجد به t-SNE. ومع ذلك، حتى مع مجموعة البيانات الصغيرة هذه، تستغرق خوارزمية t-SNE وقتًا أطول بكثير لإكمالها من PCA.


```python
%%time

from sklearn.manifold import TSNE

tsne = TSNE(random_state=17)

X_tsne = tsne.fit_transform(X)

plt.figure(figsize=(12, 10))
plt.scatter(
    X_tsne[:, 0],
    X_tsne[:, 1],
    c=y,
    edgecolor="none",
    alpha=0.7,
    s=40,
    cmap=plt.cm.get_cmap("nipy_spectral", 10),
)
plt.colorbar()
plt.title("MNIST. t-SNE projection");
```


من الناحية العملية، نختار عدد المكونات الرئيسية بحيث يمكننا تفسير 90% من تشتت البيانات الأولية (عبر `explained_variance_ratio`). وهذا يعني هنا الاحتفاظ بـ 21 مكونًا رئيسيًا؛ لذلك قمنا بتقليل الأبعاد من 64 ميزة إلى 21.


```python
pca = decomposition.PCA().fit(X)

plt.figure(figsize=(10, 7))
plt.plot(np.cumsum(pca.explained_variance_ratio_), color="k", lw=2)
plt.xlabel("Number of components")
plt.ylabel("Total explained variance")
plt.xlim(0, 63)
plt.yticks(np.arange(0, 1.1, 0.1))
plt.axvline(21, c="b")
plt.axhline(0.9, c="r")
plt.show();
```


## 2. التجميعالفكرة الرئيسية وراء التجميع واضحة جدًا. في الأساس، نقول لأنفسنا، "لدي هذه النقاط هنا، وأستطيع أن أرى أنها تنتظم في مجموعات. سيكون من اللطيف وصف هذه الأشياء بشكل أكثر تحديدًا، وعندما تأتي نقطة جديدة، قم بتعيينها للمجموعة الصحيحة." تشجع هذه الفكرة العامة على الاستكشاف وتفتح مجموعة متنوعة من الخوارزميات للتجميع.

### ك-يعني
تعد خوارزمية K-means هي الأكثر شيوعًا والأبسط من بين جميع خوارزميات التجميع. وهنا كيف يعمل:
1. حدد عدد المجموعات $k$ التي تعتقد أنها العدد الأمثل.
2. قم بتهيئة $k$ النقاط على أنها "مراكز مركزية" بشكل عشوائي في مساحة بياناتنا.
3. قم بإسناد كل ملاحظة إلى أقرب نقطة مركزية لها.
4. قم بتحديث النقط الوسطى إلى مركز كل مجموعة الملاحظات المنسوبة. 
5. كرر الخطوتين 3 و4 عددًا محددًا من المرات أو حتى تستقر جميع النقط الوسطى (أي لم تعد تتغير في الخطوة 4).
هذه الخوارزمية سهلة الوصف والتصور. دعونا نلقي نظرة.


```python
# Let's begin by allocation 3 cluster's points
X = np.zeros((150, 2))

np.random.seed(seed=42)
X[:50, 0] = np.random.normal(loc=0.0, scale=0.3, size=50)
X[:50, 1] = np.random.normal(loc=0.0, scale=0.3, size=50)

X[50:100, 0] = np.random.normal(loc=2.0, scale=0.5, size=50)
X[50:100, 1] = np.random.normal(loc=-1.0, scale=0.2, size=50)

X[100:150, 0] = np.random.normal(loc=-1.0, scale=0.2, size=50)
X[100:150, 1] = np.random.normal(loc=2.0, scale=0.5, size=50)

plt.figure(figsize=(5, 5))
plt.plot(X[:, 0], X[:, 1], "bo");
```

```python
# Scipy has function that takes 2 tuples and return
# calculated distance between them
from scipy.spatial.distance import cdist

# Randomly allocate the 3 centroids
np.random.seed(seed=42)
centroids = np.random.normal(loc=0.0, scale=1.0, size=6)
centroids = centroids.reshape((3, 2))

cent_history = []
cent_history.append(centroids)

for i in range(3):
    # Calculating the distance from a point to a centroid
    distances = cdist(X, centroids)
    # Checking what's the closest centroid for the point
    labels = distances.argmin(axis=1)

    # Labeling the point according the point's distance
    centroids = centroids.copy()
    centroids[0, :] = np.mean(X[labels == 0, :], axis=0)
    centroids[1, :] = np.mean(X[labels == 1, :], axis=0)
    centroids[2, :] = np.mean(X[labels == 2, :], axis=0)

    cent_history.append(centroids)
```

```python
# Let's plot K-means
plt.figure(figsize=(8, 8))
for i in range(4):
    distances = cdist(X, cent_history[i])
    labels = distances.argmin(axis=1)

    plt.subplot(2, 2, i + 1)
    plt.plot(X[labels == 0, 0], X[labels == 0, 1], "bo", label="cluster #1")
    plt.plot(X[labels == 1, 0], X[labels == 1, 1], "co", label="cluster #2")
    plt.plot(X[labels == 2, 0], X[labels == 2, 1], "mo", label="cluster #3")
    plt.plot(cent_history[i][:, 0], cent_history[i][:, 1], "rX")
    plt.legend(loc=0)
    plt.title("Step {:}".format(i + 1));
```


هنا، استخدمنا المسافة الإقليدية، لكن الخوارزمية سوف تتقارب مع أي مقياس آخر. لا يمكنك تغيير عدد الخطوات أو معايير التقارب فحسب، بل يمكنك أيضًا تغيير قياس المسافة بين النقاط والنقط الوسطى العنقودية.
"ميزة" أخرى لهذه الخوارزمية هي حساسيتها للمواضع الأولية للنقط الوسطى العنقودية. يمكنك تشغيل الخوارزمية عدة مرات ثم حساب متوسط ​​جميع نتائج النقطه الوسطى.



## اختيار عدد المجموعات لوسائل Kوعلى النقيض من مهام التعلم الخاضعة للإشراف مثل التصنيف والانحدار، يتطلب التجميع المزيد من الجهد لاختيار معيار التحسين. عادةً، عند العمل باستخدام وسائل k، نقوم بتحسين مجموع المسافات المربعة بين الملاحظات والنقط الوسطى الخاصة بها.
$$ J(C) = \sum_{k=1}^K\sum_{i~\in~C_k} ||x_i - \mu_k|| \rightarrow \min\limits_C,$$
حيث $C$ - عبارة عن مجموعة من المجموعات ذات الطاقة $K$، $\mu_k$ عبارة عن مركز مركزي للكتلة $C_k$.
يبدو هذا التعريف معقولًا - فنحن نريد أن تكون ملاحظاتنا قريبة من النقط الوسطى قدر الإمكان. ولكن، هناك مشكلة - يتم الوصول إلى الحد الأمثل عندما يكون عدد النقط الوسطى مساويًا لعدد الملاحظات، لذلك سينتهي بك الأمر بكل ملاحظة على أنها مجموعة منفصلة خاصة بها.
لتجنب هذه الحالة، يجب علينا اختيار عدد من المجموعات التي تتناقص بعدها الدالة $J(C_k)$ بسرعة أقل. وبشكل أكثر رسمية،
$$ D(k) = \frac{|J(C_k) - J(C_{k+1})|}{|J(C_{k-1}) - J(C_k)|}  \rightarrow \min\limits_k $$
دعونا نلقي نظرة على مثال.


```python
from sklearn.cluster import KMeans
```

```python
inertia = []
for k in range(1, 8):
    kmeans = KMeans(n_clusters=k, random_state=1).fit(X)
    inertia.append(np.sqrt(kmeans.inertia_))
```

```python
plt.plot(range(1, 8), inertia, marker="s")
plt.xlabel("$k$")
plt.ylabel("$J(C_k)$");
```


نرى أن $J(C_k)$ يتناقص بشكل ملحوظ حتى يصل عدد المجموعات إلى 3 ثم لا يتغير كثيرًا بعد الآن. وهذا يعني أن العدد الأمثل للمجموعات هو 3.

## نشر التقارب
يعد نشر التقارب مثالاً آخر على خوارزمية التجميع. على عكس وسائل K، لا يتطلب هذا النهج منا تحديد عدد المجموعات مسبقًا. الفكرة الرئيسية هنا هي أننا نرغب في تجميع بياناتنا بناءً على تشابه الملاحظات (أو كيفية "توافقها" مع بعضها البعض).
دعونا نحدد مقياس التشابه بحيث $s(x_i, x_j) > s(x_i, x_k)$ إذا كانت الملاحظة $x_i$ أكثر تشابهًا مع الملاحظة $x_j$ وأقل تشابهًا مع الملاحظة $x_k$. مثال بسيط لمقياس التشابه هذا هو مربع المسافة السلبي $s(x_i, x_j) = - ||x_i - x_j||^{2}$.
الآن، دعونا نصف "المراسلات" من خلال إنشاء مصفوفتين صفريتين. أحدهم، $r_{i,k}$، يحدد مدى جودة الملاحظة $k$ باعتبارها "نموذجًا يحتذى به" للملاحظة $i$ فيما يتعلق بجميع "نماذج الأدوار" المحتملة الأخرى. تحدد مصفوفة أخرى، $a_{i,k}$ مدى ملاءمة $i$ للملاحظة رقم $k$ باعتبارها "نموذجًا يحتذى به". قد يبدو هذا مربكًا، لكنه يصبح أكثر قابلية للفهم مع بعض التدريب العملي.
ويتم تحديث المصفوفات تسلسلياً بالقواعد التالية:
$$r_{i,k} \leftarrow s_(x_i, x_k) - \max_{k' \neq k} \left\{ a_{i,k'} + s(x_i, x_k') \right\}$$
$$a_{i,k} \leftarrow \min \left( 0, r_{k,k} + \sum_{i' \not\in \{i,k\}} \max(0, r_{i',k}) \right), \ \ \  i \neq k$$
$$a_{k,k} \leftarrow \sum_{i' \neq k} \max(0, r_{i',k})$$



## التجميع الطيفي
يجمع التجميع الطيفي بين بعض الأساليب الموضحة أعلاه لإنشاء طريقة تجميع أقوى.بداية، تتطلب منا هذه الخوارزمية تحديد مصفوفة التشابه للملاحظات والتي تسمى مصفوفة المجاورة. يمكن القيام بذلك بطريقة مشابهة كما في خوارزمية نشر التقارب: $A_{i, j} = - ||x_i - x_j||^{2}$. تصف هذه المصفوفة رسمًا بيانيًا كاملاً مع الملاحظات كرؤوس وقيمة التشابه المقدرة بين زوج من الملاحظات كأوزان حافة لهذا الزوج من القمم. بالنسبة للمقياس المحدد أعلاه والملاحظات ثنائية الأبعاد، يعد هذا أمرًا بديهيًا جدًا - فالملاحظتان متشابهتان إذا كانت الحافة بينهما أقصر. 
نود تقسيم الرسم البياني إلى رسمين بيانيين فرعيين بحيث تكون كل ملاحظة في كل رسم بياني فرعي مشابهة لملاحظة أخرى في هذا الرسم البياني الفرعي. رسميًا، هذه مشكلة تخفيضات طبيعية؛ لمزيد من التفاصيل، نوصي بقراءة [هذه المقالة](http://people.eecs.berkeley.edu/~malik/papers/SM-ncut.pdf).



## التجمعات التكتلية



الخوارزمية التالية هي الأبسط والأسهل للفهم بين جميع خوارزميات التجميع دون عدد محدد من المجموعات.
الخوارزمية بسيطة إلى حد ما:
1. نبدأ بتخصيص كل ملاحظة لمجموعتها الخاصة
2. ثم قم بفرز المسافات الزوجية بين مراكز التجمعات ترتيباً تنازلياً
3. خذ أقرب مجموعتين متجاورتين وادمجهما معًا، وأعد حساب المراكز
4. كرر الخطوتين 2 و3 حتى يتم دمج كافة البيانات في مجموعة واحدة
يمكن إجراء عملية البحث عن أقرب مجموعة باستخدام طرق مختلفة لحصر الملاحظات:
1. رابط واحد 
$d(C_i, C_j) = min_{x_i \in C_i, x_j \in C_j} ||x_i - x_j||$
2. الربط الكامل 
$d(C_i, C_j) = max_{x_i \in C_i, x_j \in C_j} ||x_i - x_j||$
3. الارتباط المتوسط 
$d(C_i, C_j) = \frac{1}{n_i n_j} \sum_{x_i \in C_i} \sum_{x_j \in C_j} ||x_i - x_j||$
4. ربط النقطه الوسطى 
$d(C_i, C_j) = ||\mu_i - \mu_j||$
أما الخيار الثالث فهو الأكثر فعالية في حساب الوقت لأنه لا يتطلب إعادة حساب المسافات في كل مرة يتم فيها دمج المجموعات.يمكن تصور النتائج على شكل شجرة عنقودية جميلة (dendogram) للمساعدة في التعرف على اللحظة التي يجب أن تتوقف فيها الخوارزمية للحصول على أفضل النتائج. هناك الكثير من أدوات بايثون لبناء هذه المخططات العمرية للتجميع التكتلي.
دعونا نفكر في مثال للمجموعات التي حصلنا عليها من وسائل K:


```python
from scipy.cluster import hierarchy
from scipy.spatial.distance import pdist

X = np.zeros((150, 2))

np.random.seed(seed=42)
X[:50, 0] = np.random.normal(loc=0.0, scale=0.3, size=50)
X[:50, 1] = np.random.normal(loc=0.0, scale=0.3, size=50)

X[50:100, 0] = np.random.normal(loc=2.0, scale=0.5, size=50)
X[50:100, 1] = np.random.normal(loc=-1.0, scale=0.2, size=50)

X[100:150, 0] = np.random.normal(loc=-1.0, scale=0.2, size=50)
X[100:150, 1] = np.random.normal(loc=2.0, scale=0.5, size=50)

# pdist will calculate the upper triangle of the pairwise distance matrix
distance_mat = pdist(X)
# linkage — is an implementation of agglomerative algorithm
Z = hierarchy.linkage(distance_mat, "single")
plt.figure(figsize=(10, 5))
dn = hierarchy.dendrogram(Z, color_threshold=0.5)
```


## مقاييس الدقة
وعلى عكس التصنيف، من الصعب تقييم جودة النتائج الناتجة عن التجميع. هنا، لا يمكن أن يعتمد المقياس على التصنيفات، بل على جودة التقسيم فقط. ثانيًا، ليس لدينا عادة تسميات حقيقية للملاحظات عندما نستخدم التجميع.
هناك مقاييس جودة *داخلية* و *خارجية*. تستخدم المقاييس الخارجية المعلومات حول الانقسام الحقيقي المعروف بينما لا تستخدم المقاييس الداخلية أي معلومات خارجية وتقوم بتقييم جودة المجموعات بناءً على البيانات الأولية فقط. عادةً ما يتم تحديد العدد الأمثل للمجموعات فيما يتعلق ببعض المقاييس الداخلية. 
يتم تنفيذ كافة المقاييس الموضحة أدناه في `sklearn.metrics`.
**مؤشر الراند المعدل (ARI)**هنا، نفترض أن التسميات الحقيقية للأشياء معروفة. لا يعتمد هذا المقياس على قيم التسميات ولكن على تقسيم مجموعة البيانات. دع $N$ هو عدد الملاحظات في العينة. افترض أن $a$ هو عدد أزواج المراقبة التي تحمل نفس التسميات وتقع في نفس المجموعة، ودع $b$ هو عدد أزواج المراقبة ذات تسميات مختلفة وتقع في مجموعات مختلفة. يمكن حساب مؤشر Rand باستخدام الصيغة التالية: $$\text{RI} = \frac{2(a + b)}{n(n-1)}.$$ 
وبعبارة أخرى، فإنه يقوم بتقييم حصة أزواج المراقبة التي تكون فيها هذه الانقسامات (النتيجة الأولية والمجموعية) متسقة. يقوم مؤشر Rand (RI) بتقييم التشابه بين الانقسامين لنفس العينة. لكي يكون هذا الفهرس قريبًا من الصفر لأي نتائج تجميعية تحتوي على $n$ وعدد المجموعات، من الضروري توسيع نطاقه، ومن ثم مؤشر Rand المعدل: $$\text{ARI} = \frac{\text{RI} - E[\text{RI}]}{\max(\text{RI}) - E[\text{RI}]}.$$
هذا المقياس متماثل ولا يعتمد على تبديل التسمية. ولذلك، فإن هذا المؤشر هو مقياس للمسافات بين تقسيمات العينة المختلفة. $\text{ARI}$ يأخذ القيم الموجودة في النطاق $[-1, 1]$. تشير القيم السالبة إلى استقلال الانقسامات، وتشير القيم الموجبة إلى أن هذه الانقسامات متسقة (وهي تطابق $\text{ARI} = 1$).
**المعلومات المتبادلة المعدلة (AMI)**يشبه هذا المقياس $\text{ARI}$. كما أنه متماثل ولا يعتمد على قيم التسميات وتباديلها. يتم تعريفه بواسطة الدالة [الإنتروبيا](https://en.wikipedia.org/wiki/Entropy_(information_theory) ويفسر تقسيم العينة على أنه توزيع منفصل (احتمال التعيين إلى مجموعة يساوي النسبة المئوية للكائنات الموجودة فيه). يتم تعريف الفهرس $MI$ على أنه [المعلومات المتبادلة](https://en.wikipedia.org/wiki/Mutual_information) لتوزيعتين، بما يتوافق مع تقسيم العينة إلى مجموعات. بشكل بديهي، تقيس المعلومات المتبادلة حصة المعلومات المشتركة لكلا الانقسامين العنقوديين، أي كيف تقلل المعلومات المتعلقة بأحدهما من عدم اليقين لدى الآخر.
على غرار $\text{ARI}$، تم تعريف $\text{AMI}$. يتيح لنا ذلك التخلص من زيادة مؤشر $MI$ مع عدد المجموعات. يقع $\text{AMI}$ في النطاق $[0, 1]$. تعني القيم القريبة من الصفر أن الانقسامات مستقلة، وتلك القريبة من 1 تعني أنها متشابهة (مع تطابق كامل في $\text{AMI} = 1$).
**التجانس، الاكتمال، قياس V**رسميًا، يتم تعريف هذه المقاييس أيضًا بناءً على دالة الإنتروبيا ودالة الإنتروبيا الشرطية، وتفسير انقسامات العينة على أنها توزيعات منفصلة: $$h = 1 - \frac{H(C\mid K)}{H(C)}, c = 1 - \frac{H(K\mid C)}{H(K)},$$
حيث $K$ هو نتيجة تجميع و$C$ هو التقسيم الأولي. ولذلك، $h$ بتقييم ما إذا كانت كل مجموعة تتكون من نفس كائنات الفئة، و$c$ يقيس مدى ملاءمة نفس كائنات الفئة للمجموعات. هذه المقاييس ليست متماثلة. يقع كلاهما في النطاق $[0, 1]$، وتشير القيم الأقرب إلى 1 إلى نتائج تجميع أكثر دقة. لا يتم قياس قيم هذه المقاييس مثل المقاييس $\text{ARI}$ أو $\text{AMI}$ وبالتالي تعتمد على عدد المجموعات. لن تحتوي نتيجة التجميع العشوائي على قيم مقاييس أقرب إلى الصفر عندما يكون عدد المجموعات كبيرًا بما يكفي وعدد الكائنات صغيرًا. في مثل هذه الحالة، سيكون من المعقول أكثر استخدام $\text{ARI}$. ومع ذلك، مع وجود عدد كبير من الملاحظات (أكثر من 100) وعدد المجموعات أقل من 10، تكون هذه المشكلة أقل أهمية ويمكن تجاهلها.
$V$- القياس عبارة عن مزيج من $h$ و$c$ وهو الوسط التوافقي لهما:
$$v = 2\frac{hc}{h+c}.$$
إنه متماثل ويقيس مدى اتساق نتائج التجميع.
**خيال**وعلى النقيض من المقاييس الموضحة أعلاه، فإن هذا المعامل لا يعني معرفة التسميات الحقيقية للأشياء. يتيح لنا تقدير جودة التجميع باستخدام العينة الأولية غير المسماة ونتائج التجميع فقط. في البداية، يتم حساب معامل الصورة الظلية لكل ملاحظة. اجعل $a$ هو متوسط ​​المسافة بين كائن وكائنات أخرى داخل مجموعة واحدة و$b$ هو متوسط ​​المسافة من كائن إلى كائنات من أقرب مجموعة (يختلف عن المجموعة التي ينتمي إليها الكائن). ثم قياس الصورة الظلية لهذا الكائن هو $$s = \frac{b - a}{\max(a, b)}.$$
الصورة الظلية للعينة هي القيمة المتوسطة لقيم الصورة الظلية من هذه العينة. ولذلك، فإن مسافة الصورة الظلية توضح إلى أي مدى تختلف المسافة بين الكائنات من نفس الفئة عن متوسط ​​المسافة بين الكائنات من مجموعات مختلفة. يأخذ هذا المعامل القيم الموجودة في النطاق $[-1, 1]$. تتوافق القيم القريبة من -1 مع نتائج التجميع السيئة بينما تتوافق القيم الأقرب إلى 1 مع مجموعات كثيفة ومحددة جيدًا. لذلك، كلما ارتفعت قيمة الصورة الظلية، كانت نتائج التجميع أفضل.
بمساعدة الصورة الظلية، يمكننا تحديد العدد الأمثل للمجموعات $k$ (إذا لم نكن نعرف ذلك بالفعل من البيانات) عن طريق أخذ عدد المجموعات التي تزيد من معامل الصورة الظلية.



في الختام، دعونا نلقي نظرة على كيفية أداء هذه المقاييس مع مجموعة بيانات الأرقام المكتوبة بخط اليد MNIST:


```python
import pandas as pd
from sklearn import datasets, metrics
from sklearn.cluster import (AffinityPropagation, AgglomerativeClustering,
                             KMeans, SpectralClustering)

data = datasets.load_digits()
X, y = data.data, data.target

algorithms = []
algorithms.append(KMeans(n_clusters=10, random_state=1))
algorithms.append(AffinityPropagation())
algorithms.append(
    SpectralClustering(n_clusters=10, random_state=1, affinity="nearest_neighbors")
)
algorithms.append(AgglomerativeClustering(n_clusters=10))

data = []
for algo in algorithms:
    algo.fit(X)
    data.append(
        (
            {
                "ARI": metrics.adjusted_rand_score(y, algo.labels_),
                "AMI": metrics.adjusted_mutual_info_score(y, algo.labels_),
                "Homogenity": metrics.homogeneity_score(y, algo.labels_),
                "Completeness": metrics.completeness_score(y, algo.labels_),
                "V-measure": metrics.v_measure_score(y, algo.labels_),
                "Silhouette": metrics.silhouette_score(X, algo.labels_),
            }
        )
    )

results = pd.DataFrame(
    data=data,
    columns=["ARI", "AMI", "Homogenity", "Completeness", "V-measure", "Silhouette"],
    index=["K-means", "Affinity", "Spectral", "Agglomerative"],
)

results
```
