## 1. مقدمة



قبل أن نتعمق في مادة مقال هذا الأسبوع، دعونا نتحدث عن نوع المشكلة التي سنقوم بحلها ومكانها في مجال التعلم الآلي المثير. يقدم كتاب تي ميتشل "التعلم الآلي" (1997) تعريفًا كلاسيكيًا وعامًا للتعلم الآلي على النحو التالي: 
> يقال إن برنامج الكمبيوتر يتعلم من الخبرة E فيما يتعلق ببعض فئات المهام T ومقياس الأداء P، إذا كان أدائه في المهام في T، كما تم قياسه بـ P، يتحسن مع الخبرة E.في إعدادات المشكلة المختلفة، يمكن أن تشير T وP وE إلى أشياء مختلفة تمامًا. فيما يلي بعض المهام الأكثر شيوعًا في التعلم الآلي:
- تصنيف المثيل إلى إحدى الفئات بناءً على ميزاته؛
- الانحدار - التنبؤ بميزة الهدف العددي بناءً على ميزات أخرى للمثيل؛
- التجميع - تحديد أقسام المثيلات بناءً على ميزات هذه المثيلات بحيث يكون الأعضاء داخل المجموعات أكثر تشابهًا مع بعضهم البعض من أولئك الموجودين في المجموعات الأخرى؛
- اكتشاف الحالات الشاذة - البحث عن الحالات "المختلفة إلى حد كبير" عن بقية العينة أو لمجموعة معينة من الحالات؛
- وغيرها الكثير.
يتم توفير نظرة عامة جيدة في فصل "أساسيات التعلم الآلي" من ["التعلم العميق"](http://www.deeplearningbook.org) (بقلم إيان جودفيلو، يوشوا بينجيو، آرون كورفيل، 2016).
تشير التجربة E إلى البيانات (لا يمكننا الذهاب إلى أي مكان بدونها). يمكن تقسيم خوارزميات التعلم الآلي إلى تلك التي يتم تدريبها بطريقة خاضعة للإشراف أو غير خاضعة للإشراف. في مهام التعلم غير الخاضعة للرقابة، يكون لدى المرء مجموعة تتكون من الحالات الموصوفة بواسطة مجموعة من الميزات. في مشاكل التعلم الخاضع للإشراف، يوجد أيضًا متغير مستهدف، وهو ما نود أن نكون قادرين على التنبؤ به، والمعروف لكل حالة في مجموعة التدريب. 


#### مثال
ويشرف على التصنيف والانحدار مشاكل التعلم. على سبيل المثال، كمؤسسة ائتمانية، قد نرغب في التنبؤ بالتخلف عن سداد القروض بناءً على البيانات المتراكمة حول عملائنا. هنا، التجربة E هي بيانات التدريب المتاحة: مجموعة من الحالات (العملاء)، ومجموعة من الميزات (مثل العمر، والراتب، ونوع القرض، والتخلف عن سداد القروض السابقة، وما إلى ذلك) لكل منها، ومتغير مستهدف (سواء تخلفوا عن سداد القرض). هذا المتغير المستهدف هو مجرد حقيقة تتعلق بالتخلف عن سداد القرض (1 أو 0)، لذا تذكر أن هذه مشكلة تصنيف (ثنائية). إذا كنت تتنبأ بدلاً من ذلك بمقدار الوقت الذي يتأخر فيه سداد القرض، فسيصبح هذا مشكلة انحدار.
 
أخيرًا، المصطلح الثالث المستخدم في تعريف التعلم الآلي هو مقياس تقييم أداء الخوارزمية P. وتختلف هذه المقاييس باختلاف المشكلات والخوارزميات، وسنناقشها أثناء دراسة الخوارزميات الجديدة. في الوقت الحالي، سنشير إلى مقياس بسيط لخوارزميات التصنيف، وهو نسبة الإجابات الصحيحة - الدقة - في مجموعة الاختبار.
 
دعونا نلقي نظرة على مشكلتين في التعلم تحت الإشراف: التصنيف والانحدار.



## 2. شجرة القرار 
نبدأ نظرة عامة على أساليب التصنيف والانحدار بإحدى أكثر الطرق شيوعًا - شجرة القرار. تُستخدم أشجار القرار في قرارات الحياة اليومية، وليس فقط في التعلم الآلي. مخططات التدفق هي في الواقع تمثيلات مرئية لأشجار القرار. على سبيل المثال، تنشر المدرسة العليا للاقتصاد مخططات معلوماتية لتسهيل حياة موظفيها. فيما يلي مقتطف من التعليمات الخاصة بنشر ورقة بحثية على بوابة المؤسسة.

### كيفية بناء شجرة القرار
لقد رأينا سابقًا أن قرار منح القرض يتم اتخاذه بناءً على العمر والأصول والدخل ومتغيرات أخرى. ولكن ما هو المتغير الذي يجب النظر إليه أولاً؟ دعونا نناقش مثالاً بسيطًا حيث تكون جميع المتغيرات ثنائية.
 
تذكر لعبة "20 سؤالًا"، والتي غالبًا ما يتم الرجوع إليها عند تقديم أشجار القرار. ربما تكون قد لعبت هذه اللعبة - يفكر أحد الأشخاص في أحد المشاهير بينما يحاول الآخر التخمين من خلال طرح أسئلة "نعم" أو "لا" فقط. ما هو السؤال الذي سيطرحه الحازر أولاً؟ بالطبع، سوف يسألون الشخص الذي يقوم بتضييق عدد الخيارات المتبقية أكثر من غيره. السؤال "هل هي أنجلينا جولي؟" من شأنه، في حالة الرد السلبي، ترك جميع المشاهير باستثناء واحد في عالم الاحتمال. في المقابل، يسأل "هل المشاهير امرأة؟" من شأنه أن يقلل الاحتمالات إلى النصف تقريبًا. وهذا يعني أن ميزة "الجنس" تفصل بين مجموعة بيانات المشاهير بشكل أفضل بكثير من الميزات الأخرى مثل "أنجلينا جولي" أو "الإسبانية" أو "تحب كرة القدم". يتوافق هذا المنطق مع مفهوم اكتساب المعلومات على أساس الإنتروبيا.


#### الانتروبيا
يتم تعريف إنتروبيا شانون لنظام يحتوي على N حالات محتملة على النحو التالي:
$$\Large S = -\sum_{i=1}^{N}p_i \log_2{p_i},$$



حيث $p_i$ هو احتمال العثور على النظام في الحالة $i$-. وهذا مفهوم مهم جدًا يستخدم في الفيزياء ونظرية المعلومات ومجالات أخرى. يمكن وصف الإنتروبيا بأنها درجة الفوضى في النظام. كلما زادت الإنتروبيا، قل ترتيب النظام، والعكس صحيح. سيساعدنا هذا في إضفاء الطابع الرسمي على "التقسيم الفعال للبيانات"، وهو ما أشرنا إليه في سياق "20 سؤالاً".



#### مثال لعبة
لتوضيح كيف يمكن للإنتروبيا أن تساعدنا في تحديد الميزات الجيدة لبناء شجرة القرار، دعونا نلقي نظرة على مثال لعبة. سنتوقع لون الكرة بناءً على موقعها.
<img align='center' src='../../img/decision_tree3.png'><br>



هناك 9 كرات زرقاء و11 كرة صفراء. إذا سحبنا كرة بشكل عشوائي، فستكون زرقاء مع احتمال $p_1=\frac{9}{20}$ وصفراء مع احتمال $p_2=\frac{11}{20}$، مما يعطينا إنتروبيا $S_0 = -\frac{9}{20}\log_2{\frac{9}{20}}-\frac{11}{20}\log_2{\frac{11}{20}} \approx 1$. هذه القيمة في حد ذاتها قد لا تخبرنا بالكثير، ولكن دعونا نرى كيف تتغير القيمة إذا قمنا بتقسيم الكرات إلى مجموعتين: مع موضع أقل من أو يساوي 12 وأكبر من 12.



<img align='center' src='../../img/topic3_entropy_balls2.png'><br>


المجموعة اليسرى بها 13 كرة، 8 زرقاء و5 صفراء. إنتروبيا هذه المجموعة هي $S_1 = -\frac{5}{13}\log_2{\frac{5}{13}}-\frac{8}{13}\log_2{\frac{8}{13}} \approx 0.96$. المجموعة اليمنى بها 7 كرات، 1 زرقاء و 6 صفراء. إنتروبيا المجموعة الصحيحة هي $S_2 = -\frac{1}{7}\log_2{\frac{1}{7}}-\frac{6}{7}\log_2{\frac{6}{7}} \approx 0.6$. كما ترون، انخفضت الإنتروبيا في كلا المجموعتين، وبشكل أكبر في المجموعة الصحيحة. وبما أن الإنتروبيا هي في الواقع درجة الفوضى (أو عدم اليقين) في النظام، فإن انخفاض الإنتروبيا يسمى اكتساب المعلومات. رسميًا، يتم تعريف كسب المعلومات (IG) للتقسيم بناءً على المتغير $Q$ (في هذا المثال هو المتغير "$x \leq 12$") على أنه



$$\Large IG(Q) = S_O - \sum_{i=1}^{q}\frac{N_i}{N}S_i,$$



حيث $q$ هو عدد المجموعات بعد التقسيم، و$N_i$ هو عدد الكائنات من العينة التي يكون فيها المتغير $Q$ مساوياً للقيمة $i$-th. في مثالنا، أدى تقسيمنا إلى مجموعتين ($q = 2$)، واحدة بها 13 عنصرًا ($N_1 = 13$)، والأخرى بها 7 ($N_2 = 7$). لذلك، يمكننا حساب كسب المعلومات كما



$$ \Large IG(x \leq 12) = S_0 - \frac{13}{20}S_1 - \frac{7}{20}S_2 \approx 0.16.$$



اتضح أن تقسيم الكرات إلى مجموعتين عن طريق التقسيم على "الإحداثيات أقل من أو يساوي 12" أعطانا نظامًا أكثر ترتيبًا. دعونا نستمر في تقسيمهم إلى مجموعات حتى تصبح الكرات في كل مجموعة بنفس اللون.

بالنسبة للمجموعة الصحيحة، يمكننا أن نرى بسهولة أننا نحتاج فقط إلى قسم إضافي واحد باستخدام "الإحداثيات أقل من أو يساوي 18". لكن بالنسبة للمجموعة اليسرى، نحتاج إلى ثلاثة آخرين. لاحظ أن إنتروبيا المجموعة التي تكون جميع الكرات فيها بنفس اللون تساوي 0 ($\log_2{1} = 0$).لقد نجحنا في إنشاء شجرة قرارات تتنبأ بلون الكرة بناءً على موضعها. قد لا تعمل شجرة القرار هذه بشكل جيد إذا أضفنا أي كرات لأنها تناسب تمامًا مجموعة التدريب (20 كرة أولية). إذا أردنا أن نحقق نتائج جيدة في هذه الحالة، فإن الشجرة التي تحتوي على عدد أقل من "الأسئلة" أو الانقسامات ستكون أكثر دقة، حتى لو لم تكن مناسبة تمامًا لمجموعة التدريب. سنناقش مشكلة التجهيز الزائد لاحقًا.

### معايير الجودة الأخرى للانقسامات في مشاكل التصنيف



ناقشنا كيف يسمح لنا الإنتروبيا بإضفاء الطابع الرسمي على الأقسام في الشجرة. ولكن هذا مجرد إرشادي واحد. هناك آخرون:- عدم يقين جيني (نجاسة جيني): $G = 1 - \sum\limits_k (p_k)^2$. يمكن تفسير تعظيم هذا المعيار على أنه تعظيم عدد أزواج الكائنات من نفس الفئة الموجودة في نفس الشجرة الفرعية (يجب عدم الخلط بينه وبين مؤشر جيني).
- خطأ في التصنيف الخاطئ: $E = 1 - \max\limits_k p_k$
من الناحية العملية، لا يتم استخدام خطأ التصنيف الخاطئ أبدًا تقريبًا، ويعمل عدم اليقين واكتساب المعلومات في جيني بشكل مماثل.
 
بالنسبة للتصنيف الثنائي، يتخذ الإنتروبيا وعدم اليقين في جيني الشكل التالي:
$ S = -p_+ \log_2{p_+} -p_- \log_2{p_-} = -p_+ \log_2{p_+} -(1 - p_{+}) \log_2{(1 - p_{+})};$
$ G = 1 - p_+^2 - p_-^2 = 1 - p_+^2 - (1 - p_+)^2 = 2p_+(1-p_+).$
حيث ($p_+$ هو احتمال وجود كائن له تسمية +).
إذا قمنا برسم هاتين الوظيفتين مقابل الوسيطة $p_+$، فسنرى أن مخطط الإنتروبيا قريب جدًا من مخطط عدم يقين جيني، وقد تضاعف. لذلك، في الممارسة العملية، هذين المعيارين متطابقان تقريبا.

### معلمات الشجرة الحاسمةمن الناحية الفنية، يمكنك إنشاء شجرة قرار حتى تحتوي كل ورقة على مثيل واحد بالضبط، ولكن هذا ليس شائعًا في الممارسة العملية عند بناء شجرة واحدة لأنه سيتم *تجهيزها بشكل زائد*، أو ضبطها بشكل كبير على مجموعة التدريب، ولن تتنبأ بتسميات البيانات الجديدة بشكل جيد. في الجزء السفلي من الشجرة، على عمق كبير، ستكون هناك أقسام على ميزات أقل أهمية (على سبيل المثال، ما إذا كان العميل قد جاء من ليدز أو نيويورك). يمكننا أن نبالغ في هذه القصة أكثر ونجد أن جميع العملاء الأربعة الذين جاءوا إلى البنك للحصول على قرض وهم يرتدون سراويل خضراء لم يعيدوا القرض. وحتى لو كان ذلك صحيحًا في التدريب، فإننا لا نريد أن يقوم نموذج التصنيف الخاص بنا بتوليد مثل هذه القواعد المحددة.
 
هناك استثناءان حيث يتم بناء الأشجار إلى أقصى عمق:
- الغابة العشوائية (مجموعة من الأشجار) تقوم بحساب متوسط الاستجابات من الأشجار الفردية التي تم بناؤها إلى أقصى عمق (سنتحدث لاحقًا عن سبب وجوب القيام بذلك)
- *تقليم* الأشجار. في هذا النهج، يتم إنشاء الشجرة أولاً إلى أقصى عمق. بعد ذلك، من الأسفل إلى الأعلى، تتم إزالة بعض عقد الشجرة من خلال مقارنة جودة الشجرة مع هذا القسم وبدونه (يتم إجراء المقارنة باستخدام *التحقق المتقاطع*، المزيد حول هذا أدناه).
الصورة أدناه هي مثال على حدود فاصلة مبنية في شجرة مفرطة التجهيز.

### Class DecisionTreeClassifier في Scikit-learn

- `min_samples_leaf` – الحد الأدنى لعدد العينات في الورقة. تمنع هذه المعلمة إنشاء أشجار حيث تحتوي أي ورقة على عدد قليل من الأعضاء فقط.
يجب تعيين معلمات الشجرة اعتمادًا على البيانات المدخلة، وعادةً ما يتم ذلك عن طريق *التحقق المتقاطع*، المزيد عن هذا أدناه.



### شجرة القرار في مشكلة الانحدار
عند التنبؤ بمتغير رقمي، تظل فكرة بناء الشجرة كما هي، ولكن تتغير معايير الجودة: 
- التباين: 
$$\Large D = \frac{1}{\ell} \sum\limits_{i =1}^{\ell} (y_i - \frac{1}{\ell} \sum\limits_{j=1}^{\ell} y_j)^2, $$
حيث $\ell$ هو عدد العينات في الورقة، و$y_i$ هي قيمة المتغير الهدف. ببساطة، من خلال تقليل التباين، نبحث عن الميزات التي تقسم مجموعة التدريب بطريقة تجعل قيم الميزة المستهدفة في كل ورقة متساوية تقريبًا.



#### مثال
لنقم بإنشاء بعض البيانات الموزعة بواسطة الدالة $f(x) = e^{-x ^ 2} + 1.5 * e^{-(x - 2) ^ 2}$ مع بعض الضوضاء. ثم سنقوم بتدريب الشجرة بهذه البيانات والتنبؤات التي تقوم بها الشجرة.


````python
n_train = 150
n_test = 1000
noise = 0.1


def f(x):
    x = x.ravel()
    return np.exp(-(x ** 2)) + 1.5 * np.exp(-((x - 2) ** 2))


def generate(n_samples, noise):
    X = np.random.rand(n_samples) * 10 - 5
    X = np.sort(X).ravel()
    y = (
        np.exp(-(X ** 2))
        + 1.5 * np.exp(-((X - 2) ** 2))
        + np.random.normal(0.0, noise, n_samples)
    )
    X = X.reshape((n_samples, 1))
    return X, y


X_train, y_train = generate(n_samples=n_train, noise=noise)
X_test, y_test = generate(n_samples=n_test, noise=noise)

from sklearn.tree import DecisionTreeRegressor

reg_tree = DecisionTreeRegressor(max_depth=5, random_state=17)

reg_tree.fit(X_train, y_train)
reg_tree_pred = reg_tree.predict(X_test)

plt.figure(figsize=(10, 6))
plt.plot(X_test, f(X_test), "b")
plt.scatter(X_train, y_train, c="b", s=20)
plt.plot(X_test, reg_tree_pred, "g", lw=2)
plt.xlim([-5, 5])
plt.title(
    "Decision tree regressor, MSE = %.2f"
    % (np.sum((y_test - reg_tree_pred) ** 2) / n_test)
)
plt.show()
````


نرى أن شجرة القرار تقارب البيانات باستخدام دالة ثابتة متعددة التعريف.



## 3. طريقة أقرب الجيران*طريقة أقرب الجيران* (k-Nearest Neighbors، أو k-NN) هي طريقة تصنيف أخرى شائعة جدًا تُستخدم أحيانًا أيضًا في مشاكل الانحدار. يعد هذا، مثل أشجار القرار، أحد أكثر طرق التصنيف قابلية للفهم. الحدس الأساسي هو أنك تبدو مثل جيرانك. وبشكل أكثر رسمية، تتبع الطريقة فرضية الاكتناز: إذا تم قياس المسافة بين الأمثلة بشكل جيد بما فيه الكفاية، فمن المرجح أن تنتمي الأمثلة المماثلة إلى نفس الفئة.



وفقًا لطريقة أقرب جيران، سيتم تصنيف الكرة الخضراء على أنها "زرقاء" بدلاً من "حمراء".

على سبيل المثال، إذا كنت لا تعرف كيفية وضع علامة على سماعة رأس Bluetooth في قائمة عبر الإنترنت، فيمكنك العثور على 5 سماعات رأس مماثلة، وإذا تم وضع علامة على 4 منها على أنها "ملحقات" وواحدة فقط على أنها "تقنية"، فسوف تقوم أيضًا بتسميتها ضمن "الملحقات".
 
لتصنيف كل عينة من مجموعة الاختبار، يحتاج المرء إلى إجراء العمليات التالية بالترتيب:
1. احسب المسافة إلى كل عينة في مجموعة التدريب.
2. حدد $k$ عينات من مجموعة التدريب مع أقل مسافة ممكنة عنها.
3. ستكون فئة عينة الاختبار هي الفئة الأكثر تكرارًا بين أقرب الجيران $k$.تتكيف الطريقة بسهولة تامة مع مشكلة الانحدار: في الخطوة 3، لا تُرجع الفئة، بل الرقم - متوسط ​​(أو متوسط) المتغير المستهدف بين الجيران.
 
ومن السمات البارزة لهذا النهج كسله - فالحسابات تتم فقط خلال مرحلة التنبؤ، عندما يلزم تصنيف عينة الاختبار. لم يتم بناء أي نموذج من أمثلة التدريب السابقة. في المقابل، تذكر أنه بالنسبة لأشجار القرار في النصف الأول من هذه المقالة، تم إنشاء الشجرة بناءً على مجموعة التدريب، ويتم تصنيف حالات الاختبار بسرعة نسبية من خلال المرور عبر الشجرة.
 
أقرب الجيران هو نهج مدروس جيدا. توجد العديد من النظريات المهمة التي تدعي أن هذه هي الطريقة المثلى للتصنيف في مجموعات البيانات "التي لا نهاية لها". يعتبر مؤلفو الكتاب الكلاسيكي "عناصر التعلم الإحصائي" أن k-NN هي خوارزمية مثالية من الناحية النظرية والتي يقتصر استخدامها فقط على قوة الحساب و[لعنة الأبعاد](https://en.wikipedia.org/wiki/Curse_of_dimensionality). 



### أقرب طريقة للجيران في التطبيقات الحقيقية
- يمكن أن تكون k-NN بمثابة نقطة بداية جيدة (خط الأساس) في بعض الحالات؛
- في مسابقات Kaggle، غالبًا ما يتم استخدام k-NN لبناء الميزات الوصفية (أي تنبؤات k-NN كمدخلات لنماذج أخرى) أو للتكديس/المزج؛
- تمتد طريقة أقرب الجيران إلى مهام أخرى مثل أنظمة التوصية. يمكن أن يكون القرار الأولي عبارة عن توصية لمنتج (أو خدمة) تحظى بشعبية بين *أقرب الجيران* للشخص الذي نريد تقديم توصية له؛
- من الناحية العملية، في مجموعات البيانات الكبيرة، غالبًا ما تُستخدم طرق البحث التقريبية لأقرب الجيران. هناك عدد من المكتبات مفتوحة المصدر التي تنفذ مثل هذه الخوارزميات؛ تحقق من مكتبة Spotify [مزعج](https://github.com/spotify/annoy).


تعتمد جودة التصنيف/الانحدار باستخدام k-NN على عدة معلمات:
- عدد الجيران $k$.
- قياس المسافة بين العينات (الشائعة تشمل مسافات هامينغ، الإقليدية، جيب التمام، ومينكوفسكي). لاحظ أن معظم هذه المقاييس تتطلب قياس البيانات. وببساطة، لا نريد لخاصية "الراتب" التي هي في حدود الآلاف أن تؤثر على المسافة أكثر من "العمر" الذي يقل عموما عن 100. 
- أوزان الجيران (قد يساهم كل جار بأوزان مختلفة، على سبيل المثال، كلما زادت العينة، انخفض الوزن).



### فئة `KNeighborsClassifier` في Scikit-learn
المعلمات الرئيسية للفئة `sklearn.neighbors.KNeighborsClassifier` هي:
- الأوزان: `uniform` (جميع الأوزان متساوية)، `distance` (الوزن يتناسب عكسيا مع المسافة من عينة الاختبار)، أو أي وظيفة أخرى يحددها المستخدم؛
- الخوارزمية (اختياري): `brute`، `ball_tree `، `KD_tree`، أو `auto`. في الحالة الأولى، يتم حساب أقرب الجيران لكل حالة اختبار من خلال البحث الشبكي على مجموعة التدريب. في الحالتين الثانية والثالثة، يتم تخزين المسافات بين الأمثلة في شجرة لتسريع العثور على أقرب الجيران. إذا قمت بتعيين هذه المعلمة على `auto`، فسيتم اختيار الطريقة الصحيحة للعثور على الجيران تلقائيًا بناءً على مجموعة التدريب.
- leaf_size (اختياري): الحد الأدنى للتبديل إلى بحث الشبكة إذا كانت خوارزمية البحث عن الجيران هي BallTree أو KDTree؛
- المقياس: `minkowski`، `manhattan `، `euclidean`، `chebyshev`، أو غير ذلك.



## 4. اختيار معلمات النموذج والتحقق من الصحةتتمثل المهمة الرئيسية لتعلم الخوارزميات في القدرة على *التعميم* على البيانات غير المرئية. نظرًا لأننا لا نستطيع التحقق فورًا من أداء النموذج على البيانات الجديدة الواردة (لأننا لا نعرف القيم الحقيقية للمتغير المستهدف بعد)، فمن الضروري التضحية بجزء صغير من البيانات للتحقق من جودة النموذج عليها.



ويتم ذلك غالبًا بإحدى طريقتين:
- تخصيص جزء من مجموعة البيانات (*مجموعة التحفظ/التحفظ*). وبالتالي، فإننا نحتفظ بجزء صغير من مجموعة التدريب (عادةً من 20% إلى 40%)، وندرب النموذج على البيانات المتبقية (60-80% من المجموعة الأصلية)، ونحسب مقاييس الأداء للنموذج (مثل الدقة) على مجموعة الانتظار.
- *التحقق المتبادل*. الحالة الأكثر شيوعًا هنا هي *التحقق المتقاطع k-fold*.



<img align='center' src='../../img/cross_validation.png'><br>



في التحقق المتبادل من k-fold، يتم تدريب النموذج $K$ مرات على مجموعات فرعية مختلفة ($K-1$) من مجموعة البيانات الأصلية (باللون الأبيض) ويتم فحصه على المجموعة الفرعية المتبقية (في كل مرة مجموعة فرعية مختلفة، كما هو موضح أعلاه باللون البرتقالي).
لقد حصلنا على $K$ تقييمات جودة النموذج التي يتم حساب متوسطها عادةً لإعطاء متوسط ​​إجمالي لجودة التصنيف/الانحدار.
يوفر التحقق المتبادل تقييمًا أفضل لجودة النموذج على البيانات الجديدة مقارنة بنهج مجموعة التجميد. ومع ذلك، يعد التحقق المتبادل مكلفًا من الناحية الحسابية عندما يكون لديك الكثير من البيانات.
 
يعد التحقق المتبادل تقنية مهمة جدًا في التعلم الآلي ويمكن تطبيقه أيضًا في الإحصاء والاقتصاد القياسي. فهو يساعد في ضبط المعلمات الفائقة، ومقارنة النماذج، وتقييم الميزات، وما إلى ذلك. يمكن العثور على مزيد من التفاصيل [هنا](https://sebastianraschka.com/blog/2016/model-evaluation-selection-part1.html) (منشور مدونة بقلم سيباستيان راشكا) أو في أي كتاب مدرسي كلاسيكي عن التعلم الآلي (الإحصائي).



## 5. أمثلة تطبيقية وحالات معقدة### طريقة أشجار القرار وأقرب الجيران في مهمة التنبؤ بتغير العملاء 
دعونا نقرأ البيانات في `DataFrame` ونقوم بمعالجتها مسبقًا. قم بتخزين *الحالة* في كائن `Series` منفصل في الوقت الحالي وإزالته من إطار البيانات. سوف نقوم بذلك
 قم بتدريب النموذج الأول بدون ميزة *الحالة*، وبعد ذلك سنرى ما إذا كان ذلك مفيدًا أم لا. 


````python
df = pd.read_csv("../../data/telecom_churn.csv")

df["International plan"] = pd.factorize(df["International plan"])[0]
df["Voice mail plan"] = pd.factorize(df["Voice mail plan"])[0]
df["Churn"] = df["Churn"].astype("int")
states = df["State"]
y = df["Churn"]
df.drop(["State", "Churn"], axis=1, inplace=True)
````


لنخصص 70% من المجموعة للتدريب (`X_train`، `y_train`) و30% لمجموعة الانتظار (`X_holdout`، `y_holdout`). لن تشارك مجموعة الإيقاف في ضبط معلمات النماذج. سنستخدمه في النهاية، بعد الضبط، لتقييم جودة النموذج الناتج. دعونا ندرب نموذجين: شجرة القرار وk-NN. نحن لا نعرف ما هي المعلمات الجيدة، لذلك سنفترض بعض المعلمات العشوائية: عمق الشجرة 5 وعدد أقرب الجيران يساوي 10.


````python
from sklearn.model_selection import StratifiedKFold, train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler

X_train, X_holdout, y_train, y_holdout = train_test_split(
    df.values, y, test_size=0.3, random_state=17
)

tree = DecisionTreeClassifier(max_depth=5, random_state=17)
knn = KNeighborsClassifier(n_neighbors=10)

tree.fit(X_train, y_train)

# for kNN, we need to scale features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_holdout_scaled = scaler.transform(X_holdout)
knn.fit(X_train_scaled, y_train)
````

````python
from sklearn.metrics import accuracy_score

tree_pred = tree.predict(X_holdout)
accuracy_score(y_holdout, tree_pred)  # 0.94
````

````python
knn_pred = knn.predict(X_holdout_scaled)
accuracy_score(y_holdout, knn_pred)  # 0.89
````


الآن، دعونا نحدد معلمات الشجرة باستخدام التحقق المتبادل. سنقوم بضبط الحد الأقصى للعمق والحد الأقصى لعدد الميزات المستخدمة في كل تقسيم. إليك جوهر كيفية عمل GridSearchCV: لكل زوج فريد من القيم `max_depth` و`max_features`، قم بحساب أداء النموذج من خلال التحقق المتبادل بخمسة أضعاف، ثم حدد أفضل مجموعة من المعلمات.


### حالة معقدة لأشجار القرار
لمواصلة مناقشة إيجابيات وسلبيات الطرق المعنية، دعونا نفكر في مهمة تصنيف بسيطة، حيث يمكن للشجرة أن تؤدي أداءً جيدًا ولكنها تفعل ذلك بطريقة "معقدة للغاية". لنقم بإنشاء مجموعة من النقاط على المستوى (ميزتان)، وستكون كل نقطة واحدة من فئتين (+1 للأحمر، أو -1 للأصفر). إذا نظرت إليها كمشكلة تصنيف، فستبدو بسيطة جدًا: يتم الفصل بين الفئات بخط. 


````python
def form_linearly_separable_data(n=500, x1_min=0, x1_max=30, x2_min=0, x2_max=30):
    data, target = [], []
    for i in range(n):
        x1 = np.random.randint(x1_min, x1_max)
        x2 = np.random.randint(x2_min, x2_max)
        if np.abs(x1 - x2) > 0.5:
            data.append([x1, x2])
            target.append(np.sign(x1 - x2))
    return np.array(data), np.array(target)


X, y = form_linearly_separable_data()

plt.scatter(X[:, 0], X[:, 1], c=y, cmap="autumn", edgecolors="black");
````


ومع ذلك، فإن الحدود التي تنشئها شجرة القرار معقدة للغاية؛ بالإضافة إلى أن الشجرة نفسها عميقة جدًا. تخيل أيضًا مدى سوء تعميم الشجرة على المساحة الواقعة خارج مربعات $30 \times 30$ التي تشكل إطار مجموعة التدريب.


````python
tree = DecisionTreeClassifier(random_state=17).fit(X, y)

xx, yy = get_grid(X)
predicted = tree.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
plt.pcolormesh(xx, yy, predicted, cmap="autumn")
plt.scatter(
    X[:, 0], X[:, 1], c=y, s=100, cmap="autumn", edgecolors="black", linewidth=1.5
)
plt.title("Easy task. Decision tree compexifies everything");
````

### أشجار القرار وk-NN في مهمة التعرف على الأرقام المكتوبة بخط اليد من MNIST
الآن دعونا نلقي نظرة على كيفية أداء هاتين الخوارزميتين في مهمة حقيقية. سوف نستخدم `sklearn` مجموعة البيانات المضمنة على الأرقام المكتوبة بخط اليد. هذه المهمة هي مثال حيث يعمل k-NN بشكل جيد بشكل مدهش.
 
الصور هنا عبارة عن مصفوفات 8x8 (كثافة اللون الأبيض لكل بكسل). ثم يتم "تكشف" كل هذه المصفوفة إلى متجه بطول 64، ونحصل على وصف مميز للكائن.

دعونا نختار 70% من مجموعة البيانات للتدريب (`X_train`، `y_train`) و30% للمحتجزين (`X_holdout`، `y_holdout`). لن تشارك مجموعة الإيقاف في ضبط معلمات النموذج؛ سنستخدمه في النهاية للتحقق من جودة النموذج الناتج.

### حالة معقدة لطريقة أقرب الجيران
دعونا نفكر في مثال بسيط آخر. في مشكلة التصنيف، ستكون إحدى الميزات متناسبة فقط مع متجه الاستجابات، لكن هذا لن يساعد في طريقة أقرب الجيران.


````python
def form_noisy_data(n_obj=1000, n_feat=100, random_seed=17):
    np.seed = random_seed
    y = np.random.choice([-1, 1], size=n_obj)

    # first feature is proportional to target
    x1 = 0.3 * y

    # other features are noise

    x_other = np.random.random(size=[n_obj, n_feat - 1])

    return np.hstack([x1.reshape([n_obj, 1]), x_other]), y


X, y = form_noisy_data()
````


كما هو الحال دائمًا، سننظر في دقة التحقق المتبادل ومجموعة الانتظار. لنقم ببناء منحنيات تعكس اعتماد هذه الكميات على المعلمة `n_neighbors` بطريقة أقرب الجيران. وتسمى هذه المنحنيات منحنيات التحقق من الصحة.يمكن للمرء أن يرى أن k-NN مع المسافة الإقليدية لا يعمل بشكل جيد على حل المشكلة، حتى عندما تقوم بتغيير عدد أقرب الجيران على نطاق واسع. في المقابل، فإن شجرة القرار "تكتشف" بسهولة التبعيات المخفية في البيانات على الرغم من وجود قيود على الحد الأقصى للعمق.


````python
X_train, X_holdout, y_train, y_holdout = train_test_split(
    X, y, test_size=0.3, random_state=17
)

from sklearn.model_selection import cross_val_score

cv_scores, holdout_scores = [], []
n_neighb = [1, 2, 3, 5] + list(range(50, 550, 50))

for k in n_neighb:

    knn_pipe = Pipeline(
        [("scaler", StandardScaler()), ("knn", KNeighborsClassifier(n_neighbors=k))]
    )
    cv_scores.append(np.mean(cross_val_score(knn_pipe, X_train, y_train, cv=5)))
    knn_pipe.fit(X_train, y_train)
    holdout_scores.append(accuracy_score(y_holdout, knn_pipe.predict(X_holdout)))

plt.plot(n_neighb, cv_scores, label="CV")
plt.plot(n_neighb, holdout_scores, label="holdout")
plt.title("Easy task. kNN fails")
plt.legend();
````

## 6. إيجابيات وسلبيات أشجار القرار وطريقة أقرب الجيران
### أشجار القرار
الايجابيات:
- إنشاء قواعد تصنيف واضحة ومفهومة للإنسان، على سبيل المثال. "إذا كان عمرك أقل من 25 عامًا ومهتمًا بالدراجات النارية، فارفض القرض". تسمى هذه الخاصية قابلية تفسير النموذج.
- يمكن تصور أشجار القرار بسهولة، أي يمكن "تفسير" النموذج نفسه (الشجرة) والتنبؤ بكائن اختبار معين (مسار في الشجرة).
- التدريب السريع والتنبؤ.
- عدد صغير من معلمات النموذج.
- يدعم كل من الميزات العددية والفئوية.السلبيات:
 
- الأشجار حساسة للغاية للضوضاء في البيانات المدخلة؛ يمكن أن يتغير النموذج بأكمله إذا تم تعديل مجموعة التدريب بشكل طفيف (على سبيل المثال، إزالة ميزة، إضافة بعض الكائنات). وهذا يضعف إمكانية تفسير النموذج.
- الحدود الفاصلة التي تم إنشاؤها بواسطة شجرة القرار لها حدودها - فهي تتكون من طائرات مفرطة متعامدة مع أحد محاور الإحداثيات، وهي أقل جودة من بعض الطرق الأخرى، في الممارسة العملية.
- يجب تجنب التجهيز الزائد عن طريق التقليم أو تحديد حد أدنى لعدد العينات في كل ورقة أو تحديد أقصى عمق للشجرة. لاحظ أن التجهيز الزائد يمثل مشكلة لجميع أساليب التعلم الآلي.
- عدم الاستقرار. يمكن للتغييرات الصغيرة في البيانات أن تغير شجرة القرار بشكل كبير. تتم معالجة هذه المشكلة من خلال مجموعات شجرة القرار (تتم مناقشتها في المرة القادمة).
- إن أفضل مشكلة للبحث عن شجرة القرار هي NP-Complete. يتم استخدام بعض الأساليب الاستدلالية في الممارسة العملية، مثل البحث الجشع عن ميزة ذات أقصى قدر من الحصول على المعلومات، ولكنها لا تضمن العثور على الشجرة المثالية عالميًا.
- صعوبات دعم القيم المفقودة في البيانات. قدر فريدمان أن الأمر استغرق حوالي 50% من الكود لدعم الفجوات في البيانات في CART (تم تنفيذ نسخة محسنة من هذه الخوارزمية في `sklearn`).
- يمكن للنموذج الاستيفاء فقط وليس الاستقراء (وينطبق الشيء نفسه على الغابات العشوائية وتعزيز الأشجار). أي أن شجرة القرار تقوم بالتنبؤ المستمر بالكائنات التي تقع خارج المربع المحيط الذي تحدده مجموعة التدريب في مساحة الميزة. في مثالنا مع الكرات الصفراء والزرقاء، فهذا يعني أن النموذج يعطي نفس التنبؤات لجميع الكرات ذات المواضع > 19 أو <0.



### طريقة أقرب الجيرانالايجابيات:
- التنفيذ البسيط.
- مدروسة جيدا؛
- عادةً ما تكون الطريقة حلاً أوليًا جيدًا ليس فقط للتصنيف أو الانحدار، ولكن أيضًا للتوصيات؛
- يمكن تكييفها مع مشكلة معينة عن طريق اختيار المقاييس أو النواة الصحيحة (باختصار، قد تقوم النواة بضبط عملية التشابه للكائنات المعقدة مثل الرسوم البيانية مع الحفاظ على نهج k-NN كما هو). بالمناسبة، [ألكسندر دياكونوف](https://www.kaggle.com/dyakonov)، أحد أفضل 1 كاجلر سابقًا، يحب أبسط k-NN ولكن مع مقياس تشابه الكائنات المضبوط؛
- قابلية تفسير جيدة. هناك استثناءات: إذا كان عدد الجيران كبيرًا، تتدهور قابلية التفسير ("لم نمنحه قرضًا، لأنه يشبه العملاء الـ 350، منهم 70 عميلًا سيئًا، وهذا أعلى بنسبة 12٪ من المتوسط ​​لمجموعة البيانات").السلبيات:
- تعتبر الطريقة سريعة مقارنة بتركيبات الخوارزميات، لكن عدد الجيران المستخدم للتصنيف عادة ما يكون كبيرا (100-150) في الحياة الواقعية، وفي هذه الحالة لن تعمل الخوارزمية بسرعة شجرة القرار.
- إذا كانت مجموعة البيانات تحتوي على العديد من المتغيرات، فمن الصعب العثور على الأوزان المناسبة وتحديد الميزات غير المهمة للتصنيف/الانحدار.
- الاعتماد على مقياس المسافة المحدد بين الكائنات. غالبًا ما يكون تحديد المسافة الإقليدية افتراضيًا لا أساس له من الصحة. يمكنك العثور على حل جيد من خلال البحث على الشبكة عبر المعلمات، ولكن هذا يستغرق وقتًا طويلاً للغاية بالنسبة لمجموعات البيانات الكبيرة.
- لا توجد طرق نظرية لاختيار عدد الجيران - فقط البحث على الشبكة (على الرغم من أن هذا ينطبق غالبًا على جميع المعلمات الفائقة لجميع النماذج). في حالة وجود عدد صغير من الجيران، تكون الطريقة حساسة للقيم المتطرفة، أي أنها تميل إلى التجاوز.
- كقاعدة عامة، لا يعمل الأمر بشكل جيد عندما يكون هناك الكثير من الميزات بسبب "لعنة الأبعاد". يتحدث البروفيسور بيدرو دومينغوس، وهو عضو معروف في مجتمع تعلم الآلة، عن ذلك [هنا](https://homes.cs.washington.edu/~pedrod/papers/cacm12.pdf) في ورقته البحثية الشهيرة، "بعض الأشياء المفيدة التي يجب معرفتها حول التعلم الآلي"؛ كما تم وصف "لعنة الأبعاد" في كتاب التعلم العميق في [هذا الفصل](http://www.deeplearningbook.org/contents/ml.html).
هذه معلومات كثيرة، ولكن نأمل أن تكون هذه المقالة مرجعًا رائعًا لك لفترة طويلة :)
