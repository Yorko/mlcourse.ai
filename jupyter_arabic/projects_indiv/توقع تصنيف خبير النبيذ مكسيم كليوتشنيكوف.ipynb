### مجموعة البيانات
البيانات مأخوذة من مجموعة بيانات Kaggle https://www.kaggle.com/zynicide/wine-reviews/home, والتي بدورها تم استخراجها بواسطة مؤلف مجموعة البيانات من https://www.winemag.com/<br>
هناك الكثير من المراجعات من خبراء مختلفين للنبيذ من جميع أنحاء العالم. كما يتم أيضًا توفير بعض المعلومات الخاصة بالنبيذ كجزء من مجموعة البيانات.<br>
تتكون مجموعة البيانات من الحقول التالية (لكل معلومات من https://github.com/zackthoutt/wine-deep-learning):### الميزات
* **النقاط**: عدد النقاط التي قام WineEnthusiast بتقييم النبيذ على مقياس من 1 إلى 100 (على الرغم من أنهم يقولون إنهم ينشرون فقط مراجعات للنبيذ الذي يسجل أكثر من = 80)
* **العنوان**: عنوان مراجعة النبيذ، والذي غالبًا ما يحتوي على النبيذ القديم إذا كنت مهتمًا باستخراج هذه الميزة
* **التنوع**: نوع العنب المستخدم لصنع النبيذ (أي بينوت نوير)
* **الوصف**: بضع جمل من الساقي تصف طعم النبيذ ورائحته وشكله وملمسه وما إلى ذلك.
* **البلد**: البلد الذي يأتي منه النبيذ
* **المقاطعة**: المقاطعة أو الولاية التي ينتمي إليها النبيذ
* **المنطقة 1**: منطقة زراعة النبيذ في مقاطعة أو ولاية (أي نابا)
* **المنطقة 2**: في بعض الأحيان توجد مناطق أكثر تحديدًا محددة داخل منطقة زراعة النبيذ (على سبيل المثال، روثرفورد داخل وادي نابا)، ولكن قد تكون هذه القيمة فارغة في بعض الأحيان
* **مصنع النبيذ**: مصنع النبيذ الذي يصنع النبيذ
* **المسمى**: الكرم الموجود داخل مصنع النبيذ حيث يأتي العنب الذي يصنع النبيذ
* **السعر**: تكلفة زجاجة النبيذ بالدولار الأمريكي
* **اسم المتذوق**: اسم الشخص الذي تذوق النبيذ وراجعه
* ** Taster Twitter Handle **: مقبض Twitter للشخص الذي تذوق النبيذ وقام بمراجعته
### الهدف
لدينا تصنيف النبيذ (** النقاط **) كهدف. يقدم المراجعون من الموقع الأصلي تصنيفًا لأنواع النبيذ التي تتراوح من 80 إلى 100، وفيما يلي تفاصيل النطاقات المختلفة:| النطاق | مارك | الوصف |
|--------|-----------|--------------------------------------------------------|
| 98–100 | كلاسيك | قمة الجودة |

| 83–86 | جيد | مناسبة للاستهلاك اليومي. قيمة جيدة في كثير من الأحيان |
| 80–82 | مقبول | يمكن توظيفهم في ظروف غير رسمية وأقل خطورة |


نرى أن لدينا الكثير من التقييمات للنبيذ القادم من الولايات المتحدة، وهو ما يمكن تفسيره بحقيقة أن معظم المراجعين موجودون في الولايات المتحدة.<br>

لكي يعرض Choropleth التلوين بطريقة أكثر قابلية للفهم، فلنقم ```log1p```- بتحويل عدد المراجعات لكل بلد:


````python
countries = df.groupby('country').size().reset_index()
countries.columns = ['name', 'size']
countries.name = countries.name.replace({ # making the country names compatible with pycountry
    'England': 'United Kingdom',
    'Czech Republic': 'United Kingdom',
    'Macedonia': 'Macedonia, Republic of',
    'Moldova': 'Moldova, Republic of',
    'US': 'United States'
})

data = pd.DataFrame(index=countries.index)
data['name'] = countries.name
data['size'] = countries['size']
data['code'] = countries.apply(lambda x: pycountry.countries.get(name=x['name']), axis=1)
data['code'] = data.code.apply(lambda x: x.alpha_3 if x else None)
data = data.dropna()

choropleth_data = [dict(
    type='choropleth',
    locations=data['code'],
    z=np.log1p(data['size']),
    #showscale=False,
    text=data['name'],
    marker=dict(
        line=dict(
            color='rgb(180,180,180)',
            width=0.5
        )),
)]

layout = dict(
    title='Number of wine reviews per country, log-transformed',
    geo=dict(
        showframe=False,
        showcoastlines=True,
        projection=dict(
            type='natural earth'
        )
    ))

fig = dict(data=choropleth_data, layout=layout)
py.iplot(fig, validate=False)
````


#### التنظيف والتحويل



دعونا نرى ما إذا كان بإمكاننا الحصول على شيء من العنوان:


````python
df[['region_1', 'title']].head()
````


كما نرى، تتكرر بعض المناطق في العنوان، وحتى إذا كانت المنطقة NaN، فمن الممكن ملؤها بالقيمة من العنوان، لذلك دعونا نفعل ذلك:


````python
def extract_region_1(row):
    if row.region_1 == 'nan':
        return row.region_1
    if not row.title.endswith(')'):
        return None
    return row.title[row.title.rindex('(')+1:-1]

df.region_1 = df.apply(extract_region_1, axis=1)
````

````python
df[['region_1', 'title']].head()
````


رائع، فلنعد الآن إنشاء **الموقع**:


````python
df['location'] = df.apply(lambda x: ' / '.join([y for y in [str(x['trans_country']), str(x['province']), str(x['region_1']), str(x['region_2'])] if y != 'nan']), axis=1)
df.location.head()
````


لنستبدل الآن المواقع التي تحتوي على عدد أقل من المراجعات بالاسم "أخرى"

````python
vc = df.location.value_counts()
df.location = df.location.replace(vc[vc < 2].index, 'Other')
````

لذلك دعونا نستخدم Ridge ونرى كيف سيعمل:


````python
MODEL = Ridge(random_state=RANDOM_SEED)
````


## اختيار التحقق من الصحة
نظرًا لأن بياناتنا لا تحتوي على أي تفاصيل ثقيلة، فيمكننا اختيار التحقق البسيط من صحة KFold لمدة 10 طيات، مع التبديل العشوائي:


````python
CV = KFold(n_splits=10, shuffle=True, random_state=RANDOM_SEED)
````


## المعالجة المسبقة للبيانات


````python
df_full = df.copy(deep=True)
df_full = df_full.drop(['country', 'price', 'taster_name', 'taster_twitter_handle', 'variety', 'province', 'region_1', 'region_2'], axis=1)
df_full.columns = [x.replace('trans_', '') for x in df_full.columns]
df_full.info()
````


### التعامل مع القيم الخالية


````python
# Fill missing countries with "Other"
df_full.country = df_full.country.fillna('Other')

# Fill missing locations with "Other"
df_full.location = df_full.location.fillna('Other')

# Remove samples with missing prices since there are not so much of them and it's and important feature
df_full = df_full[~df_full.price.isna()]

df_full.info()
````


### تقسيم اختبار التدريب


````python
df_train, df_test, y_train, y_test = train_test_split(df_full.drop(['points'], axis=1), df_full.points, test_size=0.25, random_state=RANDOM_SEED)
df_train.shape, df_test.shape, y_train.shape, y_test.shape
````


لاحظ أننا سنقوم بمعالجة مجموعات القطار والاختبار بشكل منفصل، حتى لا نقدم مشكلة "التطلع إلى المستقبل".



### ترميز الميزات الفئوية


````python
categorical_features = ['country', 'variety', 'location']

for feature in categorical_features:
    categorical = pd.Categorical(df_train[feature].unique())
    
    df_train[feature] = df_train[feature].astype(categorical)
    df_test[feature] = df_test[feature].astype(categorical)
    
X_train_cat = pd.get_dummies(df_train[categorical_features], sparse=True)
X_test_cat = pd.get_dummies(df_test[categorical_features], sparse=True)

X_train_cat.shape, X_test_cat.shape
````


### تحويل النص باستخدام TF-IDF


````python
tv = TfidfVectorizer(stop_words=stopwords, max_features=10000)
X_train_desc = tv.fit_transform(df_train.description)
X_test_desc = tv.transform(df_test.description)
X_train_desc.shape, X_test_desc.shape
````


### قياس الميزات العددية
نموذجنا حساس للميزات الرقمية غير المركزية، لذلك نحن بحاجة إلى توسيع نطاقها:


````python
ss = StandardScaler()
df_train.price = ss.fit_transform(df_train[['price']])
df_test.price = ss.transform(df_test[['price']])
````


### تجميع الميزات معًا


````python
X_train = csr_matrix(hstack([
    df_train[['price']],
    X_train_cat,
    X_train_desc,
]))
X_test = csr_matrix(hstack([
    df_test[['price']],
    X_test_cat,
    X_test_desc,
]))
X_train.shape, X_test.shape
````

### تجميع خطوات المعالجة المسبقة معًا


````python
def prepare_data(df_full, categorical_features):
    df_train, df_test, y_train, y_test = train_test_split(df_full.drop(['points'], axis=1), df_full.points, test_size=0.25, random_state=RANDOM_SEED)
    df_train.shape, df_test.shape, y_train.shape, y_test.shape

    print('processing categorical features')
    for feature in categorical_features:
        categorical = pd.Categorical(df_train[feature].unique())

        df_train[feature] = df_train[feature].astype(categorical)
        df_test[feature] = df_test[feature].astype(categorical)

    print('preparing dummies')
    X_train_cat = pd.get_dummies(df_train[categorical_features], sparse=True)
    X_test_cat = pd.get_dummies(df_test[categorical_features], sparse=True)
    
    print('extracting word vectors')
    tv = TfidfVectorizer(stop_words=stopwords, max_features=10000)
    X_train_desc = tv.fit_transform(df_train.description)
    X_test_desc = tv.transform(df_test.description)
    X_train_desc.shape, X_test_desc.shape
    
    print('scaling')
    ss = StandardScaler()
    df_train.price = ss.fit_transform(df_train[['price']])
    df_test.price = ss.transform(df_test[['price']])
    df_train.describe()

    print('combining features')
    X_train = csr_matrix(hstack([
        df_train[['price']],
        X_train_cat,
        X_train_desc,
    ]))
    X_test = csr_matrix(hstack([
        df_test[['price']],
        X_test_cat,
        X_test_desc,
    ]))

    return X_train, X_test, y_train, y_test

X_train, X_test, y_train, y_test = prepare_data(df_full, ['country', 'variety', 'location'])
X_train.shape, X_test.shape, y_train.shape, y_test.shape
````


## تدريب النموذج



دعونا نلائم نموذجنا لأول مرة ونرى كيف سيتم تنفيذه:


````python
def train_and_cv(model, X_train, y_train, X_test, y_test):
    cvs = cross_val_score(model, X_train, y_train, cv=CV, scoring=SCORING, n_jobs=-1)
    print('MSE and STD on CV:\t', -cvs.mean(), cvs.std())
    
    model.fit(X_train, y_train)
    print('MSE on holdout:\t\t', mean_squared_error(MODEL.predict(X_test), y_test))
    
    return model

train_and_cv(MODEL, X_train, y_train, X_test, y_test);
````


وكانت النتائج جيدة جدًا، ولدينا بالفعل نسبة خطأ جيدة ومنخفضة نسبيًا. ولكن يمكننا بالتأكيد تحسينه إلى أبعد من ذلك.



الآن دعونا نرى منحنى التعلم:


````python
def plot_learning_curve(model, X_train, y_train):
    plt.figure(figsize=(10, 5))
    viz = LearningCurve(model, cv=CV, train_sizes=np.linspace(.1, 1.0, 10), scoring=SCORING, n_jobs=-1)
    viz.fit(X_train, y_train)
    viz.poof()
    
plot_learning_curve(MODEL, X_train, y_train);
````


نرى صورة نموذجية عندما تتناقص درجة التدريب مع زيادة عدد العينات المعطاة للنموذج وتزايد الدرجات المتقاطعة.<br>
ولكن لا تزال هناك فجوة بينهما، لذلك سيتم تحسين نموذجنا بعدد أكبر من العينات المقدمة.<br>
ومن الجدير بالذكر أيضًا أن التباين في الدرجات المتقاطعة منخفض جدًا، مما يخبرنا أن نموذجنا يعطي تنبؤات مستقرة جدًا.



## ضبط المعلمة الفائقة
إحدى الخصائص الرئيسية لنموذجنا هي بساطته، لذلك لدينا فقط معلمة لضبطها: [alpha](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.Ridge.html)، وهي قوة التنظيم لتنظيم التلال.<br>
دعونا نحاول أن نرى ما هي أفضل قيمة لها بصريًا:
