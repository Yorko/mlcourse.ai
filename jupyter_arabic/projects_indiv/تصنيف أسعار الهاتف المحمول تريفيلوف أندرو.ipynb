# الجزء 5. اختيار المقاييس <a id='part5'></a>


لدينا مشكلة التصنيف متعدد الفئات. من الضروري التنبؤ بالفئة نفسها، وليس احتمال الانتماء إلى الفئة، لذلك نستخدم المقاييس من مشكلة التصنيف، وهي `accuracy`، `precision`، `recall  `، `f1`. المقياس الأساسي الذي سنحصل عليه هو `accuracy` لكننا سنستخدم `classification_report` لتقدير المقاييس الأخرى.
يمكننا استخدام `accuracy`، لأن لدينا توزيعًا موحدًا للمتغير المستهدف.
$$\mathcal accuracy = \dfrac{1}{l}\sum_{i=1}^l [a(x_{i})=y_{i}]$$
سنأخذ في الاعتبار أيضًا `confusion matrix`، الأعمدة `i` - تسمية الفئة الحقيقية، السطر `j` - تقييم عضوية الفصل من الخوارزمية الخاصة بنا، حيث $q_{ij}$: 
$$\mathcal q_{ij} = \sum_{m=1}^l [a(x_{m})=i][y_{m}=j]$$

# الجزء 7. المعالجة المسبقة للبيانات <a id='part7'></a>



نقسم عينتنا إلى مصفوفة من الميزات ومتجه من الإجابات:


```python
X = data_train.drop('price_range', axis=1)
y = data_train.price_range
```


لنقم بالتقسيم إلى عينة قطار وعينة انتظار:


```python
X_train_part, X_valid, y_train, y_valid = train_test_split(X, y, test_size=0.3, stratify=y, random_state=17)
```


لا ينبغي تحجيم بعض النماذج، ولكن بالنسبة للنماذج الأخرى فمن الضروري:


```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_train_part_scaled, X_valid_scaled, y_train, y_valid = train_test_split(X_scaled, y,\
                                                        test_size=0.3, stratify=y, random_state=17)
```


# الجزء 8. التحقق من الصحة وتعديل المعلمات الفائقة للنموذج <a id='part8'></a>



## `LogisticRegression` مع ميزات متدرجة:


```python
lr = LogisticRegression(random_state=17)
lr.fit(X_train_part_scaled, y_train);
print(accuracy_score(y_valid, lr.predict(X_valid_scaled)))
print(classification_report(y_valid, lr.predict(X_valid_scaled)))
```


#### مصفوفة الارتباك لـ `LogisticRegression`:


```python
tab = pd.crosstab(y_valid, lr.predict(X_valid_scaled), margins=True)
tab.index = ['low cost', 'medium cost', 'high cost', 'very high cost', 'all']
tab.columns = tab.index
tab
```

بالنسبة إلى `GridSearchCV` نختار معلمات `LogisticRegression`: <b>C</b> - على عكس قوة التنظيم، تحدد القيم الأصغر تنظيمًا أقوى. <b>solver</b> - الخوارزمية المستخدمة في مشكلة التحسين. <b>class_weight</b> - الأوزان المرتبطة بالفئات في النموذج {class_label:weight}. 


```python
params = {'C': np.logspace(-5, 5, 11),
          'solver' : ['newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga'],
          'class_weight' : ['balanced', None]}


lr_grid = GridSearchCV(lr, params, n_jobs=-1, cv=5, scoring='accuracy', verbose=1)
lr_grid.fit(X_train_part_scaled, y_train);
```

```python
print(accuracy_score(y_valid, lr_grid.predict(X_valid_scaled)))
print(classification_report(y_valid, lr_grid.predict(X_valid_scaled)))
lr_grid.best_params_, lr_grid.best_score_
```


جميل، بعد `GridSearchCV` نرى هذه النتيجة تزيد.



#### مصفوفة الارتباك لـ `LogisticRegression` بعد `GridSearchCV`:


```python
tab = pd.crosstab(y_valid, lr_grid.predict(X_valid_scaled), margins=True)
tab.index = ['low cost', 'medium cost', 'high cost', 'very high cost', 'all']
tab.columns = tab.index
tab
```


## `KNeighborsClassifier` بميزات غير قابلة للقياس:


#### مصفوفة الارتباك لـ `KNeighborsClassifier`:


```python
tab = pd.crosstab(y_valid, kneigh.predict(X_valid), margins=True)
tab.index = ['low cost', 'medium cost', 'high cost', 'very high cost', 'all']
tab.columns = tab.index
tab
```


## `OneVsOneClassifier` مع ميزات متدرجة:


```python
clf = OneVsOneClassifier(LogisticRegression(random_state=17))
clf.fit(X_train_part_scaled, y_train);

print(accuracy_score(y_valid, clf.predict(X_valid_scaled)))
print(classification_report(y_valid, clf.predict(X_valid_scaled)))
```


#### القيام بـ `GridSearchCV` لـ `OneVsOneClassifier` مع `LogisticRegression`:

# الجزء 9. إنشاء ميزات جديدة ووصف لهذه العملية <a id='part9'></a>


`inch` (اختصار: in أو ″) هي وحدة طول في أنظمة القياس الإمبراطورية (البريطانية) والولايات المتحدة العرفية. وهو يساوي1⁄36 ياردة أو1⁄12 قدم. مشتقة من uncia الرومانية ("الثاني عشر")، تُستخدم كلمة بوصة أيضًا أحيانًا لترجمة وحدات مماثلة في أنظمة قياس أخرى، تُفهم عادةً على أنها مشتقة من عرض الإبهام البشري. تباينت معايير الطول الدقيق للبوصة في الماضي، ولكن منذ اعتماد الساحة الدولية خلال الخمسينيات والستينيات من القرن الماضي، فقد تم الاعتماد عليها على النظام المتري وتم تعريفها على أنها بالضبط <b>2.54 سم</b>. 



البكسل في البوصة (`ppi`) أو البكسل في السنتيمتر (ppcm) عبارة عن قياسات لكثافة البكسل (الدقة) لجهاز صورة إلكتروني، مثل شاشة الكمبيوتر أو شاشة التلفزيون، أو جهاز رقمنة الصورة مثل الكاميرا أو الماسح الضوئي للصور. عادة ما تكون الكثافة الأفقية والرأسية هي نفسها، حيث أن معظم الأجهزة تحتوي على بكسلات مربعة، ولكنها تختلف عن الأجهزة التي تحتوي على بكسلات غير مربعة. 



لذا، أعتقد أن `ppi` هي ميزة جيدة، لأنه كلما زادت القيمة، زادت وضوح الصورة.
دعونا نتحقق من هذا.

# الجزء 10. رسم منحنيات التدريب والتحقق <a id='part10'></a>


```python
def plot_learning_curve(estimator, title, X, y, ylim=None, cv=None,
                        n_jobs=None, train_sizes=np.linspace(.1, 1.0, 5)):

    plt.figure()
    plt.title(title)
    if ylim is not None:
        plt.ylim(*ylim)
    plt.xlabel("Training examples")
    plt.ylabel("Score")
    train_sizes, train_scores, test_scores = learning_curve(
        estimator, X, y, cv=cv, n_jobs=n_jobs, train_sizes=train_sizes)
    train_scores_mean = np.mean(train_scores, axis=1)
    train_scores_std = np.std(train_scores, axis=1)
    test_scores_mean = np.mean(test_scores, axis=1)
    test_scores_std = np.std(test_scores, axis=1)
    plt.grid()

    plt.fill_between(train_sizes, train_scores_mean - train_scores_std,
                     train_scores_mean + train_scores_std, alpha=0.1,
                     color="r")
    plt.fill_between(train_sizes, test_scores_mean - test_scores_std,
                     test_scores_mean + test_scores_std, alpha=0.1, color="g")
    plt.plot(train_sizes, train_scores_mean, 'o-', color="r",
             label="Training score")
    plt.plot(train_sizes, test_scores_mean, 'o-', color="g",
             label="Cross-validation score")

    plt.legend(loc="best")
    return plt
```


رسم منحنيات التدريب والتحقق من صحة نموذج الشبكة مع ميزات جديدة `SVC`:


```python
svc3 = SVC(C=0.1, kernel='linear', probability=True, class_weight='balanced', random_state=17)
title = "Learning Curves (SVM, Linear kernel, C=0.1)"
plot_learning_curve(svc3, title, X_train_part2, y_train2, (0.7, 1.01), cv=20, n_jobs=4)
plt.show()
```


رسم منحنيات التدريب والتحقق من صحة نموذج الشبكة مع الميزات الجديدة `OneVsOneClassifier` مع `LogisticRegression`:


```python
clf3 = OneVsOneClassifier(LogisticRegression(C=100,\
                                            class_weight='balanced', solver='newton-cg', random_state=17))
title = "Learning Curves (OneVsOneClassifier, LogisticRegression base model, C=100)"
plot_learning_curve(clf3, title, X_train_part_scaled2, y_train2, (0.7, 1.01), cv=20, n_jobs=4)
plt.show()
```
