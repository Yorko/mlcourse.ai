# مشروع فردي: توقع ظهور مرض السكري بناءً على التدابير التشخيصية



هذه الحالة من تحليل البيانات الطبية. لدينا في هذا المجال بعض الخصائص التي يجب التطرق إليها قبل البدء بالمشروع.
المشاكل في علم البيانات الطبية هي:
1. مجموعات البيانات الصغيرة؛
2. البيانات المفقودة؛
3. البيانات التالفة (المعلومات الكاذبة: الكذب، الاختبار الإيجابي/السلبي، الاختبار السيئ وما إلى ذلك)
4. الخوارزميات القابلة للتفسير فقط هي التي تكون قابلة للتطبيق (في بعض البلدان ينص على ذلك القانون)
لذلك نحن بحاجة إلى نموذج قوي (بحيث لا يتم الإفراط في التجهيز والاعتماد بشكل كبير على البيانات) مع تفسير يمكنه التعامل مع البيانات المفقودة والبيانات صغيرة الحجم.
بالنظر إلى جميع النماذج المذكورة أعلاه، سأحاول استخدام الخوارزمية الكلاسيكية (التي تستخدم غالبًا في التطبيقات الطبية) - شبكة بايز.
حول [مجموعة البيانات](https://www.kaggle.com/uciml/pima-indians-diabetes-database):
يؤثر داء السكري على 422 مليون شخص (بتاريخ 2014) في العالم أو 8.5% من البالغين (أكثر من 18 عامًا)، ويتسبب في 1.5-5.0 مليون سنويًا (1.6 مليون بسبب مرض السكري و2.2 بسبب ارتفاع نسبة الجلوكوز في الدم في عام 2016).كان السكان في هذه الدراسة هم سكان هنود بيما بالقرب من فينيكس، أريزونا. كان هذا السكان قيد الدراسة المستمرة منذ عام 1965 من قبل المعهد الوطني للسكري وأمراض الجهاز الهضمي والكلى بسبب ارتفاع معدل الإصابة بمرض السكري. طُلب من كل مقيم في المجتمع يزيد عمره عن 5 سنوات الخضوع لفحص موحد كل عامين،
والتي تضمنت اختبار تحمل الجلوكوز عن طريق الفم. تم تشخيص مرض السكري وفقاً لمعايير منظمة الصحة العالمية؛ أي إذا كان مستوى الجلوكوز في البلازما بعد الحمل لمدة ساعتين لا يقل عن 200 مجم/ديسيلتر (11.1 مليمول/لتر) في أي فحص مسحي أو إذا وجدت مستشفى الخدمة الصحية الهندية التي تخدم المجتمع تركيز جلوكوز لا يقل عن 200 مجم/ديسيلتر خلال فترة الرعاية الطبية الروتينية. بالإضافة إلى كونها قاعدة بيانات مألوفة للباحثين، فقد وفرت مجموعة البيانات هذه مصدر بيانات تم التحقق من صحته بشكل جيد لاستكشاف التنبؤ بتاريخ ظهور مرض السكري بطريقة طولية.
تم اختيار ثمانية متغيرات لتشكل الأساس للتنبؤ بظهور مرض السكري في غضون خمس سنوات لدى نساء هنود بيما (تم اختيار هذه المتغيرات لأنه وجد أنها عوامل خطر كبيرة للإصابة بمرض السكري بين بيماس أو السكان الآخرين)
1. عدد مرات الحمل
2. تركيز الجلوكوز في البلازما بعد ساعتين في اختبار تحمل الجلوكوز عن طريق الفم (GTIT)
3. ضغط الدم الانبساطي ($mmHg$)
4. سماكة جلد ثلاثية الرؤوس ($mm$)
5. مصل الأنسولين لمدة ساعتين ($\mu U/ml$)
6. مؤشر كتلة الجسم ($Weight(kg) / Height (m)^2$)
7. وظيفة نسب مرض السكري
8. العمر (سنوات)



## فهم المهمة


1. عدد مرات الحمل: خلال فترة الحمل يمكن أن تصاب المرأة بسكري الحمل: ارتفاع مستويات السكر في الدم، ولكن كانت تلك المستويات طبيعية قبل الحمل. بعد الولادة، عادة ما يختفي سكري الحمل. لكن سكري الحمل يجعل المرأة أكثر عرضة للإصابة بمرض السكري من النوع الثاني. [[المصدر]](https://www.webmd.com/diabetes/gestational-diabetes-guide/gestational-diabetes)
2. تركيز الجلوكوز في البلازما بعد ساعتين في اختبار تحمل الجلوكوز عن طريق الفم (GTIT): يقيس اختبار تحمل الجلوكوز مدى قدرة خلايا الجسم على امتصاص الجلوكوز (السكر) بعد استهلاك كمية محددة من السكر. يستخدم الأطباء مستويات السكر في الدم أثناء الصيام وقيم الهيموجلوبين A1c لتشخيص مرض السكري من النوع 1 والنوع 2 بالإضافة إلى مرض السكري. يتم استخدام اختبار تحمل الجلوكوز عن طريق الفم (OGTT) لمدة ساعتين لمدة ساعتين لاختبار مرض السكري أو تحمل الجلوكوز. هذه هي المخططات لتقييم النتائج [[المصدر]](https://www.healthline.com/health/glucose-tolerance-test)
   - لمقدمات السكري: 140-199 ملجم/ديسيلتر
    
   - لمرض السكري: 200 ملجم/ديسيلتر أو أكثر
    
   - لسكري الحمل: >153 ملغم/ديسيلتر
 
3. ضغط الدم الانبساطي: ارتفاع ضغط الدم، أو ارتفاع ضغط الدم، هو حالة تظهر لدى الأشخاص المصابين بداء السكري من النوع الثاني. من غير المعروف سبب وجود علاقة مهمة بين المرضين. ويعتقد أن ما يلي يساهم في كلتا الحالتين: السمنة، واتباع نظام غذائي غني بالدهون والصوديوم، والالتهاب المزمن، والخمول. [[المصدر]](https://www.healthline.com/health/type-2-diabetes/hypertension)
4. سماكة ثنية الجلد ثلاثية الرؤوس: يوفر قياس سماكة ثنية الجلد حجمًا تقديريًا للدهون تحت الجلد، وهي طبقة الأنسجة تحت الجلد وتتكون من الخلايا الشحمية. الدهون تحت الجلد هي المحدد الرئيسي لحساسية الأنسولين ولها علاقة قوية بمقاومة الأنسولين. ومع ذلك، فإن الأدلة التي يمكن التنبؤ بها، فإن تأثير مدة الإصابة بمرض السكري على سمك طيات الجلد لا يزال غير واضح. [[المصدر]](https://pdfs.semanticscholar.org/5d68/b7a7391272feb9a737f4d69539483deb2556.pdf)- المجموعة الضابطة لديها حوالي 16.7
     - مجموعة أقل من 5 سنوات من مرض السكري 30.45
     - المجموعة 5-10 سنوات من مرض السكري 31
     - المجموعة > 10 سنوات من مرض السكري 38.09.
5. الأنسولين في الدم لمدة ساعتين: يبدو أنه مؤشر جيد لمقاومة الأنسولين. يمكن أن تكون أداة مفيدة، خاصة في ظل ظروف انخفاض الموارد حيث يمكن لعينة واحدة تأكيد التشخيص، وبالتالي تقليل التكلفة وتكرار الزيارات. [[المصدر]](https://search.proquest.com/openview/f218efe6291008a23ac0b3b57e332e60/1?pq-origsite=gscholar&cbl=226481)
6. مؤشر كتلة الجسم (الوزن بالكيلو جرام / (الطول بالمتر)^2) [[المصدر]](https://en.wikipedia.org/wiki/Body_mass_index): 
    - نقص الوزن = <18.5
    
    - الوزن الطبيعي = 18.5 – 24.9 
    
    - الوزن الزائد = 25-29.9 
    
    - السمنة = مؤشر كتلة الجسم 30 أو أكثر
7. وظيفة نسب مرض السكري (DPF): توفير توليفة من تاريخ مرض السكري لدى الأقارب والعلاقة الوراثية لهؤلاء الأقارب بالموضوع. يستخدم DPF المعلومات من الآباء والأجداد والأشقاء الأشقاء ونصف العمات والأعمام وأبناء العمومة من الدرجة الأولى. وهو يوفر مقياسًا للتأثير الجيني المتوقع للأقارب المتأثرين وغير المتأثرين على خطر الإصابة بمرض السكري في نهاية المطاف. **لم يتم التحقق من صحة هذه الوظيفة.** [[المصدر]](https://europepmc.org/backend/ptpmcrender.fcgi?accid=PMC2245318&blobtype=pdf)
8. العمر (بالسنوات): لا يزال الأشخاص في منتصف العمر وكبار السن هم الأكثر عرضة للإصابة بمرض السكري من النوع الثاني. وفقًا لتقرير إحصاءات مرض السكري الوطني لعام 2017 الصادر عن مركز السيطرة على الأمراض، كان هناك حوالي 1.5 مليون حالة إصابة جديدة بالسكري بين البالغين في عام 2015. وفي عام 2015، كان البالغون الذين تتراوح أعمارهم بين 45 و64 عامًا هم الفئة العمرية الأكثر تشخيصًا لمرض السكري. تم توزيع الحالات الجديدة لمرض السكري من النوع 1 والنوع 2 لدى الأشخاص الذين تبلغ أعمارهم 18 عامًا فما فوق على النحو التالي [[المصدر]] (https://www.healthline.com/health/type-2-diabetes-age-of-onset):
    - الأعمار من 18 إلى 44: 355 ألف حالة جديدة
    
    - الأعمار من 45 إلى 64: 809 آلاف حالة جديدة
    
    - العمر 65 فما فوق: 366 ألف حالة جديدة
    
   وبالتالي فإن مجموعة المخاطر هي **45-64** سنة


### هندسة الميزات: QUICKI



دعونا ننشئ مؤشرًا بديلًا للتنبؤ بمرض السكري. [QUICKI](http://diabetes.diabetesjournals.org/content/54/7/1914.full-text.pdf) هو مقياس بسيط ودقيق وذو معنى، فلماذا لا تستخدمه


```python
df['QUICKI'] = 1 / (np.log(df['Insulin']) + np.log(df['Glucose']))
```

```python
plt.hist(df[df['Outcome'] == 1]['QUICKI'].dropna(), color='red', alpha=0.3, bins=30);
plt.hist(df[df['Outcome'] == 0]['QUICKI'].dropna(), color='green', alpha=0.5, bins=30);
```

```python
from sklearn.metrics import precision_recall_curve, average_precision_score
from sklearn.utils.fixes import signature
```

```python
precision_recall_curve(y_test, probes, pos_label=1)
average_precision = average_precision_score(y_test, probes)
```

```python
precision, recall, threshold = precision_recall_curve(y_test, probes, pos_label=1)

# In matplotlib < 1.5, plt.fill_between does not have a 'step' argument
step_kwargs = ({'step': 'post'}
               if 'step' in signature(plt.fill_between).parameters
               else {})
plt.step(recall, precision, color='b', alpha=0.2,
         where='post')
plt.fill_between(recall, precision, alpha=0.2, color='b', **step_kwargs)

plt.xlabel('Recall')
plt.ylabel('Precision')
plt.ylim([0.0, 1.05])
plt.xlim([0.0, 1.0])
plt.title('2-class Precision-Recall curve: AP={0:0.2f}'.format(
          average_precision))
```

```python
def plot_precision_recall_vs_threshold(precisions, recalls, thresholds):
    """
    Modified from:
    Hands-On Machine learning with Scikit-Learn
    and TensorFlow; p.89
    """
    plt.figure(figsize=(20, 10))
    plt.title("Precision and Recall Scores as a function of the decision threshold")
    plt.plot(thresholds, precisions[:-1], "b--", label="Precision")
    plt.plot(thresholds, recalls[:-1], "g-", label="Recall")
    plt.ylabel("Score")
    plt.xlabel("Decision Threshold")
    plt.legend(loc='best')
```

```python
plot_precision_recall_vs_threshold(precision, recall, thresholds=threshold)
```
