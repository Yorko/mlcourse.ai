### مقدمة
   النجم النابض هو نجم نيوتروني دوار ممغنط للغاية ينبعث منه شعاع من الإشعاع الكهرومغناطيسي. يمكن ملاحظة هذا الإشعاع فقط عندما يكون شعاع الانبعاث موجهًا نحو الأرض (يشبه إلى حد كبير الطريقة التي يمكن بها رؤية المنارة فقط عندما يتم توجيه الضوء في اتجاه الراصد)، وهو مسؤول عن المظهر النبضي للانبعاث. النجوم النيوترونية كثيفة جدًا، ولها فترات دوران قصيرة ومنتظمة. وينتج عن ذلك فاصل زمني دقيق جدًا بين النبضات يتراوح من المللي ثانية إلى الثواني بالنسبة للنجم النابض الفردي. يُعتقد أن النجوم النابضة هي أحد المرشحين لمصدر الأشعة الكونية ذات الطاقة العالية جدًا (انظر أيضًا آلية الطرد المركزي للتسارع).
ينتج كل نجم نابض نمط انبعاث مختلف قليلاً، والذي يختلف قليلاً مع كل دورة. وبالتالي، يتم حساب متوسط ​​الكشف المحتمل للإشارة، المعروف باسم "المرشح"، على مدار العديد من دورات النجم النابض، وفقًا لما يحدده طول المراقبة. في غياب معلومات إضافية، يمكن لكل مرشح أن يصف نجمًا نابضًا حقيقيًا. ولكن من الناحية العملية، فإن جميع الاكتشافات تقريبًا تنتج عن تداخل الترددات الراديوية (RFI) والضوضاء، مما يجعل من الصعب العثور على الإشارات الشرعية.



### معلومات مجموعة البيانات

### شرح البيانات
يتم وصف كل كائن بواسطة 8 متغيرات مستمرة، ومتغير هدف فئة واحدة. 
الأربعة الأولى هي إحصائيات بسيطة تم الحصول عليها من ملف النبض المتكامل (ملف مطوي). هذه عبارة عن مجموعة من المتغيرات المستمرة التي تصف نسخة محددة من خط الطول للإشارة التي تم حساب متوسطها في كل من الوقت والتردد. يتم الحصول على المتغيرات الأربعة المتبقية بالمثل من منحنى DM-SNR.
|               № | الوصف |

#### توضيح بسيط حول الانحراف والتفرطح
   يقيم التواء مدى تماثل توزيع المتغير. إذا كان توزيع الاستجابات لمتغير يمتد نحو الذيل الأيمن أو الأيسر للتوزيع، فيشار إلى التوزيع على أنه منحرف. التفرطح هو مقياس لما إذا كان التوزيع قد وصل إلى ذروته (توزيع ضيق للغاية مع وجود معظم الاستجابات في المركز).
   
#### توضيح صغير حول DM-SNR<center>
<img src="../../img/dmsnr.PNG">
    مثال على منحنى DM-SNR
</center>
   أحد معلمات المساحة الموصوفة هو قياس التشتت (DM) للمساحة. يحدث التشتت بسبب الوسط بين النجمي، ويختلف بالنسبة لكل نجم نابض، اعتمادًا على المسافة التي يفصلها وعدد الإلكترونات الموجودة في الوسط بين النجمي في اتجاه النجم النابض. يؤدي التشتت إلى وصول الترددات المنخفضة للإشارة في وقت متأخر عن الترددات الأعلى. هذا يشوه أو يشتت النبض. سيؤدي هذا التلطيخ إلى طمس النبض تمامًا إذا لم يتم تشتيت الإشارة قبل الطي.
SNR - نسبة الإشارة إلى الضوضاء في ملف تعريف النبض.

## الجزء السابع: المعالجة المسبقة للبيانات



#### تقسيم الميزة والهدف


```python
y = df.target_class.values
df.drop(["target_class"],axis=1,inplace=True)
X = df.values
```


#### التحجيم
قم بتوسيع نطاق البيانات الخاصة بالنموذج الخطي، وقم بإنشاء كائن مقياس الحجم وتدريبه على الجزء التدريبي. تطبيق التحول الذي تم الحصول عليه على كلا العينات


```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```


#### تقسيم القطار والاختبار
نقوم بتقسيم البيانات إلى عينات تدريب واختبار. في اختبار العينات سنأخذ 30% من البيانات. تأكد من أن عينات التدريب والاختبار تحتوي على أجسام النجم النابض.


```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=512)
print('Propotion of target class for train sample: ', sum(y_train)/X_train.shape[0])
print('Propotion of target class for test sample: ',sum(y_test)/X_test.shape[0]) 
```


## الجزء 8: التحقق من الصحة وتعديل المعلمات الفائقة للنموذج



#### سوف نقوم بتقسيم العينة مع الأخذ بعين الاعتبار رصيد الصفوف مع التحقق المتبادل بمقدار 5 أضعاف


```python
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=512)
```


#### تحديد معلمات الغابة العشوائية


```python
param_grid_for_forest = { "n_estimators": [10, 50, 70],
                          "max_depth": [5, 10, 20],
                          "min_samples_split": [5, 10, 20],
                          "min_samples_leaf": [5, 10, 20]}
```

```python
gs_forest = GridSearchCV(RandomForestClassifier(random_state=512), param_grid = param_grid_for_forest,
                         scoring='roc_auc', cv = skf, n_jobs=-1)
gs_forest.fit(X_train, y_train)
```


#### دعونا نلقي نظرة على أفضل المعلمات التي تم العثور عليها من GridSearchCV وأفضل نتيجة لهذه المعلمات


```python
print('Best parameters: ', gs_forest.best_params_)
print('Best score: ', gs_forest.best_score_)
```


#### دعونا نرى ما هي نتيجة النموذج في عينة الاختبار.


```python
model_rf = RandomForestClassifier(max_depth=10, min_samples_leaf=5, min_samples_split=5,
                               n_estimators=70, random_state=512)
model_rf.fit(X_train, y_train)
print('ROC-AUC on validation set: ', roc_auc_score(y_test, model_rf.predict_proba(X_test)[:,1]))
```


#### تحديد معلمات LightGBM


```python
param_grid_for_lightgbm = {
    'learning_rate': [0.01, 0.1, 0.5],
    'n_estimators': [40, 200, 1000],
    'num_leaves': [6, 10, 16],
    'boosting_type' : ['gbdt'],
    'objective' : ['binary'],
    'random_state' : [512], 
    'colsample_bytree' : [0.66],
    'subsample' : [0.75],
    'reg_alpha' : [1,1.2],
    'reg_lambda' : [1,1.4],
    }
```

```python
gs_lightgbm = GridSearchCV(lgb.LGBMClassifier(random_state=512), param_grid_for_lightgbm, verbose=3, cv=skf, n_jobs=-1)
gs_lightgbm.fit(X_train, y_train)
```

#### دعونا نلقي نظرة على أفضل المعلمات التي تم العثور عليها من GridSearchCV وأفضل نتيجة لهذه المعلمات


```python
print('Best parameters: ', gs_lightgbm.best_params_)
print('Best score: ', gs_lightgbm.best_score_)
```


#### دعونا نرى النتيجة التي حصل عليها النموذج في عينة الاختبار


```python
model_gbm = lgb.LGBMClassifier(**gs_lightgbm.best_params_)
model_gbm.fit(X_train, y_train)
print('ROC-AUC on validation set: ', roc_auc_score(y_test, model_gbm.predict_proba(X_test)[:, 1]))
```

## الجزء العاشر: رسم منحنيات التدريب والتحقق من الصحة



#### سيتم رسم منحنيات التدريب والتحقق من أفضل نموذج - LightGBM


```python
def plot_learning_curve(estimator, title, X, y, ylim=None, cv=None,
                        n_jobs=1, train_sizes=np.linspace(.1, 1.0, 5)):

    plt.figure()
    plt.title(title)
    if ylim is not None:
        plt.ylim(*ylim)
    plt.xlabel("Training examples")
    plt.ylabel("Score")
    train_sizes, train_scores, test_scores = learning_curve(
        estimator, X, y, cv=cv, n_jobs=n_jobs, train_sizes=train_sizes, scoring='roc_auc')
    train_scores_mean = np.mean(train_scores, axis=1)
    train_scores_std = np.std(train_scores, axis=1)
    test_scores_mean = np.mean(test_scores, axis=1)
    test_scores_std = np.std(test_scores, axis=1)
    plt.grid()

    plt.fill_between(train_sizes, train_scores_mean - train_scores_std,
                     train_scores_mean + train_scores_std, alpha=0.1,
                     color="r")
    plt.fill_between(train_sizes, test_scores_mean - test_scores_std,
                     test_scores_mean + test_scores_std, alpha=0.1, color="g")
    plt.plot(train_sizes, train_scores_mean, 'o-', color="r",
             label="Training score")
    plt.plot(train_sizes, test_scores_mean, 'o-', color="g",
             label="Cross-validation score")

    plt.legend(loc="best")
    return plt
```

```python
plt.figure(figsize=(6, 4))
plot_learning_curve(model_gbm, 'Training curve', X_train, y_train, cv=skf, n_jobs=-1);
```

```python
a = np.arange(2, 20)
val_train, val_test = validation_curve(model_gbm, X_train, y_train, param_name='num_leaves',
                                                       param_range=a, cv=skf, scoring='roc_auc')

def plot_with_err(x, data, **kwargs):
    mu, std = data.mean(1), data.std(1)
    lines = plt.plot(x, mu, '-', **kwargs)
    plt.fill_between(x, mu - std, mu + std, edgecolor='none',
                     facecolor=lines[0].get_color(), alpha=0.2)
```

```python
fig = plt.figure(figsize=(20, 5))
ax1 = fig.add_subplot(121)
ax1.set_title('Validation curve')
plot_with_err(a, val_train, label='training scores')
plot_with_err(a, val_test, label='validation scores')
plt.xlabel(r'num_leaves'); plt.ylabel('ROC-AUC')
plt.legend();
```
