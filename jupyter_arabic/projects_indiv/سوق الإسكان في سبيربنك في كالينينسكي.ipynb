## نظرة عامة على مسابقة Kaggleتتطلب تكاليف الإسكان استثمارات كبيرة من المستهلكين والمطورين على حد سواء. وعندما يتعلق الأمر بالتخطيط لميزانية - سواء كانت شخصية أو مؤسسية - فإن آخر ما يحتاجه أي شخص هو عدم اليقين بشأن أحد أكبر نفقاته. يساعد بنك سبيربنك، وهو أقدم وأكبر بنك في روسيا، عملائه من خلال وضع تنبؤات حول أسعار العقارات بحيث يكون المستأجرون والمطورون والمقرضون أكثر ثقة عند توقيع عقد إيجار أو شراء مبنى.
على الرغم من أن سوق الإسكان مستقر نسبيًا في روسيا، إلا أن الاقتصاد المتقلب في البلاد يجعل التنبؤ بالأسعار كدالة لخصائص الشقق تحديًا فريدًا. التفاعلات المعقدة بين ميزات السكن مثل عدد غرف النوم والموقع كافية لجعل توقعات الأسعار معقدة. إن إضافة اقتصاد غير مستقر إلى هذا المزيج يعني أن سبيربنك وعملائه بحاجة إلى أكثر من مجرد نماذج انحدار بسيطة في ترسانتهم.
في هذه المنافسة، يتحدى Sberbank شركة Kagglers لتطوير خوارزميات تستخدم مجموعة واسعة من الميزات للتنبؤ بأسعار العقارات. سيعتمد المنافسون على مجموعة بيانات غنية تتضمن بيانات الإسكان وأنماط الاقتصاد الكلي. سيسمح نموذج التنبؤ الدقيق لسبيربنك بتوفير المزيد من اليقين لعملائه في اقتصاد غير مؤكد.

#### 6.1 إنشاء ميزات جديدةسأضيف الميزات الأساسية المشتقة من الطابع الزمني: *السنة*، *الشهر*، و*يوم من الأسبوع*. 
ثم سأحسب كمية الصفقات العقارية في شهر معين (*month_year_cnt*) والأسبوع (*week_year_cnt*). سأفعل ذلك، لأن سعر المنزل قد يتأثر أيضًا بتوفر منازل أخرى في نفس الفترة الزمنية. لذا فإن إنشاء متغير عدد على عدد المنازل في فترة زمنية معينة قد يكون مفيدًا.
وبعد ذلك، سأحسب نسبة الطابق المحدد إلى إجمالي عدد الطوابق في المبنى (*rel_floor*) ونسبة مربع المطبخ إلى المربع الكامل (*rel_kitch_sq*). وأخيرًا، سيتم حذف العمود *الطابع الزمني* لتجنب التجهيز الزائد
وبما أن المدارس بشكل عام تلعب دورًا مهمًا في البحث عن المنازل، فسوف أقوم بإنشاء بعض المتغيرات حول المدرسة.


```python
# Add year, month and day-of-week
full_df['year'] = full_df.timestamp.dt.year
full_df['month'] = full_df.timestamp.dt.month
full_df['dow'] = full_df.timestamp.dt.dayofweek

# Add month-year
month_year = (full_df.timestamp.dt.month + full_df.timestamp.dt.year * 100)
month_year_cnt_map = month_year.value_counts().to_dict()
full_df['month_year_cnt'] = month_year.map(month_year_cnt_map)

# Add week-year count
week_year = (full_df.timestamp.dt.weekofyear + full_df.timestamp.dt.year * 100)
week_year_cnt_map = week_year.value_counts().to_dict()
full_df['week_year_cnt'] = week_year.map(week_year_cnt_map)

# Other feature engineering
full_df['rel_floor'] = full_df['floor'] / full_df['max_floor'].astype(float)
full_df['rel_kitch_sq'] = full_df['kitch_sq'] / full_df['full_sq'].astype(float)
full_df["extra_sq"] = full_df["full_sq"] - full_df["life_sq"]
full_df["floor_from_top"] = full_df["max_floor"] - full_df["floor"]

full_df["ratio_preschool"] = full_df["children_preschool"] / full_df["preschool_quota"].astype("float")
full_df["ratio_school"] = full_df["children_school"] / full_df["school_quota"].astype("float")

# Remove timestamp column (may overfit the model in train)
full_df.drop(['timestamp'], axis=1, inplace=True)
```


### الجزء 7. التحقق من الصحة وضبط المعلمات الفائقة



سأقوم بتعيين قيمة أولية عشوائية = **111**، التكرارات الأولية (الجولات = 1500).
أريد أيضًا التأكيد على دور التحقق المتبادل. لقد واجهت مناقشة حول السيرة الذاتية وهنا بيان مهم:
ميخايلو ليسوفي (https://stackoverflow.com/users/9640384/mykhailo-lisovyi):
"بشكل عام، الغرض من السيرة الذاتية هو عدم القيام بتحسين المعلمات الفائقة. والغرض من ذلك هو تقييم أداء إجراءات بناء النموذج.يتطابق تقسيم التدريب/الاختبار الأساسي من الناحية المفاهيمية مع السيرة الذاتية ذات الطية الواحدة (مع حجم مخصص للانقسام على عكس حجم القطار 1/K في السيرة الذاتية ذات الطية k). تتمثل ميزة إجراء المزيد من الانقسامات (أي k> 1 CV) في الحصول على مزيد من المعلومات حول تقدير خطأ التعميم. هناك المزيد من المعلومات بمعنى الحصول على الخطأ + عدم اليقين في الإحصائيات. هناك مناقشة ممتازة (https://stats.stackexchange.com/questions/244907/how-to-get-hyper-parameters-in-nested-cross-validation?rq=1) على CrossValidated (ابدأ بالروابط المضافة إلى السؤال، والتي تغطي نفس السؤال، ولكن تمت صياغتها بطريقة مختلفة). وهو يغطي التحقق المتقاطع المتداخل وهو ليس واضحًا على الإطلاق. ولكن إذا كنت ستلتف حول المفهوم بشكل عام، فهذا سيساعدك في مختلف المواقف غير التافهة. الفكرة التي يجب عليك استخلاصها هي: **الغرض من السيرة الذاتية هو تقييم أداء إجراءات بناء النموذج.**
مع أخذ هذه الفكرة في الاعتبار، كيف يمكن للمرء التعامل مع تقدير المعلمة الفائقة بشكل عام (ليس فقط في LightGBM)؟تريد تدريب نموذج بمجموعة من المعلمات على بعض البيانات وتقييم كل شكل من أشكال النموذج على مجموعة مستقلة (التحقق من الصحة). ثم تنوي اختيار أفضل المعلمات عن طريق اختيار المتغير الذي يوفر أفضل مقياس تقييم من اختيارك.
** يمكن القيام بذلك من خلال تقسيم بسيط للتدريب/الاختبار. ** لكن الأداء المقيم، وبالتالي اختيار معلمات النموذج الأمثل، قد يكون مجرد تقلب في تقسيم معين.
وبالتالي، يمكنك **تقييم كل نموذج من هذه النماذج بتقييم متوسط أكثر قوة من الناحية الإحصائية عبر عدة أقسام للتدريب/الاختبار، على سبيل المثال k-fold CV.**
بعد ذلك، يمكنك اتخاذ خطوة أبعد والقول إن لديك مجموعة احتجاز إضافية، تم فصلها قبل بدء تحسين المعلمة الفائقة. بهذه الطريقة يمكنك تقييم أفضل نموذج تم اختياره في تلك المجموعة لقياس خطأ التعميم النهائي. ومع ذلك، يمكنك المضي قدمًا وبدلاً من الحصول على عينة اختبار واحدة، يمكنك الحصول على حلقة السيرة الذاتية الخارجية، مما يقودنا إلى التحقق المتبادل المتداخل.
من الناحية الفنية، *lightbgm.cv()* يسمح لك فقط بتقييم الأداء على تقسيم k-fold مع معلمات النموذج الثابتة. لضبط المعلمات الفائقة، ستحتاج إلى تشغيلها في حلقة توفر معلمات مختلفة وإعادة ترميز متوسط ​​الأداء لاختيار أفضل مجموعة معلمات. بعد اكتمال الحلقة. تختلف هذه الواجهة عن *sklearn*، والتي توفر لك وظائف كاملة للقيام بتحسين المعلمات الفائقة في حلقة السيرة الذاتية. شخصيًا، **أوصي باستخدام sklearn-API الخاص بـlightgbm**. إنه مجرد غلاف حول وظيفة *lightgbm.train()* الأصلية، وبالتالي فهو ليس أبطأ. ولكنه يسمح لك باستخدام المجموعة الكاملة من مجموعة أدوات *sklearn*، مما يجعل حياتك أسهل بكثير."
