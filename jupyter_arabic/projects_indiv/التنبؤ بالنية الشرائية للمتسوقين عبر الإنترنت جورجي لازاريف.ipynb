## المعالجة المسبقة للبيانات 


```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```


أولاً، سنقوم بالتحويل من السمات المنطقية إلى النوع الثنائي


```python
df[['Weekend','Revenue']]=df[['Weekend','Revenue']].apply(lambda x:x.astype(int))
```


بعد ذلك، سيتم حذف الحالات التي تحتوي على *VisitorType* كـ "Returning Visitor":


```python
df=df.drop(df.loc[df.VisitorType=='Other'].index)
```

```python
df.shape
```

```python
vt=df.VisitorType.map({'New_Visitor':0,'Returning_Visitor':1})
```


ستكون هناك عمليات معالجة مسبقة مختلفة أثناء عملنا مع نماذج تعتمد على أساليب مختلفة.
بالنسبة للانحدار اللوجستي، من الجيد توسيع نطاق ميزاتنا الرقمية وإجراء تشفير One_Hot على الميزات الفئوية. لتجنب تسرب البيانات، سيتم إجراء القياس بعد تقسيم البيانات. أما بالنسبة لـ OHE وLabelEncoding (للنماذج القائمة على الشجرة)، أفترض أنه يمكننا القيام بذلك قبل التقسيم لأننا نعرف نطاقًا لجميع القيم الممكنة للميزات الفئوية، لذلك لا يوجد تسرب للبيانات لمنعه.


```python
dummies=pd.concat([pd.get_dummies(df.Month,drop_first=True),
                   pd.get_dummies(df.Browser,drop_first=True,prefix='Browser'),
                   pd.get_dummies(df.Region,drop_first=True,prefix='Region'),
                   pd.get_dummies(df.OperatingSystems,drop_first=True,prefix='OS'),
                   pd.get_dummies(df.TrafficType,drop_first=True,prefix='TT')],axis=1)
```

```python
dummies.shape
```

```python
target=df.Revenue
```


*feats_logreg* سيحتوي على كافة ميزات الانحدار اللوجستي. *feats_tb* مخصص للنماذج المبنية على الأشجار 


```python
feats_logreg=pd.concat([df[num_feats],dummies,df['Weekend'],vt],axis=1)
```

```python
feats_logreg.shape
```


الآن سنقوم بتقسيم بياناتنا.  *طبقية* تستخدم بسبب عدم التوازن في الفصول الدراسية.

```python
X_train_logreg_,X_test_logreg_,y_train_logreg,y_test_logreg=train_test_split(feats_logreg,
                                        target,test_size=0.3,random_state=17,stratify=target)
```


دعونا نتحقق من توزيع الفصول في مجموعات التدريب والاختبار:


```python
plt.subplot(121)
y_train_logreg.value_counts(normalize=True).plot.bar()
plt.subplot(122)
y_test_logreg.value_counts(normalize=True).plot.bar()
```


نعم، هذا يبدو صحيحا.
سيتم الآن تقسيم مجموعة الاختبار إلى مجموعتين بنفس الحجم: واحدة للتحقق من الصحة والأخرى للاختبار النهائي. لن نختبر نماذجنا في النموذج الثاني حتى النهاية.


```python
X_valid_logreg_,X_test_logreg_,y_valid_logreg,y_test_logreg=train_test_split(X_test_logreg_,
                                                            y_test_logreg,test_size=0.5,random_state=17)
```

```python
scaler=StandardScaler()
```

```python
X_train_logreg=X_train_logreg_.copy(deep=True)
X_valid_logreg=X_valid_logreg_.copy(deep=True)
X_test_logreg=X_test_logreg_.copy(deep=True)


X_train_logreg[num_feats]=scaler.fit_transform(X_train_logreg[num_feats])
X_valid_logreg[num_feats]=scaler.transform(X_valid_logreg[num_feats])
X_test_logreg[num_feats]=scaler.transform(X_valid_logreg[num_feats])
```

```python
X_train_logreg.shape
```


بالنسبة للنماذج المستندة إلى الشجرة، ستتضمن المعالجة المسبقة لدينا فقط LabelEncoding لـ *month*. الميزات الفئوية الأخرى باستثناء الميزة المنطقية تم ترميزها بالفعل. التقسيم إلى مجموعات ***3*** هو نفسه.


```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
```

```python
feats_tb=pd.concat([df[num_feats],df[['Weekend','TrafficType','OperatingSystems','Browser','Region']],vt],axis=1)
```

```python
feats_tb['month_enc']=le.fit_transform(df.Month)
```

```python
feats_tb.shape
```


## رسم منحنيات التدريب والتحقق من الصحة


```python
from sklearn.model_selection import learning_curve,validation_curve
```

```python
def plot_learning_curve(estimator, title, X, y, ylim=None, cv=None,
                        n_jobs=1, train_sizes=np.linspace(.1, 1.0, 5)):
    
    plt.figure()
    plt.title(title)
    if ylim is not None:
        plt.ylim(*ylim)
    plt.xlabel("Training examples")
    plt.ylabel("ROC AUC")
    train_sizes, train_scores, test_scores = learning_curve(
        estimator, X, y, cv=cv, n_jobs=n_jobs, train_sizes=train_sizes, scoring='roc_auc')
    train_scores_mean = np.mean(train_scores, axis=1)
    train_scores_std = np.std(train_scores, axis=1)
    test_scores_mean = np.mean(test_scores, axis=1)
    test_scores_std = np.std(test_scores, axis=1)
    plt.grid()

    plt.fill_between(train_sizes, train_scores_mean - train_scores_std,
                     train_scores_mean + train_scores_std, alpha=0.1,
                     color="r")
    plt.fill_between(train_sizes, test_scores_mean - test_scores_std,
                     test_scores_mean + test_scores_std, alpha=0.1, color="g")
    plt.plot(train_sizes, train_scores_mean, 'o-', color="r",
             label="Training score")
    plt.plot(train_sizes, test_scores_mean, 'o-', color="g",
             label="Cross-validation score")

    plt.legend(loc="best")
    return plt
```

```python
plt.figure(figsize=(10, 7))
plot_learning_curve(lr, 'Logistic Regression', X_train_logreg_wn, y_train_logreg, cv=skf, n_jobs=-1);
```

نلاحظ شيئًا جيدًا - تميل منحنيات التحقق من الصحة للتدريب إلى التقارب. لا يوجد نقص.


```python
plt.figure(figsize=(10,7))
param_range=np.array([0.01, 0.05, 0.1, 0.25, 0.5, 1, 5])
train_scores, test_scores = validation_curve(lr, X_train_logreg_wn, y_train_logreg, param_name="C",
                                             param_range=param_range, cv=skf, scoring="roc_auc", n_jobs=-1)
train_scores_mean = np.mean(train_scores, axis=1)
train_scores_std = np.std(train_scores, axis=1)
test_scores_mean = np.mean(test_scores, axis=1)
test_scores_std = np.std(test_scores, axis=1)

plt.title("Validation Curve")
plt.xlabel("C")
plt.ylabel("ROC AUC")
#plt.ylim(0.0, 1.1)
lw = 2
plt.plot(param_range, train_scores_mean, label="Training score",
             color="darkorange", lw=lw)
plt.fill_between(param_range, train_scores_mean - train_scores_std,
                 train_scores_mean + train_scores_std, alpha=0.2,
                 color="darkorange", lw=lw)
plt.plot(param_range, test_scores_mean, label="Cross-validation score",
             color="navy", lw=lw)
plt.fill_between(param_range, test_scores_mean - test_scores_std,
                 test_scores_mean + test_scores_std, alpha=0.2,
                 color="navy", lw=lw)
plt.legend(loc="best")
plt.show()
```


لذا فإن نتائجنا متسقة إلى حد كبير على طول النطاق *C*. على الرغم من ارتفاع ROC AUC بشكل كبير مع زيادة *C* من *0* إلى *~0.4*

إذن هذه هي النهاية. استخدمنا ثلاثة نماذج للبيانات غير المتوازنة وأظهر كل منها درجة عالية جدًا من ROC AUC. يمكننا أن نلاحظ أن النتيجة في المجموعة الصالحة قد تغيرت وفقًا لـ *cross_val_score*. 
سألتزم بالانحدار اللوجستي. يمكنك ملاحظة ذلك من خلال الأجزاء الأخيرة من المشروع. علاوة على ذلك، أشعر أنه يمكن القيام بالمزيد في حالة RandomForest وXGBoost.
عند دمجه مع وحدة أخرى لتحديد احتمالية مغادرة الزائر للموقع (ذكرت ذلك في البداية)، يمكن للشركة استخدام نموذج التصنيف هذا لإظهار العروض الخاصة الفردية لهؤلاء الزوار قبل مغادرة موقع المتجر.
- تم جمع البيانات خلال عام واحد ولدينا خاصية *الشهر*. لذلك لدينا في الأساس نوع من الجدول الزمني. ومع ذلك فقد استخدمت *الشهر* فقط كميزة فئوية دون أي سياق زمني. ولست متأكدًا من أن العكس سيكون له أي معنى.- ساعد اختيار الميزات والهندسة قليلاً على زيادة نقاط LogReg. باستثناء تجربة التصورات، لا أرى طريقة أخرى لتحسين هذه العملية. لم يساعد الإفراط في أخذ العينات (لكنه لم يتفاقم أيضًا).
- أما بالنسبة لاختيار نطاق المعلمات: فأنا جديد جدًا في هذا الأمر، لذا ليس لدي خبرة كبيرة في ضبط النماذج. لذلك أقوم فقط بجعل نطاق القيم قريبًا من الافتراضي أو فقط حدد نطاقًا واسعًا ولكن صغيرًا أولاً ثم قم بالتكرار على كمية أكبر من القيم في المناطق الأصغر المختارة التي تبدو مثالية. شبكات المعلمات المستخدمة في الكود أعلاه هي ما توصلت إليه بعد مرور بعض الوقت. يحتاج ضبط المعلمات الفائقة بالتأكيد إلى نهج أكثر حكمة.
- إنها أيضًا فكرة مثيرة للاهتمام إنشاء نظام تعليمي عبر الإنترنت يمكن تحديثه مع كل مثال جديد.
