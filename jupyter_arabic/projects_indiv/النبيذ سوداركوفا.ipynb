**الحموضة الثابتة**<br> معظم الأحماض المتضمنة في النبيذ أو الأحماض الثابتة أو غير المتطايرة (لا تتبخر بسهولة).<br>النوع: رقمي<br><br>
**الحموضة المتطايرة**<br> كمية حمض الأسيتيك في النبيذ، والتي عند مستويات عالية جدًا يمكن أن تؤدي إلى طعم الخل الكريه.<br>النوع: رقمي<br><br>
**حمض الستريك**<br> يمكن أن يضيف حمض الستريك بكميات صغيرة "نضارة" ونكهة إلى النبيذ.<br>النوع: رقمي<br><br>
**السكر المتبقي**<br> كمية السكر المتبقية بعد توقف التخمير، من النادر العثور على نبيذ يحتوي على أقل من 1 جرام/لتر، ويعتبر النبيذ الذي يحتوي على أكثر من 45 جرامًا/لتر حلوًا.<br>النوع: رقمي<br><br>
**الكلوريدات**<br> كمية الملح في النبيذ.<br>النوع: رقمي<br><br>
**ثاني أكسيد الكبريت الحر**<br> الشكل الحر لثاني أكسيد الكبريت موجود في حالة توازن بين ثاني أكسيد الكبريت الجزيئي (كغاز مذاب) وأيون ثنائي الكبريتيت؛ فهو يمنع نمو الميكروبات وأكسدة النبيذ.<br>النوع: رقمي<br><br>
**إجمالي ثاني أكسيد الكبريت**<br> كمية الأشكال الحرة والمقيدة من S02؛ في التركيزات المنخفضة، لا يمكن اكتشاف ثاني أكسيد الكبريت في الغالب في النبيذ، ولكن عند تركيزات ثاني أكسيد الكبريت الحرة التي تزيد عن 50 جزء في المليون، يصبح ثاني أكسيد الكبريت واضحًا في الأنف ومذاق النبيذ.<br>النوع: رقمي<br><br>
**الكثافة**<br> كثافة الماء قريبة من كثافة الماء اعتمادًا على نسبة الكحول والسكر.<br>النوع: رقمي<br><br>

**الكبريتات**<br> مادة مضافة للنبيذ يمكن أن تساهم في زيادة مستويات غاز ثاني أكسيد الكبريت (S02)، والتي تعمل كمضاد للميكروبات ومضاد للأكسدة.<br>النوع: رقمي<br><br>
**الكحول**<br> النسبة المئوية لمحتوى الكحول في النبيذ.<br>النوع: رقمي<br><br>
**الجودة**<br> متغير الإخراج (استنادًا إلى البيانات الحسية، النتيجة بين 0 و10).<br>النوع: رقمي<br><br>

### <center> 8. التحقق من الصحة وتعديل المعلمات الفائقة للنموذج


```python
# Start Modeling
# Split Training and Testing data (within training data file only)
# Hold out method validation

from sklearn.model_selection import train_test_split

X_train = wine.iloc[:,1:len(wine.columns)-2] # this represents the input Features
Y_train = wine.loc[:,'quality']


# Scaling features (only feature NOT observation)
from sklearn.preprocessing import MinMaxScaler, StandardScaler
# Scaling - brings to range (0 to 1)
ScaleFn = MinMaxScaler()
X_Scale = ScaleFn.fit_transform(X_train)
# Standardise - brings to Zero mean, Unit variance
ScaleFn = StandardScaler()
X_Strd = ScaleFn.fit_transform(X_train)
```


#### تابع تنفيذ PCA:


```python
from sklearn.decomposition import PCA
pca = PCA()
x_pca = pca.fit_transform(X_Strd)
```


ارسم الرسم البياني للعثور على المكونات الرئيسية


```python
plt.figure(figsize=(10,10))
plt.plot(np.cumsum(pca.explained_variance_ratio_), 'ro-')
plt.grid()
```


كما هو موضح في الرسم البياني، يمكننا أن نرى أن 7 مكونات رئيسية مسؤولة عن 90% من التباين في البيانات. <br>
يجب علينا اختيار المكونات السبعة الأولى لتوقعاتنا.


```python
pca_new = PCA(n_components=8)
x_new = pca_new.fit_transform(X_Strd)
print(x_new)
```

```python
test_size = .30
seedNo = 11


X_train,X_test,Y_train,Y_test = train_test_split(x_new,Y_train,test_size = test_size, random_state = seedNo)

print("train X", X_train.shape)
print("train Y", Y_train.shape)
print("test X", X_test.shape)
print("test Y", Y_test.shape)

# Choose algorithm and train
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from xgboost import XGBClassifier


from sklearn.model_selection import KFold
from sklearn.model_selection import cross_val_score

mymodel = []
mymodel.append(('LogReg', LogisticRegression()))
mymodel.append(('KNN', KNeighborsClassifier()))
mymodel.append(('DeciTree', DecisionTreeClassifier()))
mymodel.append(('RandForest', RandomForestClassifier()))
mymodel.append(('SVM', SVC()))
mymodel.append(('XGBoost', XGBClassifier()))



All_model_result = []
All_model_name = []
for algoname, algorithm in mymodel:    
    kfoldFn = KFold(n_splits = 11, random_state = seedNo)
    Eval_result = cross_val_score(algorithm, X_train, Y_train, cv = kfoldFn, scoring = 'accuracy')
    
    All_model_result.append(Eval_result)
    All_model_name.append(algoname)
    print("Modelname and Model accuracy:", algoname, 100*Eval_result.mean(),"%")
```


## غابة عشوائية
