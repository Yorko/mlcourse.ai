# شرح بديهي لخوارزمية تعظيم التوقعات



تُستخدم النماذج الاحتمالية، مثل نماذج ماركوف المخفية أو الشبكات الافتراضية، بشكل شائع لنمذجة البيانات البيولوجية. كثيرا
يمكن أن تعزى شعبيتها إلى وجود إجراءات فعالة وقوية لتعلم المعلمات من الملاحظات. 
ومع ذلك، في كثير من الأحيان، تكون البيانات الوحيدة المتاحة لتدريب النموذج الاحتمالي غير كاملة. يمكن أن تحدث القيم المفقودة، على سبيل المثال، في التشخيص الطبي، حيث يتضمن تاريخ المريض بشكل عام نتائج من مجموعة محدودة من الاختبارات.
وبدلاً من ذلك، في تجميع التعبير الجيني، تنشأ البيانات غير المكتملة من الإغفال المتعمد لتعيينات الجينات إلى المجموعة في النموذج الاحتمالي. 
**تتيح خوارزمية تعظيم التوقعات تقدير المعلمات في النماذج الاحتمالية ذات البيانات غير المكتملة.**



### ما هو EM؟
- تعد خوارزمية تعظيم التوقعات (EM) طريقة للعثور على **تقديرات الاحتمالية القصوى** لمعلمات النموذج عندما تكون بياناتك غير مكتملة، أو تحتوي على نقاط بيانات مفقودة، أو تحتوي على متغيرات كامنة (مخفية) غير ملحوظة.
- إنها **طريقة تكرارية لتقريب دالة الاحتمالية القصوى.**
- على الرغم من أن تقدير الاحتمالية القصوى يمكن أن يجد النموذج "الأفضل ملاءمة" لمجموعة من البيانات، إلا أنه لا يعمل بشكل جيد مع مجموعات البيانات غير المكتملة.
- يمكن لخوارزمية EM الأكثر تعقيدًا العثور على معلمات النموذج حتى لو كانت لديك بيانات مفقودة. وهو يعمل عن طريق اختيار قيم عشوائية لنقاط البيانات المفقودة، واستخدام تلك التخمينات لتقدير مجموعة ثانية من البيانات. يتم استخدام القيم الجديدة لإنشاء تخمين أفضل للمجموعة الأولى، وتستمر العملية حتى تتقارب الخوارزمية عند نقطة ثابتة.



### كيف يختلف EM عن MLE؟على الرغم من أن تقدير الاحتمال الأقصى (MLE) وEM يمكنهما العثور على المعلمات "الأفضل ملاءمة"، إلا أن كيفية العثور على النماذج مختلفة تمامًا.
- يقوم MLE بتجميع كافة البيانات أولاً ثم يستخدم تلك البيانات لإنشاء النموذج الأكثر احتمالاً.
- يقوم EM بتخمين المعلمات أولاً - مع الأخذ في الاعتبار البيانات المفقودة - ثم يقوم بتعديل النموذج ليناسب التخمينات والبيانات المرصودة.



### كيف يعمل EM؟
الخطوات الأساسية للخوارزمية هي:
1. يتم إجراء تخمين أولي لمعلمات النموذج ويتم إنشاء التوزيع الاحتمالي. يُطلق على هذا أحيانًا اسم "الخطوة الإلكترونية" للتوزيع "المتوقع".
2. يتم إدخال البيانات التي تمت ملاحظتها حديثًا في النموذج.
3. تم تعديل التوزيع الاحتمالي من الخطوة الإلكترونية ليشمل البيانات الجديدة. يُطلق على هذا أحيانًا اسم "الخطوة M".
4. يتم تكرار الخطوات من 2 إلى 4 حتى يتم الوصول إلى الاستقرار (أي التوزيع الذي لا يتغير من الخطوة E إلى الخطوة M).



سيحاول هذا استخدام قدر أقل من الرياضيات وإعطاء فكرة عن كيفية تقارب EM.



### تعظيم التوقعات باستخدام مثال على عملتين: 



- لدينا عملتان متحيزتان ولا نعرف احتمال الحصول على وجه لكل قطعة نقدية. افترض $\theta_A$ و$\theta{_B}$
- نختار عملة معدنية واحدة بشكل عشوائي ونرميها 10 مرات ونكرر هذه التجربة خمس مرات.
- هناك طريقة بسيطة لتقدير $\theta_A$ و$\theta_B$ وهي إرجاع النسب الملحوظة لصور كل عملة:
  $$\hat{\theta_A} = \frac{Number ~of~ heads ~using ~coin~ A}{total ~number~ flips~ using~ coin ~A }$$ 
  
  $$\hat{\theta_B} = \frac{Number~ of~ heads ~using ~coin ~B}{total number~ flips ~using~ coin~ B }$$
  
  
هذا التخمين البديهي معروف في الواقع في الأدبيات الإحصائية باسم <font color ='red'>تقدير الاحتمال الأقصى </font>. **كما هو موضح في الشكل (أ) أدناه**



<img src="@@KEEP_00000@@ width="500" height="500">


**مشكلة أكثر تعقيدًا**:
- إذا كنا نعرف في التجربة عدد الرأس فقط، ولكن العملة المستخدمة في التجربة غير معروفة.
- كيفية تقدير احتمالات الرأس لكل عملة؟ يمكن حل هذه المشكلة باستخدام <font color='red'>تعظيم التوقعات.</font>
الخطوات موضحة أعلاه **الشكل (ب)**:
    1. افترض بعض احتمالية ظهور الرأس لكلتا القطعتين.
    2. إنشاء التوزيع الاحتمالي (باستخدام التوزيع ذي الحدين). (التوقع)
        مثال: في التجربة الثانية لدينا 9 رؤوس وذيل واحد.
        
  $$p(A) = \theta_A^9(1-\theta_A)^{10-9} \approx 0.004$$
  
  $$p(B) = \theta_B^9(1-\theta_B)^{10-9} \approx 0.001$$
       
       هذا يعطينا:
  $$\frac{0.004}{0.004+0.001}=0.8$$
  
  $$\frac{0.001}{0.004+0.001}=0.2$$
    
  وبالمثل يمكننا حساب توزيع العملة A والعملة B لكل تجربة.
  
    3 أ. بناءً على التوزيع، قم بحساب الاحتمالية المقدرة الجديدة. (التعظيم)
    3ب. كرر الخطوتين أعلاه.
    4. أعط الاحتمال المقدر بعد التقارب.

### تعظيم التوقعات للتوزيع الرقمي



في هذه التجربة سنحاول إيجاد الوسط الحسابي والتوزيع الطبيعي لاثنين.



#### يتيح إنشاء بعض البيانات


```python
import warnings

import numpy as np
from scipy import stats

warnings.filterwarnings("ignore")

np.random.seed(110)  # for reproducible random results

# set parameters
red_mean = 3
red_std = 0.8

blue_mean = 7
blue_std = 2

# draw 20 samples from normal distributions with red/blue parameters
red = np.random.normal(red_mean, red_std, size=20)
blue = np.random.normal(blue_mean, blue_std, size=20)

both_colours = np.sort(np.concatenate((red, blue)))
```

```python
%matplotlib inline
import matplotlib.mlab as mlab
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (20, 3)
```

```python
y = [1 for i in range(len(red))]
plt.plot(red, y, "ro", blue, y, "bo")
plt.title("Data(With Label)")
```


**إذا تم إعطاء التسمية، فهي مشكلة سهلة ويمكننا حساب المتوسط والقياس باستخدام MLE**


```python
print("Red Mean is {}".format(np.mean(red)))
print("Red Std is {}".format(np.std(red)))
print("")
print("Blue Mean is {}".format(np.mean(blue)))
print("Blue Std is {}".format(np.std(blue)))
```


** إذا لم يتم إعطاء التسمية. استخدم م** 


```python
y = [1 for i in range(len(both_colours))]
plt.plot(both_colours, y, "mo")
plt.title("Data(Without Label)")
```


## الخطوة 1 (التخمين)


```python
# estimates for the mean
red_mean_guess = 1.1
blue_mean_guess = 9

# estimates for the standard deviation
red_std_guess = 2
blue_std_guess = 1.7
```


### سوف تبدو المؤامرة


```python
x = np.linspace(-3, 12, 100)
plt.plot(
    x,
    mlab.normpdf(x, red_mean_guess, red_std_guess),
    "r",
    x,
    mlab.normpdf(x, blue_mean_guess, blue_std_guess),
    "b",
    alpha=1,
)
plt.ylim(-0.02, 0.28)
y = [0.001 for i in range(len(both_colours))]
plt.plot(both_colours, y, "mo")
ymaxr = max(mlab.normpdf(x, red_mean_guess, red_std_guess)) / 0.28
ymaxb = max(mlab.normpdf(x, blue_mean_guess, blue_std_guess)) / 0.28
plt.axvline(red_mean_guess, color="r", linestyle="--", ymin=0.067, ymax=ymaxr)
plt.axvline(blue_mean_guess, color="b", linestyle="--", ymin=0.067, ymax=ymaxb);
```


## الخطوة 2.
### احتمالية كل نقطة تحت التخمين الحالي


```python
likelihood_of_red = stats.norm(red_mean_guess, red_std_guess).pdf(both_colours)
likelihood_of_blue = stats.norm(blue_mean_guess, blue_std_guess).pdf(both_colours)
```

```python
print(
    "likelihood of point {} being red {}".format(both_colours[1], likelihood_of_red[1])
)
print(
    "likelihood of point {} being blue {}".format(
        both_colours[1], likelihood_of_blue[1]
    )
)
```


## الخطوة 3
### تحويل الاحتمالية إلى أوزان


```python
likelihood_total = likelihood_of_red + likelihood_of_blue

red_weight = likelihood_of_red / likelihood_total
blue_weight = likelihood_of_blue / likelihood_total
```


## الخطوة 4
### باستخدام التقديرات الحالية والأوزان الجديدة يمكننا حساب تقديرات جديدة


الجزء الأساسي من الحدس هو أنه كلما زاد وزن اللون على نقطة البيانات، زاد تأثير نقطة البيانات على التقديرات التالية لمعلمات ذلك اللون. وهذا له تأثير "سحب" المعلمات في الاتجاه الصحيح.


```python
def estimate_mean(data, weight):
    return np.sum(data * weight) / np.sum(weight)


def estimate_std(data, weight, mean):
    variance = np.sum(weight * (data - mean) ** 2) / np.sum(weight)
    return np.sqrt(variance)


# new estimates for standard deviation
new_blue_std_guess = estimate_std(both_colours, blue_weight, blue_mean_guess)
new_red_std_guess = estimate_std(both_colours, red_weight, red_mean_guess)

# new estimates for mean
new_red_mean_guess = estimate_mean(both_colours, red_weight)
new_blue_mean_guess = estimate_mean(both_colours, blue_weight)
```

```python
print(
    "Red Mean New: {} and Blue Mean New: {}".format(
        new_red_mean_guess, new_blue_mean_guess
    )
)
```


## الخطوة 5 (تكرار)
### لنعمل لمدة 5 تكرارات


```python
# estimates for the mean
red_mean_guess = 1.1
blue_mean_guess = 9

# estimates for the standard deviation
red_std_guess = 2
blue_std_guess = 1.7


y = [0.001 for i in range(len(both_colours))]
plt.plot(both_colours, y, "mo")
plt.ylim(-0.02, 0.40)

for i in range(5):
    likelihood_of_red = stats.norm(red_mean_guess, red_std_guess).pdf(both_colours)
    likelihood_of_blue = stats.norm(blue_mean_guess, blue_std_guess).pdf(both_colours)

    likelihood_total = likelihood_of_red + likelihood_of_blue

    red_weight = likelihood_of_red / likelihood_total
    blue_weight = likelihood_of_blue / likelihood_total

    blue_std_guess = estimate_std(both_colours, blue_weight, blue_mean_guess)
    red_std_guess = estimate_std(both_colours, red_weight, red_mean_guess)

    red_mean_guess = estimate_mean(both_colours, red_weight)
    blue_mean_guess = estimate_mean(both_colours, blue_weight)

    print(red_mean_guess, blue_mean_guess)

    x = np.linspace(-3, 12, 100)
    ymaxr = max(mlab.normpdf(x, red_mean_guess, red_std_guess)) / 0.39
    ymaxb = max(mlab.normpdf(x, blue_mean_guess, blue_std_guess)) / 0.39

    plt.plot(
        x,
        mlab.normpdf(x, red_mean_guess, red_std_guess),
        "r",
        x,
        mlab.normpdf(x, blue_mean_guess, blue_std_guess),
        "b",
        alpha=0.2 * i,
    )
    plt.axvline(
        red_mean_guess, color="r", linestyle="--", ymin=0.067, ymax=ymaxr, alpha=0.2 * i
    )
    plt.axvline(
        blue_mean_guess,
        color="b",
        linestyle="--",
        ymin=0.067,
        ymax=ymaxb,
        alpha=0.2 * i,
    )
```


وكما نرى فإن المتوسط المقدر يقترب من المتوسط الحقيقي.


```python
# estimates for the mean
red_mean_guess = 1.1
blue_mean_guess = 9

# estimates for the standard deviation
red_std_guess = 2
blue_std_guess = 1.7

for i in range(20):
    likelihood_of_red = stats.norm(red_mean_guess, red_std_guess).pdf(both_colours)
    likelihood_of_blue = stats.norm(blue_mean_guess, blue_std_guess).pdf(both_colours)

    likelihood_total = likelihood_of_red + likelihood_of_blue

    red_weight = likelihood_of_red / likelihood_total
    blue_weight = likelihood_of_blue / likelihood_total

    blue_std_guess = estimate_std(both_colours, blue_weight, blue_mean_guess)
    red_std_guess = estimate_std(both_colours, red_weight, red_mean_guess)

    red_mean_guess = estimate_mean(both_colours, red_weight)
    blue_mean_guess = estimate_mean(both_colours, blue_weight)

y = [0.001 for i in range(len(both_colours))]
plt.plot(both_colours, y, "mo")
plt.ylim(-0.02, 0.50)

x = np.linspace(-3, 12, 100)
ymaxr = max(mlab.normpdf(x, red_mean_guess, red_std_guess)) / 0.5
ymaxb = max(mlab.normpdf(x, blue_mean_guess, blue_std_guess)) / 0.5

plt.plot(
    x,
    mlab.normpdf(x, red_mean_guess, red_std_guess),
    "r",
    x,
    mlab.normpdf(x, blue_mean_guess, blue_std_guess),
    "b",
    alpha=1,
)
plt.axvline(red_mean_guess, color="r", linestyle="--", ymin=0.06, ymax=ymaxr, alpha=1)
plt.axvline(blue_mean_guess, color="b", linestyle="--", ymin=0.06, ymax=ymaxb, alpha=1)
```
