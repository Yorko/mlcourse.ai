**1. التقييس**



نتيجة التوحيد (أو تطبيع درجة Z) هي أنه سيتم إعادة قياس الميزات بهذه الطريقة 
أنه سيكون لديهم خصائص التوزيع الطبيعي القياسي بمتوسط m = 0، والانحراف المعياري d = 0.
يتم حساب الدرجات القياسية (وتسمى أيضًا الدرجات z) للعينات على النحو التالي:
**ض=(س−م)/د**
إن توحيد الميزات بحيث تتمركز حول 0 مع انحراف معياري قدره 1 ليس مهمًا فقط 
إذا كنا نقارن القياسات التي تحتوي على وحدات مختلفة، ولكنه أيضًا مطلب عام للعديد من التعلم الآلي 
خوارزميات. بشكل بديهي، يمكننا التفكير في النسب المتدرج كمثال بارز (خوارزمية تحسين تستخدم غالبًا في
الانحدار اللوجستي، SVMs، الإدراك الحسي، الشبكات العصبية وما إلى ذلك)؛ مع وجود ميزات بمقاييس مختلفة وأوزان معينة 
قد يتم تحديثه بشكل أسرع من الآخرين نظرًا لأن قيم الميزات xj تلعب دورًا في تحديثات الوزن.
تشمل الأمثلة البديهية الأخرى خوارزميات K-Nearest Neighbor وخوارزميات التجميع التي تستخدم، على سبيل المثال، مقاييس المسافة الإقليدية - في الواقع، ربما يكون المصنف القائم على الشجرة هو المصنف الوحيد الذي لا يُحدث فيه قياس الميزات فرقًا.في الواقع، عائلة الخوارزميات الوحيدة التي يمكن أن أفكر في كونها ثابتة الحجم هي الأساليب المبنية على الشجرة. لنأخذ خوارزمية شجرة القرار العامة CART. دون الخوض في الكثير من العمق فيما يتعلق بكسب المعلومات ومقاييس الشوائب، يمكننا أن نفكر في القرار على أنه "هل الميزة x_i >= some_val؟" بشكل بديهي، يمكننا أن نرى أنه لا يهم حقًا على أي مقياس تكون هذه الميزة (سم، فهرنهايت، مقياس موحد - لا يهم حقًا).
بعض الأمثلة على الخوارزميات التي يكون فيها قياس الميزات مهمًا هي:
-k- أقرب الجيران بمقياس المسافة الإقليدية إذا كنت تريد أن تساهم جميع الميزات بالتساوي
-k-يعني (انظر أقرب الجيران)
- الانحدار اللوجستي، وSVMs، والإدراك الحسي، والشبكات العصبية وما إلى ذلك. إذا كنت تستخدم التحسين القائم على النسب المتدرج/الصعود، وإلا فسيتم تحديث بعض الأوزان بشكل أسرع بكثير من غيرها
- التحليل التمييزي الخطي، تحليل المكون الرئيسي، تحليل المكون الرئيسي للنواة لأنك تريد العثور على اتجاهات لتعظيم التباين (تحت القيود التي تكون فيها تلك الاتجاهات/المتجهات الذاتية/المكونات الرئيسية متعامدة)؛ تريد الحصول على ميزات على نفس المقياس نظرًا لأنك تركز على المتغيرات على "مقاييس قياس أكبر" أكثر. هناك العديد من الحالات التي لا يمكنني سردها هنا... أوصيك دائمًا بالتفكير في الخوارزمية وما تفعله، ومن ثم يصبح من الواضح ما إذا كنا نريد توسيع نطاق ميزاتك أم لا.بالإضافة إلى ذلك، نود أيضًا أن نفكر فيما إذا كنا نريد "توحيد" أو "تطبيع" (هنا: القياس إلى النطاق [0, 1]) بياناتنا. تفترض بعض الخوارزميات أن بياناتنا تتمركز عند 0. على سبيل المثال، إذا قمنا بتهيئة أوزان إدراك صغير متعدد الطبقات مع وحدات تنشيط تانه إلى 0 أو قيم عشوائية صغيرة تتمركز حول الصفر، فإننا نريد تحديث أوزان النموذج "بالتساوي". كقاعدة عامة، أود أن أقول: عندما تكون في شك، ما عليك سوى توحيد البيانات، فلن يضر ذلك.



**2. تحجيم الحد الأدنى-الأقصى**



هناك نهج بديل لتطبيع (أو توحيد النقاط Z) وهو ما يسمى مقياس Min-Max (غالبًا ما يسمى أيضًا ببساطة "التطبيع" - وهو سبب شائع للغموض).
في هذا الأسلوب، يتم قياس البيانات إلى نطاق ثابت - عادة من 0 إلى 1.
تكلفة وجود هذا النطاق المحدد - على النقيض من التوحيد القياسي - هي أننا سننتهي بانحرافات معيارية أصغر، والتي يمكن أن تقمع تأثير القيم المتطرفة.
عادةً ما يتم إجراء مقياس Min-Max عبر المعادلة التالية:
**Xnorm=(X−Xmin)/(Xmax−Xmin)**



"التوحيد القياسي أو تحجيم Min-Max؟" - لا توجد إجابة واضحة على هذا السؤال: فالأمر يعتمد حقًا على التطبيق.
على سبيل المثال، في التحليلات التجميعية، قد يكون التوحيد حاسمًا بشكل خاص من أجل مقارنة أوجه التشابه بين الميزات بناءً على مقاييس مسافة معينة. مثال بارز آخر هو تحليل المكونات الرئيسية، حيث نفضل عادة التوحيد القياسي على مقياس Min-Max، لأننا مهتمون بالمكونات التي تزيد التباين إلى الحد الأقصى.ومع ذلك، هذا لا يعني أن مقياس Min-Max غير مفيد على الإطلاق! أحد التطبيقات الشائعة هو معالجة الصور، حيث يجب ضبط شدة البكسل لتتناسب مع نطاق معين (على سبيل المثال، من 0 إلى 255 لنطاق ألوان RGB). أيضًا، تتطلب خوارزمية الشبكة العصبية النموذجية بيانات بمقياس 0-1.



دعونا ننظر، كيف يمكننا أن نفعل التوحيد أو القياس Min-Max. 
بالطبع، يمكننا الاستفادة من إمكانات التوجيه الخاصة بـ NumPy لحساب درجات z للتوحيد القياسي وتطبيع البيانات باستخدام المعادلات التي تم ذكرها في الأقسام السابقة. ومع ذلك، هناك طريقة أكثر ملاءمة باستخدام وحدة المعالجة المسبقة من إحدى مكتبات التعلم الآلي مفتوحة المصدر في Python scikit-learn.
بالنسبة للأمثلة والمناقشة التالية، سنلقي نظرة على مجموعة بيانات "Wine" المجانية المودعة في مستودع التعلم الآلي الخاص بشركة UCI
(http://archive.ics.uci.edu/ml/datasets/Wine).
تتكون مجموعة بيانات النبيذ من 3 فئات مختلفة حيث يتوافق كل صف مع عينة نبيذ معينة.
يتم إدراج تسميات الفئات (1، 2، 3) في العمود الأول، وتتوافق الأعمدة من 2 إلى 14 مع 13 سمة (ميزات) مختلفة


```python
import numpy as np
import pandas as pd

df = pd.io.parsers.read_csv(
    "https://raw.githubusercontent.com/rasbt/pattern_classification/master/data/wine_data.csv",
    header=None,
    usecols=[0, 1, 2],
)

df.columns = ["Class label", "Alcohol", "Malic acid"]

df.head()
```


كما نرى في الجدول أعلاه، يتم قياس خصائص الكحول (النسبة المئوية/الحجم) وحمض الماليك (جم/لتر) بمقاييس مختلفة،
لذلك يعد قياس الميزات ضروريًا قبل أي مقارنة أو مجموعة من هذه البيانات.


```python
from sklearn import preprocessing

std_scale = preprocessing.StandardScaler().fit(df[["Alcohol", "Malic acid"]])
df_std = std_scale.transform(df[["Alcohol", "Malic acid"]])

minmax_scale = preprocessing.MinMaxScaler().fit(df[["Alcohol", "Malic acid"]])
df_minmax = minmax_scale.transform(df[["Alcohol", "Malic acid"]])
print(
    "Mean after standardization:\nAlcohol={:.2f}, Malic acid={:.2f}".format(
        df_std[:, 0].mean(), df_std[:, 1].mean()
    )
)
print(
    "\nStandard deviation after standardization:\nAlcohol={:.2f}, Malic acid={:.2f}".format(
        df_std[:, 0].std(), df_std[:, 1].std()
    )
)
```

```python
print(
    "Min-value after min-max scaling:\nAlcohol={:.2f}, Malic acid={:.2f}".format(
        df_minmax[:, 0].min(), df_minmax[:, 1].min()
    )
)
print(
    "\nMax-value after min-max scaling:\nAlcohol={:.2f}, Malic acid={:.2f}".format(
        df_minmax[:, 0].max(), df_minmax[:, 1].max()
    )
)
```

```python
%matplotlib inline
from matplotlib import pyplot as plt


def plot():
    plt.figure(figsize=(8, 6))

    plt.scatter(
        df["Alcohol"], df["Malic acid"], color="green", label="input scale", alpha=0.5
    )

    plt.scatter(
        df_std[:, 0],
        df_std[:, 1],
        color="red",
        label="Standardized [N  (m=0, ; d=1)]",
        alpha=0.3,
    )

    plt.scatter(
        df_minmax[:, 0],
        df_minmax[:, 1],
        color="blue",
        label="min-max scaled [min=0, max=1]",
        alpha=0.3,
    )

    plt.title("Alcohol and Malic Acid content of the wine dataset")
    plt.xlabel("Alcohol")
    plt.ylabel("Malic Acid")
    plt.legend(loc="upper left")
    plt.grid()

    plt.tight_layout()


plot()
plt.show()
```


بالطبع، يمكننا أيضًا ترميز المعادلات الخاصة بالتوحيد القياسي وقياس 0-1 Min-Max "يدويًا". ومع ذلك، لا تزال أساليب scikit-Learn مفيدة إذا كنت تعمل مع مجموعات بيانات الاختبار والتدريب وترغب في توسيع نطاقها بالتساوي.


**4. التحول غير الخطي. التعيين إلى التوزيع الغوسي**



في العديد من سيناريوهات النمذجة، يكون من المرغوب فيه أن تكون الميزات الطبيعية في مجموعة البيانات أمرًا مرغوبًا فيه. تحويلات الطاقة هي مجموعة من التحويلات البارامترية الرتيبة التي تهدف إلى تعيين البيانات من أي توزيع إلى أقرب توزيع غاوسي قدر الإمكان من أجل تثبيت التباين وتقليل الانحراف.
يوفر PowerTransformer حاليًا اثنين من تحويلات الطاقة، تحويل Yeo-Johnson وتحويل Box-Cox.


```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PowerTransformer, QuantileTransformer

N_SAMPLES = 1000
FONT_SIZE = 16
BINS = 30


rng = np.random.RandomState(304)
bc = PowerTransformer(method="box-cox")
yj = PowerTransformer(method="yeo-johnson")
qt = QuantileTransformer(output_distribution="normal", random_state=rng)
size = (N_SAMPLES, 1)


# lognormal distribution
X_lognormal = rng.lognormal(size=size)

# create plots
distributions = [("Lognormal", X_lognormal)]

colors = ["firebrick"]

fig, axes = plt.subplots(nrows=4, ncols=1, figsize=(18, 20))
axes = axes.flatten()
axes_idxs = [(0, 1, 2, 3)]

axes_list = [(axes[i], axes[j], axes[k], axes[l]) for (i, j, k, l) in axes_idxs]


for distribution, color, axes in zip(distributions, colors, axes_list):
    name, X = distribution
    X_train, X_test = train_test_split(X, test_size=0.5)

    # perform power transforms and quantile transform
    X_trans_bc = bc.fit(X_train).transform(X_test)
    lmbda_bc = round(bc.lambdas_[0], 2)
    X_trans_yj = yj.fit(X_train).transform(X_test)
    lmbda_yj = round(yj.lambdas_[0], 2)
    X_trans_qt = qt.fit(X_train).transform(X_test)

    ax_original, ax_bc, ax_yj, ax_qt = axes

    ax_original.hist(X_train, color=color, bins=BINS)
    ax_original.set_title(name, fontsize=FONT_SIZE)
    ax_original.tick_params(axis="both", which="major", labelsize=FONT_SIZE)

    for ax, X_trans, meth_name, lmbda in zip(
        (ax_bc, ax_yj, ax_qt),
        (X_trans_bc, X_trans_yj, X_trans_qt),
        ("Box-Cox", "Yeo-Johnson", "Quantile transform"),
        (lmbda_bc, lmbda_yj, None),
    ):
        ax.hist(X_trans, color=color, bins=BINS)
        title = "After {}".format(meth_name)
        if lmbda is not None:
            title += "\n$\lambda$ = {}".format(lmbda)
        ax.set_title(title, fontsize=FONT_SIZE)
        ax.tick_params(axis="both", which="major", labelsize=FONT_SIZE)
        ax.set_xlim([-3.5, 3.5])


plt.tight_layout()
plt.show()
```


لاحظ أيضًا أنه على الرغم من أن أداء Box-Cox يبدو أفضل من Yeo-Johnson فيما يتعلق بالتوزيعات اللوغاريتمية الطبيعية ومربع كاي، ضع في اعتبارك أن Box-Cox لا يدعم المدخلات ذات القيم السالبة.
يمكنك أيضًا تجربة التحويل وتطبيقه على توزيعات أخرى.

لا يمكن للعديد من خوارزميات التعلم الآلي العمل على بيانات التسمية مباشرة. أنها تتطلب أن تكون جميع متغيرات الإدخال ومتغيرات الإخراج رقمية.

إزالة كلمات التوقف: الكلمات التي لها أهمية ضئيلة أو معدومة خاصة عند إنشاء ميزات ذات معنى من النص تُعرف باسم كلمات التوقف أو كلمات التوقف. عادةً ما تكون هذه الكلمات هي التي ينتهي بها الأمر بالحصول على الحد الأقصى من التكرار إذا قمت بعمل مصطلح بسيط أو تكرار الكلمات في مجموعة نصية. تعتبر الكلمات مثل a وan وthe وما إلى ذلك بمثابة كلمات توقف. لا توجد قائمة كلمات توقف عالمية ولكننا نستخدم قائمة كلمات توقف قياسية باللغة الإنجليزية من nltk. يمكنك أيضًا إضافة كلمات التوقف الخاصة بنطاقك حسب الحاجة.


```python
import re
import unicodedata

import spacy

nlp = spacy.load("en", parse=False, tag=False, entity=False)


def remove_accented_chars(text):
    text = (
        unicodedata.normalize("NFKD", text)
        .encode("ascii", "ignore")
        .decode("utf-8", "ignore")
    )
    return text


def remove_special_characters(text):
    text = re.sub("[^a-zA-z0-9\s]", "", text)
    return text


def lemmatize_text(text):
    text = nlp(text)
    text = " ".join(
        [word.lemma_ if word.lemma_ != "-PRON-" else word.text for word in text]
    )
    return text
```


** 8. ميزة التحولات مع مجموعات من الأشجار **



قم بتحويل ميزاتك إلى مساحة متفرقة ذات أبعاد أعلى. ثم تدريب نموذج خطي على هذه الميزات.
قم أولاً بتركيب مجموعة من الأشجار (أشجار عشوائية تمامًا، أو غابة عشوائية، أو أشجار معززة متدرجة) في مجموعة التدريب. ثم يتم تعيين فهرس ميزات تعسفي ثابت لكل ورقة من كل شجرة في المجموعة في مساحة ميزات جديدة. يتم بعد ذلك تشفير مؤشرات الأوراق هذه بطريقة واحدة ساخنة.تمر كل عينة بقرارات كل شجرة في المجموعة وينتهي الأمر بورقة واحدة لكل شجرة. يتم ترميز العينة عن طريق تعيين قيم المعالم لهذه الأوراق على 1 وقيم الميزات الأخرى على 0.
وقد تعلم المحول الناتج بعد ذلك تضمينًا فئويًا خاضعًا للإشراف ومتفرقًا وعالي الأبعاد للبيانات.


```python
import numpy as np

np.random.seed(10)

import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.ensemble import (GradientBoostingClassifier,
                              RandomForestClassifier, RandomTreesEmbedding)
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_curve
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import OneHotEncoder

n_estimator = 10
X, y = make_classification(n_samples=80000)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5)

# It is important to train the ensemble of trees on a different subset
# of the training data than the linear regression model to avoid
# overfitting, in particular if the total number of leaves is
# similar to the number of training samples
X_train, X_train_lr, y_train, y_train_lr = train_test_split(
    X_train, y_train, test_size=0.5
)

# Unsupervised transformation based on totally random trees
rt = RandomTreesEmbedding(max_depth=3, n_estimators=n_estimator, random_state=0)

rt_lm = LogisticRegression(solver="lbfgs", max_iter=1000)
pipeline = make_pipeline(rt, rt_lm)
pipeline.fit(X_train, y_train)
y_pred_rt = pipeline.predict_proba(X_test)[:, 1]
fpr_rt_lm, tpr_rt_lm, _ = roc_curve(y_test, y_pred_rt)

# Supervised transformation based on random forests
rf = RandomForestClassifier(max_depth=3, n_estimators=n_estimator)
rf_enc = OneHotEncoder(categories="auto")
rf_lm = LogisticRegression(solver="lbfgs", max_iter=1000)
rf.fit(X_train, y_train)
rf_enc.fit(rf.apply(X_train))
rf_lm.fit(rf_enc.transform(rf.apply(X_train_lr)), y_train_lr)

y_pred_rf_lm = rf_lm.predict_proba(rf_enc.transform(rf.apply(X_test)))[:, 1]
fpr_rf_lm, tpr_rf_lm, _ = roc_curve(y_test, y_pred_rf_lm)

# Supervised transformation based on gradient boosted trees
grd = GradientBoostingClassifier(n_estimators=n_estimator)
grd_enc = OneHotEncoder(categories="auto")
grd_lm = LogisticRegression(solver="lbfgs", max_iter=1000)
grd.fit(X_train, y_train)
grd_enc.fit(grd.apply(X_train)[:, :, 0])
grd_lm.fit(grd_enc.transform(grd.apply(X_train_lr)[:, :, 0]), y_train_lr)

y_pred_grd_lm = grd_lm.predict_proba(grd_enc.transform(grd.apply(X_test)[:, :, 0]))[
    :, 1
]
fpr_grd_lm, tpr_grd_lm, _ = roc_curve(y_test, y_pred_grd_lm)

# The gradient boosted model by itself
y_pred_grd = grd.predict_proba(X_test)[:, 1]
fpr_grd, tpr_grd, _ = roc_curve(y_test, y_pred_grd)

# The random forest model by itself
y_pred_rf = rf.predict_proba(X_test)[:, 1]
fpr_rf, tpr_rf, _ = roc_curve(y_test, y_pred_rf)

plt.figure(1)
plt.plot([0, 1], [0, 1], "k--")
plt.plot(fpr_rt_lm, tpr_rt_lm, label="RT + LR")
plt.plot(fpr_rf, tpr_rf, label="RF")
plt.plot(fpr_rf_lm, tpr_rf_lm, label="RF + LR")
plt.plot(fpr_grd, tpr_grd, label="GBT")
plt.plot(fpr_grd_lm, tpr_grd_lm, label="GBT + LR")
plt.xlabel("False positive rate")
plt.ylabel("True positive rate")
plt.title("ROC curve")
plt.legend(loc="best")
plt.show()

plt.figure(2)
plt.xlim(0, 0.2)
plt.ylim(0.8, 1)
plt.plot([0, 1], [0, 1], "k--")
plt.plot(fpr_rt_lm, tpr_rt_lm, label="RT + LR")
plt.plot(fpr_rf, tpr_rf, label="RF")
plt.plot(fpr_rf_lm, tpr_rf_lm, label="RF + LR")
plt.plot(fpr_grd, tpr_grd, label="GBT")
plt.plot(fpr_grd_lm, tpr_grd_lm, label="GBT + LR")
plt.xlabel("False positive rate")
plt.ylabel("True positive rate")
plt.title("ROC curve (zoomed in at top left)")
plt.legend(loc="best")
plt.show()
```
