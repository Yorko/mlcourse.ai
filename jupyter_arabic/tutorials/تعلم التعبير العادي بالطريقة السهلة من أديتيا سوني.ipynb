# يُرجع المحرك الموجه بواسطة Regex دائمًا التطابق الموجود في أقصى اليسار
هذه نقطة مهمة جدًا يجب فهمها: سيُرجع المحرك الموجه بالتعبير العادي دائمًا المطابقة الموجودة في أقصى اليسار،
حتى لو أمكن العثور على تطابق **أفضل** لاحقًا. عند تطبيق regex على سلسلة، سيبدأ المحرك عند
الحرف الأول من السلسلة. سيحاول كل التباديل الممكنة للتعبير العادي عند الحرف الأول.
فقط إذا تم تجربة كل الاحتمالات وفشلت، سيستمر المحرك مع الحرف الثاني
النص. مرة أخرى، سيحاول كل التباديل الممكنة للتعبير العادي، بنفس الترتيب تمامًا. والنتيجة هي ذلك
سيعيد المحرك الموجه إلى regex المطابقة الموجودة في أقصى اليسار.
- عند تطبيق **<cat\>** على **التقط سمك السلور لقطته.**، سيحاول المحرك مطابقة الأول
الرمز المميز في التعبير العادي **<c\>** للحرف الأول في المباراة **H**. هذا فشل. لا يوجد أي شيء آخر ممكن
التباديل لهذا التعبير العادي، لأنه يتكون فقط من سلسلة من الأحرف الحرفية. لذلك محرك ريكس
يحاول مطابقة **<c\>** مع **e**. يفشل هذا أيضًا، كما هو الحال مع مطابقة **c** مع المسافة. الوصول إلى الرابع
الشخصية في المباراة، **<c\>** تتطابق مع **c**. سيحاول المحرك بعد ذلك مطابقة الرمز المميز الثاني **<a\>** بالرمز الخامس
حرف **أ**. وهذا ينجح أيضا. ولكن بعد ذلك، فشل **<t\>** في مطابقة **p**. عند هذه النقطة، يعرف المحرك التعبير العادي
لا يمكن مطابقتها بدءًا من الشخصية الرابعة في المباراة. لذلك سوف يستمر مع الخامس: **أ**. مرة أخرى، **<c\>**
فشل في التطابق هنا واستمر المحرك. عند الحرف الخامس عشر من المباراة، **<c\>** يطابق **c** مرة أخرى. ال
يشرع المحرك بعد ذلك في محاولة مطابقة ما تبقى من التعبير العادي عند الحرف 15 ويجد أن **<a\>**
يطابق **a** و **<t\>** يتطابق مع **t**.
- يمكن مطابقة التعبير العادي بأكمله بدءًا من الحرف 15. المحرك **"حريص"** على الإبلاغ عن
مباراة. **لذلك سيتم الإبلاغ عن الأحرف الثلاثة الأولى من سمك السلور باعتبارها مطابقة صالحة**. المحرك ** أبدا ** يستمر
بعد هذه النقطة لمعرفة ما إذا كان هناك أي تطابقات **أفضل**. *المباراة الأولى* تعتبر جيدة بما فيه الكفاية. 



# أساسيات Regex



## مجموعات/فئات الأحرف



تسمى مجموعات الأحرف أيضًا فئة الأحرف. **الأقواس المربعة** تستخدم لتحديد مجموعات الأحرف. استخدم **واصلة** داخل مجموعة أحرف لتحديد نطاق الأحرف. لا يهم ترتيب نطاق الأحرف داخل الأقواس المربعة. على سبيل المثال، التعبير العادي `[Tt]he` يعني: `an uppercase T or lowercase t, followed by the letter h, followed by the letter e.`
- **<[Tt]he>** => <font color=red>The</font> السيارة متوقفة في مرآب <font color=red>the</font>.
ومع ذلك، فإن الفترة الموجودة داخل مجموعة الأحرف تعني فترة حرفية. التعبير العادي **<ar[.]>** يعني: حرف صغير a، متبوعًا بحرف r، متبوعًا بنقطة **.** حرف.
- **<ar[.]>** => يعتبر المرآب مكانًا جيدًا لركن السيارة c<font color=red>ar.</font>
- **<[0-9]>** => يطابق **رقمًا واحدًا بين 0 و9**. يمكنك استخدام أكثر من نطاق واحد.
- **<[0-9a-fA-F]>** => يطابق **رقمًا سداسيًا عشريًا واحدًا**، بدون حساسية لحالة الأحرف. 
- يمكنك الجمع بين النطاقات والأحرف الفردية. **<[0-9a-fxA-FX]>** يتطابق مع رقم سداسي عشري أو الحرف X.* مرة أخرى، لا يهم ترتيب الأحرف والنطاقات.*
- ابحث عن كلمة، حتى لو كانت بها أخطاء إملائية، مثل **<sep[ae]r[ae]te>** أو **<li[cs]en[cs]e>**. 



### مجموعات/فئات الأحرف السلبية


سيؤدي كتابة **علامة الإقحام (^)** بعد القوس المربع المفتوح إلى إلغاء فئة الحرف. ** والنتيجة هي أن الشخصية
سوف تتطابق الفئة مع أي حرف <font color = red> وليس </font> في فئة الأحرف.**
- من المهم أن تتذكر أن فئة الأحرف الملغاة **يجب أن تتطابق مع الحرف**. **<q[^u]>** لا
يعني: **<font color= red> a q لا يتبعه u </font>**. يعني: **<font color= red a q followed by a character that is not a u </font>**. وسوف ** لا ** ** تطابق
$q$ في السلسلة $Iraq$. سوف يطابق $q$ و$the space$ بعد $q$ في **العراق بلد**.

## التكراريتم استخدام أحرف التعريف التالية `+` أو `*` أو `?` لتحديد عدد مرات
يمكن أن يحدث النمط الفرعي. تعمل هذه الشخصيات التعريفية بشكل مختلف باختلاف
المواقف.



### النجمة *
يتطابق الرمز `*` مع صفر أو أكثر من تكرارات المطابق السابق. ال
التعبير العادي `a*` يعني: صفر أو أكثر من تكرار الأحرف الصغيرة السابقة
الحرف `a`. ولكن إذا ظهر بعد مجموعة أحرف أو فئة، فسيتم العثور عليه
تكرار مجموعة الأحرف بأكملها. 
على سبيل المثال، التعبير العادي
- `[a-z]*` يعني: أي عدد من الأحرف الصغيرة على التوالي.
يمكن استخدام الرمز `*` مع حرف التعريف `.` لمطابقة أي سلسلة من
الأحرف `.*`. يمكن استخدام الرمز `*` مع مسافة بيضاء `\s`
لمطابقة سلسلة من أحرف المسافات البيضاء. على سبيل المثال، التعبير
`\s*cat\s*` تعني: صفر مسافات أو أكثر، متبوعة بحرف صغير `c`،
متبوعًا بحرف صغير `a`، متبوعًا بحرف صغير `t`،
متبوعة بصفر أو أكثر من المسافات.



### بلس +
يتطابق الرمز `+` مع تكرار واحد أو أكثر للحرف السابق. ل
على سبيل المثال، التعبير العادي `c.+t` يعني: حرف صغير `c`، متبوعًا بـ
حرف واحد على الأقل، متبوعاً بالحرف الصغير `t`. يجب أن يكون
أوضح أن `t` هو `t` الأخير في الجملة.
- **<c.+t>** => جلست القطة السمينة <font color='red'> على السجادة</font>.


### علامة الاستفهام؟
في التعبير العادي، يجعل حرف التعريف `?` الحرف السابق اختياريًا. يطابق هذا الرمز صفرًا أو مثيلًا واحدًا للحرف السابق. على سبيل المثال، التعبير العادي `[T]?he` يعني: `Optional the uppercase letter T, followed by the lowercase character h, followed by the lowercase character e.`
- **<[Tt]he>** => <font color=red>The</font> السيارة متوقفة في مرآب <font color=red>the</font>.



### النجم الكسول *؟ 
يكرر العنصر السابق صفر مرات أو أكثر. كسول، لذلك يحاول المحرك أولاً تخطي

البند.

**<[^\r\n]>** (نكهات Windows regex).<font color = 'Red'> <b> استخدم النقطة باعتدال <b></font>

## بداية السلسلة ونهاية مرساة السلسلة ($ و^)



المراسي هي سلالة مختلفة. أنها لا تتطابق مع أي شخصية على الإطلاق. بدلا من ذلك، فإنها تطابق موقفا من قبل،
بعد أو بين الحروف. يمكن استخدامها `anchor` لمطابقة التعبير العادي في موضع معين. 
- علامة الإقحام **<^>**
يطابق الموضع قبل الحرف الأول في السلسلة. تطبيق **<^a>** على `abc` يطابق `a`. **<^ب>** سوف
لا يتطابق مع `abc` على الإطلاق، لأنه لا يمكن مطابقة **<b\>** مباشرة بعد بداية السلسلة، ومطابقته بـ **<^>**.
- وبالمثل، يتطابق **<\$>** مباشرةً بعد الحرف الأخير في السلسلة. **<c\$>** يطابق `c` في `abc`، بينما **<a\$>** `does not` يتطابق مع `abc` على الإطلاق....

## حدود الكلمات



يطابق حد الكلمة \b المواضع التي يكون فيها أحد الجانبين حرف كلمة (عادةً حرف أو رقم أو شرطة سفلية - ولكن انظر أدناه لمعرفة الاختلافات عبر المحركات) والجانب الآخر ليس حرف كلمة (على سبيل المثال، قد يكون بداية السلسلة أو حرف مسافة).
ولذلك فإن `\bcat\b` سيتطابق مع `cat` في `black cat`، ولكنه لن يطابقه في `catatonic`، أو `tomcat`، أو `certificate`. إزالة أحد الحدود، `\bcat` سيطابق `cat` في `catfish`، و`cat\b` سيطابق `cat` في `tomcat`، ولكن ليس العكس. كلاهما بالطبع يتطابقان مع `cat` بمفردهما. 
تكون حدود الكلمات مفيدة عندما تريد مطابقة سلسلة من الأحرف (أو الأرقام) بمفردها، أو التأكد من ظهورها في بداية أو نهاية سلسلة من الأحرف.
ومع ذلك، كن على علم بأن `\bcat\b` لن يتطابق مع `cat` في `_cat` أو في `cat25` لأنه لا يوجد حدود بين الشرطة السفلية والحرف، ولا بين الحرف والرقم: تنتمي جميعها إلى ما يعرفه التعبير العادي على أنه أحرف كلمة.

** التعبير العادي **: `<[^>]+>`

** التعبير العادي **: `https?:\/\/[\w\.\/\-?=&%,]+`
- ماذا يفعل: سيجد هذا عنوان URL. سيتم التقاط معظم عناوين URL التي تبدأ بـ http:// أو https://.
** التعبير العادي **: `'\w+?'`
- **ماذا يفعل**: يبحث هذا عن الكلمات المفردة المحاطة بفواصل عليا.** التعبير العادي **: `([-A-Za-z0-9_]*?([-A-Za-z_][0-9]|[0-9][-A-Za-z_])[-A-Za-z0-9_]*)`
- **ماذا يفعل**: أرقام الأجزاء والمراجع الأبجدية الرقمية مثل: 1111_A، أو AA1AAA أو 1-1-1-A، أو 21A1، و10UC10P-BACW، أو abcd-1234، أو 1234-pqtJK، أو sft-0021 أو 21-1_AB و55A أو AK7_GY.
يمكن أن يكون هذا مفيدًا جدًا إذا كنت تترجم مستندات تحتوي على الكثير من الرموز الأبجدية الرقمية أو المراجع، وتحتاج إلى أن تكون قادرًا على العثور عليها بسهولة.
** التعبير العادي **: `\b(the|The)\b.*?\b(?=\W?\b(is|are|was|can|shall| must|that|which|about|by|at|if|when|should|among|above|under|$)\b)`
- **ما يفعله**: يبحث هذا عن نص يبدأ بـ أو The وينتهي بكلمات توقف مثل is، are، been، can، must، must، that، about، by، at، if، متى، يجب، بين، فوق أو تحت، أو نهاية المقطع.

```python
%load_ext autoreload
%autoreload
import re
import time
```

```python
s = "How do you do this"
print(
    "After applying re.sub -- ",
    re.sub(r"How do you", "How do I", s),
    "\nOriginal Text is still -- ",
    s,
)
```


فهل هذا يعني أنه يتعين علينا كتابة تعبير عادي واحد في كل مرة، وتشغيله والتحقق منه وبعد ذلك سيحدث الاستبدال؟ أي ألا يمكننا تكديس `re.sub(), re.sub(), re.sub()....`
بالتأكيد لا، Remeber `re.sub()` يقوم بإرجاع سلسلة بعد إجراء التغييرات التي تطابق النمط الذي طلبته أكثر..


```python
s_old = "How do you do this"
print("After applying re.sub -- ", end="")
s_new = re.sub(r"How do you", "How do I", s_old)
print(f"\nOriginal Text isn't still **{s_old}** but it's now **{s_new}**")
# Obviously s_old and s_new are different, I am just trying to show that we can stack the operations....
```

```python
tweet = "#fingerprint #Pregnancy Test https://goo.gl/h1MfQV #android +#apps +#beautiful \
         #cute #health #igers #iphoneonly #iphonesia #iphone \
             <3 ;D :( :-("

# Let's take care of emojis and the #(hash-tags)...

print(f"Original Tweet ---- \n {tweet}")

## Replacing #hashtag with only hashtag
tweet = re.sub(r"#(\S+)", r" \1 ", tweet)
# this gets a bit technical as here we are using Backreferencing and Character Sets Shorthands and replacing the captured Group.
# \S = [^\s] Matches any charachter that isn't white space
print(f"\n Tweet after replacing hashtags ----\n  {tweet}")

## Love -- <3, :*
tweet = re.sub(r"(<3|:\*)", " EMO_POS ", tweet)
print(f"\n Tweet after replacing Emojis for Love with EMP_POS ----\n  {tweet}")

# The parentheses are for Grouping, so we search (remeber the raw string (`r`))
# either for <3 or(|) :\* (as * is a meta character, so preceeded by the backslash)

## Wink -- ;-), ;), ;-D, ;D, (;,  (-;
tweet = re.sub(r"(;-?\)|;-?D|\(-?;)", " EMO_POS ", tweet)
print(f"\n Tweet after replacing Emojis for Wink with EMP_POS ----\n  {tweet}")

# The parentheses are for Grouping as usual, then we first focus on `;-), ;),`, so we can see that 1st we need to have a ;
# and then we can either have a `-` or nothing, so we can do this via using our `?` clubbed with `;` and hence we have the very
# starting with `(;-?\)` and simarly for others...

## Sad -- :-(, : (, :(, ):, )-:
tweet = re.sub(r"(:\s?\(|:-\(|\)\s?:|\)-:)", " EMO_NEG ", tweet)
print(f"\n Tweet after replacing Emojis for Sad with EMP_NEG ----\n  {tweet}")
```

```python
##See the Output Carefully, there are Spaces inbetween un-necessary...
## Replace multiple spaces with a single space
tweet = re.sub(r"\s+", " ", tweet)
print(f"\n Tweet after replacing xtra spaces ----\n  {tweet}")

##Replace the Puctuations (+,;)
tweet = re.sub(r"[^\w\s]", "", tweet)
print(f"\n Tweet after replacing Punctuation + with PUNC ----\n  {tweet}")
```
