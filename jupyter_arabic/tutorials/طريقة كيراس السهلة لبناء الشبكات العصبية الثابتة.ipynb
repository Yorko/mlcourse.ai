## مقدمة



Keras هي واجهة برمجة تطبيقات للشبكات العصبية عالية المستوى، مكتوبة بلغة بايثون.
مميزات كيراس الرئيسية:
- قادر على العمل على TensorFlow أو CNTK أو Theano.
- يتيح Keras إنشاء نماذج أولية سهلة وسريعة ويدعم كلاً من شبكات Perceptrons والشبكات التلافيفية والشبكات المتكررة (بما في ذلك LSTM)، بالإضافة إلى مجموعاتها. 
- كيراس متوافق مع: بايثون 2.7-3.6.



لجعل العملية أكثر إثارة للاهتمام، دعونا نفكر في مثال التصنيف من الحياة الواقعية.
## وصف المثاللنأخذ المهمة من أحد هاكاتون، الذي نظمه بعض منتجي البولي بروبيلين هذا العام. لذا، دعونا نفكر في إنتاج حبيبات البولي بروبيلين بواسطة الطارد. الطارد هو نوع من "مفرمة اللحم" التي تحتوي على سكاكين في نهاية العملية والتي تقوم بتقطيع المنتج الناتج إلى الحبيبات.  
المشكلة هي أنه في بعض الأحيان تكون كتلة الإنتاج غير منتظمة وتلتصق بالسكاكين. عندما يكون هناك الكثير من الكتلة العالقة - لم تعد السكاكين قادرة على العمل. وفي هذه الحالة لا بد من وقف عملية الإنتاج، وهي مكلفة للغاية. إذا لاحظنا بداية عملية الالتصاق هذه - فهناك طريقة لتنظيف السكاكين بسرعة كبيرة وغير مؤلمة ومواصلة الإنتاج دون توقف.
لذا، تتمثل المهمة في إرسال إشارة التوقف إلى المشغل مقدمًا (دعنا نقول ليس بعد 15 دقيقة من هذا الحدث) - حتى يكون لديه الوقت لإجراء التلاعبات اللازمة.

لقد قمنا الآن بالفعل بمعالجة مجموعة البيانات المقيسة مسبقًا باستخدام متجهات قيم مستشعرات النظام (5160 ميزة) وأهداف 0/1. لقد تم تقسيمه بالفعل إلى [القطار](https://drive.google.com/open?id=1TMlClLguxcXTOAJt8VKe-iLrndJuFShl) و[الاختبار](https://drive.google.com/open?id=1JonMu0wmMbUqcbSd17Qr2A3AhVF3nutZ). 
فلنقم بتنزيل مجموعات البيانات الخاصة بنا والاستعداد لها. توجد في مجموعات البيانات أهداف في العمود الصفري والطوابع الزمنية في العمود الأول. لذا، دعونا نستخرج مصفوفة التدريب والاختبار بالإضافة إلى أهدافنا. سنقوم أيضًا بتحويل الأهداف إلى أهداف فئوية - بحيث تكون نتيجة لذلك أهدافنا كمتجهات ثنائية الأبعاد، أي متجهات الاحتمالات 0 و1.


```python
import numpy as np
import pandas as pd
from keras.utils import np_utils

df_train = pd.read_csv("train2.tsv", sep="\t", header=None)
df_test = pd.read_csv("test2.tsv", sep="\t", header=None)

Y_train = np.array(df_train[0].values.astype(int))
Y_test = np.array(df_test[0].values.astype(int))
X_train = np.array(df_train.iloc[:, 2:].values.astype(float))
X_test = np.array(df_test.iloc[:, 2:].values.astype(float))

Y_train = Y_train.astype(np.int)
Y_test = Y_test.astype(np.int)

Y_train = np_utils.to_categorical(Y_train, 2)
Y_test = np_utils.to_categorical(Y_test, 2)

print(X_train.shape)
print(Y_train.shape)
print(X_test.shape)
print(Y_test.shape)
```


## بناء الشبكة العصبية


دعونا نفكر في كيف يمكن لشبكة Newral Network (NN) البسيطة، وMultilayer Perceptron (MLP)، ذات الثلاث طبقات المخفية (كخط أساسي)، التي أنشأها Keras، أن تساعدنا في حل هذه المشكلة.
بما أن لدينا طبقات مخفية، ستكون هذه شبكة عصبية عميقة. 
أيضًا، يمكننا أن نرى أننا بحاجة إلى 5160 خلية عصبية في طبقة الإدخال، لأن هذا هو حجم المتجه X و2 خلية عصبية في الطبقة الأخيرة - لأن هذا هو حجم هدفنا (مقابل الصورة أدناه، حيث يوجد 4 خلايا عصبية في طبقة الإخراج).
يمكنك قراءة، على سبيل المثال، [هنا](https://en.wikipedia.org/wiki/Multilayer_perceptron) أو [هنا](https://towardsdatascience.com/meet-artificial-neural-networks-ae5939b1dd3a) بعض المعلومات الإضافية حول بنية MLP.

بنية البيانات الأساسية لـ Keras هي **_model_** - طريقة لتنظيم الطبقات. أبسط نوع من النماذج هو النموذج **_Sequential_**، وهو عبارة عن مجموعة خطية من الطبقات، وهو مناسب لبناء MLP (للمعماريات الأكثر تعقيدًا، يجب عليك استخدام واجهة برمجة التطبيقات الوظيفية Keras، والتي تسمح ببناء رسوم بيانية عشوائية للطبقات).


```python
import keras
from keras import Sequential

model1 = Sequential()
```


بعد تحديد نوع النموذج، نحتاج إلى إضافة طبقات باستمرار كـ **_Dense_**. يعد تكديس الطبقات أمرًا سهلاً مثل **_.add()_**.
أثناء إضافة الطبقة، نحتاج إلى تحديد **عدد الخلايا العصبية** و***التنشيط*** **الوظائف** التي يمكننا ضبطها بعد ذلك. بالنسبة للطبقة القبضة، نحتاج أيضًا إلى إضافة أبعاد المتجهات X (***input_dim***). في حالتنا هذا هو 5160. تتكون الطبقة الأخيرة من خليتين عصبيتين تمامًا كما يفعل السترات المستهدفة Y_train وY_test.
يمكن أيضًا ضبط **عدد الطبقات**.


```python
from keras.layers import Activation, Dense

model1.add(Dense(64, input_dim=5160))
model1.add(Activation("relu"))

model1.add(Dense(64))
model1.add(Activation("sigmoid"))

model1.add(Dense(128))
model1.add(Activation("tanh"))

model1.add(Dense(2))
model1.add(Activation("softmax"))
```


بمجرد أن يبدو نموذجنا جيدًا، نحتاج إلى تكوين عملية التعلم الخاصة به باستخدام **_.compile()_**.هنا يجب علينا أيضًا وصف **_loss_** **الوظيفة** و **_metrics_** التي نريد استخدامها بالإضافة إلى **_optimizer_** (نوع نزول التدرج الذي سيتم استخدامه) والذي يبدو مناسبًا في كل حالة معينة.


```python
model1.compile(loss="categorical_crossentropy", optimizer="adam", metrics=["accuracy"])
```


يمكننا الآن تكرار بيانات التدريب الخاصة بنا على دفعات باستخدام ***batch_size*** الذي نريده، حيث X_train وy_train عبارة عن مصفوفات Numpy تمامًا كما هو الحال في Scikit-Learn API.
كما يمكننا تحديد عدد ***العصور*** (أي الحد الأقصى لعدد الدورات الكاملة لتدريب النموذج). 
***Verbose=1*** يتيح لنا فقط الاطلاع على ملخص المرحلة الحالية من العمليات الحسابية.
يمكننا أيضًا طباعة معلمات النموذج باستخدام ***model.summary()***. 
قد يكون من المفيد أيضًا رؤية **أشكال** X_train وy_train وX_test وy_test
يمكننا أيضًا حفظ **أفضل إصدار للنموذج** أثناء عملية التدريب عبر خيار ***callback_save***.
 
وهناك خيار ***callback_early stop*** لإيقاف عملية التدريب عندما لا يكون لدينا تحسن كبير (محدد بواسطة ***min_delta***) خلال عدد معين من العصور (***الصبر***). 
 
الآن نموذجنا الأول جاهز:


```python
from keras.callbacks import EarlyStopping, ModelCheckpoint

model1.summary()

print(X_train.shape)
print(Y_train.shape)
print(X_test.shape)
print(Y_test.shape)

callback_save = ModelCheckpoint(
    "best_model1.model1", monitor="val_acc", verbose=1, save_best_only=True, mode="auto"
)
callback_earlystop = EarlyStopping(
    monitor="val_loss", min_delta=0, patience=10, verbose=1, mode="auto"
)


model1.fit(
    X_train,
    Y_train,
    batch_size=20,
    epochs=10000,
    verbose=1,
    validation_data=(X_test, Y_test),
    callbacks=[callback_save, callback_earlystop],
)
```


** إذن، حصلنا على خط أساس بالدقة = 0.79 **. يبدو رائعًا، لأننا لم نقم بضبط أي شيء بعد!



دعونا نحاول تحسين هذه النتيجة. على سبيل المثال، يمكننا تقديم **Dropout** - وهو نوع من التنظيم للشبكات Neral.
**مستوى التسرب** (بين قوسين، جنبًا إلى جنب مع ***بذرة***) هو احتمال الاستبعاد من طبقة معينة لخلايا الاختيار العشوائي أثناء الحسابات الحالية. لذلك، تساعد عمليات التسرب على منع فرط تجهيز NN.  
لنقم بإنشاء النموذج الجديد:


```python
from keras.layers import Dropout

model2 = Sequential()

model2.add(Dense(64, input_dim=5160))
model2.add(Activation("relu"))
model2.add(Dropout(0.3, seed=123))

model2.add(Dense(64))
model2.add(Activation("sigmoid"))
model2.add(Dropout(0.4, seed=123))

model2.add(Dense(128))
model2.add(Activation("tanh"))
model2.add(Dropout(0.5, seed=123))

model2.add(Dense(2))
model2.add(Activation("softmax"))


model2.compile(loss="categorical_crossentropy", optimizer="adam", metrics=["accuracy"])

model2.summary()

print(X_train.shape)
print(Y_train.shape)
print(X_test.shape)
print(Y_test.shape)

callback_save = ModelCheckpoint(
    "best_model2.model2", monitor="val_acc", verbose=1, save_best_only=True, mode="auto"
)
callback_earlystop = EarlyStopping(
    monitor="val_loss", min_delta=0, patience=10, verbose=1, mode="auto"
)


model2.fit(
    X_train,
    Y_train,
    batch_size=20,
    epochs=10000,
    verbose=1,
    validation_data=(X_test, Y_test),
    callbacks=[callback_save, callback_earlystop],
)
```


وبالتالي، بإضافة المتسربين، قمنا **بزيادة الدقة في الاختبار حتى 0.86830**


يمكننا أيضًا **ضبط جميع معلمات gyper** مثل **عدد الطبقات**، **مستويات المتسربين**، **وظائف التنشيط**، **المُحسِّن**، **عدد الخلايا العصبية** وما إلى ذلك.
لهذا الغرض، يمكننا استخدام، على سبيل المثال، مكتبة أخرى ودية للغاية وسهلة التطبيق - مكتبة Hyperas. يمكنك العثور على الوصف مع الأمثلة [هنا](https://github.com/maxpumperla/hyperas).
ونتيجة لهذا الضبط، حصلنا على تكوين النموذج التالي:


```python
model3 = Sequential()

model3.add(Dense(64, input_dim=5160))
model3.add(Activation("relu"))
model3.add(Dropout(0.11729755246044238, seed=123))

model3.add(Dense(256))
model3.add(Activation("relu"))
model3.add(Dropout(0.8444244099007299, seed=123))

model3.add(Dense(1024))
model3.add(Activation("linear"))
model3.add(Dropout(0.41266207281071243, seed=123))

model3.add(Dense(256))
model3.add(Activation("relu"))
model3.add(Dropout(0.4844455237320119, seed=123))

model3.add(Dense(2))
model3.add(Activation("softmax"))

model3.compile(
    loss="categorical_crossentropy", optimizer="rmsprop", metrics=["accuracy"]
)

model3.summary()

print(X_train.shape)
print(Y_train.shape)
print(X_test.shape)
print(Y_test.shape)

callback_save = ModelCheckpoint(
    "best_model3.model3", monitor="val_acc", verbose=1, save_best_only=True, mode="auto"
)
callback_earlystop = EarlyStopping(
    monitor="val_loss", min_delta=0, patience=10, verbose=1, mode="auto"
)

model3.fit(
    X_train,
    Y_train,
    batch_size=60,
    epochs=10000,
    verbose=1,
    validation_data=(X_test, Y_test),
    callbacks=[callback_save, callback_earlystop],
)
```


**الآن، مع المعلمات المضبوطة، تمكنا من تحسين الدقة حتى 0.88073**



مع Keras، من الممكن أيضًا استخدام **تنظيمات الوزن L1/L2** التي تسمح بتطبيق عقوبات على معلمات الطبقة أو نشاط الطبقة أثناء التحسين. تم دمج هذه العقوبات في وظيفة الخسارة التي يقوم بها مُحسِّنو الشبكة. دعنا نضيف بعض التنظيم إلى الطبقة الأولى.


```python
from keras import regularizers

model4 = Sequential()
model4.add(
    Dense(
        64,
        input_dim=5160,
        kernel_regularizer=regularizers.l2(0.0015),
        activity_regularizer=regularizers.l1(0.0015),
    )
)
model3.add(Activation("relu"))
model4.add(Dropout(0.11729755246044238, seed=123))

model4.add(Dense(256))
model4.add(Activation("relu"))
model4.add(Dropout(0.8444244099007299, seed=123))

model4.add(Dense(1024))
model4.add(Activation("linear"))
model4.add(Dropout(0.41266207281071243, seed=123))

model4.add(Dense(256))
model4.add(Activation("relu"))
model4.add(Dropout(0.4844455237320119, seed=123))

model4.add(Dense(2))
model4.add(Activation("softmax"))

model4.compile(
    loss="categorical_crossentropy", optimizer="rmsprop", metrics=["accuracy"]
)

model4.summary()

print(X_train.shape)
print(Y_train.shape)
print(X_test.shape)
print(Y_test.shape)

callback_save = ModelCheckpoint(
    "best_model4.model4", monitor="val_acc", verbose=1, save_best_only=True, mode="auto"
)
callback_earlystop = EarlyStopping(
    monitor="val_loss", min_delta=0, patience=10, verbose=1, mode="auto"
)

model4.fit(
    X_train,
    Y_train,
    batch_size=60,
    epochs=10000,
    verbose=1,
    validation_data=(X_test, Y_test),
    callbacks=[callback_save, callback_earlystop],
)
```


لذلك، يمكننا أن نرى أن إضافة إعادة التنظيم مع المعاملات الحالية إلى الطبقة الأولى حصلنا على **دقة 0.84421** والتي لم تحسن النتيجة. هذا يعني أنه كالعادة يجب ضبطها بعناية :) 



عندما **نرغب في استخدام أفضل نموذج مدرب** حصلنا عليه، يمكننا فقط تنزيل أفضل نموذج تم حفظه مسبقًا (تلقائيًا) (عبر ***load_model***) وتطبيقه على البيانات المطلوبة.
دعونا نرى ما سنحصل عليه في مجموعة الاختبار:
