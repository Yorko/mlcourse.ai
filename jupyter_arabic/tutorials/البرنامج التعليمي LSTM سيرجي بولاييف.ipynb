### مقدمة
لكي نفهم بشكل أفضل معنى "تنبؤات السلاسل الزمنية"، دعونا نقسم المصطلح إلى قسمين:
1. السلسلة الزمنية هي سلسلة من الملاحظات المأخوذة بالتتابع في الوقت المناسب.
2. التنبؤ يعني عمل تنبؤات حول حدث مستقبلي.
لقد تم استخدام التنبؤ بالسلاسل الزمنية لبعض الوقت في العديد من الصناعات؛ غالبًا ما يتم استخدامه في أي صناعة لتوجيه القرارات المستقبلية، على سبيل المثال، يعد التنبؤ في مجال البيع بالتجزئة مهمًا جدًا بحيث يمكن شراء المواد الخام وفقًا لذلك. أفضل مثال معروف هو توقعات الطقس، حيث يمكن التنبؤ بالمستقبل بناءً على النمط الموجود في الماضي والتغيرات الأخيرة. هذه التنبؤات مهمة جدًا وعادةً ما تكون الخطوة الأولى لحل المشكلات الأخرى.
يتضمن تحليل السلاسل الزمنية التقليدية تحليل البيانات إلى مكوناتها مثل مكون الاتجاه والمكون الموسمي والضوضاء.
<تم استخدام تقنيات مثل ARIMA(p,d,q) والمتوسط ​​المتحرك والانحدار التلقائي لتحليل السلاسل الزمنية. تم العثور على RNN ذات الحالة مثل LSTM لتكون فعالة جدًا في تحليل السلاسل الزمنية في الماضي القريب.
تعد الشبكات العصبية عائلة شاملة جدًا من نماذج التعلم الآلي، وفي السنوات الأخيرة، زادت تطبيقاتها في التمويل والاقتصاد بشكل كبير.
##### ما هي RNN أو الشبكات العصبية المتكررة؟
فكرة الشبكات العصبية المتكررة (RNNs) هي استخدام المعلومات المتسلسلة. في الشبكة العصبية التقليدية، نفترض أن جميع المدخلات مستقلة. ولكن بالنسبة للعديد من المهام، فهذه ليست فكرة مثالية. تُسمى شبكات RNN المتكررة لأنك تؤدي نفس المهمة لكل عنصر من عناصر التسلسل، ويعتمد الإخراج على الحسابات السابقة.RNN هي حالة خاصة من الشبكات العصبية تشبه الشبكات العصبية التلافيفية، والفرق هو أن RNN يمكنها الاحتفاظ بحالة المعلومات الخاصة بها.
##### ما هو LSTM؟
<img src="@@KEEP_00000@@ />
LSTM هو أحد أشكال بنية RNN. كان من الصعب تدريب النماذج باستخدام بنيات RNN التقليدية. توضح التطورات الحديثة أحدث النتائج باستخدام LSTM (الذاكرة طويلة المدى). تحتوي خلية LSTM على 5 مكونات أساسية تسمح لها بنمذجة البيانات طويلة المدى وقصيرة المدى: حالة الخلية، والحالة المخفية، وبوابة الإدخال، وبوابة النسيان، وبوابة الإخراج. إحدى المزايا الحاسمة لـ LSTMs هي قدرتها على التذكر من التسلسلات طويلة المدى.
كانت بنية LSTM قادرة على الاهتمام بمشكلة التدرج المتلاشي في شبكة RNN التقليدية.
الهدف من هذا البرنامج التعليمي هو توفير دليل عملي للشبكات العصبية للتنبؤ ببيانات السلاسل الزمنية المالية.



###### استخدام مرونة LSTM لاستخدام مجموعات متعددة من المعلمات:
1. العديد من النماذج (مفيدة إذا كنت تريد التنبؤ اليوم، مع مراعاة جميع المدخلات السابقة)،
1. نماذج متعدد إلى متعدد (مفيدة عند التنبؤ بخطوات مستقبلية متعددة في نفس الوقت، مع مراعاة جميع المدخلات السابقة)
وعدد من الانحرافات الأخرى عنه.
2. يمكننا تخصيص عدة أشياء، على سبيل المثال - حجم نافذة المراقبة التي يجب التنبؤ بها في المرحلة الحالية،
3. عدد الخطوات التي نريد التنبؤ بها في المستقبل
4. أعد التنبؤ الحالي إلى النافذة لإجراء التنبؤ في الخطوة التالية (تُعرف هذه التقنية أيضًا بنافذة التقدم).
5. تحديد أفضل بنية LSTM
5. وهكذا



تم عرض شبكات LSTM في ورقة Hochreiter and Schmidhuber, 1997،أول عمل حول شبكات LSTM كان في عام 1997 من قبل Hochreiter وSchmidhuber لحل مشكلة التبعيات طويلة المدى.
جميع شبكات RNN لها شكل سلسلة من الوحدات المتكررة.

```python
# Making a function to break data, scale data, window data


def get_window_data(data, window):
    # Get window data and scale
    scaler = MinMaxScaler(feature_range=(0, 1))
    data = scaler.fit_transform(data.reshape(-1, 1))

    X = []
    y = []

    for i in range(len(data) - window - 1):
        X.append(data[i : i + window])
        y.append(data[i + window + 1])

    X = np.asarray(X)
    y = np.asarray(y)
    return X, y, scaler


window_size = 50
X, y, scaler = get_window_data(df.values, window_size)

test_split = int(len(df) * 0.8)

X_train = X[:test_split]
X_test = X[test_split:]

y_train = y[:test_split]
y_test = y[test_split:]
```

```python
from keras.layers import LSTM, Activation, Dense, Dropout
from keras.models import Sequential

# ------- Building the LSTM model --------- #

model = Sequential()
model.add(LSTM(50, input_shape=(window_size, 1)))  # 50neurons, and and window_size = 50
model.add(Dropout(0.2))
model.add(Dense(1))
model.add(Activation("linear"))
model.compile(loss="mse", optimizer="adam")

# ------- Building the LSTM model --------- #
```

```python
# We can just fit or write it to history of learning. Then we will see the process of learning

history = model.fit(
    X_train,
    y_train,
    epochs=20,
    batch_size=10,
    validation_data=(X_test, y_test),
    verbose=2,
    shuffle=False,
)
```

```python
mse_lstm = mean_squared_error(
    scaler.inverse_transform(y_test), scaler.inverse_transform(model.predict(X_test))
)
print("RMSE for LSTM {:.2f}".format(np.sqrt(mse_lstm)))
```

```python
# make functions for plotting history and predictions


def plot_history(history):
    plt.figure(figsize=(6, 4))
    plt.plot(history.history["loss"], label="Train")
    plt.plot(history.history["val_loss"], label="Test")
    plt.title("Loss over epoch")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.legend()
    plt.show()


def plot(X_test):
    pred = model.predict(X_test)
    pred_inverse = scaler.inverse_transform(pred.reshape(-1, 1))
    y_test_inverse = scaler.inverse_transform(y_test.reshape(-1, 1))

    # Calculate mean_squared_error. Previosly we did MinMax scale, so apply inverse_transform to recover values
    rmse = sqrt(mean_squared_error(y_test_inverse, pred_inverse))
    print("Test RMSE: %.3f" % rmse)
    plt.figure(figsize=(15, 8))
    plt.plot(pred_inverse, label="predict")
    plt.plot(y_test_inverse, label="actual")
    plt.legend()
    plt.show()
```

```python
df = pd.read_csv("../../data/ads_hour.csv", index_col=["Date"], parse_dates=["Date"])
```

```python
from sklearn.preprocessing import MinMaxScaler

X = df["ads"].values.reshape(-1, 1)
scaler = MinMaxScaler(feature_range=(0, 1))
X = scaler.fit_transform(X)

X.shape, df["ads"].shape
```

```python
# test_index = round(0.8*df.shape[0])

train_size = round(len(X) * 0.7)
test_size = len(X) - train_size
train, test = X[0:train_size, :], X[train_size : len(X), :]
print(len(train), len(test))
```

```python
def create_dataset(dataset, look_back=360):
    X, Y = [], []
    for i in range(len(dataset) - look_back):
        a = dataset[i : (i + look_back), 0]
        X.append(a)
        Y.append(dataset[i + look_back, 0])
    return np.array(X), np.array(Y)
```

```python
look_back = 5
X_train, y_train = create_dataset(train, look_back)
X_test, y_test = create_dataset(test, look_back)

X_train.shape, y_train.shape, X_test.shape, y_test.shape
```

```python
# Making 3D vector
X_train = np.reshape(X_train, (X_train.shape[0], 1, X_train.shape[1]))
X_test = np.reshape(X_test, (X_test.shape[0], 1, X_test.shape[1]))

X_train.shape, X_test.shape
```

```python
# ------- Building the LSTM model --------- #

model = Sequential()
model.add(LSTM(100, input_shape=(X_train.shape[1], X_train.shape[2])))
model.add(Dense(1))
model.compile(loss="mae", optimizer="adam")
history = model.fit(
    X_train,
    y_train,
    epochs=300,
    batch_size=100,
    validation_data=(X_test, y_test),
    verbose=0,
    shuffle=False,
)

# ------- Building the LSTM model --------- #
```

```python
plot_history(history)
plot(X_test)
```

يمكن أن يعطي نتائج جيدة جدًا لأن LSTM يتذكر الفترات السابقة والقمم والتوترات واللحظات الأخرى. 
أكثر من 1000 معلمة لضبط النموذج تجعل من الصعب اختيار LSTM في التنبؤ بالوقت البسيط. يمكنك تغيير عدد الطبقات، واختيار التسرب، وحجم النافذة، والمحسن، وحجم الحمام، وعدد العصور، أو إرجاع التسلسلات أو عدم إعادة إنتاجها.
<يبدو أن اختيار LSTM أمر جيد، عندما لا يقدم ARIMA وSARIMA و Prophet مقاييس ونتائج مقبولة. أو عندما يكون لدينا الكثير من المعلومات الإضافية للسلاسل الزمنية. على سبيل المثال: استخدام الأخبار للتنبؤ بحركات المخزون، للمبيعات مع الكثير من الشروط من الأعمال التي تصف التغييرات المختلفة في السلاسل الزمنية.
