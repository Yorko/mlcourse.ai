## مواضيع نموذجية  
فيما يلي العشرات من المواضيع النموذجية (ولكن هذا مجرد مثال، يمكنك/ينبغي عليك أن تتوصل إلى مواضيع خاصة بك):
#### الباندا وتحليل البيانات
- جمع البيانات، والزحف، والعمل مع XML، JSON وما إلى ذلك.
- العمل مع DataFrames الكبيرة، Dask. 
- تحسين استخدام الذاكرة أثناء العمل مع NumPy وPandas وPython النقي
- العمل مع JSON وXML في Pandas
- متى تستخدم SQL مع Pandas أو SQL فقط
- تنسيق بيانات الريشة
- تقليل استهلاك الذاكرة أثناء تحليل البيانات: النصائح والحيل
- تحليل البيانات باستخدام bash (باستخدام أدوات سطر الأوامر الفعالة للغاية)
 
#### التصور
 - نظرة عامة على Bokeh أو مكتبة أخرى
 - العمل مع البيانات الجغرافية المكانية بلغة بايثون
 - إنشاء الرسوم المتحركة لتحليل البيانات
 - تصور الشجرة في بايثون (أكثر من الرسم البياني القياسي)
 - تقليل الأبعاد للتصور (هناك بعض الطرق إلى جانب t-SNE، على سبيل المثال UMAP)
 
#### أشجار القرار وkNN
 - أشجار القرار مع الاختبارات الإحصائية في العقد
 - نظرة عامة على مكتبة H2O
 - kNN في أنظمة الإنتاج : الإزعاج 
 - kNN كخط أساس قوي في أنظمة التوصية
 
#### النماذج الخطية
 - بواسون/كمية أو أي نوع آخر من الانحدار
 - تصنيف متعدد التسمية
 - التنفيذ الفعال للنماذج الخطية (تحليل QR أو ما شابه ذلك)
 - الانحدار المتوسط والكمي
 - تفسير النماذج الخطية: SHAP، eli5#### الميزات والتحقق من الصحة
 - العدادات في مهام التعلم الخاضعة للإشراف: WOE، والاحتمالية الميسرة وطرق أخرى لهندسة الميزات بناءً على الميزة المستهدفة
 - هناك الكثير في Scikit-learn لم نغطيه (NestedCV، GroupKFold، إلخ)
#### التعبئة والغابات العشوائية
 - تفسير الترددات اللاسلكية عن طريق تحويل الغابات إلى أشجار
 - ضغط الغابات العشوائية
 - الاستدلال المختلفة لتقييم أهمية الميزة في الغابات وتعزيزها
 
#### التعلم غير الخاضع للرقابة
 - مراجعة بعض أساليب التجميع (مع التحفيز، وسبب الحاجة إليه)
 - تنفيذ PCA الفعال، PCA العشوائي، PCA عبر الإنترنت
 - التعلم غير الخاضع للرقابة في المهام الحقيقية
 - تطبيق Word2Vec على البيانات المتسلسلة (حتى جلسات موقع الويب كما هو الحال في مسابقة "Alice")
#### دولار سنغافوري وفوبال وابيت
 - شيء يغطي المادة الدراسية بشكل أفضل، على سبيل المثال. مراجعة أوسع لـ Vowpal Wabbit 
 - تحليل المصفوفة وتنفيذ FMs وVowpal Wabbit
 - خوارزمية FTRL: اتبع القائد المنظم
#### سلسلة زمنية
 - التنبؤ بسلاسل زمنية متعددة في نفس الوقت
 - كشف الانحرافات في السلاسل الزمنية
 
#### التعزيز
 - نظرة عامة على CatBoost + أمثلة حيث يعمل بشكل جيد حقًا 
 - نظرة عامة على مكتبة H2O
 - تعزيز التدرج وتسريع GPU
#### أخرى
 - اختبارات أ/ب والتشذير
 - الطرق البايزية لتحسين المعلمات الفائقة
 - مجموعات بيانات الإصدار (مثل DVC.org)
 - تحسين المقاييس غير التافهة: النصائح والحيل
 - أقرب إلى مقاييس الأعمال: نمذجة الارتقاء (الرفع)
