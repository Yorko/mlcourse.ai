### 1.3. فهم موجز لـ XML وXpath.



إن _المفتاح_ والفرق الوحيد بين الاثنين الذي سنحتاجه هو أن **XML** يستخدم لتخزين ونقل البيانات، بينما **HTML** يستخدم لتنسيق وعرض نفس البيانات.<br>
ما هو **XML**؟
* على الرغم من أن XML يشبه إلى حد كبير HTML، إلا أن له غرضًا وشجاعة مختلفة تمامًا. XML هو اختصار لـ `eXtensible Markup Language`، وهو ما يفسر نفسه في الواقع.<br>
* من المثير للدهشة أن لغة XML لا تفعل أي شيء، فهي تقوم فقط ببناء البيانات وتخزينها ونقلها عند الطلب.<br>
* أحد أسباب تسميته بـ eXtensible هو أنه يمكنك اختراع علاماتك الخاصة، مما يساعدك على التنقل بين البيانات بالطريقة التي تريدها، بينما يحتوي HTML على علامات محددة مسبقًا وتستند جميع مستندات HTML إلى علامات موحدة، مثل `<body>, <p>, <li>` وما إلى ذلك.<br>
* يساعد هذا المطور على ابتكار العلامات الخاصة به وتنظيم البيانات بالطريقة التي تناسب طبيعة المستند. ومع ذلك، فإن XML ليس بديلاً لـ HTML، ولكنه امتداد (بجدية يا رجل؟).<br>لذلك، في معظم حلول الويب، يتم صياغة الكلمات بشكل تآزري، حيث ينقل XML وتنسيقات HTML ويعرض البيانات بشكل جيد.
كل هذا جعل XML أداة حيوية للإنترنت ويتم استخدامها في كل مكان، حيث يتعين على المرء نقل البيانات بين جميع أنواع التطبيقات.<br>



هكذا يبدو **رمز XML**:


```python
<?xml version="1.0" encoding="UTF-8"?>
<bookstore>
   <book category="COOKING">
      <title lang="en">Everyday Italian</title>
      <author>Giada De Laurentiis</author>
      <year>2005</year>
      <price>30.00</price>
   </book>
   <book category="CHILDREN">
      <title lang="en">Harry Potter</title>
      <author>J K. Rowling</author>
      <year>2005</year>
      <price>29.99</price>
   </book>
   <book category="WEB">
      <title lang="en">XQuery Kick Start</title>
      <author>James McGovern</author>
      <author>Per Bothner</author>
      <author>Kurt Cagle</author>
      <author>James Linn</author>
      <author>Vaidyanathan Nagarajan</author>
      <year>2003</year>
      <price>49.99</price>
   </book>
   <book category="WEB">
      <title lang="en">Learning XML</title>
      <author>Erik T. Ray</author>
      <year>2003</year>
      <price>39.95</price>
   </book>
</bookstore>
```

ما هو **XPath**؟<br>
* XPath هي لغة خاصة لتحديد أجزاء من مستندات XML والبحث عن المعلومات وتحديدها.<br>
* يستخدم تعبيرات المسار للتنقل، والتي تشبه إلى حد كبير الاستعلامات.<br>
* كما تحتوي على قائمة من الدوال (المنطقية والعددية) لاختبار البيانات.<br>
وهذه هي الطريقة التي تبدو بها هياكل الاستعلام. بمساعدة هؤلاء يمكننا التعمق في تدوين **XML** باستخدام **Xpath** لعناصر البيانات، باستخدام المحددات الهرمية:


|تعبير XPath | النتيجة |
|------|------|
|   **/bookstore/book[1]** |تحديد عنصر الكتاب الأول الذي يكون تابعًا لعنصر المكتبة |
|**/bookstore/book[last( )]**|تحديد عنصر الكتاب الأخير الذي يعتبر فرعًا لعنصر المكتبة|
|**/bookstore/book[last() -1]**|تحديد آخر عنصر في الكتاب باستثناء عنصر واحد وهو تابع لعنصر المكتبة|
|**/bookstore/book[position() <3]**|تحديد أول عنصرين من عناصر الكتاب وهما فرعان لعنصر المكتبة|
|**//title[@lang]**|تحديد كافة عناصر العنوان التي لها سمة تسمى lang|
|**//title[@lang='eng']**|تحديد كافة عناصر العنوان التي تحتوي على سمة "lang" بقيمة "en"|
|**/bookstore/book[price>35.00]**|تحديد كافة عناصر الكتاب لعنصر المكتبة التي تحتوي على عنصر سعر بقيمة أكبر من 35.00|
|**/bookstore/book[price>35.00]/title**|تحديد كافة عناصر عنوان عناصر الكتاب لعنصر المكتبة التي تحتوي على عنصر سعر بقيمة أكبر من 35.00|



في الوقت الحالي، يكفي معرفة أن البيانات النصية الموجودة على موقع الويب مخزنة بتنسيق **XML** ويمكننا استردادها بكفاءة عن طريق الاستعلام باستخدام **Xpath**. <br>
**Xpath** بدورها، هي لغة بحثية، تعتمد على بنية العلامات الهرمية، وتشكل بنية قائمة على الشجرة، والتي يمكن تحليلها واختيارها ومعالجتها بسهولة لأغراضنا.<br>

```python
import re

sales = [response.xpath(".//div[@class='b-product-block__gain']").extract_first() if 'b-product-block__gain' in i else '0' for i in response.xpath(".//div[@class='b-product-block__price']").extract()]
sales = [sale.replace("\xa0", "") for sale in sales]  # cleaning from non-breaking space in Latin1(ISO 8859-1)
sales = [sale.strip() for sale in sales] # cleaning from unwanted spaces 
sales = [re.findall("\d+", sale) for sale in sales] # finding all objects that contain digits in our list of ojects
sales = [item for sublist in sales for item in sublist] # flatten the list of lists 
sales = [int(sale) for sale in sales] # turning string objects to integers
```

## 6. كل ذلك معًا. نص كامل.



يمكن أن يكون البرنامج النصي أدناه مجرد _"Ctl+C/Ctrl-V"_ إلى برنامج نصي عنكبوت/زاحف رئيسي - `svy.py`. لا تنس التعديلات الإضافية في `settings.py` <br>
**_ملاحظة_** ضع في اعتبارك مشكلة المسافة البادئة (4 مسافات أو علامة تبويب) عند كتابة/تصحيح التعليمات البرمجية في برامج تحرير النصوص/IDE. <br>
في حالتي، أختار علامات التبويب.<br>

```python
import re

import scrapy


class SvzSpider(scrapy.Spider):


	custom_settings = {'FEED_URI' : 'results/svyaznoy.csv'}

	name = 'svyaznoy'

	""" Making a proper list of pages. """
	allowed_domains = ['http://www.svyaznoy.ru/']
	first_page = ['http://www.svyaznoy.ru/catalog/phone/224/']
	all_others = ['http://www.svyaznoy.ru/catalog/phone/224/page-'+str(x) for x in range(2,109)]
	
	""" Inserting the 1st page. """
	start_urls = first_page+all_others


	def parse(self, response):
        
		# Retrieving objects
		imgs = response.xpath("//div[@class='b-product-block__image']//img[@class='lazy']/@data-original").extract()
		titles = response.xpath("//div[@class='b-product-block__image']//img[@class='lazy']/@title").extract()
		prices = response.xpath(".//div[@class='b-product-block__price']//span[@class='b-product-block__visible-price']/text()").extract()
		sales = [response.xpath(".//div[@class='b-product-block__gain']").extract_first() if 'b-product-block__gain' in i else '0' for i in response.xpath(".//div[@class='b-product-block__price']").extract()]

		# Processing prices
		prices = [price.replace("\n", "") for price in response.xpath(".//div[@class='b-product-block__price']//span[@class='b-product-block__visible-price']/text()").extract()]
		prices = [price.replace("\xa0", "") for price in prices]
		prices = [price.strip() for price in prices]
		prices = [int(price) for price in prices if price]

		# Processing sale offers
		sales = [sale.replace("\xa0", "") for sale in sales] 
		sales = [sale.strip() for sale in sales] 
		sales = [re.findall("\d+", sale) for sale in sales]
		sales = [item for sublist in sales for item in sublist] 
		sales = [int(sale) for sale in sales] 
        
		# Yielding objects    
		for item in zip(titles,prices,sales,imgs):
				scraped_info = {
					'title' : item[0],
					'price' : item[1],
					'sale_offer': sales[2],
					'image_urls' : [item[3]]}
				yield scraped_info
  
		# Pagination loop       
		next_page = response.xpath(".//li[@class='next']//a/@data-page").extract()
		next_page = str(int(next_page[0])+1)
		if next_page is not None:
			next_page = response.urljoin(next_page)
			yield scrapy.Request(next_page, callback=self.parse)
```
