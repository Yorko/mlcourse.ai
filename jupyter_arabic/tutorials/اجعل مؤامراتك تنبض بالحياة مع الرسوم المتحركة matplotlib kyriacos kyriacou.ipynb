### 2. الرسوم المتحركة ماتبلوتليب



لحسن الحظ بالنسبة لنا، يوفر matplotlib بالفعل مكتبات ممتازة لإنشاء الرسوم المتحركة.



يوفر Matplotlib فئتين لإنشاء الرسوم المتحركة: [FuncAnimation](https://matplotlib.org/api/_as_gen/matplotlib.animation.FuncAnimation.html#matplotlib.animation.FuncAnimation) و[ArtistAnimation](https://matplotlib.org/api/_as_gen/matplotlib.animation.ArtistAnimation.html#matplotlib.animation.ArtistAnimation).
مع الأول، يتم إنشاء الرسوم المتحركة بشكل متكرر عن طريق استدعاء وظيفة التحديث، بينما مع الإطارات الأخيرة يتم حسابها مسبقًا ثم تحريكها. سنلقي نظرة على كليهما في هذا البرنامج التعليمي.عند استخدام أي من فئات الرسوم المتحركة، من الضروري الاحتفاظ بمرجع إلى كائن المثيل. يتم تقديم الرسوم المتحركة بواسطة مؤقت يحمل الكائن `Animation` المرجع الوحيد له. إذا لم تحتفظ بهذا المرجع، فسيتم جمع البيانات المهملة مما يؤدي إلى إيقاف الرسوم المتحركة. وهذا يتعارض مع وظائف التخطيط الأخرى لـ `matplotlib`، حيث يمكنك استخدام `plt.plot()` دون تخزين النتيجة، لذا فهذه نقطة مهمة يجب تذكرها.
دعونا نلقي نظرة على بعض التعليمات البرمجية لتحريك وظيفة <i>cosine</i>، وبعد ذلك سنتعمق في شرح كل وظيفة وواجهة برمجة التطبيقات `FuncAnimation`.


````python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

sns.set(rc={"figure.figsize": (10, 7)})

import warnings

warnings.filterwarnings("ignore")

# to display animation in notebook
from IPython.display import HTML
from matplotlib.animation import ArtistAnimation, FuncAnimation
````

````python
fig, ax = plt.subplots()
ax.set_xlim((0, 2))
ax.set_ylim((-2, 2))
ax.grid(True)
# create a line which we'll animate
(line,) = plt.plot([], [], lw=2)

# function used to draw a clear frame
def init():
    line.set_data([], [])
    return (line,)


# called sequentialy on each frame
def update(frame):
    x = np.linspace(0, 2, 1000)
    y = np.cos(2 * np.pi * (x - 0.01 * frame))
    line.set_data(x, y)
    return (line,)


# hide static plot
plt.close()

ani = FuncAnimation(fig, update, frames=100, interval=20, init_func=init, blit=True)
HTML(ani.to_html5_video())
# ani.save('cosine_function.mp4', fps=30)
````


دعونا نلقي نظرة على المعلمات المهمة لـ `FuncAnimation` API:



`matplotlib.animation.FuncAnimation(fig, func, frames=None, interval=None, init_func=None, blit=True)`
- **الشكل**: الكائن الشكلي المستخدم في الرسم 
- **func**: الوظيفة التي يتم الاتصال بها في كل إطار
- **الإطارات**: كم عدد الإطارات المراد تحريكها. إذا لم يكن هناك شيء، قم بتمرير itertools.count
- **الفاصل الزمني**: التأخير بين الإطارات المتتالية بالمللي ثانية
- **init_func**: الوظيفة المستخدمة لرسم إطار واضح. تم الاتصال به مرة واحدة قبل الإطار الأول
- **blitting**: يتحكم في ما إذا كان سيتم استخدام blitting أم لا. يعمل Blitting على تحسين الرسوم المتحركة من خلال إعادة رسم الأجزاء التي تغيرت فقط


### 3. تحريك كائنات متعددة



الآن بعد أن أصبحنا على دراية بتحريك كائن واحد، فلنذهب إلى أبعد من ذلك من خلال تحريك كائنات متعددة. دعونا نضيف سطر <i>sine</i> ونعرض رقم الإطار:


````python
fig, ax = plt.subplots()
ax.set_xlim((0, 2))
ax.set_ylim((-2, 2))
ax.grid(True)

frame_text = ax.text(0.05, 1.9, "", bbox=dict(facecolor="white", alpha=1))
(cos_line,) = plt.plot([], [], lw=2, c="b", label="cos")
(sin_line,) = plt.plot([], [], lw=2, c="r", label="sin")
plt.legend()

# function used to draw a clear frame
def init():
    cos_line.set_data([], [])
    sin_line.set_data([], [])
    frame_text.set_text("")
    return cos_line, sin_line, frame_text


# called sequentialy on each frame
def update(frame):
    x = np.linspace(0, 2, 1000)
    cos_x = np.cos(2 * np.pi * (x - 0.01 * frame))
    sin_x = np.sin(2 * np.pi * (x - 0.01 * frame))
    cos_line.set_data(x, cos_x)
    sin_line.set_data(x, sin_x)
    frame_text.set_text("frame = %d" % frame)
    return cos_line, sin_line, frame_text


# hide static plot
plt.close()

ani = FuncAnimation(fig, update, frames=100, interval=20, init_func=init, blit=True)
ani
````


### 4. تحريك KMeans



على الرغم من أنني متأكد من أن النظر إلى مؤامرة الرسوم المتحركة cos/sin أمر ممتع للغاية، فلنقم بإنشاء بعض الرسوم المتحركة التي يمكن أن تكون مفيدة لنا في رحلة تعلم الآلة.في الموضوع السابع، تعلمنا عن خوارزميات التجميع، وعلى وجه الخصوص K-Means. ومع ذلك، لم نتمكن من رسم المجموعات النهائية إلا بعد اكتمال جميع التكرارات. سيكون أكثر إثارة للاهتمام أن نرى كيف تتشكل المجموعات وتتغير خلال كل تكرار. يبدو أن الرسوم المتحركة مثالية لهذا!
سنستخدم مجموعة بيانات القزحية لهذه المهمة. نظرًا لأن مجموعة بيانات القزحية تتكون من ثلاث ميزات، فلنستخدم PCA لتقليل ذلك إلى 2، حتى نتمكن من رسمها في بعدين، تمامًا كما فعلنا في الموضوع 7:


````python
from sklearn import datasets, decomposition
from sklearn.cluster import KMeans

iris = datasets.load_iris()
X = iris.data
y = iris.target

pca = decomposition.PCA(n_components=2)
X_centered = X - X.mean(axis=0)
pca.fit(X_centered)
X_pca = pca.transform(X_centered)
X_pca.shape
````


دعونا أولاً نلقي نظرة على بياناتنا باستخدام التصنيفات الحقيقية:


````python
plt.plot(X_pca[y == 0, 0], X_pca[y == 0, 1], "bo", label="Setosa")
plt.plot(X_pca[y == 1, 0], X_pca[y == 1, 1], "go", label="Versicolour")
plt.plot(X_pca[y == 2, 0], X_pca[y == 2, 1], "yo", label="Virginica")
plt.grid(False)
plt.legend(loc=0);
````


لسوء الحظ، لا يوفر sklearn طريقة لعرض البيانات التاريخية للنقط الوسطى - أي أنه لا يمكننا رؤية مواقع النقط الوسطى النهائية إلا بعد <i>n</i> التكرارات، ولكننا نريد رؤية النقط الوسطى في كل تكرار. هناك طريقتان للتغلب على ذلك، يمكننا إما إنشاء تطبيقنا الخاص لـ KMeans الذي يخزن ويعيد موقع النقطه الوسطى بعد كل تكرار أو مع القليل من البراعة يمكننا استخدام KMeans من sklearn للحصول على السنتروا في كل تكرار.
سأقوم بتهيئة النقط الوسطى بنفسي، ثم سأتصل على التوالي بـ KMeans بحالة عشوائية عشوائية ثابتة والنقط الوسطى الأولية، وفي كل تكرار سأزيد المعلمة `max_iter` بمقدار 1 وأتناسب مع البيانات. بهذه الطريقة، في كل تكرار للحلقة، يمكننا أن ننظر إلى النقط الوسطى. سيكون الأمر أكثر منطقية عندما تقرأ الكود.
بالنسبة لهذه الرسوم المتحركة سنستخدم فئة `ArtistAnimation`.


````python
fig, ax = plt.subplots()
ax.set_xlim((-3.5, 4))
ax.set_ylim((-1.5, 1.5))
ax.grid(False)
ims = []

# initial centroids
init = np.array([-1, 0.5, 3, 1, 3, -0.5]).reshape((3, 2))

for i in range(10):
    txt = plt.text(
        -0.5, 1.4, "Iteration = {}".format(i + 1), bbox=dict(facecolor="white", alpha=1)
    )

    kmeans = KMeans(n_clusters=3, init=init, max_iter=i + 1, random_state=6)
    y = kmeans.fit_predict(X_pca, y)

    centers = plt.scatter(
        kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], c="r", marker="X"
    )
    (c1,) = plt.plot(X_pca[y == 0, 0], X_pca[y == 0, 1], "bo")
    (c2,) = plt.plot(X_pca[y == 1, 0], X_pca[y == 1, 1], "go")
    (c3,) = plt.plot(X_pca[y == 2, 0], X_pca[y == 2, 1], "yo")
    ims.append([centers, c1, c2, c3, txt])

# hide static plot
plt.close()

ani = ArtistAnimation(fig, ims, interval=1500, repeat_delay=1000, blit=False)
ani
````


matplotlib.animation.ArtistAnimation (الشكل، الفنانين، الفاصل الزمني = 200، blit = خطأ، كرر_delay = لا شيء)- **الشكل**: الكائن الشكلي المستخدم في الرسم 
- **الفنانون**: تُدخل كل قائمة مجموعة من فناني matplotlib الذين يمثلون ما يجب تمكينه في كل إطار. مزيد من المعلومات عن الفنانين: https://matplotlib.org/users/artists.html
- **الفاصل الزمني**: التأخير بين الإطارات المتتالية بالمللي ثانية
- **blit**: يتحكم في ما إذا كان سيتم استخدام blitting أم لا. يعمل Blitting على تحسين الرسوم المتحركة من خلال إعادة رسم الأجزاء التي تغيرت فقط
- **repeat_delay**: إذا كانت الرسوم المتحركة متكررة، فسيتم إضافة تأخير بالمللي ثانية قبل تكرار الرسوم المتحركة

### 5. تحريك حدود شجرة القرار



الآن دعونا نلقي نظرة على كيفية تغير حدود شجرة القرار أثناء تغيير المعلمة `max_depth`. سنكون قادرين على رؤية رسم متحرك للشجرة التي تتلاءم مع البيانات وفي النهاية تتلاءم مع البيانات!
لإنشاء النقاط وعرضها، سأستخدم الكود من الموضوع 3 في mlcourse.ai.


````python
# generate points
np.random.seed(6)

class_1 = np.random.normal(size=(50, 2))
class_1_lbl = np.zeros(50)
class_2 = np.random.normal(size=(50, 2), loc=1.8)
class_2_lbl = np.ones(50)

train_data = np.r_[class_1, class_2]
train_labels = np.r_[class_1_lbl, class_2_lbl]

plt.scatter(
    train_data[:, 0],
    train_data[:, 1],
    c=train_labels,
    s=100,
    cmap="autumn",
    edgecolors="black",
    linewidth=1,
);
````

````python
from sklearn.tree import DecisionTreeClassifier


def get_grid(data):
    x_min, x_max = data[:, 0].min() - 1, data[:, 0].max() + 1
    y_min, y_max = data[:, 1].min() - 1, data[:, 1].max() + 1
    return np.meshgrid(np.arange(x_min, x_max, 0.01), np.arange(y_min, y_max, 0.01))


fig, ax = plt.subplots()
ax.grid(False)
ims = []
xx, yy = get_grid(train_data)

for i in range(8):
    txt = plt.text(
        0, 4.5, "DT max depth = {}".format(i + 1), bbox=dict(facecolor="white", alpha=1)
    )

    pred = (
        DecisionTreeClassifier(criterion="gini", max_depth=i + 1, random_state=17)
        .fit(train_data, train_labels)
        .predict(np.c_[xx.ravel(), yy.ravel()])
        .reshape(xx.shape)
    )
    pcm = ax.pcolormesh(xx, yy, pred, cmap="autumn")

    pts = plt.scatter(
        train_data[:, 0],
        train_data[:, 1],
        c=train_labels,
        s=100,
        cmap="autumn",
        edgecolors="black",
        linewidth=1,
    )
    ims.append([pcm, pts, txt])

# hide static plot
plt.close()

ani = ArtistAnimation(fig, ims, interval=2000, repeat_delay=1000, blit=False)
ani
````

كما ترون أعلاه، يبدأ الثلاثة في التجهيز عند أقصى عمق = 4. الكود هو تقريبًا نفس كود تصور KMeans، وعلى هذا النحو لن أشرح كل جزء.

لدى Matplotlib العديد من الأمثلة على الرسوم المتحركة [هنا](https://matplotlib.org/api/animation_api.html#examples). سأعرض أدناه واحدًا يعرض إمكانيات ثلاثية الأبعاد:


````python
# taken from https://matplotlib.org/gallery/animation/random_walk.html

import mpl_toolkits.mplot3d.axes3d as p3

# Fixing random state for reproducibility
np.random.seed(19680801)


def Gen_RandLine(length, dims=2):
    """
    Create a line using a random walk algorithm

    length is the number of points for the line.
    dims is the number of dimensions the line has.
    """
    lineData = np.empty((dims, length))
    lineData[:, 0] = np.random.rand(dims)
    for index in range(1, length):
        # scaling the random numbers by 0.1 so
        # movement is small compared to position.
        # subtraction by 0.5 is to change the range to [-0.5, 0.5]
        # to allow a line to move backwards.
        step = (np.random.rand(dims) - 0.5) * 0.1
        lineData[:, index] = lineData[:, index - 1] + step

    return lineData


def update_lines(num, dataLines, lines):
    for line, data in zip(lines, dataLines):
        # NOTE: there is no .set_data() for 3 dim data...
        line.set_data(data[0:2, :num])
        line.set_3d_properties(data[2, :num])
    return lines


# Attaching 3D axis to the figure
fig = plt.figure()
ax = p3.Axes3D(fig)

# Fifty lines of random 3-D lines
data = [Gen_RandLine(25, 3) for index in range(50)]

# Creating fifty line objects.
# NOTE: Can't pass empty arrays into 3d version of plot()
lines = [ax.plot(dat[0, 0:1], dat[1, 0:1], dat[2, 0:1])[0] for dat in data]

# Setting the axes properties
ax.set_xlim3d([0.0, 1.0])
ax.set_xlabel("X")

ax.set_ylim3d([0.0, 1.0])
ax.set_ylabel("Y")

ax.set_zlim3d([0.0, 1.0])
ax.set_zlabel("Z")

ax.set_title("3D Test")

plt.close()

# Creating the Animation object
ani = FuncAnimation(fig, update_lines, 25, fargs=(data, lines), interval=50, blit=False)

ani
````
