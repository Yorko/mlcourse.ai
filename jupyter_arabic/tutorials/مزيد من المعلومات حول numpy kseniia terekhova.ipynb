### 3. بث مصفوفة NumPy



<p>أحد الأشياء التي يجب ملاحظتها في المثال أعلاه هو أنه يتعامل مع مصفوفات ذات أبعاد مختلفة. تحتوي np_point على الشكل (2،) وnp_other_points - (10000000، 2). ومع ذلك، يتم إجراء عمليات عنصرية باستخدامها، مما يؤدي إلى إنتاج مصفوفة ذات شكل (10000000،). هذا ممكن بفضل آلية <b>@KEEP_00036@@matrix البث</i></b>.</p>
<p>بث Matrix في NumPy عبارة عن مجموعة من القواعد التي تسمح بتوسيع مصفوفتين بأبعاد مختلفة لمطابقة أشكال بعضها البعض، من أجل تنفيذ عمليات عنصر بعنصر. الأهم من ذلك، أن مجموعة القواعد هذه هي آلية "افتراضية"، تسمح فقط بفهم كيفية تفاعل المصفوفات. لم يتم إجراء أي توسيع حقيقي أو تخصيص للذاكرة.</p>
<p> أبسط حالة هي جمع مصفوفة برقم قياسي، سيتم إضافته إلى كل عنصر من عناصر المصفوفة:</p>


```python
M = np.ones((3, 3))
print("M:")
print(M)

scalar = 2
M_added = M + scalar

print()
print("M +", scalar, ":")
print(M_added)
```


يمكن جمع المصفوفة بالمتجه بالطريقة المماثلة:


```python
M = np.ones((3, 3))
print(M)

v = np.array([0, 1, 2])
print()
print(v)

M_added = M + v
print()
print(M_added)
```


في حالة أكثر تعقيدًا، يتم بث المصفوفتين:


```python
a = np.arange(3)
b = np.arange(3)[:, np.newaxis]

print("a:", a)
print("b:\n", b)

summed = a + b
print("\nSummed:")
print(summed)
```

قواعد بث المصفوفة هي بهذه الطريقة:<br>
<b>القاعدة 1:</b> إذا كان الصفيفان يختلفان في عدد أبعادهما، فسيتم تعبئة شكل الصفيف ذي الأبعاد الأقل بتلك الموجودة على جانبه الأمامي (الأيسر).<br>
<b>القاعدة 2:</b> إذا كان شكل الصفيفين غير متطابق في أي بعد، فسيتم تمديد المصفوفة التي لها شكل يساوي 1 في ذلك البعد لتتناسب مع الشكل الآخر.<br>
<b>القاعدة 3:</b> إذا كانت الأحجام مختلفة في أي بُعد ولم يكن أي منها يساوي 1، فسيظهر خطأ.<br>



لذلك، تتم إضافة مصفوفة ثنائية الأبعاد إلى مصفوفة أحادية البعد بهذه الطريقة:

### 4. ملاحظة سريعة حول ضرب المصفوفات


بالطبع، الجمع ليس العملية الوحيدة التي يمكن تطبيقها على المصفوفات؛ ويمكن استخدام عدد من العمليات الحسابية، إلى جانب العديد من "الوظائف العامة". العملية التي أود أن أوليها بعض الاهتمام هي الضرب، أي '*'. مثل العمليات الحسابية الأخرى، يتم تطبيقها في NumPy بشكل عنصري:


```python
A = np.full((2, 2), 2)
print("A:")
print(A)
print()

B = np.full((2, 2), 3)
print("B:")
print(B)
print()

print("A*B:")
print(A*B)
```


ولكن عند الحديث عن ضرب المصفوفات، خاصة في الجبر الخطي (وتعلم الآلة)، غالبًا ما يتم الإشارة إلى عملية أخرى، <i>matrix Product</i>، والتي يتم تعريفها بهذه الطريقة (صيغة من <a href="@@KEEP_00004@@ target=__blank>wikipedia</a>):



$$ A = \begin{pmatrix} a_{11}, a_{12} & \cdots & a_{1m}\\a_{21}, a_{22} & \cdots & a_{2m} \\ \vdots & \ddots & \vdots \\ a_{n1}, a_{n2} & \cdots & a_{nm} \end{pmatrix},  B = \begin{pmatrix} b_{11}, b_{12} & \cdots & b_{1p}\\b_{21}, b_{22} & \cdots & b_{2p} \\ \vdots & \ddots & \vdots \\ b_{m1}, b_{m2} & \cdots & b_{mp} \end{pmatrix} $$
منتج المصفوفة <b>C</b> = <b>AB</b>:
$$C = \begin{pmatrix} c_{11}, c_{12} & \cdots & c_{1p} \\c_{21}, c_{22} & \cdots & c_{2p} \\ \vdots & \ddots & \vdots \\ c_{n1}, c_{n2} & \cdots & c_{np} \end{pmatrix} $$ 
$c_{ij} = a_{i1}b_{1j} + ... +  a_{im}b_{mj} = \sum_{k=1}^{m} a_{ik}b_{kj}$



يتم استخدام الدالة <i>numpy.dot()</i> أو الاختصار "@" لهذا الغرض في NumPy. من غير الجيد الخلط بين هاتين العمليتين، خاصة عند وجود بث مصفوفي:


```python
A = np.arange(1, 10).reshape(3,3)
print("A:")
print(A)
print()

B = np.full((3,), 3)
print("B:")
print(B)
print()

print("A*B:")
print(A*B)

print()
print("A@B:")
print(A@B)
```

### 5. لا فوضى مع np.meshgrid



لقد استخدمنا الدالة np.meshgrid في مكان ما من الدورة، ولكن دون أي تفسيرات. هذا أمر مؤسف في رأيي، لأنه ليس من السهل فهم ما يفعله من NumPy <a href="@@KEEP_00005@@ target=__blank>documentation</a>:
<blockquote cite="@@KEEP_00006@@>
numpy.meshgrid(*xi, **kwargs)<br>
إرجاع المصفوفات الإحداثية من متجهات الإحداثيات.<br>
قم بإنشاء صفائف إحداثيات ND للتقييمات الموجهة للحقول العددية/المتجهة ND عبر شبكات ND، نظرًا لصفائف الإحداثيات أحادية البعد x1، x2،...، xn.
</blockquote>


ومع ذلك، فإن المرة الأولى التي رأيت فيها استخدامه كانت في <a href="@@KEEP_00007@@ target=__blank>article</a> حول شبكات المحولات المكانية، في خطوة "شبكة الهوية" و"الشبكة المتداخلة المحولة". لذلك، يمكن أن يكون مادة مفيدة. <br>
في الواقع، تقوم هذه الوظيفة فقط بإنشاء مجموعة من الشبكات بإحداثيات x وy وما إلى ذلك في مواقع الشبكة المقابلة. 


```python
import matplotlib.pyplot as plt
from matplotlib import cm

xvalues = np.arange(-1, 1.05, 0.5)
yvalues = np.arange(-1, 1.05, 0.5)

xx, yy = np.meshgrid(xvalues, yvalues)
print(xx)
print()
print(yy)
grid = plt.plot(xx, yy, marker='.', color='k', linestyle='none')
```


بعد ذلك، يمكن استخدام شبكات الإحداثيات هذه لحساب قيم الدوال متعددة المتغيرات، أو لتصور شيء جميل.


```python
x = np.arange(-8, 8, 0.01)
y = np.arange(-8, 8, 0.01)
xx, yy = np.meshgrid(x, y, sparse=True)
z = np.sin(xx**2 + yy**2) / (xx**2 + yy**2)

h = plt.contourf(x,y,z, cmap=cm.PuBu_r)
```

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(10, 8))
ax = fig.gca(projection='3d')
ax.view_init(60, 35)

# Make data.
X = np.arange(-5, 5, 0.25)
Y = np.arange(-5, 5, 0.25)
X, Y = np.meshgrid(X, Y)
R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)

# Plot the surface.
surf = ax.plot_surface(X, Y, Z, cmap=cm.coolwarm, linewidth=0, antialiased=False)
```

#### 6.1 ك-أقرب الجيران



كانت هناك عينة أعلاه، والتي تحسب المسافات من نقطة واحدة إلى مجموعة من النقاط الأخرى. أما إذا كان الأمر بحاجة إلى إيجاد أقرب النقاط لكل نقطة في المجموعة فيمكن إجراء ذلك بهذه الطريقة:


```python
%matplotlib inline
import matplotlib.pyplot as plt

import seaborn; seaborn.set()

rand = np.random.RandomState(42)
X = rand.rand(10, 2)
plt.scatter(X[:, 0], X[:, 1], s=100);
```

```python
 dist_sq = np.sum((X[:, np.newaxis, :] - X[np.newaxis, :, :]) ** 2, axis=-1)
```


هنا، تسمح إضافة محور جديد بتحويل مصفوفة X ذات الشكل (10، 2) إلى مصفوفتين (10، 1، 2) و (1، 10، 2)، تحتويان على عشر نقاط ثنائية الأبعاد في الصفوف والأعمدة على التوالي؛ ومن ثم، يسمح البث بحساب الفرق بين إحداثيات النقاط في الصف i والعمود j؛ ثم تعطي العمليات المربعة والمجموع مسافة إقليدية مربعة بين تلك النقاط. <br>
تحتوي مصفوفة dist_qs على أصفار على قطرها، مما يثبت صحة الحساب. المسافة بين العنصر ii هي المسافة بين النقطة i ونفسها؛ هذا هو الصفر.


```python
dist_sq.diagonal()
```


ثم تسمح الدالة <i>np.argsort</i> بفرز العناصر في كل صف وطباعة فهارس النقاط الأخرى بترتيب بعدها عن النقطة i:


```python
nearests = np.argsort(dist_sq, axis=1)
print(nearests)
```

إذا كانت هناك حاجة فقط إلى أقرب نقاط K (غير المصنفة)، فإن الدالة <i>np.argpartition</i> تسمح بأخذها فقط، دون فرز الصفوف بأكملها: 


```python
K = 2
nearest_partition = np.argpartition(dist_sq, K + 1, axis=1)
print(nearest_partition[:, :K+1])
```


ويمكن تصور ذلك:


```python
cmap = plt.get_cmap('viridis')
colors = cmap(np.linspace(0, 1, 10))
plt.scatter(X[:, 0], X[:, 1], s=150, color=colors)

# draw lines from each point to its two nearest neighbors
K = 2

for i in range(X.shape[0]):
    for j in nearest_partition[i, :K+1]:
        # plot a line from X[i] to X[j]
        # the lines colors correspond to outgoing point color, but some lines obviously overlap
        # using some zip magic:
        plt.plot(*zip(X[j], X[i]), c=colors[i])
```


#### 6.2 لعبة الحياة لكونواي



<a href="@@KEEP_00008@@ target=__blank>Conway's Game of Life</a> هي نموذج كلاسيكي لنموذج الإنسان الآلي الخلوي ولعبة بدون لاعبين. في حالته الأولية، يبدأ في عيش حياته الخاصة من خلال تطبيق القواعد التالية في كل خطوة:<ul>
<li>كل خلية في شبكة ثنائية الأبعاد تكون "حية"(1) أو "ميتة"(0)</li>
<li>أي خلية حية لها 2 أو 3 جيران تبقى على قيد الحياة، وإلا ماتت [0,1 أو أكثر من 4 جيران]</li>
<li>أي خلية بها 3 جيران بالضبط تصبح حية (إذا كانت ميتة)</li></ul>



ما الذي يتطلبه تنفيذ هذه اللعبة باستخدام صفائف NumPy؟ إنه أمر صعب، ولكن ليس طويلاً:


```python
def iterate(Z):
    # Count neighbours
    N = (Z[0:-2,0:-2] + Z[0:-2,1:-1] + Z[0:-2,2:] +
         Z[1:-1,0:-2]                + Z[1:-1,2:] +
         Z[2:  ,0:-2] + Z[2:  ,1:-1] + Z[2:  ,2:])

    # Apply rules
    birth = (N==3) & (Z[1:-1,1:-1]==0)
    survive = ((N==2) | (N==3)) & (Z[1:-1,1:-1]==1)
    Z[...] = 0
    Z[1:-1,1:-1][birth | survive] = 1
    return Z
```


<p> أولاً، تقوم بتقطيع المصفوفة الأصلية إلى شرائح بخطوات رأسية 0 و1 و2، وبالتالي الحصول على مصفوفات ذات صفوف "متوسطة"، وصفوف مُزاحة بمقدار -1 عن الصفوف "الوسطى"، وصفوف مُزاحة بمقدار +1. يتم تنفيذ الإجراء المماثل مع الأعمدة. يؤدي دمج هذه الخطوات في اتجاهات مختلفة إلى إنشاء 8 مصفوفات مُزاحة؛ مجموع العناصر الخاص بهم يعطي مقدارًا من العناصر المجاورة "الأحياء" لكل عنصر من عناصر مصفوفة N.</p>
<p>ثم، من خلال تطبيق قواعد اللعبة على مصفوفات N وZ، يمكن حساب مصفوفتي القناع المنطقي "الميلاد" و"البقاء على قيد الحياة". بعد ذلك، تسمح الفهرسة المنطقية NumPy بتعيين الخلايا التي ولدت أو نجت فقط على "1".</p>



يسمح الكود التالي بتحريك عملية اللعبة عند تشغيل الكمبيوتر المحمول. بالقفز، سيعرض برنامج nbviewer الصورة الناتجة على الأقل :)


```python
import matplotlib.pyplot as plt
import numpy as np

%matplotlib notebook
plt.ion()

#initialize game field
Z = np.random.choice([0,1],size=(100,100))

fig = plt.figure()
ax = fig.add_subplot(111)
fig.show()

for _ in range(1000):    
    #update
    Z = iterate(Z)
    
    #re-draw image
    ax.clear()
    ax.imshow(Z,cmap='gray')
    fig.canvas.draw()
```

#### 6.3 الطبقات التلافيفية في الشبكات العصبية التلافيفية



الموضوع أعلاه قريب جدًا من مجال ML، حيث أن الإجراء الذي يتم تنفيذه هناك يشبه مبدأ الطبقات التلافيفية في الشبكات العصبية التلافيفية (CNN).<br>
عندما يتم تطبيق الالتفاف، على سبيل المثال، على صورة في رؤية الكمبيوتر، فإنه يأخذ المصفوفة مع بكسلات الصورة وينزلق عبرها بنافذة صغيرة تسمى "الفلتر" أو "النواة". النواة عبارة عن مصفوفة وزن صغيرة (3x3 أو 5x5 أو 7x7 أو شيء من هذا القبيل) تحتوي على معاملات رقمية. يجب ضرب كل من هذه المعاملات في قيمة البكسل الموجودة حاليًا ضمن خلية النواة المقابلة. بعد ذلك، يتم جمع كل هذه العناصر المضروبة، مما يعطي قيمة أحد عناصر الخريطة الناتجة. بعد ذلك، يتم نقل النواة بخطوة واحدة أو أكثر أفقيًا أو رأسيًا إلى مجموعة البكسل التالية، ويتم حساب المجموع المرجح التالي.<br>
توضح هذه الصورة عملية تطبيق المرشح وحساب أحد عناصر الخريطة الناتجة:

في الواقع، فإن حساب المصفوفة N لعدد الخلايا المجاورة هو نتيجة تلافي مماثلة، وذلك باستخدام مرشح بوزن "0" للعنصر الأوسط و"1" لجميع جيرانه.<br>


```python
filters = np.array([[1,1,1],
                    [1,0,1],
                    [1,1,1]])
```


تقدم الشبكات العصبية عادةً مثل هذه العملية كوظيفة جاهزة (<a href="@@KEEP_00009@@ target=__blank>Pytorch على سبيل المثال</a>). وحتى نموذج scipy لديه مثل <a href="@@KEEP_00010@@ target=__blank>function</a>. <br>
دعونا نرى ما إذا كان يفعل الشيء نفسه بالفعل:


```python
from scipy import signal

Z = np.random.choice([0,1],size=(10,10))

N1 = (Z[0:-2,0:-2] + Z[0:-2,1:-1] + Z[0:-2,2:] +
      Z[1:-1,0:-2]                + Z[1:-1,2:] +
      Z[2:  ,0:-2] + Z[2:  ,1:-1] + Z[2:  ,2:]) 

N2 = signal.convolve2d(Z, filters, mode='valid')

print("Are N1 and N2 identical?")
print(np.array_equal(N1, N2))
```
