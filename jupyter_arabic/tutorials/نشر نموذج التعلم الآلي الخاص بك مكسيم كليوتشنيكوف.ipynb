في معظم الحالات، يتعين علينا أخذ أجزاء **معالجة الميزات** و**النموذج المستمر** و**عمل التنبؤات** معنا.
في الحالات التي يدعم فيها نموذج تعلم الآلة [التدريب عبر الإنترنت](https://en.wikipedia.org/wiki/Online_machine_learning)، يمكننا أيضًا أخذ جزء **التدريب النموذجي**. بشكل أساسي، يسمح التدريب عبر الإنترنت بتدريب النموذج جزئيًا باستخدام البيانات الجديدة فقط، دون الحاجة إلى إعادة تدريبه على مجموعة البيانات بأكملها، وهو أمر مفيد بشكل خاص لمجموعات البيانات الكبيرة والنماذج المعقدة. ومن أجل التبسيط، سنأخذ النموذج غير المتصل بالإنترنت.
ما هو **النموذج المستمر**؟ بعد التدريب فإن أهم جزء في النموذج هو الأوزان التي قام هذا النموذج بحسابها. إلى جانب الأوزان، من الجيد دائمًا أن يكون لديك بعض البيانات الوصفية التي تصف الخوارزمية التي ستستخدم هذه الأوزان، وخط أنابيب معالجة الميزات، وإصدار النموذج، وما إلى ذلك - كل هذه الأشياء يجب تفريغها من ذاكرة الكمبيوتر ووضعها في مكان ما (القرص / قاعدة البيانات / السحابة). تسمى هذه البيانات بالنموذج المستمر ويمكن تحميلها مرة أخرى إلى الذاكرة لاستخدامها في المستقبل.


##2.2. النموذج
نظرًا لأننا مهتمون في الغالب بالأشياء المتعلقة بنموذجنا، فلنأخذ مجموعة البيانات الأكثر _Hello World_'ish في التعلم الآلي: [The Iris Dataset](https://archive.ics.uci.edu/ml/datasets/iris). ولحسن الحظ، فإن برنامج scikit-learn قد تم تضمينه بالفعل، لذلك لا نحتاج حتى إلى تنزيل أي شيء.
لنقوم باستيراد التبعيات أولاً:


````python
import warnings

import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from pandas.tools.plotting import parallel_coordinates
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.pipeline import FeatureUnion, Pipeline
from sklearn.preprocessing import PolynomialFeatures, StandardScaler

warnings.filterwarnings("ignore")

RANDOM_SEED = 17  # because it's the most popular random number between 1 and 20
````


الآن قم بتحميل مجموعة البيانات وإلقاء نظرة على خصائصها:


````python
dataset = load_iris()
print("Feature names:", dataset.feature_names)
print("Iris names:", dataset.target_names)
print("Number of instances:", dataset.data.shape[0])
````


أسماء الميزات بهذا التنسيق ليست مناسبة للاستخدام المستقبلي، فلنحولها:


````python
feature_names = list(
    map(lambda x: x.replace(" (cm)", "").replace(" ", "_"), dataset.feature_names)
)
feature_names
````


قم بتحميل مجموعة البيانات في DataFrame:


````python
target = np.array([dataset.target_names[x] for x in dataset.target]).reshape(-1, 1)
df_full = pd.DataFrame(
    np.concatenate([dataset.data, target], axis=1), columns=feature_names + ["target"]
)
df_full[feature_names] = df_full[feature_names].astype(float)

df_full.sample(5, random_state=RANDOM_SEED)
````

````python
df_full.describe()
````


بالطبع لا يمكننا الحصول على برنامج تعليمي متعلق بعلوم البيانات بدون مخطط رائع، لذلك إليك أحد هذه البرامج:


````python
plt.figure(figsize=(15, 7))
parallel_coordinates(
    df_full, "target", colormap=plt.get_cmap("cool")
)  # see - it's cool as promised
plt.title("Iris species features with their values")
plt.ylabel("cm")
# semicolon to suppress the last line output. Plot will be shown anyway, even without plt.show()
````


الآن، دعونا نقسم بياناتنا للتدريب والاختبار، بحيث تكون نسبة الرفض 40%:


````python
df_train, df_test, y_train, y_test = train_test_split(
    df_full.drop("target", axis=1), df_full.target, test_size=0.4
)
df_train.shape, y_train.shape, df_test.shape, y_test.shape
````


نظرًا لطبيعة مجموعة البيانات، فنحن مقيدون في هندسة الميزات، ولكن لكي نتمكن من **معالجة الميزات** ** في مكانها الصحيح، فلنضيف ميزات متعددة الحدود إلى بياناتنا ثم نقوم بقياس جميع الميزات.
وأخيرا، كمقدر سوف نستخدم الانحدار اللوجستي البسيط.
للحصول على كل هذه الخطوات في مكان واحد، دعونا نضعها في المسار الصحيح - فهذا سيجعل حياتنا أسهل لاحقًا.


````python
pipeline = Pipeline(
    [
        (
            "features",
            FeatureUnion(
                [("poly", PolynomialFeatures()), ("scaler", StandardScaler())]
            ),
        ),
        ("logreg", LogisticRegression(random_state=RANDOM_SEED)),
    ]
)
````


لاحظ أنه باستخدام الإعدادات الافتراضية، ستضيف PolynomialFeatures 15 ميزة جديدة:


````python
", ".join(PolynomialFeatures().fit(df_train).get_feature_names())
````


قم بملائمة النموذج ورسم مصفوفة الارتباك:


````python
pipeline.fit(df_train, y_train)
pred = pipeline.predict(df_test)
plt.figure(figsize=(7, 5))
sns.heatmap(
    confusion_matrix(pred, y_test),
    annot=True,
    xticklabels=dataset.target_names,
    yticklabels=dataset.target_names,
);
````


وياى ، لقد حصلنا على نتائج جيدة جدًا :)
دعونا نقوم ببعض التنبؤات بناءً على الرسم البياني الرائع أعلاه، كما لو قمنا بإجراء قياسات حقيقية. خذ القيم بطريقة نحصل فيها على فئة _setosa_:


````python
pipeline.predict(
    pd.DataFrame(
        [
            {
                "sepal_length": 4.0,
                "sepal_width": 5.0,
                "petal_length": 1.0,
                "petal_width": 0.5,
            }
        ]
    )
)
````


هممم، إنه ليس _setosa_، فماذا حدث؟ دعونا نتحقق من كيفية إنشاء DataFrame:


````python
pd.DataFrame(
    [{"sepal_length": 4.0, "sepal_width": 5.0, "petal_length": 1.0, "petal_width": 0.5}]
)
````

هنا تكمن المشكلة: عند التهيئة من قاموس، لا يمكن ضمان أن يكون ترتيب الأعمدة في DataFrame هو نفسه كما هو الحال في القاموس الذي تم تمريره. على الرغم من ملاحظة أننا في الواقع لا نحتاج إلى DataFrame هنا - يمكننا تمرير قائمة بسيطة بدون أسماء الأعمدة، لكن DataFrame أفضل من حيث سهولة القراءة وراحتنا.
الحل البديل هو تمرير ```columns=dictionary.keys()``` إلى مُنشئ DataFrame أو فقط تحديد أعمدة DataFrame قبل القيام بالتنبؤات:


````python
df_predict = pd.DataFrame(
    [{"sepal_length": 4.0, "sepal_width": 5.0, "petal_length": 1.0, "petal_width": 0.5}]
)
pipeline.predict(df_predict[feature_names])
````


حسنًا، نحن الآن في حالة جيدة، ولكن دعونا نتذكر هذا السلوك لحله لاحقًا أيضًا.



##2.3. استمرار النموذج
بعد نموذج التدريب يجب أن يستمر. يتم ذلك بشكل أساسي عن طريق إجراء تسلسل لحالة النموذج من الذاكرة إلى ملف على القرص بتنسيق معين. من الشائع استخدام [pickle](https://docs.python.org/3/library/pickle.html) التسلسل الثنائي في Python لذلك، ولكن نظرًا لأسباب [performance](https://scikit-learn.org/stable/modules/model_persistence.html) سنستخدم [joblib](https://joblib.readthedocs.io/en/latest/persistence.html) (التي لا تزال تستخدم المخلل داخليًا).
لاحظ أن كلا من Pickle وjoblib يواجهان مشكلات تتعلق بالأمان - عليك التأكد من أن الملف الذي تحاول إلغاء تسلسله لم يتم استبداله بمهاجم لأنه من الممكن إجبار Python على تنفيذ تعليمات برمجية ضارة أثناء إلغاء التسلسل. يمكنك معرفة كيفية عمله بمزيد من التفاصيل [هنا](https://rushter.com/blog/pickle-serialization-internals/).
ما الذي يجب أن يستمر بالضبط في حالتنا؟ من الواضح أنه يجب حفظ الأشياء التالية:
* النموذج نفسه وهو ```LogisticRegression``` المقدر في حالتنا - فهو يحتوي على الأوزان والمنطق حول كيفية معالجة هذه الأوزان؛
* خطوات معالجة الميزات التي تمثلها ```PolynomialFeatures``` و```StandardScaler``` - سيسمح لنا حفظها بمعالجة تاريخ التنبؤات بنفس الطريقة تمامًا كما فعلنا في التدريب.ليس واضحًا جدًا، ولكن لا يزال من الجيد أن يكون لديك:
* إصدار النموذج - لراحتنا ولنكون قادرين على التمييز بين النماذج المختلفة؛
* أسماء الميزات الواردة أو أي معلومات أخرى تصف بيانات الإدخال و/أو الإخراج.
### 2.3.1. scikit-Learn خطوط الأنابيب
قبل أن نواصل، دعونا نتظاهر بأن لدينا ```PolynomialFeatures``` و```StandardScaler``` ككائنات منفصلة. في هذه الحالة، سنحتاج إلى حفظها و```LogisticRegression``` ككيانات منفصلة مما يسبب الكثير من المتاعب، خاصة إذا تطور نموذجنا في النهاية وستتم إضافة خطوات معالجة الميزات الأخرى.
لذا، هناك العديد من الأسباب الوجيهة لاستخدام مسارات scikit-Learn وهنا سبب آخر: يمكننا الاحتفاظ بكائننا ```Pipeline``` فقط وتخزين جميع الخطوات ككيان واحد في ملف واحد.
### 2.3.2. إصدارات بايثون والمكتبة
أيضًا، عند إجراء تسلسل لنموذجك، يجب أن تكون على دراية بأن الكائن المتسلسل قد لا يتم تحميله في إصدار مختلف من Python أو مع إصدار آخر من المكتبة، والتي تم إجراء تسلسل لفئاتها. لهذا السبب، من الجيد دائمًا إصلاح لغة Python وجميع إصدارات المكتبات ذات الصلة والاحتفاظ بها بالإضافة إلى النموذج. سنفعل ذلك في هذا البرنامج التعليمي.
### 2.3.3. التفاف كافة البيانات النموذجية
لنقم بإنشاء فصل دراسي يحتوي على نموذج وجميع البيانات الوصفية الإضافية:


````python
import numpy as np
import pandas as pd
from sklearn.pipeline import Pipeline


class IrisModel:
    def __init__(
        self, pipeline: Pipeline, version="unknown", input_features=[], class_names=[]
    ):
        self.pipeline = pipeline

        self.version = version
        self.input_features = input_features
        self.class_names = class_names

    def predict(self, data: pd.DataFrame) -> np.ndarray:
        data = data[
            self.input_features
        ]  # this is the workaround for the problem with DataFrame fields order, see above
        return self.pipeline.predict(data)

    # pretty-print our class in Jupyter notebook and when we're converting it to string
    def __repr__(self):
        return (
            f"{self.__class__.__name__}("
            f"{self.pipeline!r}, "
            f"version={self.version!r}, "
            f"input_features={self.input_features!r}, "
            f"class_names={self.class_names!r})"
        )
````


ضع محتوى الخلية أعلاه في ملف يسمى **iris_model.py** واحفظه في نفس الدليل الذي تقوم بتشغيل هذا الكمبيوتر الدفتري منه. يعد حفظ هذه الفئة في ملف (الذي أصبح وحدة Python في الواقع) أمرًا ضروريًا لحل مشكلة استمرار الفئات التي ليست جزءًا من أي وحدة نمطية.


##3.1. إنشاء نقطة نهاية للتنبؤات
Flask، باعتباره إطار عمل نموذجي للويب، يسمح لنا بكشف _endpoints_ مع عناوين URL الخاصة بها، _routes_، وربط هذه المسارات بطرقنا المخصصة التي ستنفذ الإجراءات التي نحتاجها وتعيد نتائجها.
لنقم بإنشاء نقطة النهاية ```/predict```:


````python
app = Flask(__name__)


@app.route(
    "/predict", methods=["POST"]
)  # Flask decorator to mark the method which will be called when application recieved
# an HTTP request to /predict URL using the POST method
def predict():
    # here we expect an HTTP request body with the following data in JSON format:
    # [ {"sepal_length": 4.0, "sepal_width": 5.0, "petal_length": 1.0, "petal_width": 0.5},
    #   {"sepal_length": 3.0, "sepal_width": 3.0, "petal_length": 2.0, "petal_width": 4} ]
    data = request.get_json()  # get the request body as a Python list of dictionaries

    df_predict = pd.DataFrame(data)
    predictions = IRIS_MODEL.predict(df_predict)
    predictions = predictions.tolist()

    return jsonify(
        {
            "version": IRIS_MODEL.version,  # return model version to calling side
            "status": "success",  # indicate that the call went well
            "predictions": predictions,  # actual predictions, ["setosa", "virginica"]
        }
    )
````

كما ترون، يتم تمثيل نقطة النهاية الخاصة بنا بواسطة ```/predict``` المسار المرتبط بالطريقة ```predict()```.

* نأخذ JSON من نص طلب HTTP، ونحوله إلى قائمة قواميس Python، وندخلها في DataFrame ونقوم بالتنبؤات؛

* تقوم القارورة بتكوين استجابة HTTP وإعادتها مرة أخرى إلى جانب الاتصال (المتصفح أو تطبيق آخر).
كقاعدة عامة، من الأفضل أن تجعل نقاط نهاية التنبؤ الخاصة بك تعمل مع تنبؤات متعددة في وقت واحد، وذلك للأسباب التالية:

والتي، اعتمادًا على النظام الأساسي الخاص بك، افتح متصفحًا لإجراء تسجيل الدخول أو اطلب تسجيل الدخول وكلمة المرور الخاصة بـ Heroku في سطر الأوامر.



##4.2. قم بإعداد تطبيقنا لـ Heroku
عندما يتلقى Heroku ملفات التطبيق الخاص بك، فإنه يحتاج إلى التعرف على نوع التطبيق الذي تقوم بنشره وكيفية تشغيله. غالبًا ما يتم ذلك تلقائيًا عن طريق تحليل ملفات التطبيق، ولكن من الأفضل تقديم تعريف دقيق لما نحتاج إلى Heroku للقيام به.
لنقم بإنشاء ملفين في الدليل الجذر للتطبيق والذي سيخبر Heroku بكيفية العمل مع تطبيقنا.
```
|   Procfile
|   requirements.txt
```
### 4.2.1. ملف تعريفي
هذا هو الملف الذي يستخدمه Heroku للحصول على الأوامر المطلوب تنفيذها لبدء التطبيق الخاص بك والقيام ببعض إجراءات ما بعد البدء.
هنا هذا الملف لتطبيقنا:
```
web: gunicorn iris_app:app --log-file -
```
نعم سطر واحد فقط يخبر Heroku ما يلي:
* يجب تشغيل تطبيقنا كتطبيق ```web```؛
* استخدم ```gunicorn``` كخادم ويب (تذكر أن خادم Flask مخصص للتطوير فقط، لذلك قمنا بالتبديل إلى [gunicorn](https://gunicorn.org/)) وقم بتمرير المسار إلى تطبيقنا داخل الوحدة النمطية المحددة؛
* ```--log-file -``` هو خيار لـ gunicorn لإخراج معلومات السجل مباشرة إلى STDOUT - سيتم بعد ذلك التقاطها بواسطة Heroku وتوفيرها كجزء من آلية التسجيل الخاصة بـ Heroku.
### 4.2.2. المتطلبات.txt
هذا الملف ليس خاصًا بـ Heroku - إنه ملف شائع يستخدم في نظام Python البيئي للاحتفاظ بقائمة التبعيات المطلوبة لتشغيل التطبيق أو المكتبة. Heroku قادر على التعرف على هذا الملف وسيطلب التبعيات أثناء نشر التطبيق الخاص بك.وهنا هو:
```
pandas==0.22.0
numpy==1.15.4
scikit-learn==0.20.1
joblib==0.13.0
Flask==1.0.2
gunicorn==19.9.0
```
لاحظ أنني أضفت ```gunicorn``` يدويًا - ليس من الضروري تثبيته على جهاز محلي، ولكننا نحتاج إلى تثبيته على Heroku.
يمكنك رؤية المكتبات الموجودة لديك وحفظها في تثبيت Python الخاص بك (أو، وهو أفضل بكثير، في [virtualenv](https://virtualenv.pypa.io/en/latest/)) عن طريق تشغيل الأمر التالي:
> تجميد النقاط > المتطلبات.txt
ومع ذلك، فإنه سيحفظ جميع التبعيات التي لم يتم استخدامها في تطبيقك والتبعيات الفرعية أيضًا، والتي قد لا تسبب أي ضرر، ولكن من الأفضل مراجعة الملف الذي تم إنشاؤه وإزالة تلك التبعيات التي ليست ضرورية لتطبيقك للعمل.



##4.3. تهيئة مستودع Git المحلي
يسمح لك Heroku بنشر التطبيق الخاص بك عن طريق إجراء عملية دفع بسيطة. لبدء استخدام Git، نحتاج إلى تشغيل الأوامر التالية في الدليل باستخدام تطبيق الويب الخاص بنا:
تهيئة مستودع Git الفارغ:
> بوابة الحرف الأول
إضافة جميع الملفات الموجودة إلى Git:
> بوابة إضافة .
تنفيذ الملفات المضافة إلى مستودع Git المحلي الذي تم إنشاؤه للتو:
> git الالتزام -m "الالتزام الأولي لتطبيق iris"



##4.4. إنشاء تطبيق Heroku
قم بتشغيل الأمر التالي في الدليل الجذر للتطبيق:
> إنشاء هيروكو
سيتم إنشاء تطبيق Heroku ذو اسم غريب لك وإضافة مستودع Git البعيد (المستضاف على Heroku) إلى تكوين Git لتطبيقك:
> بوابة البعيد -v
```

heroku  https://git.heroku.com/blooming-eyrie-32543.git (push)
```
لا تقلق من تسمية تطبيقك مثل ```blooming-eyrie-32543``` - يمكنك [إعادة تسميته](https://devcenter.heroku.com/articles/renaming-apps) لاحقًا :) يتم ذلك عن عمد لأن تطبيقات Heroku يجب أن يكون لها أسماء فريدة ولا نريد قضاء الوقت للعثور على اسم لا يستخدمه شخص آخر. للإشارة فقط، من الممكن استخدام [نطاقك المخصص](https://devcenter.heroku.com/articles/custom-domains) على Heroku، إذا لزم الأمر.

# 5. الأجزاء المفقودة
يمكنك هنا العثور على بعض الأفكار والتوجيهات حول ما يمكن القيام به بشكل أفضل باستخدام ما فعلناه في البرنامج التعليمي وما يمكن أن يكون مجالات اهتمام محتملة لمزيد من التعلم.
هذا القسم شخصي تمامًا، لذلك قد تختلف المسافة المقطوعة.



##5.1. الإصدار
في كل مرة تقوم فيها بإعادة التدريب أو تغيير العناصر الداخلية لنموذجك، يجب أن يكون له إصدار مختلف.
يجب أن تحتوي واجهة برمجة التطبيقات (API) المبنية على النموذج على نسختها وتحديثها وفقًا لذلك إذا تم تغيير مجموعة حقول الإدخال أو بيانات الإخراج، حتى لا يؤدي ذلك إلى تعطيل العملاء الحاليين. الإصدار + نقاط النهاية + تنسيق بيانات الإدخال والإخراج هو **عقد** بين الخدمة وعملائها، ولا يجوز انتهاكه بحرية. يجب أن يتمتع العملاء الأقدم، إذا لم يكونوا تحت سيطرتك، بالقدرة على التحدث إلى واجهة برمجة التطبيقات القديمة الخاصة بك والتي يجب نشرها بشكل منفصل عن الواجهة الجديدة.من الجيد أيضًا استخدام بنية Git المتفرعة المناسبة، مثل [GitFlow](https://www.atlassian.com/git/tutorials/comparing-workflows/gitflow-workflow) أو أنظمة التحكم في إصدار التعلم الآلي المتخصصة، مثل [DVC](https://dvc.org).



##5.2. إدارة التبعية
تواجه العديد من أنظمة المكتبات المتعددة مشاكلها الخاصة عندما تحتاج إلى إصدارات متعددة من نفس المكتبة تستخدمها تطبيقات مختلفة: [DLL Hell](https://en.wikipedia.org/wiki/DLL_Hell)، [JAR Hell](https://dzone.com/articles/what-is-jar-hell)، [Python Dependency Hell](https://medium.com/knerd/the-nine-circles-of-python-dependency-hell-481d53e3e025)، ولكل منها <img src="../../img/tutorial-deployment-special-hell.jpg" width="400" />،
عندما يتعلق الأمر بإمكانية النقل بين أجهزة كمبيوتر مختلفة وإصدارات Python والبيئات المختلفة، فمن الضروري إصلاح تبعيات مشروعك بحيث يكون لديك نفس المكتبات التي كانت لديك أثناء التطوير.
لقد تطرقنا إلى هذا المجال قليلاً من خلال إنشاء ```requirements.txt```، ولكن يجب أن تتعلم كيفية استخدام [virtualenv](https://virtualenv.pypa.io/en/latest/)، إذا لم تتعلم ذلك بعد.
كشكل متطرف من أشكال إدارة التبعية، من الممكن استخدام Docker واستخدامه كبديل لـ virtualenv - سيوفر لك أيضًا تجريدًا من نظام التشغيل المستهدف حيث يمكنك إصلاحه في حاوية Docker بالإضافة إلى مكتبات Python.



##5.3. حاوية التطبيق الخاص بك
كما ذكرنا سابقًا، يوفر Docker فكرة جيدة عن البيئة حيث يمكنك إصلاح نظام التشغيل ومكتبات النظام والبرامج المثبتة ومكتبات Python داخل Docker cotainer.
لاحقًا، يمكن تسليم حاوية Docker هذه إلى الإنتاج بنفس الحالة التي كانت عليها أثناء وقت التطوير.
بالإضافة إلى ذلك، يوجد حاليًا العديد من موفري الخدمات السحابية الذين يسمحون لك بنشر حاويات Docker كتطبيقات.
في الواقع، يعد وضع تطبيقات ML في حاويات موضوعًا كبيرًا جدًا ويستحق برنامجًا تعليميًا آخر.


##5.4. قابلية التوسع
من المفيد دائمًا معرفة مقدار موارد الأجهزة التي تحتاجها حتى يعمل نموذجك بشكل صحيح ودون إبطاء عملائه.
الخطوة الأولى في هذا الاتجاه هي قياس أداء تطبيقك النهائي عن طريق إرسال مجموعة من الاستعلامات إليه ([ApacheBench](https://en.wikipedia.org/wiki/ApacheBench)، [JMeter](https://en.wikipedia.org/wiki/Apache_JMeter)).
نظرًا لأن تطبيقنا عديم الحالة (أي أنه لا ينتج أي شيء يجب تخزينه بين الطلبات الواردة من نفس العميل)، فيمكننا توسيع نطاقه ببساطة عن طريق إضافة المزيد من المثيلات على خدمة الاستضافة الخاصة بنا، Heroku.
بالإضافة إلى ذلك يمكننا [ضبط gunicorn لمعالجة استعلامات متعددة في وقت واحد](https://devcenter.heroku.com/articles/optimizing-dyno-usage#python) عن طريق إنتاج المزيد من العمال، حتى نتمكن من الاستفادة من موارد المثيل الفردي بشكل كامل.



##5.5. بيانات الإدخال المستمرة
في بعض الأحيان تكون هناك حاجة لجمع بيانات حقيقية يتلقاها نموذجك بعد نشره. عادةً ما يتم الاحتفاظ بهذه البيانات جنبًا إلى جنب مع التنبؤات التي قام بها نموذجك.
على سبيل المثال، يمكن استخدامه في الحالات التالية:
* تحتاج إلى تحليل أن البيانات المرسلة من قبل العملاء لا تزال من نفس التوزيع الذي كان لديك أثناء وقت التدريب - لذلك بعد جمع البيانات الحقيقية، يمكنك تحليلها، وتصنيفها، وإضافتها إلى البيانات الأصلية وإعادة تدريب النموذج الخاص بك؛
* تبين أن النموذج الخاص بك يعطي نتائج غريبة على بعض البيانات المدخلة - فوجود سجلات بهذه البيانات وتنبؤات النموذج الخاص بك سيسمح لك بتصحيح الأخطاء.
الخيار الواضح للاستمرارية هو استخدام قواعد البيانات، وهو أمر سهل جدًا في بايثون.انتبه إلى أن اختيار SQLite (وهو مجرد ملف على القرص المحلي)، أثناء استضافة النموذج الخاص بك على خدمة PaaS (مثل Heroku) قد يسبب لك مشكلة لأنه في معظم الحالات يكون نظام الملفات المحلي سريع الزوال ويمكن محوه في أي وقت. لذا من الأفضل اختيار خادم قاعدة البيانات المستضاف على جهاز آخر - عادةً ما تكون خدمات PaaS بمثابة [خيار](https://devcenter.heroku.com/categories/data-management).



##5.6. أعباء العمل والنماذج الثقيلة
النموذج الذي استخدمناه في البرنامج التعليمي هو نموذجي للغاية ولن يعاني كثيرًا حتى في ظل الحمل العالي. ولكن في بعض السيناريوهات، قد لا تكون الأدوات التي استخدمناها كافية للمواكبة. في مثل هذه الحالات، يجب مراعاة التقنيات المختلفة التي تم إنشاؤها خصيصًا للبيانات الضخمة والأحمال العالية.
باعتبارك أحد ممارسي التعلم الآلي، ألقِ نظرة على [Spark MLlib](https://spark.apache.org/mllib/) و[H2O](https://www.h2o.ai/products/h2o/).



##5.7. خيارات النشر الأخرى
هناك بعض خيارات النشر الأخرى التي لم يتم تناولها في هذا البرنامج التعليمي ولكن الفكرة الأساسية لا تزال كما هي. خذ وقتًا لتعلمها أو إنشاء برنامج تعليمي آخر :)
* حاويات Docker (انظر أعلاه حول إيجابياتها)؛
* عمليات النشر بدون خادم باستخدام [AWS Lambda](https://aws.amazon.com/lambda/) أو [Azure Functions](https://docs.microsoft.com/en-us/azure/azure-functions/) - ستسمح لك بتشغيل النموذج الخاص بك عند الطلب، وعادةً ما تستخدم موارد وأموال أقل بكثير. إنها طريقة رائعة أخرى لنشر إثبات المفاهيم والنماذج الصغيرة في الإنتاج؛
* مستضافة ذاتيًا، عندما تريد التحكم الكامل في أجهزتك وبرامجك أو عندما يتعين عليك استخدام النماذج الخاصة بك داخل شبكة المؤسسة. تتطلب هذه الطريقة الكثير من الجهود ومن الأفضل أن يكون لديك خبرة في DevOps في الفريق للحصول على المساعدة.
