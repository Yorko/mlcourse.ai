## تفسير اختبار  
قبل أن تتمكن من تطبيق الاختبارات الإحصائية، يجب أن تعرف كيفية تفسير النتائج.
سيُرجع كل اختبار شيئين على الأقل:
* __الإحصاء:__ الكمية المحسوبة بالاختبار والتي يمكن تفسيرها في سياق الاختبار من خلال مقارنتها بالقيم الحرجة من توزيع إحصائية الاختبار.
* __$p-value$:__ يستخدم لتفسير الاختبار، في الحالة الأولى، سواء كانت العينة مأخوذة من توزيع غاوسي.  
يحسب كل اختبار إحصائية خاصة بالاختبار. يمكن أن تساعد هذه الإحصائية في تفسير النتيجة، على الرغم من أنها قد تتطلب إتقانًا أعمق للإحصائيات ومعرفة أعمق بالاختبار الإحصائي المحدد. وبدلاً من ذلك، يمكن استخدام _p-value_ لتفسير الإحصائيات بسرعة ودقة في التطبيقات العملية.
في الجزء الأول سنقوم باختبار البيانات التي تم سحب العينة من توزيع غاوسي. من الناحية الفنية يسمى هذا بالفرضية الصفرية، أو H0. ويتم اختيار مستوى العتبة المسمى ألفا، وعادةً ما يكون 5% (أو 0.05)، والذي يُستخدم لتفسير القيمة الاحتمالية. تفسير القيمة p على النحو التالي:  
* __$p <= alpha$__: رفض H0، ليس عاديًا.  
* __$p > alpha$__: فشل في رفض H0، عادي.  
وهذا يعني أننا، بشكل عام، __نسعى للحصول على نتائج ذات قيمة p أكبر للتأكد من أنه من المحتمل أن تكون عينتنا مأخوذة من توزيع غاوسي__.النتيجة التي تزيد عن 5% __لا تعني__ أن الفرضية الصفرية __صحيحة__. وهذا يعني أنه من المحتمل جدًا أن يكون صحيحًا في ضوء الأدلة المتاحة. القيمة p ليست احتمالية ملاءمة البيانات لتوزيع غاوسي؛ ويمكن اعتبارها قيمة تساعدنا في تفسير الاختبار الإحصائي.

### اختبار داجوستينو $K^2$ <a id='norm2'></a>
يحسب اختبار D'Agostino's K^2 إحصائيات موجزة من البيانات، وهي التفرطح والتواء، لتحديد ما إذا كان توزيع البيانات يخرج عن التوزيع الطبيعي، المسمى باسم رالف داجوستينو.
* __Skew__ هو قياس كمي لمدى دفع التوزيع إلى اليسار أو اليمين، وهو مقياس لعدم التماثل في التوزيع.
* __التفرطح__ يحدد مقدار التوزيع الموجود في الذيل. إنه اختبار إحصائي بسيط وشائع الاستخدام للحياة الطبيعية.  
يتوفر اختبار D’Agostino’s K^2 عبر الدالة Normaltest() SciPy ويقوم بإرجاع إحصائيات الاختبار والقيمة p.المثال الكامل لاختبار D’Agostino’s K^2 على مجموعة البيانات مدرج أدناه.


```python
from scipy.stats import normaltest

stat, p = normaltest(normal)
print("Statistics=%.3f, p=%.3f" % (stat, p))

alpha = 0.05
if p > alpha:
    print("Sample looks Gaussian (fail to reject H0)")
else:
    print("Sample does not look Gaussian (reject H0)")
```

```python
from scipy.stats import normaltest

stat, p = normaltest(log_normal)
print("Statistics=%.3f, p=%.3f" % (stat, p))

alpha = 0.05
if p > alpha:
    print("Sample looks Gaussian (fail to reject H0)")
else:
    print("Sample does not look Gaussian (reject H0)")
```


مزيد من المعلومات:
 * [scipy.stats.normaltest](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.normaltest.html)



### اختبار أندرسون-دارلينج <a id='norm3'></a>



* H0: العينة ذات توزيع غاوسي.  
* H1: العينة ليس لها توزيع غاوسي.


```python
from scipy.stats import anderson

result = anderson(normal)
print("Statistic: %.3f" % result.statistic)
p = 0
for i in range(len(result.critical_values)):
    sl, cv = result.significance_level[i], result.critical_values[i]
    if result.statistic < result.critical_values[i]:
        print("%.3f: %.3f, data looks normal (fail to reject H0)" % (sl, cv))
    else:
        print("%.3f: %.3f, data does not look normal (reject H0)" % (sl, cv))
```

```python
from scipy.stats import anderson

result = anderson(uniform)
print("Statistic: %.3f" % result.statistic)
p = 0
for i in range(len(result.critical_values)):
    sl, cv = result.significance_level[i], result.critical_values[i]
    if result.statistic < result.critical_values[i]:
        print("%.3f: %.3f, data looks normal (fail to reject H0)" % (sl, cv))
    else:
        print("%.3f: %.3f, data does not look normal (reject H0)" % (sl, cv))
```

## اختبارات الارتباط <a id='cor_t'></a>
يسرد هذا القسم الاختبارات الإحصائية التي يمكنك استخدامها للتحقق من وجود عينتين مرتبطتين.


```python
x = np.random.normal(0, 1, 1000)
y = (3 * x) - np.random.normal(0, 2, 1000)
```


### معامل ارتباط بيرسون <a id='cor1'></a>
يختبر ما إذا كانت هناك علاقة خطية بين عينتين.  
* H0: العينتان مستقلتان.
*H1: هناك تبعية بين العينات.


```python
from scipy.stats import pearsonr

corr, p = pearsonr(x, y)
corr, p
```


مزيد من المعلومات:

## اختبارات الفرضيات الإحصائية البارامترية <a id='par_t'></a>
يسرد هذا القسم الاختبارات الإحصائية التي يمكنك استخدامها لمقارنة عينات البيانات. غالبًا ما تعني الأساليب الإحصائية البارامترية تلك الطرق التي تفترض أن عينات البيانات لها توزيع غاوسي.  اختبارات الدلالة الإحصائية البارامترية التي تحدد الفرق بين متوسطي عينتين أو أكثر من البيانات.



بالنسبة إلى __الاختبارات الإحصائية البارامترية__، تفترض أنه تم سحب عينة بيانات من توزيع سكاني محدد.
غالبًا ما تشير إلى الاختبارات الإحصائية التي تفترض التوزيع الغوسي. نظرًا لأنه من الشائع جدًا أن تتناسب البيانات مع هذا التوزيع، فإن الأساليب الإحصائية البارامترية هي الأكثر شيوعًا.السؤال النموذجي الذي قد يكون لدينا حول عينتين أو أكثر من البيانات هو ما إذا كانت لهما نفس التوزيع. اختبارات الأهمية الإحصائية البارامترية هي تلك الأساليب الإحصائية التي تفترض أن البيانات تأتي من نفس التوزيع الغوسي، أي توزيع البيانات بنفس المتوسط ​​والانحراف المعياري: معلمات التوزيع.
يمكن اعتبار القيمة p بمثابة احتمالية مراقبة عينتي البيانات بالنظر إلى الافتراض الأساسي (فرضية العدم) بأن العينتين تم سحبهما من مجتمع له نفس التوزيع.
__p <= alpha__: رفض الفرضية الصفرية، توزيع مختلف.  
__p > alpha__: فشل في رفض الفرضية الصفرية، نفس التوزيع


```python
x1 = 3 * np.random.randn(1000) + 20
x2 = 3 * np.random.randn(1000) + 21
print("x1: mean=%.3f stdv=%.3f" % (np.mean(x1), np.std(x1)))
print("x2: mean=%.3f stdv=%.3f" % (np.mean(x2), np.std(x2)))
sns.distplot(x1)
sns.distplot(x2);
```


### اختبار الطالب <a id='par1'></a>
اختبار ما إذا كانت وسائل عينتين مستقلتين مختلفة بشكل كبير.
أحد اختبارات t الأكثر استخدامًا هو اختبار العينات المستقلة. يمكنك استخدام هذا الاختبار عندما تريد مقارنة متوسط عينتين مستقلتين على متغير معين
الافتراضات
* الملاحظات في كل عينة مستقلة وموزعة بشكل مماثل (iid).
* يتم توزيع الملاحظات في كل عينة بشكل طبيعي.
* الملاحظات في كل عينة لها نفس التباين.  
التفسير
* H0: متوسطات العينات متساوية.
* H1: متوسطات العينات غير متساوية.


```python
from scipy.stats import ttest_ind

stat, p = ttest_ind(x1, x2)
print("P value:", p)
alpha = 0.05
if p > alpha:
    print("Same distributions (fail to reject H0)")
else:
    print("Different distributions (reject H0)")
```


ويخلص تفسير الإحصائية إلى أن متوسطات العينة مختلفة، مع دلالة لا تقل عن 5٪.



مزيد من المعلومات:
 * [scipy.stats.tttest_ind](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.ttest_ind.html)



### اختبار t للطالب المقترن <a id='par2'></a>
يختبر ما إذا كانت وسائل العينات المقترنة مختلفة بشكل كبير. اختبار الطالب المقترن لقياس الفرق بين متوسط ​​عينتين من البيانات التابعة.
الافتراضات* الملاحظات في كل عينة مستقلة وموزعة بشكل مماثل (iid).
* يتم توزيع الملاحظات في كل عينة بشكل طبيعي.
* الملاحظات في كل عينة لها نفس التباين.
* يتم إقران الملاحظات عبر كل عينة.
التفسير
* H0: متوسطات العينات متساوية.
* H1: متوسطات العينات غير متساوية.


```python
from scipy.stats import ttest_rel

stat, p = ttest_rel(x1, x2)
print("P value:", p)
alpha = 0.05
if p > alpha:
    print("Same distributions (fail to reject H0)")
else:
    print("Different distributions (reject H0)")
```


يشير تفسير النتيجة إلى أن العينات لها وسائل مختلفة وبالتالي توزيعات مختلفة



مزيد من المعلومات:
 * [scipy.stats.tttest_ind](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.ttest_rel.html)  
 * [ويكيبيديا](https://en.wikipedia.org/wiki/Student%27s_t-test)



# تحليل اختبار التباين (ANOVA) <a id='par3'></a> 
اختبار ما إذا كانت وسائل عينتين مستقلتين أو أكثر مختلفة بشكل كبير. ANOVA والتدابير المتكررة ANOVA للتحقق من التشابه أو الاختلاف بين وسيلة عينتين أو أكثر من البيانات.
* H0: متوسطات العينات متساوية.
* H1: واحد أو أكثر من متوسطات العينات غير متكافئة.


```python
from scipy.stats import f_oneway

stat, p = f_oneway(x1, x2)
print("P value:", p)
alpha = 0.05
if p > alpha:
    print("Same distributions (fail to reject H0)")
else:
    print("Different distributions (reject H0)")
```


مزيد من المعلومات:
 * [scipy.stats.f_oneway](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.f_oneway.html)  
 * [ويكيبيديا](https://en.wikipedia.org/wiki/Analysis_of_variance)

* يمكن ترتيب الملاحظات في كل عينة.
التفسير

*H1: توزيعات العينتين غير متساوية.


### اختبار كولموجوروف-سميرنوف <a id='nonpar5'></a>


يعتمد على ما تريد تحديده. إذا كنت مهتمًا بتحديد ما إذا كانت التوزيعات لها __نفس المتوسط__، ولا تهتم بالباقي، فإن __K-S__ ليس الأفضل. يعد K-S مفيدًا جدًا لتحديد ما إذا كانت عينتان في الأساس تأتيان من نفس المجموعة السكانية.


```python
import numpy as np
from scipy.stats import ks_2samp

np.random.seed(123456)
x = np.random.normal(0, 1, 1000)
y = np.random.normal(0, 1, 1000)
z = np.random.normal(1.1, 0.9, 1000)

ks_2samp(x, y), ks_2samp(x, z)
```


في ظل الفرضية الصفرية فإن التوزيعين متطابقان. إذا كانت إحصائية K-S صغيرة أو كانت القيمة p عالية (أكبر من مستوى الأهمية، على سبيل المثال 5٪)، فلا يمكننا رفض الفرضية القائلة بأن توزيعات العينتين متماثلة. وعلى العكس من ذلك، يمكننا رفض فرضية العدم إذا كانت القيمة p منخفضة.
