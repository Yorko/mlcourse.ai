# دمج DataFrames مع الباندا



هنا، سوف تتعلم كل شيء عن دمج DataFrames الباندا. سوف تستكشف تقنيات مختلفة للدمج، وتتعرف على الصلات اليسرى، والصلات اليمنى، والصلات الداخلية، والصلات الخارجية، بالإضافة إلى متى يتم استخدامها. ستتعرف أيضًا على الدمج المرتب، وهو أمر مفيد عندما تريد دمج DataFrames التي تحتوي أعمدتها على ترتيب طبيعي، مثل أعمدة التاريخ والوقت.


```python
import pandas as pd
```


***[البرنامج التعليمي لمجموعات البيانات](https://www.dropbox.com/sh/a0f9yschd7wu2ls/AAAQGizvP5JpV0hYMYg2fdKYa?dl=0)***



### دمج DataFrames للشركة
لنفترض أن شركتك لديها عمليات في عدة مدن مختلفة تحت إشراف عدة مديرين مختلفين. تحتوي DataFrames `revenue` و`managers` على معلومات جزئية تتعلق بالشركة. وهذا يعني أن صفوف الأعمدة `city` غير متطابقة تمامًا في `revenue` و`managers` (ليس لدى فرع Mendocino أي إيرادات حتى الآن منذ افتتاحه للتو وترك مدير فرع Springfield الشركة مؤخرًا).


```python
revenue = pd.read_csv("revenue.csv")
managers = pd.read_csv("managers.csv")
print(revenue)
print(managers)
```


تمت طباعة DataFrames في IPython Shell. إذا قمت بتشغيل الأمر `combined = pd.merge(revenue, managers, on='city')`، فكم عدد الصفوف مجتمعة؟


```python
combined = pd.merge(revenue, managers, on="city")
print(combined)
```


صحيح! نظرًا لأن الإستراتيجية الافتراضية لـ `pd.merge()` هي *صلة داخلية*، فإن `combined` سيكون لها صفين.



أمر الدمج هو الهدف التعليمي الرئيسي لهذا البرنامج التعليمي. تأخذ عملية الدمج في أبسط صورها إطار بيانات أيسر (الوسيطة الأولى)، وإطار بيانات أيمن (الوسيطة الثانية)، ثم اسم عمود الدمج، أو عمود للدمج "على". في الإخراج/النتيجة، تتم مطابقة الصفوف من إطارات البيانات اليمنى واليسرى حيث توجد قيم مشتركة لعمود الدمج المحدد بواسطة "on".


يحتفظ *الدمج الداخلي (أو الصلة الداخلية)* فقط بالقيم المشتركة في إطاري البيانات الأيمن والأيسر للنتيجة.



### الدمج في عمود محدد  
تتوقع أن تنمو شركتك، وأن تعمل في نهاية المطاف في مدن تحمل نفس الاسم في ولايات مختلفة. على هذا النحو، عليك أن تقرر أن كل فرع يجب أن يكون له معرف فرع رقمي. وبالتالي، يمكنك إضافة عمود `branch_id` إلى كلا إطاري البيانات. علاوة على ذلك، تمت إضافة مدن جديدة إلى كل من `revenue` و`managers` DataFrames أيضًا.


```python
revenue = pd.read_csv("revenue_branch_id.csv")
managers = pd.read_csv("managers_branch_id.csv")
print(revenue)
print(managers)
```


باستخدام `pd.merge()`، قم بدمج DataFrames `revenue` و`managers` في عمود `'city'` لكل منهما


```python
merge_by_city = pd.merge(revenue, managers, on="city")
print(merge_by_city)
```


قم بدمج DataFrames `revenue` و`managers` في عمود `'branch_id'` لكل منهما.


```python
merge_by_id = pd.merge(revenue, managers, on="branch_id")
print(merge_by_id)
```


أحسنت! لاحظ أنه عند الدمج في `'city'`، يكون لـ DataFrame الناتج نتيجة غريبة: في الصف 2، تحتوي مدينة Springfield على معرفين فرعيين مختلفين. وذلك لأن هناك في الواقع مدينتين مختلفتين تدعى سبرينغفيلد - واحدة في ولاية إلينوي، والأخرى في ولاية ميسوري. يحتوي `revenue` DataFrame على الإطار من إلينوي، و`managers` DataFrame يحتوي على الإطار من ولاية ميسوري. وبالتالي، عند الدمج في `'branch_id'`، يتم إسقاط كل منهما من DataFrame المدمج.



### دمج الأعمدة ذات التسميات غير المتطابقة
نواصل العمل مع `revenue` & `managers` DataFrames من قبل. هذه المرة، قام شخص ما بتغيير اسم الحقل `'city'` إلى `'branch'` في الجدول `managers`. الآن، عند محاولة دمج DataFrames، يتم طرح استثناء:


```python
revenue = pd.read_csv("revenue_branch_id_2.csv")
managers = pd.read_csv("managers_branch_id_2.csv")
print(revenue)
print(managers)
```

> `pd.merge(revenue, managers, on='city')  
 التتبع (آخر مكالمة أخيرة):  
     ... <text deleted> ...  
     pd.merge(revenue, managers, on='city')  
     ... <text deleted> ...  
 خطأ رئيسي: "المدينة" `



نظرًا لهذا، سيستغرق الأمر المزيد من العمل للانضمام أو الدمج في اسم المدينة/الفرع. يجب عليك تحديد المعلمات `left_on` و`right_on` في استدعاء `pd.merge()`.


```python
combined = pd.merge(revenue, managers, left_on="city", right_on="branch")
print(combined)
```


عمل عظيم! من المهم ** الانتباه إلى ** كيفية تسمية الأعمدة ** في DataFrames المختلفة.



### الدمج في أعمدة متعددة  
هناك إستراتيجية أخرى لتوضيح المدن ذات الأسماء المتطابقة وهي إضافة معلومات عن الولايات التي تقع فيها المدن. ولتحقيق هذه الغاية، يمكنك إضافة عمود يسمى `state` إلى كل من DataFrames من التمارين السابقة.  
  
هدفنا في هذا التمرين هو استخدام `pd.merge()` لدمج DataFrames باستخدام أعمدة متعددة (باستخدام `'branch_id'`، و`'city'`، و`'state'` في هذه الحالة).


```python
revenue = pd.read_csv("revenue_branch_id.csv")
managers = pd.read_csv("managers_branch_id.csv")

# Add 'state' column to revenue
revenue["state"] = ["TX", "CO", "IL", "CA"]
# Add 'state' column to managers
managers["state"] = ["TX", "CO", "CA", "MO"]

print(revenue)
print(managers)
```

```python
# Merge revenue & managers on 'branch_id', 'city', & 'state'
combined = pd.merge(revenue, managers, on=["branch_id", "city", "state"])
print(combined)
```


عمل ممتاز!



## أنواع الدمج الأخرى



هناك ثلاثة أنواع مختلفة من عمليات الدمج المتوفرة في Pandas. تعد أنواع الدمج هذه شائعة عبر معظم قواعد البيانات واللغات الموجهة نحو البيانات (SQL وR وSAS) ويُشار إليها عادةً باسم "الانضمامات". إذا كنت لا تعرفهم، تعلمهم الآن.



* **الدمج الداخلي / الانضمام الداخلي** - سلوك Pandas الافتراضي، احتفظ فقط بالصفوف التي توجد فيها قيمة الدمج "تشغيل" في إطاري البيانات الأيمن والأيسر.    



* ** الدمج الأيسر / الصلة الخارجية اليسرى ** - (المعروف أيضًا باسم الدمج الأيسر أو الصلة اليسرى) احتفظ بكل صف في إطار البيانات الأيسر. عندما تكون هناك قيم مفقودة للمتغير "on" في إطار البيانات الصحيح، قم بإضافة قيم فارغة/NaN في النتيجة.  


* **الدمج الأيمن / الصلة الخارجية اليمنى** - (المعروف أيضًا باسم الدمج الأيمن أو الصلة اليمنى) احتفظ بكل صف في إطار البيانات الصحيح. عندما تكون هناك قيم مفقودة للمتغير "on" في العمود الأيسر، قم بإضافة قيم فارغة / NaN في النتيجة.  



* **دمج خارجي / صلة خارجية كاملة** - تُرجع الصلة الخارجية الكاملة جميع الصفوف من إطار البيانات الأيسر، وجميع الصفوف من إطار البيانات الأيمن، وتطابق الصفوف حيثما أمكن، مع NaNs في مكان آخر.



يتم تحديد النوع `merge` الذي سيتم استخدامه باستخدام المعلمة `how` في أمر الدمج، مع أخذ القيم `left`، أو `right`، أو `inner` (افتراضي)، أو `outer`.



تُستخدم مخططات Venn بشكل شائع لتوضيح أنواع الدمج والربط المختلفة.



![مخططات Venn](https://raw.githubusercontent.com/Raigred/mlcourse.ai/master/img/join-types-merge-names.jpg "مخططات Venn")



### دمج اليسار واليمين في أعمدة متعددة  
لدينا الآن، بالإضافة إلى `revenue` و`managers`، DataFrame `sales` الذي يلخص الوحدات المباعة من فروع محددة (يتم تحديدها بواسطة `city` و`state` ولكن ليس `branch_id`).



من خلال دمج `revenue` و`sales` مع الدمج *الصحيح*، يمكننا تحديد قيم `revenue` المفقودة. هنا، لا نحتاج إلى تحديد `left_on` أو `right_on` لأن الأعمدة المراد دمجها لها تسميات متطابقة.


```python
managers = pd.read_csv("managers_branch_id_2.csv")
managers["state"] = ["TX", "CO", "CA", "MO"]
sales = pd.read_csv("sales.csv")
print(sales)
```

```python
revenue_and_sales = pd.merge(revenue, sales, how="right", on=["city", "state"])
print(revenue_and_sales)
```


من خلال دمج `sales` و`managers` مع الدمج *الأيسر*، يمكننا تحديد المدير المفقود. هنا، تحتوي الأعمدة المراد الدمج عليها على تسميات متعارضة، لذا يجب علينا تحديد `left_on` و`right_on`. في كلتا الحالتين، نحن نتطلع إلى معرفة كيفية ربط الحقول في الصفوف التي تحتوي على `Springfield`.


```python
sales_and_managers = pd.merge(
    sales, managers, how="left", left_on=["city", "state"], right_on=["branch", "state"]
)
print(sales_and_managers)
```


أحسنت! هذه طريقة جيدة للاحتفاظ بكلا إدخالات `Springfield`.


### دمج DataFrames مع الصلة الخارجية  
تحتوي DataFrames المدمجة على معلومات كافية لإنشاء DataFrame مع 5 صفوف مع محاذاة جميع المعلومات المعروفة بشكل صحيح وإدراج كل فرع مرة واحدة فقط. سنحاول دمج DataFrames المدمجة على كافة المفاتيح المطابقة (التي تحسب الصلة الداخلية افتراضيًا). يمكننا مقارنة النتيجة بصلة خارجية وأيضًا بصلة خارجية مع مجموعة فرعية مقيدة من الأعمدة كمفاتيح.



دمج `sales_and_managers` مع `revenue_and_sales`


```python
merge_default = pd.merge(sales_and_managers, revenue_and_sales)
print(merge_default)
```


دمج `sales_and_managers` مع `revenue_and_sales` باستخدام `how='outer'`


```python
merge_outer = pd.merge(sales_and_managers, revenue_and_sales, how="outer")
print(merge_outer)
```


ادمج `sales_and_managers` مع `revenue_and_sales` فقط على `['city','state']` باستخدام صلة *خارجية*.


```python
merge_outer_on = pd.merge(
    sales_and_managers, revenue_and_sales, how="outer", on=["city", "state"]
)
print(merge_outer_on)
```


عمل رائع! **لاحظ** كيف يقوم الدمج الافتراضي بإسقاط الصفوف `Springfield`، بينما يتضمن الدمج الخارجي الافتراضي الصفوف مرتين.



## عمليات الدمج المطلوبة



### استخدام merge_ordered()  
يستخدم هذا التمرين DataFrames `austin` و`houston` التي تحتوي على بيانات الطقس من مدينتي أوستن وهيوستن على التوالي. 
تم تسجيل الظروف الجوية في أيام منفصلة ونحتاج إلى دمج إطاري البيانات هذين معًا بحيث يتم ترتيب التواريخ. للقيام بذلك، سوف نستخدم `pd.merge_ordered()`. لاحظ ترتيب الصفوف قبل الدمج وبعده.


```python
austin = pd.read_csv("austin.csv")
houston = pd.read_csv("houston.csv")
print(austin)
print(houston)
```


قم بإجراء عملية دمج مرتبة على `austin` و`houston` باستخدام `pd.merge_ordered()`


```python
tx_weather = pd.merge_ordered(austin, houston)
print(tx_weather)
```


قم بإجراء عملية دمج مرتبة أخرى على `austin` و`houston`.
هذه المرة، حدد وسيطات الكلمات الرئيسية `on='date'` و`suffixes=['_aus','_hus']` بحيث يمكن تمييز الصفوف.


```python
tx_weather_suff = pd.merge_ordered(
    austin, houston, on="date", suffixes=["_aus", "_hus"]
)
print(tx_weather_suff)
```

قم بإجراء عملية دمج ثالثة مرتبة على `austin` و`houston`.
هذه المرة، بالإضافة إلى المعلمات `on` و`suffixes`، حدد وسيطة الكلمة الأساسية `fill_method='ffill'` لاستخدام التعبئة الأمامية لاستبدال الإدخالات `NaN` بأحدث إدخال غير فارغ


```python
tx_weather_ffill = pd.merge_ordered(
    austin, houston, on="date", suffixes=["_aus", "_hus"], fill_method="ffill"
)
print(tx_weather_ffill)
```


أحسنت! لاحظ أنه بعد استخدام طريقة التعبئة، لم تعد هناك إدخالات `NaN`.
