### <center>خريطة التنظيم الذاتي.</center>



خريطة التنظيم الذاتي هي نوع من [الشبكة العصبية](https://en.wikipedia.org/wiki/Artificial_neural_network) يتم تدريبها باستخدام [التعلم غير الخاضع للرقابة](https://en.wikipedia.org/wiki/Unsupervised_learning) وهي أيضًا إحدى شبكات Kohonen العصبية. تعود فكرة إنشاء هذه الشبكات إلى العالم الفنلندي [Teuvo Kohonen](https://en.wikipedia.org/wiki/Teuvo_Kohonen). تقوم الشبكات بشكل أساسي بمهام التجميع وتصور البيانات. ولكنها تسمح أيضًا بتقليل البيانات متعددة الأبعاد إلى مساحة ذات بُعد أصغر، وتُستخدم للبحث عن الأنماط في البيانات.
سنرى في هذه المقالة كيفية حل مشكلة التجميع بمساعدة شبكة Kohonen وسنقوم ببناء خريطة التنظيم الذاتي.



### <center>شبكة كوهونين العصبية.</center>



العنصر الرئيسي للشبكة هو طبقة كوهونين التي تتكون من عدد من العناصر الخطية لها مدخلات م. تقوم كل طبقة بإدخال المتجه $x = (x_1, x_2, ... , x_n)$ من بيانات الإدخال. الناتج من كل طبقة هو
$$y_j = \sum_i^n w_{ij}x_i $$
- $i$ - عدد الإدخال.
- $j$ - عدد الطبقة (عدد الخلايا العصبية).
- $w_{ij}$ - ​​وزن طبقة j-input.  
بعد حساب $y$ لكل خلية عصبية، سيتم تحديد الخلية العصبية للفائز وفقًا لقاعدة "الفائز يأخذ كل شيء". يتم البحث عن الحد الأقصى $$y_{max} = argmax\{y_j\}$$ بين الكل ومن ثم سيكون مخرج هذه الخلية العصبية هو $1$، وستكون جميع المخرجات الأخرى $0$. 
إذا تم الوصول إلى الحد الأقصى في عدة خلايا عصبية:  
- جميع الإشارات ستكون مساوية $1$.
- الأول في قائمة الحد الأقصى سيكون $1$.



### <center>خريطة كوهونين.</center>


- **التهيئة**
 
     الطرق الأكثر شيوعًا لتعيين أوزان العقدة الأولية هي:- تعيين القيم العشوائية لجميع الأوزان.
     - لأخذ $x_i$ بشكل عشوائي من البيانات المدخلة ووضعها في الأوزان.
     - اختيار متجهات الوزن من المكونات الرئيسية [(PCA)](https://en.wikipedia.org/wiki/Principal_component_analysis) لمجموعة بيانات الإدخال.
  
  ونتيجة لذلك، نحصل على $M_k$ هي خريطة الخلايا العصبية. $k$- الخلايا العصبية، يتم تحديد عددها بواسطة التحليلات.
  
      $N$ - ​​عدد البيانات المدخلة.
    
 - **التدريب**
 
    التهيئة $t=0$ هي عدد مرات التكرار وخلط بيانات الإدخال.
    
     - اختيار $x(t)$ من البيانات المدخلة. 
     - حساب المسافة $$d_i(x(t), m_i(t)) = \sqrt{(x_1(t) - m_{1i}(t))^2 + (x_2(t) - m_{2i}(t))^2 + ... + (x_n(t) - m_{ni}(t))^2}$$ من متجه الإدخال إلى جميع الخلايا العصبية في الخريطة. هنا:
          - $d_i(x(t), m_i(t))$ هي [المسافة الإقليدية](https://en.wikipedia.org/wiki/Euclidean_distance).
          - $x(t)$ هو ناقل الإدخال.
          - $m_i \in M_k$ 
     
     من بين جميع الخلايا العصبية، تم تحديده الأقرب إلى المتجه الوارد $d_{min} = argmin\{d_i\}$. وستكون الخلية العصبية المرتبطة بـ $d_{min}$ هي الفائزة. إذا تم الوصول إلى $d_{min}$ في عدة خلايا عصبية، فسيتم اختيار الفائز عشوائيًا. $m_w$ هي الخلية العصبية الفائزة.
     
    تستخدم خرائط كوهونين، على عكس الشبكات، خوارزمية "الفائز يأخذ معظم" في التدريب. بهذه الطريقة، لا يتم تحديث أوزان الخلايا العصبية الفائزة فحسب، بل يتم أيضًا تحديث أوزان الخلايا العصبية القريبة [طوبولوجيًا](https://en.wikipedia.org/wiki/Topology).
   
     - حساب الدالة $h$ التي تحدد "مقياس الجوار" للخلايا العصبية. عادةً ما تكون $h$ هي [الدالة الغوسية](https://en.wikipedia.org/wiki/Gaussian_function). $$h_{wi} = \alpha(t)  \exp(-\frac{||m_w - m_i||^2}{2 \gamma(t) ^ 2})  $$ 
          - $m_w$ هي الخلية العصبية الفائزة.
          - $m_i \in M_k$ عبارة عن خلية عصبية من الخريطة.
          - $\gamma(t)$ هي دالة القيمة لعدد الجيران. الوظيفة الأكثر استخدامًا هي التناقص الخطي من رقم تكرار التدريب. يتم تعيين القيمة في التكرار الأول بواسطة thه محلل. أيضًا في إصدار بسيط $\gamma(t) = \alpha(t)$.
          - $\alpha(t)$ هي وظيفة معدل التعلم. في الإصدار البسيط ثابت. ولكن للحصول على أفضل نتيجة، غالبًا ما يتم استخدام الوظيفة $$\alpha(t) = \frac{A}{t + B}$$ هنا 
              - $t$ هو رقم التكرار.
              - $A and B$ ثابت.
          
  - تغيير الأوزان.
      
      احسب $m_i(t) = m_i(t-1) + h_i(t) (x(t) - m_i(t-1)),     i = 1,2,..., k$.
      
      
  قم بتحديث أوزان جميع الخلايا العصبية المجاورة للخلية العصبية للفائز.
  زيادة $t$ وتكرار التعلم.
  يستمر التدريب حتى $t < N$ أو حتى يصبح الخطأ صغيرًا.

```python
import matplotlib.pylab as plt

%matplotlib inline
import warnings

warnings.filterwarnings("ignore")
from time import time

import numpy as np
import pandas as pd
import sompy

np.random.seed(17)
```


إنشاء مجموعة بيانات "لعبة".


```python
data_len = 200

data_frame_1 = pd.DataFrame(data=np.random.rand(data_len, 4))
data_frame_1.values[:, 1] = (
    data_frame_1.values[:, 1] + 0.42 * np.random.rand(data_len, 1)
)[:, 0]

data_frame_2 = pd.DataFrame(data=np.random.rand(data_len, 4) + 1)
data_frame_2.values[:, 1] = (
    -1 * data_frame_2.values[:, 1] + 0.62 * np.random.rand(data_len, 1)
)[:, 0]

data_frame_3 = pd.DataFrame(data=np.random.rand(data_len, 4) + 2)
data_frame_3.values[:, 1] = (
    0.5 * data_frame_3.values[:, 1] + 1 * np.random.rand(data_len, 1)
)[:, 0]

data_frame_4 = pd.DataFrame(data=np.random.rand(data_len, 4) + 3.5)
data_frame_4.values[:, 1] = (
    -0.1 * data_frame_4.values[:, 1] + 0.5 * np.random.rand(data_len, 1)
)[:, 0]

data_full = np.concatenate((data_frame_1, data_frame_2, data_frame_3, data_frame_4))

fig = plt.figure()
plt.plot(data_full[:, 0], data_full[:, 1], "ob", alpha=0.2, markersize=4)
fig.set_size_inches(7, 7)
```

```python
data_full
```


نحتاج أولاً إلى ضبط حجم الخريطة، حيث أن مجموعة بيانات اللعبة صغيرة، لذا سنقوم أولاً بتعيين الحجم الصغير للخريطة.


```python
mapsize = [2, 2]
```


تقوم طريقة <code>build</code> من SOMFactory بإنشاء خريطة ذاتية التنظيم، وإعطائها حجم الخريطة والبيانات. تأخذ الطريقة حجم الخريطة والبيانات.
<code>initialization='عشوائي'</code> هو نوع من أوزان العقدة الأولية، والقيم العشوائية لجميع الأوزان.


```python
som = sompy.SOMFactory.build(data_full, mapsize, initialization="random")
som.train(n_job=1, verbose="info")
```

للتصور استخدم Mapview.View2DPacked.


```python
v = sompy.mapview.View2DPacked(10, 10, "example", text_size=8)
v.show(som)
```


يمكن للسوم التعرف على أربع مجموعات. على الرغم من أن نطاق المجموعة أبعد ما يكون عن المثالية.
تستخدم طريقة "الكتلة" [sklearn.Kmeans](https://scikit-learn.org/stable/modules/generated/sklearn.cluster.KMeans.html) للتنبؤ بالمجموعات على البيانات الأولية.


```python
v = sompy.mapview.View2DPacked(5, 5, "test", text_size=8)
som.cluster(n_clusters=4)
som.cluster_labels
```

```python
v.show(som, what="cluster");
```


دعونا نلقي نظرة على تصور المجموعات على الشبكة. لهذا استخدم HitMapView.


```python
h = sompy.hitmap.HitMapView(8, 8, "hitmap", text_size=8, show_text=True)
h.show(som);
```


شبكة خريطة التنظيم الذاتي لها نوعان:

سنقوم الآن بإنشاء SOM جديد وإضافة بعض الوسائط للحصول على أفضل نتيجة.
زيادة حجم الخريطة.


```python
mapsize = [20, 20]
```


<code>lattice='rect'</code> عبارة عن شبكة مربعة من SOM.
<code>normalization='var'</code> هو نوع [التطبيع](https://en.wikipedia.org/wiki/Normalization_(statistics)) لبيانات الإدخال. 'var' هو [t-statistic](https://en.wikipedia.org/wiki/T-statistic).
$$\frac{X-\bar{X}}{s}$$
- $X$ هي بيانات الإدخال.
- $\bar{X}$ هو متوسط ​​البيانات المدخلة.
- $s$ هو [الانحراف المعياري](https://en.wikipedia.org/wiki/Standard_deviation).
<code>initialization='pca'</code> هو نوع من أوزان العقدة الأولية، وتهيئة المكون الرئيسي.
<code>neighborhood='gaussian'</code> استخدم الدالة "gaussian" لـ "قياس الحي".


```python
som = sompy.SOMFactory.build(
    data_full,
    mapsize,
    lattice="rect",
    normalization="var",
    initialization="random",
    neighborhood="gaussian",
)
som.train(n_job=1, verbose="info")
```

```python
v = sompy.mapview.View2DPacked(10, 10, "example", text_size=8)
v.show(som)
```

```python
v = sompy.mapview.View2DPacked(5, 5, "test", text_size=8)
som.cluster(n_clusters=4)
som.cluster_labels
```

```python
h = sompy.hitmap.HitMapView(8, 8, "hitmap", text_size=8, show_text=True)
h.show(som);
```
