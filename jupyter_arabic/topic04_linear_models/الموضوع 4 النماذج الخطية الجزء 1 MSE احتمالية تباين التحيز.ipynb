## 1. مقدمة 
سنبدأ بدراسة النماذج الخطية ذات الانحدار الخطي. أولاً، يجب عليك تحديد نموذج يربط المتغير التابع $y$ بالعوامل (أو الميزات) الموضحة؛ بالنسبة للنماذج الخطية، ستتخذ دالة التبعية الشكل التالي: $\large y = w_0 + \sum_{i=1}^m w_i x_i$، حيث $m$ هو عدد الميزات. إذا أضفنا بُعدًا وهميًا $x_0 = 1$ (يُسمى _bias_ أو _intercept_ term) لكل ملاحظة، فيمكن إعادة كتابة النموذج الخطي بطريقة أكثر إحكاما قليلاً عن طريق سحب الحد المطلق $w_0$ إلى المجموع: $\large y = \sum_{i=0}^m w_i x_i = \textbf{w}^\text{T} \textbf{x}$. إذا كانت لدينا مصفوفة من الملاحظات $n$، حيث تكون الصفوف عبارة عن ملاحظات من مجموعة بيانات، فنحن بحاجة إلى إضافة عمود واحد من الملاحظات الموجودة على اليسار. نحدد النموذج على النحو التالي:$$\large \textbf y = \textbf X \textbf w + \epsilon,$$
أين
- $\textbf w \in \mathbb{R}^{m+1}$ – هو $(m+1) \times 1$ ناقل عمود لمعلمات النموذج (في التعلم الآلي، غالبًا ما يشار إلى هذه المعلمات باسم *الأوزان*)؛
- $\textbf X \in \mathbb{R}^{n \times (m+1)}$ - عبارة عن مصفوفة $n \times (m+1)$ للملاحظات وميزاتها، (بما في ذلك العمود الوهمي الموجود على اليسار) مع العمود الكامل [الرتبة](https://en.wikipedia.org/wiki/Rank_(linear_algebra)): $\text{rank}\left(\textbf X\right) = m + 1 $;
- $\epsilon \in \mathbb{R}^n$ - هو $n \times 1$ ناقل عمود عشوائي، يشار إليه باسم *خطأ* أو *ضوضاء*؛
- $\textbf y \in \mathbb{R}^n$ - هو $n \times 1$ عمود متجه - المتغير التابع (أو *الهدف*).
يمكننا أيضًا كتابة هذا التعبير لكل ملاحظة
$$\large y_i = \sum_{j=0}^m w_j X_{ij} + \epsilon_i$$
سيتم تطبيق القيود التالية على مجموعة الأخطاء العشوائية $\epsilon_i$ (انظر نظرية [غاوس ماركوف](https://en.wikipedia.org/wiki/Gauss%E2%80%93Markov_theorem) للحصول على دوافع إحصائية أعمق):
- توقع جميع الأخطاء العشوائية هو صفر: $\forall i: \mathbb{E}\left[\epsilon_i\right] = 0 $;
- جميع الأخطاء العشوائية لها نفس التباين المحدود، وتسمى هذه الخاصية <a href="@@KEEP_00010@@>homoscedasticity</a>: $\forall i: \text{Var}\left(\epsilon_i\right) = \sigma^2 < \infty $;
- الأخطاء العشوائية غير مرتبطة: $\forall i \neq j: \text{Cov}\left(\epsilon_i, \epsilon_j\right) = 0 $.
[المقدر](https://en.wikipedia.org/wiki/Estimator) $\widehat{w}_i $ من الأوزان $w_i $ يسمى *خطي* إذا
$$\large \widehat{w}_i = \omega_{1i}y_1 + \omega_{2i}y_2 + \cdots + \omega_{ni}y_n,$$
حيث $\forall\ k\ $، $\omega_{ki} $
تعتمد فقط على العينات من $\textbf X$، ولكن ليس على $\textbf w$ - نحن لا نعرف المعاملات الحقيقية $w_i$ (ولهذا السبب نحاول تقديرها باستخدام البيانات المقدمة). وبما أن الحل لإيجاد الأوزان المثالية هو المقدر الخطي، فإن النموذج يسمى *الانحدار الخطي*. 
**إخلاء المسؤولية:** في هذه المرحلة، حان الوقت للقول إننا لا نخطط لدمج دورة إحصائية كاملة في مقالة واحدة، لذلك نحاول تقديم الفكرة العامة للرياضيات وراء OLS.


دعونا نقدم تعريفا آخر. يُطلق على المقدر $\widehat{w}_i $ اسم *غير متحيز* عندما يكون توقعه مساويًا للقيمة الحقيقية ولكن غير المعروفة للمعلمة المقدرة:
$$\Large \mathbb{E}\left[\widehat{w}_i\right] = w_i$$
إحدى طرق حساب تلك الأوزان هي طريقة المربعات الصغرى العادية (OLS)، والتي تقلل متوسط مربع الخطأ بين القيمة الفعلية للمتغير التابع والقيمة المتوقعة التي يعطيها النموذج:
$$\Large \begin{array}{rcl}\mathcal{L}\left(\textbf X, \textbf{y}, \textbf{w} \right) &=& \frac{1}{2n} \sum_{i=1}^n \left(y_i - \textbf{w}^\text{T} \textbf{x}_i\right)^2 \\
&=& \frac{1}{2n} \left\| \textbf{y} - \textbf X \textbf{w} \right\|_2^2 \\
&=& \frac{1}{2n} \left(\textbf{y} - \textbf X \textbf{w}\right)^\text{T} \left(\textbf{y} - \textbf X \textbf{w}\right)
\end{array}$$
لحل مشكلة التحسين هذه، نحتاج إلى حساب المشتقات فيما يتعلق بمعلمات النموذج. لقد وضعناها على الصفر وحلنا المعادلة الناتجة لـ $\textbf w$ (قد يبدو التمايز بين المصفوفات صعبًا؛ حاول القيام بذلك من حيث المجاميع للتأكد من الإجابة).
<br>
<details>
<summary> ورقة معلومات صغيرة حول مشتقات المصفوفات (انقر فوق المثلث للتوسيع)</summary>
<p>
$$\large \begin{array}{rcl} 
\frac{\partial}{\partial \textbf{X}} \textbf{X}^{\text{T}} \textbf{A} &=& \textbf{A} \\
\frac{\partial}{\partial \textbf{X}} \textbf{X}^{\text{T}} \textbf{A} \textbf{X} &=& \left(\textbf{A} + \textbf{A}^{\text{T}}\right)\textbf{X} \\
\frac{\partial}{\partial \textbf{A}} \textbf{X}^{\text{T}} \textbf{A} \textbf{y} &=&  \textbf{X}^{\text{T}} \textbf{y}\\
\frac{\partial}{\partial \textbf{X}} \textbf{A}^{-1} &=& -\textbf{A}^{-1} \frac{\partial \textbf{A}}{\partial \textbf{X}} \textbf{A}^{-1} 
\end{array}$$
</p>
</details>
ما نحصل عليه هو:
$$\Large \begin{array}{rcl} \frac{\partial \mathcal{L}}{\partial \textbf{w}} &=& \frac{\partial}{\partial \textbf{w}} \frac{1}{2n} \left( \textbf{y}^{\text{T}} \textbf{y} -2\textbf{y}^{\text{T}} \textbf{X} \textbf{w} + \textbf{w}^{\text{T}} \textbf{X}^{\text{T}} \textbf{X} \textbf{w}\right) \\
&=& \frac{1}{2n} \left(-2 \textbf{X}^{\text{T}} \textbf{y} + 2\textbf{X}^{\text{T}} \textbf{X} \textbf{w}\right)
\end{array}$$
$$\Large \begin{array}{rcl} \frac{\partial \mathcal{L}}{\partial \textbf{w}} = 0 &\Leftrightarrow& \frac{1}{2n} \left(-2 \textbf{X}^{\text{T}} \textbf{y} + 2\textbf{X}^{\text{T}} \textbf{X} \textbf{w}\right) = 0 \\
&\Leftrightarrow& -\textbf{X}^{\text{T}} \textbf{y} + \textbf{X}^{\text{T}} \textbf{X} \textbf{w} = 0 \\
&\Leftrightarrow& \textbf{X}^{\text{T}} \textbf{X} \textbf{w} = \textbf{X}^{\text{T}} \textbf{y} \\
&\Leftrightarrow& \textbf{w} = \left(\textbf{X}^{\text{T}} \textbf{X}\right)^{-1} \textbf{X}^{\text{T}} \textbf{y}
\end{array}$$
مع الأخذ في الاعتبار جميع التعريفات والشروط الموضحة أعلاه، يمكننا القول أنه استنادًا إلى نظرية <a href="@@KEEP_00012@@>Gauss–Markov</a>، فإن مقدرات OLS لمعلمات النموذج هي الأمثل بين جميع المقدرين الخطيين وغير المتحيزين، أي أنهم يقدمون أقل التباين.



## 2. تقدير الاحتمال الأقصى
يمكن للمرء أن يتساءل لماذا نختار تقليل متوسط ​​الخطأ المربع بدلاً من شيء آخر؟ بعد كل شيء، يمكن للمرء تقليل متوسط ​​القيمة المطلقة للباقي. الشيء الوحيد الذي سيحدث، إذا قمنا بتغيير القيمة المصغرة، هو أننا سوف نتجاوز شروط نظرية غاوس ماركوف، وبالتالي فإن تقديراتنا لن تكون الأمثل على تلك الخطية وغير المتحيزة.
قبل أن نواصل، دعونا نستطرد لتوضيح الحد الأقصى لتقدير الاحتمالية بمثال بسيط.ربما يتذكر العديد من الأشخاص صيغة الكحول الإيثيلي، لذلك قررت إجراء تجربة لتحديد ما إذا كان الناس يتذكرون صيغة أبسط للميثانول: $CH_3OH$. لقد قمنا باستطلاع آراء 400 شخص ووجدنا أن 117 شخصًا فقط يتذكرون الصيغة. ومن ثم، فمن المعقول افتراض أن احتمال أن يعرف المستجيب التالي صيغة كحول الميثيل هو $\frac{117}{400} \approx 29\%$. دعونا نبين أن هذا التقييم البديهي ليس جيدًا فحسب، بل هو أيضًا تقدير الحد الأقصى للاحتمالية. ومن أين يأتي هذا التقدير؟ تذكر تعريف توزيع برنولي: المتغير العشوائي له <a href="@@KEEP_00013@@>توزيع برنولي</a> إذا كان يأخذ قيمتين فقط ($1$ و$0$ مع احتمال $\theta$ و$1 - \theta$، على التوالي) وله ما يلي دالة التوزيع الاحتمالي:
$$\Large p\left(\theta, x\right) = \theta^x \left(1 - \theta\right)^\left(1 - x\right), x \in \left\{0, 1\right\}$$
هذا التوزيع هو بالضبط ما نحتاجه، ومعلمة التوزيع $\theta$ هي تقدير احتمال أن يعرف الشخص صيغة كحول الميثيل. في $400$ تجاربنا *المستقلة*، دعنا نشير إلى نتائجها بالرمز $\textbf{x} = \left(x_1, x_2, \ldots, x_{400}\right)$. دعنا نكتب احتمالية بياناتنا (الملاحظات)، أي احتمال ملاحظة هذه الإدراكات الـ 117 للمتغير العشوائي $x = 1$ و 283 إدراكًا لـ $x = 0$:
$$\Large p(\textbf{x}; \theta) = \prod_{i=1}^{400} \theta^{x_i} \left(1 - \theta\right)^{\left(1 - x_i\right)} = \theta^{117} \left(1 - \theta\right)^{283}$$
بعد ذلك، سنقوم بتكبير التعبير فيما يتعلق بـ $\theta$. في أغلب الأحيان لا يتم ذلك مع الاحتمال $p(\textbf{x}; \theta)$ ولكن مع اللوغاريتم الخاص به (التحويل الرتيب لا يغير الحل ولكنه يبسط الحساب إلى حد كبير):
$$\Large \log p(\textbf{x}; \theta) = \log \prod_{i=1}^{400} \theta^{x_i} \left(1 - \theta\right)^{\left(1 - x_i\right)} = $$
$$ \large = \log \theta^{117} \left(1 - \theta\right)^{283} =  117 \log \theta + 283 \log \left(1 - \theta\right)$$الآن، نريد العثور على قيمة $\theta$ التي من شأنها زيادة الاحتمالية إلى أقصى حد. لهذا الغرض، سنأخذ المشتقة بالنسبة إلى $\theta$ ونضعها على الصفر ونحل المعادلة الناتجة:
$$\Large  \frac{\partial \log p(\textbf{x}; \theta)}{\partial \theta} = \frac{\partial}{\partial \theta} \left(117 \log \theta + 283 \log \left(1 - \theta\right)\right) = \frac{117}{\theta} - \frac{283}{1 - \theta};$$
لقد اتضح أن تقييمنا البديهي هو بالضبط الحد الأقصى لتقدير الاحتمالية. الآن دعونا نطبق نفس المنطق على مشكلة الانحدار الخطي ونحاول معرفة ما يكمن وراء الخطأ التربيعي المتوسط. للقيام بذلك، سنحتاج إلى النظر إلى الانحدار الخطي من منظور احتمالي. ومن الطبيعي أن يظل نموذجنا كما هو:
$$\Large \textbf y = \textbf X \textbf w + \epsilon,$$
لكن لنفترض الآن أن الأخطاء العشوائية تتبع [التوزيع الطبيعي](https://en.wikipedia.org/wiki/Normal_distribution) المتمركز (احذر، إنه افتراض إضافي، وليس شرطًا أساسيًا لنظرية غاوس-ماركوف):
$$\Large \epsilon_i \sim \mathcal{N}\left(0, \sigma^2\right)$$
دعونا نعيد كتابة النموذج من منظور جديد:
$$\Large \begin{array}{rcl} 
y_i &=& \sum_{j=1}^m w_j X_{ij} + \epsilon_i \\
&\sim& \sum_{j=1}^m w_j X_{ij} + \mathcal{N}\left(0, \sigma^2\right) \\
p\left(y_i \mid \textbf X; \textbf{w}\right) &=& \mathcal{N}\left(\sum_{j=1}^m w_j X_{ij}, \sigma^2\right)
\end{array}$$
نظرًا لأن الأمثلة تم أخذها بشكل مستقل (الأخطاء غير المترابطة هي أحد شروط نظرية غاوس-ماركوف)، فإن الاحتمالية الكاملة للبيانات ستبدو كمنتج لوظائف الكثافة $p\left(y_i\right)$. دعونا نفكر في احتمالية السجل، والتي تسمح لنا بتبديل المنتجات إلى مبالغ:
$$\Large \begin{array}{rcl} 
\log p\left(\textbf{y}\mid \textbf X; \textbf{w}\right) &=& \log \prod_{i=1}^n \mathcal{N}\left(\sum_{j=1}^m w_j X_{ij}, \sigma^2\right) \\
&=& \sum_{i=1}^n \log \mathcal{N}\left(\sum_{j=1}^m w_j X_{ij}, \sigma^2\right) \\
&=& -\frac{n}{2}\log 2\pi\sigma^2 -\frac{1}{2\sigma^2} \sum_{i=1}^n \left(y_i - \textbf{w}^\text{T} \textbf{x}_i\right)^2
\end{array}$$
نريد العثور على فرضية الاحتمالية القصوى، أي أننا بحاجة إلى تعظيم التعبير $p\left(\textbf{y} \mid \textbf X; \textbf{w}\right)$ للحصول على $\textbf{w}_{\text{ML}}$، وهو نفس تعظيم اللوغاريتم الخاص به. لاحظ أنه أثناء تعظيم الدالة على بعض المعلمات، يمكنك التخلص من جميع الأعضاء الذين لا يعتمدون على هذا المعلم:
$$\Large \begin{array}{rcl} 
\textbf{w}_{\text{ML}} &=& \arg \max_{\textbf w} p\left(\textbf{y}\mid \textbf X; \textbf{w}\right) = \arg \max_{\textbf w} \log p\left(\textbf{y}\mid \textbf X; \textbf{w}\right)\\
&=& \arg \max_{\textbf w} -\frac{n}{2}\log 2\pi\sigma^2 -\frac{1}{2\sigma^2} \sum_{i=1}^n \left(y_i - \textbf{w}^{\text{T}} \textbf{x}_i\right)^2 \\
&=& \arg \max_{\textbf w} -\frac{1}{2\sigma^2} \sum_{i=1}^n \left(y_i - \textbf{w}^{\text{T}} \textbf{x}_i\right)^2 \\
&=&  \arg \min_{\textbf w} \mathcal{L}\left(\textbf X, \textbf{y}, \textbf{w} \right)
\end{array}$$
وهكذا، فقد رأينا أن تعظيم احتمالية البيانات هو نفس التقليل من متوسط الخطأ التربيعي (في ضوء الافتراضات المذكورة أعلاه). وتبين أن دالة التكلفة هذه هي نتيجة لتوزيع الأخطاء بشكل طبيعي.


## 3. تحليل التحيز والتباين
دعونا نتحدث قليلاً عن خصائص الخطأ للتنبؤ بالانحدار الخطي (في الواقع، هذه المناقشة صالحة لجميع خوارزميات التعلم الآلي). لقد قمنا فقط بتغطية ما يلي:
- القيمة الحقيقية للمتغير الهدف هي مجموع الدالة الحتمية $f\left(\textbf{x}\right)$ والخطأ العشوائي $\epsilon$: $y = f\left(\textbf{x}\right) + \epsilon$;
- يتم توزيع الخطأ عادة بمتوسط ​​صفر وبعض التباين: $\epsilon \sim \mathcal{N}\left(0, \sigma^2\right)$؛
- يتم أيضًا توزيع القيمة الحقيقية للمتغير الهدف بشكل طبيعي: $y \sim \mathcal{N}\left(f\left(\textbf{x}\right), \sigma^2\right)$;
- نحاول تقريب دالة حتمية ولكنها غير معروفة $f\left(\textbf{x}\right)$ باستخدام دالة خطية للمتغيرات المشتركة $\widehat{f}\left(\textbf{x}\right)$، والتي بدورها هي تقدير نقطي للدالة $f$ في مساحة الوظيفة (على وجه التحديد، عائلة الدوال الخطية التي حددنا مساحتنا لها)، أي متغير عشوائي له متوسط ​​وتباين.
لذا فإن الخطأ عند النقطة $\textbf{x}$ يتحلل على النحو التالي:
$$\Large \begin{array}{rcl} 
\text{Err}\left(\textbf{x}\right) &=& \mathbb{E}\left[\left(y - \widehat{f}\left(\textbf{x}\right)\right)^2\right] \\
&=& \mathbb{E}\left[y^2\right] + \mathbb{E}\left[\left(\widehat{f}\left(\textbf{x}\right)\right)^2\right] - 2\mathbb{E}\left[y\widehat{f}\left(\textbf{x}\right)\right] \\
&=& \mathbb{E}\left[y^2\right] + \mathbb{E}\left[\widehat{f}^2\right] - 2\mathbb{E}\left[y\widehat{f}\right] \\
\end{array}$$
من أجل الوضوح، سوف نحذف تدوين حجة الوظائف. دعونا نفكر في كل عضو على حدة. يتم تحليل الأولين بسهولة وفقًا للصيغة $\text{Var}\left(z\right) = \mathbb{E}\left[z^2\right] - \mathbb{E}\left[z\right]^2$:
$$\Large \begin{array}{rcl} 
\mathbb{E}\left[y^2\right] &=& \text{Var}\left(y\right) + \mathbb{E}\left[y\right]^2 = \sigma^2 + f^2\\
\mathbb{E}\left[\widehat{f}^2\right] &=& \text{Var}\left(\widehat{f}\right) + \mathbb{E}\left[\widehat{f}\right]^2 \\
\end{array}$$
لاحظ أن:
$$\Large \begin{array}{rcl} 
\text{Var}\left(y\right) &=& \mathbb{E}\left[\left(y - \mathbb{E}\left[y\right]\right)^2\right] \\
&=& \mathbb{E}\left[\left(y - f\right)^2\right] \\
&=& \mathbb{E}\left[\left(f + \epsilon - f\right)^2\right] \\
&=& \mathbb{E}\left[\epsilon^2\right] = \sigma^2
\end{array}$$
$$\Large \mathbb{E}[y] = \mathbb{E}[f + \epsilon] = \mathbb{E}[f] + \mathbb{E}[\epsilon] = f$$
وأخيرًا، نصل إلى الحد الأخير في المجموع. تذكر أن الخطأ والمتغير الهدف مستقلان عن بعضهما البعض:
$$\Large \begin{array}{rcl} 
\mathbb{E}\left[y\widehat{f}\right] &=& \mathbb{E}\left[\left(f + \epsilon\right)\widehat{f}\right] \\
&=& \mathbb{E}\left[f\widehat{f}\right] + \mathbb{E}\left[\epsilon\widehat{f}\right] \\
&=& f\mathbb{E}\left[\widehat{f}\right] + \mathbb{E}\left[\epsilon\right] \mathbb{E}\left[\widehat{f}\right]  = f\mathbb{E}\left[\widehat{f}\right]
\end{array}$$
وأخيرا، دعونا نجمع كل هذا معا:
$$\Large \begin{array}{rcl} 
\text{Err}\left(\textbf{x}\right) &=& \mathbb{E}\left[\left(y - \widehat{f}\left(\textbf{x}\right)\right)^2\right] \\
&=& \sigma^2 + f^2 + \text{Var}\left(\widehat{f}\right) + \mathbb{E}\left[\widehat{f}\right]^2 - 2f\mathbb{E}\left[\widehat{f}\right] \\
&=& \left(f - \mathbb{E}\left[\widehat{f}\right]\right)^2 + \text{Var}\left(\widehat{f}\right) + \sigma^2 \\
&=& \text{Bias}\left(\widehat{f}\right)^2 + \text{Var}\left(\widehat{f}\right) + \sigma^2
\end{array}$$
وبهذا نكون قد وصلنا إلى هدفنا النهائي -- تخبرنا الصيغة الأخيرة أن خطأ التنبؤ لأي نموذج من النوع $y = f\left(\textbf{x}\right) + \epsilon$ يتكون من:
- الانحياز التربيعي: $\text{Bias}\left(\widehat{f}\right)$ هو متوسط الخطأ لجميع مجموعات البيانات؛
- التباين: $\text{Var}\left(\widehat{f}\right)$ هو تباين الأخطاء، أو بمقدار الخطأ الذي سيختلف إذا قمنا بتدريب النموذج على مجموعات مختلفة من البيانات؛
- خطأ غير قابل للإزالة: $\sigma^2$.على الرغم من أننا لا نستطيع فعل أي شيء باستخدام المصطلح $\sigma^2$، إلا أنه يمكننا التأثير على المصطلحين الأولين. من الناحية المثالية، نود إلغاء كلا هذين المصطلحين (المربع الأيسر العلوي من الصورة)، ولكن من الناحية العملية، غالبًا ما يكون من الضروري الموازنة بين التقديرات المتحيزة وغير المستقرة (التباين العالي).
<img src="../../img/bvtf.png" width="480">
بشكل عام، عندما يصبح النموذج أكثر حسابية (على سبيل المثال عندما ينمو عدد المعلمات الحرة)، يزداد أيضًا تباين (تشتت) التقدير، ولكن يتناقص التحيز. نظرًا لأن مجموعة التدريب يتم حفظها بالكامل بدلاً من التعميم، فإن التغييرات الصغيرة تؤدي إلى نتائج غير متوقعة (التركيب الزائد). على الجانب الآخر، إذا كان النموذج ضعيفًا جدًا، فلن يتمكن من تعلم النمط، مما يؤدي إلى تعلم شيء مختلف يتم تعويضه فيما يتعلق بالحل الصحيح.
<img src="../../img/biasvariance.png" width="480">
تؤكد نظرية غاوس-ماركوف أن مقدر OLS لمعلمات النموذج الخطي هو الأفضل لفئة المقدر الخطي غير المتحيز. هذا يعني أنه في حالة وجود أي نموذج غير متحيز آخر $g$، من نفس فئة النماذج الخطية، فيمكننا التأكد من أن $Var\left(\widehat{f}\right) \leq Var\left(g\right)$.



## 4. تنظيم الانحدار الخطيهناك مواقف قد نقوم فيها عن قصد بزيادة انحياز النموذج من أجل الاستقرار، أي لتقليل تباين النموذج $\text{Var}\left(\widehat{f}\right)$. أحد شروط نظرية غاوس-ماركوف هو رتبة العمود الكامل للمصفوفة $\textbf{X}$. وإلا، فإن حل OLS $\textbf{w} = \left(\textbf{X}^\text{T} \textbf{X}\right)^{-1} \textbf{X}^\text{T} \textbf{y}$ غير موجود لأن المصفوفة العكسية $\left(\textbf{X}^\text{T} \textbf{X}\right)^{-1}$ غير موجودة. بمعنى آخر، المصفوفة $\textbf{X}^\text{T} \textbf{X}$ ستكون فردية أو منحلة. تسمى هذه المشكلة <a href="@@KEEP_00015@@> مشكلة سيئة الطرح</a>. يجب تصحيح مثل هذه المشاكل، وهي أن المصفوفة $\textbf{X}^\text{T} \textbf{X}$ يجب أن تصبح غير متدهورة، أو منتظمة (وهذا هو سبب تسمية هذه العملية بالتسوية). غالبًا ما نلاحظ ما يسمى بالتعددية الخطية في البيانات: عندما تكون ميزتان أو أكثر مرتبطتين بقوة، يتجلى ذلك في المصفوفة $\textbf{X}$ في شكل اعتماد خطي "تقريبًا" بين الأعمدة. على سبيل المثال، في مشكلة التنبؤ بأسعار المنازل من خلال معلماتها، سيكون للخاصيتين "منطقة بها شرفة" و"منطقة بدون شرفة" علاقة خطية "تقريبًا". رسميًا، المصفوفة $\textbf{X}^\text{T} \textbf{X}$ لمثل هذه البيانات قابلة للعكس، ولكن بسبب تعدد الخطية، فإن بعض قيمها الذاتية ستكون قريبة من الصفر. في المصفوفة العكسية $\textbf{X}^\text{T} \textbf{X}$، ستظهر بعض القيم الذاتية الكبيرة للغاية، حيث أن القيم الذاتية للمصفوفة العكسية هي $\frac{1}{\lambda_i}$. نتيجة هذا التذبذب في القيم الذاتية هو تقدير غير مستقر لمعلمات النموذج، أي أن إضافة مجموعة جديدة من الملاحظات إلى بيانات التدريب سيؤدي إلى حل مختلف تمامًا. 
إحدى طرق التسوية هي <a href="@@KEEP_00016@@>Tikhonov Regularization</a>، والتي تبدو عمومًا وكأنها إضافة عضو جديد إلى متوسط الخطأ المربع:
$$\Large \begin{array}{rcl} 
\mathcal{L}\left(\textbf{X}, \textbf{y}, \textbf{w} \right) &=& \frac{1}{2n} \left\| \textbf{y} - \textbf{X} \textbf{w} \right\|_2^2 + \left\| \Gamma \textbf{w}\right\|^2\\
\end{array}$$غالبًا ما يتم التعبير عن مصفوفة تيخونوف على أنها حاصل ضرب رقم بواسطة مصفوفة الهوية: $\Gamma = \frac{\lambda}{2} E$. في هذه الحالة، تصبح مشكلة تقليل متوسط ​​الخطأ التربيعي مشكلة تتعلق بتقييد القاعدة $L_2$. إذا قمنا بتمييز دالة التكلفة الجديدة فيما يتعلق بمعلمات النموذج، وقمنا بتعيين الدالة الناتجة على الصفر، وأعدنا ترتيب $\textbf{w}$، فسنحصل على الحل الدقيق للمشكلة.
$$\Large \begin{array}{rcl} 
\textbf{w} &=& \left(\textbf{X}^{\text{T}} \textbf{X} + \lambda \textbf{E}\right)^{-1} \textbf{X}^{\text{T}} \textbf{y}
\end{array}$$
ويسمى هذا النوع من الانحدار انحدار ريدج. التلال هي المصفوفة القطرية التي نضيفها إلى المصفوفة $\textbf{X}^\text{T} \textbf{X}$ لضمان حصولنا على مصفوفة عادية نتيجة لذلك.
<img src="../../img/ridge.png">
مثل هذا الحل يقلل من التشتت ولكنه يصبح متحيزًا لأن معيار ناقل المعلمات يتم تقليله أيضًا، مما يجعل الحل يتحول نحو الصفر. في الشكل أدناه، يقع حل OLS عند تقاطع الخطوط البيضاء المنقطة. تمثل النقاط الزرقاء حلولاً مختلفة لانحدار التلال. يمكن ملاحظة أنه من خلال زيادة معلمة التنظيم $\lambda$، نقوم بتحويل الحل نحو الصفر.
