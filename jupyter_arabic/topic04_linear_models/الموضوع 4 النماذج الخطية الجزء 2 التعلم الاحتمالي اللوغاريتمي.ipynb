## 1. المصنف الخطي



الفكرة الأساسية وراء المصنف الخطي يمكن فصل فئتين مستهدفتين بواسطة طائرة مفرطة في مساحة الميزة. إذا كان من الممكن القيام بذلك دون خطأ، فإن مجموعة التدريب تسمى *قابلة للفصل خطيًا*.
<img src="../../img/logit.png"/>
لقد رأينا بالفعل الانحدار الخطي والمربعات الصغرى العادية (OLS). لنفكر في مشكلة التصنيف الثنائي، ونشير إلى الفئات المستهدفة لتكون "+1" (أمثلة إيجابية) و"-1" (أمثلة سلبية). يمكن تعريف أحد أبسط المصنفات الخطية باستخدام الانحدار على النحو التالي:
$$\Large a(\textbf{x}) = \text{sign}(\textbf{w}^\text{T}\textbf x),$$أين
 - $\textbf{x}$ – هو ناقل الميزات (مع الهوية)؛
 - $\textbf{w}$ – هو ناقل للأوزان في النموذج الخطي (مع انحياز $w_0$)؛
 - $\text{sign}(\bullet)$ - هي دالة الإشارة التي تُرجع إشارة الوسيط الخاص بها؛
 - $a(\textbf{x})$ - عبارة عن استجابة مصنفة لـ $\textbf{x}$.



## 2. الانحدار اللوجستي كمصنف خطي



يعد الانحدار اللوجستي حالة خاصة للمصنف الخطي، ولكن له فائدة إضافية تتمثل في التنبؤ باحتمال $p_+$ لإحالة المثال $\textbf{x}_\text{i}$ إلى الفئة "+":
$$\Large p_+ = P\left(y_i = 1 \mid \textbf{x}_\text{i}, \textbf{w}\right) $$
إن القدرة على التنبؤ ليس فقط بالرد ("+1" أو "-1") ولكن *احتمالية* التعيين إلى الفئة "+1" هي مطلب مهم للغاية في العديد من مشاكل العمل، على سبيل المثال. التصنيف الائتماني حيث يتم استخدام الانحدار اللوجستي بشكل تقليدي. يتم تصنيف العملاء الذين تقدموا بطلب للحصول على قرض بناءً على هذا الاحتمال المتوقع (بترتيب تنازلي) للحصول على لوحة نتائج تصنف العملاء من سيئ إلى جيد. فيما يلي مثال على لوحة النتائج هذه.
<img src='../../img/toy_scorecard_eng.png' width=60% />
يختار البنك حدًا $p_*$ للتنبؤ باحتمالية التخلف عن سداد القرض (في الصورة هو $0.15$) ويتوقف عن الموافقة على القروض بدءًا من تلك القيمة. علاوة على ذلك، من الممكن ضرب هذا الاحتمال المتوقع بمبلغ القرض للحصول على توقعات الخسائر من العميل، والتي يمكن أن تشكل أيضًا مقاييس أعمال جيدة (قد يكون لدى خبراء التسجيل المزيد لإضافته، ولكن الجوهر الرئيسي هو هذا).


للتنبؤ بالاحتمال $p_+ \in [0,1]$، يمكننا البدء بإنشاء تنبؤ خطي باستخدام OLS: $b(\textbf{x}) = \textbf{w}^\text{T} \textbf{x} \in \mathbb{R}$. لكن تحويل القيمة الناتجة إلى الاحتمالية ضمن النطاق [0، 1] يتطلب بعض الوظائف $f: \mathbb{R} \rightarrow [0,1]$. يستخدم الانحدار اللوجستي وظيفة محددة لهذا: $\sigma(z) = \frac{1}{1 + \exp^{-z}}$. الآن دعونا نفهم ما هي المتطلبات الأساسية.
<img src='../../img/sigmoid.png' width=50% />



دعنا نشير إلى احتمال وقوع حدث $X$ كـ $P(X)$. ثم يتم تحديد نسبة الأرجحية $OR(X)$ بواسطة $\frac{P(X)}{1-P(X)}$، وهي نسبة احتمالات وقوع حدث ما أم لا. من الواضح أن نسبة الاحتمالية والأرجحية تحتوي على نفس المعلومات، ولكن بينما $P(X)$ يتراوح من 0 إلى 1، فإن $OR(X)$ يقع في النطاق من 0 إلى $\infty$.
إذا قمنا بحساب لوغاريتم $OR(X)$ (لوغاريتم الاحتمالات أو نسبة احتمالية السجل)، فمن السهل ملاحظة ذلك $\log{OR(X)} \in \mathbb{R}$. وهذا ما سنستخدمه مع OLS.
دعونا نرى كيف سيؤدي الانحدار اللوجستي إلى التنبؤ $p_+ = P\left(y_i = 1 \mid \textbf{x}_\text{i}, \textbf{w}\right)$. (في الوقت الحالي، لنفترض أننا حصلنا بطريقة ما على أوزان $\textbf{w}$، أي قمنا بتدريب النموذج. لاحقًا، سننظر في كيفية القيام بذلك.)
**الخطوة 1.** احسب $w_{0}+w_{1}x_1 + w_{2}x_2 + ... = \textbf{w}^\text{T}\textbf{x}$. (المعادلة $\textbf{w}^\text{T}\textbf{x} = 0$ تحدد المستوى الزائدي الذي يفصل الأمثلة إلى فئتين)؛
**الخطوة 2.** حساب نسبة احتمالات السجل: $ \log(OR_{+}) = \textbf{w}^\text{T}\textbf{x}$.
**الخطوة 3.** الآن بعد أن أتيحت لنا الفرصة لتعيين مثال لفئة "+" - $OR_{+}$، احسب $p_{+}$ باستخدام العلاقة البسيطة:
$$\large p_{+} = \frac{OR_{+}}{1 + OR_{+}} = \frac{\exp^{\textbf{w}^\text{T}\textbf{x}}}{1 + \exp^{\textbf{w}^\text{T}\textbf{x}}} = \frac{1}{1 + \exp^{-\textbf{w}^\text{T}\textbf{x}}} = \sigma(\textbf{w}^\text{T}\textbf{x})$$
على الجانب الأيمن، يمكنك أن ترى أن لدينا الدالة السيني.لذلك، يتنبأ الانحدار اللوجستي باحتمالية تعيين مثال للفئة "+" (على افتراض أننا نعرف ميزات النموذج وأوزانه) كتحويل سيني لمجموعة خطية من متجه الوزن ومتجه الميزة:
$$\large p_+(\textbf{x}_\text{i}) = P\left(y_i = 1 \mid \textbf{x}_\text{i}, \textbf{w}\right) = \sigma(\textbf{w}^\text{T}\textbf{x}_\text{i}). $$
بعد ذلك، سنرى كيف يتم تدريب النموذج. سوف نعتمد مرة أخرى على تقدير الاحتمال الأقصى.



## 3. تقدير الاحتمال الأقصى والانحدار اللوجستي
الآن دعونا نرى كيف يتم الحصول على مشكلة تحسين الانحدار اللوجستي من MLE، وهي تقليل وظيفة الخسارة *اللوجستية*. لقد رأينا للتو أن الانحدار اللوجستي يمثل احتمالية تعيين مثال للفئة "+" على النحو التالي:
$$\Large p_+(\textbf{x}_\text{i}) = P\left(y_i = 1 \mid \textbf{x}_\text{i}, \textbf{w}\right) = \sigma(\textbf{w}^T\textbf{x}_\text{i})$$
بعد ذلك، بالنسبة للفئة "-"، يكون التعبير المقابل كما يلي:
$$\Large p_-(\textbf{x}_\text{i})  = P\left(y_i = -1 \mid \textbf{x}_\text{i}, \textbf{w}\right)  = 1 - \sigma(\textbf{w}^T\textbf{x}_\text{i}) = \sigma(-\textbf{w}^T\textbf{x}_\text{i}) $$
يمكن دمج هذين التعبيرين بذكاء في تعبير واحد (راقب بعناية، ربما يتم خداعك):
$$\Large P\left(y = y_i \mid \textbf{x}_\text{i}, \textbf{w}\right) = \sigma(y_i\textbf{w}^T\textbf{x}_\text{i})$$
يُعرف التعبير $M(\textbf{x}_\text{i}) = y_i\textbf{w}^T\textbf{x}_\text{i}$ بهامش التصنيف على الكائن $\textbf{x}_\text{i}$ (يجب عدم الخلط بينه وبين الفجوة، والتي تسمى أيضًا الهامش، في سياق SVM). إذا كانت غير سلبية، فإن النموذج صحيح في اختيار فئة الكائن $\textbf{x}_\text{i}$؛ إذا كانت سالبة، فهذا يعني أن الكائن $\textbf{x}_\text{i}$ تم تصنيفه بشكل خاطئ. لاحظ أنه يتم تعريف الهامش للكائنات الموجودة في مجموعة التدريب فقط حيث تكون تسميات فئة الهدف الحقيقية $y_i$ معروفة.
لكي نفهم بالضبط لماذا وصلنا إلى مثل هذا الاستنتاج، دعونا ننتقل إلى التفسير الهندسي للمصنف الخطي.



أولاً، أوصي بالنظر إلى مشكلة تمهيدية كلاسيكية في الجبر الخطي: العثور على المسافة من النقطة ذات متجه نصف القطر $\textbf{x}_A$ إلى المستوى المحدد بواسطة المعادلة $\textbf{w}^\text{T}\textbf{x} = 0.$
الجواب:
$$\rho(\textbf{x}_A, \textbf{w}^\text{T}\textbf{x} = 0) = \frac{\textbf{w}^\text{T}\textbf{x}_A}{||\textbf{w}||}$$



<img src='../../img/simple_linal_task.png' width=60% />


عندما نصل إلى الإجابة، سنفهم أنه كلما زادت القيمة المطلقة للتعبير $\textbf{w}^\text{T}\textbf{x}_\text{i}$، كلما ابتعدت النقطة $\textbf{x}_\text{i}$ عن المستوى $\textbf{w}^\text{T}\textbf{x} = 0.$.
ومن ثم، فإن تعبيرنا $M(\textbf{x}_\text{i}) = y_i\textbf{w}^\text{T}\textbf{x}_\text{i}$ هو نوع من "الثقة" في تصنيف نموذجنا للكائن $\textbf{x}_\text{i}$:
- إذا كان الهامش كبيرًا (بالقيمة المطلقة) وإيجابيًا، يتم تعيين تسمية الفئة بشكل صحيح، ويكون الكائن بعيدًا عن المستوى الزائد الفاصل، أي يتم تصنيفه بثقة. انظر النقطة $x_3$ في الصورة؛
- إذا كان الهامش كبيرًا (بالقيمة المطلقة) وسالبًا، فسيتم تعيين تسمية الفئة بشكل غير صحيح، ويكون الكائن بعيدًا عن المستوى الزائد الفاصل (من المرجح أن يكون الكائن شذوذًا؛ على سبيل المثال، قد يتم تصنيفه بشكل غير صحيح في مجموعة التدريب). انظر النقطة $x_1$ في الصورة؛
- إذا كان الهامش صغيرًا (بالقيمة المطلقة)، فإن الكائن يكون قريبًا من المستوى الزائد الفاصل، وتحدد علامة الهامش ما إذا كان الكائن قد تم تصنيفه بشكل صحيح. راجع النقاط $x_2$ و$x_4$ على قطعة الأرض؛



<img src='../../img/margin.png' width=60% />



لنحسب الآن احتمالية مجموعة البيانات، أي احتمالية مراقبة المتجه المحدد $\textbf{y}$ من مجموعة البيانات $X$. سنفترض افتراضًا قويًا: الكائنات تأتي بشكل مستقل عن توزيع واحد (*i.i.d.*). ومن ثم يمكننا أن نكتب
$$\Large P\left(\textbf{y} \mid \textbf{X}, \textbf{w}\right) = \prod_{i=1}^{\ell} P\left(y = y_i \mid \textbf{x}_\text{i}, \textbf{w}\right),$$
حيث $\ell$ هو طول مجموعة البيانات $\textbf{X}$ (عدد الصفوف).
كالعادة، لنأخذ لوغاريتم هذا التعبير لأن تحسين المجموع أسهل بكثير من تحسين المنتج:
$$\Large \log P\left(\textbf{y} \mid \textbf{X}, \textbf{w}\right) = \log \prod_{i=1}^{\ell} P\left(y = y_i \mid \textbf{x}_\text{i}, \textbf{w}\right) = \log \prod_{i=1}^{\ell} \sigma(y_i\textbf{w}^\text{T}\textbf{x}_\text{i})   = $$
$$\Large  = \sum_{i=1}^{\ell} \log \sigma(y_i\textbf{w}^\text{T}\textbf{x}_\text{i}) = \sum_{i=1}^{\ell} \log \frac{1}{1 + \exp^{-y_i\textbf{w}^\text{T}\textbf{x}_\text{i}}} = - \sum_{i=1}^{\ell} \log (1 + \exp^{-y_i\textbf{w}^\text{T}\textbf{x}_\text{i}})$$



إن تعظيم الاحتمال يعادل تقليل التعبير:
$$\Large \mathcal{L_{\log}} (\textbf X, \textbf{y}, \textbf{w}) = \sum_{i=1}^{\ell} \log (1 + \exp^{-y_i\textbf{w}^\text{T}\textbf{x}_\text{i}}).$$
هذه هي وظيفة الخسارة *اللوجستية* التي يتم تلخيصها على جميع الكائنات في مجموعة التدريب.دعونا نلقي نظرة على الوظيفة الجديدة كدالة للهامش $L(M) = \log (1 + \exp^{-M})$ ونرسمها جنبًا إلى جنب مع الرسم البياني *خسارة صفر*، والذي يعاقب ببساطة نموذج الخطأ في كل كائن بمقدار 1 (هامش سلبي): $L_{1/0}(M) = [M < 0]$.



<img src='../../img/logloss_margin_eng.png' width=60% />



تعكس الصورة فكرة أننا إذا لم نتمكن من تقليل عدد الأخطاء في مشكلة التصنيف بشكل مباشر (على الأقل ليس عن طريق طرق التدرج - مشتقة دالة الخسارة صفر واحد عند الصفر يتحول إلى ما لا نهاية)، فيمكننا تقليل حدودها العليا. بالنسبة لوظيفة الخسارة اللوجستية (حيث يكون اللوغاريتم ثنائيًا، لكن هذا لا يهم)، يكون ما يلي صالحًا:
$$\Large \mathcal{L_{1/0}} (\textbf X, \textbf{y}, \textbf{w}) = \sum_{i=1}^{\ell} [M(\textbf{x}_\text{i}) < 0] \leq \sum_{i=1}^{\ell} \log (1 + \exp^{-y_i\textbf{w}^\text{T}\textbf{x}_\text{i}}) = \mathcal{L_{\log}} (\textbf X, \textbf{y}, \textbf{w}), $$
حيث $\mathcal{L_{1/0}} (\textbf X, \textbf{y})$ هو ببساطة عدد أخطاء الانحدار اللوجستي مع الأوزان $\textbf{w}$ في مجموعة البيانات $(\textbf X, \textbf{y})$.
وبالتالي، من خلال تقليل الحد الأعلى لـ $\mathcal{L_{log}}$ بعدد أخطاء التصنيف، نأمل في تقليل عدد الأخطاء نفسها.
## 4. $L_2$- تسوية الخسارة اللوجستية
$L_2$ - إن تنظيم الانحدار اللوجستي هو نفسه تقريبًا كما في حالة انحدار التلال. بدلاً من تصغير الوظيفة $\mathcal{L_{\log}} (\textbf X, \textbf{y}, \textbf{w})$، نقوم بتصغير ما يلي:
$$\Large \mathcal{J}(\textbf X, \textbf{y}, \textbf{w}) = \mathcal{L_{\log}} (\textbf X, \textbf{y}, \textbf{w}) + \lambda |\textbf{w}|^2$$
في حالة الانحدار اللوجستي، يتم عادةً تقديم معامل التنظيم العكسي $C = \frac{1}{\lambda}$. ثم سيكون حل المشكلة:
$$\Large \widehat{\textbf w}  = \arg \min_{\textbf{w}} \mathcal{J}(\textbf X, \textbf{y}, \textbf{w}) =  \arg \min_{\textbf{w}}\ (C\sum_{i=1}^{\ell} \log (1 + \exp^{-y_i\textbf{w}^\text{T}\textbf{x}_\text{i}})+ |\textbf{w}|^2)$$ 
بعد ذلك، سنلقي نظرة على مثال يتيح لنا أن نفهم بشكل حدسي أحد تفسيرات التنظيم.
