```python
import numpy as np
import pandas as pd

pd.set_option("display.max.columns", 100)
# to draw pictures in jupyter notebook
%matplotlib inline
# we don't like warnings
# you can comment the following 2 lines if you'd like to
import warnings

import matplotlib.pyplot as plt
import seaborn as sns

warnings.filterwarnings("ignore")
```

```python
data = pd.read_csv("../../data/adult.data.csv")
data.head()
```


**1. كم عدد الرجال والنساء (*ميزة الجنس*) الممثلين في مجموعة البيانات هذه؟** 


```python
data["sex"].value_counts()
```


**2. ما هو متوسط العمر (*ميزة العمر*) للنساء؟**


```python
data[data["sex"] == "Female"]["age"].mean()
```

**3. ما هي نسبة المواطنين الألمان (*ميزة البلد الأصلي*)؟**


```python
float((data["native-country"] == "Germany").sum()) / data.shape[0]
```


** 4-5. ما هي القيمة المتوسطة والانحراف المعياري لعمر أولئك الذين يحصلون على أكثر من 50 ألفًا سنويًا (*ميزة الراتب*) وأولئك الذين يحصلون على أقل من 50 ألفًا سنويًا؟ **


```python
ages1 = data[data["salary"] == ">50K"]["age"]
ages2 = data[data["salary"] == "<=50K"]["age"]
print(
    "The average age of the rich: {0} +- {1} years, poor - {2} +- {3} years.".format(
        round(ages1.mean()),
        round(ages1.std(), 1),
        round(ages2.mean()),
        round(ages2.std(), 1),
    )
)
```


**6. هل صحيح أن الأشخاص الذين يحصلون على أكثر من 50 ألفًا حصلوا على التعليم الثانوي على الأقل؟ (*التعليم - البكالوريوس، أو الأستاذ الجامعي، أو Assoc-acdm، أو Assoc-voc، أو الماجستير* أو *الدكتوراه*)**


```python
data[data["salary"] == ">50K"]["education"].unique()  # No
```


**7. عرض إحصائيات العمر لكل سباق (*خاصية السباق*) ولكل جنس. استخدم *groupby()* و*describe()*. ابحث عن الحد الأقصى لعمر الرجال من عرق *الأمريكيين الهنود والإسكيمو*.**


```python
for (race, sex), sub_df in data.groupby(["race", "sex"]):
    print("Race: {0}, sex: {1}".format(race, sex))
    print(sub_df["age"].describe())
```


** 8. من بين من تكون نسبة أولئك الذين يكسبون كثيرًا (> 50 ألفًا) أكثر: بين الرجال المتزوجين أو غير المتزوجين (*ميزة الحالة الاجتماعية*)؟ خذ بعين الاعتبار الأشخاص المتزوجين الذين لديهم *الحالة الاجتماعية* بدءًا من *متزوج* (زوج متزوج أو زوج غائب أو متزوج بـ AF)، أما الباقي فيعتبرون عزابًا.**


```python
# married men
data[(data["sex"] == "Male") 
     & (data["marital-status"].str.startswith("Married"))][
    "salary"
].value_counts(normalize=True)
```

```python
# single men
data[
    (data["sex"] == "Male")
    & ~(data["marital-status"].str.startswith("Married"))
]["salary"].value_counts(normalize=True)
```


من الجيد أن نتزوج :)



**9. ما هو الحد الأقصى لعدد الساعات التي يعملها الشخص في الأسبوع (*خاصية الساعات في الأسبوع*)؟ كم عدد الأشخاص الذين يعملون هذا العدد من الساعات وما هي نسبة الذين يكسبون الكثير بينهم؟**


```python
max_load = data["hours-per-week"].max()
print("Max time - {0} hours./week.".format(max_load))

num_workaholics = data[data["hours-per-week"] == max_load].shape[0]
print("Total number of such hard workers {0}".format(num_workaholics))

rich_share = (
    float(
        data[(data["hours-per-week"] == max_load) & (data["salary"] == ">50K")].shape[0]
    )
    / num_workaholics
)
print("Percentage of rich among them {0}%".format(int(100 * rich_share)))
```


**10. احسب متوسط وقت العمل (*ساعات في الأسبوع*) لأولئك الذين يكسبون القليل والكثير (*الراتب*) لكل بلد (*البلد الأصلي*).**



طريقة بسيطة:


```python
for (country, salary), sub_df in data.groupby(["native-country", "salary"]):
    print(country, salary, round(sub_df["hours-per-week"].mean(), 2))
```


طريقة أنيقة:

```python
pd.crosstab(
    data["native-country"],
    data["salary"],
    values=data["hours-per-week"],
    aggfunc=np.mean,
).T
```
