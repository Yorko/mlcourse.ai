```python
import warnings

warnings.filterwarnings("ignore")
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import Lasso, LassoCV, LinearRegression
from sklearn.metrics.regression import mean_squared_error
from sklearn.model_selection import (GridSearchCV, cross_val_score,
                                     train_test_split)
from sklearn.preprocessing import StandardScaler
```


**نحن نعمل مع مجموعة بيانات جودة UCI Wine (لا داعي لتنزيلها - فهي موجودة بالفعل، في الريبو وفي Kaggle Dataset).**


```python
data = pd.read_csv("../../data/winequality-white.csv", sep=";")
```

```python
data.head()
```

```python
data.info()
```


**افصل الميزة المستهدفة، وقسم البيانات بنسبة 7:3 (30% من مجموعة الإيقاف، استخدم Random_state=17)، وقم بمعالجة البيانات مسبقًا باستخدام `StandardScaler`.**


```python
y = data["quality"]
X = data.drop("quality", axis=1)

X_train, X_holdout, y_train, y_holdout = train_test_split(
    X, y, test_size=0.3, random_state=17
)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_holdout_scaled = scaler.transform(X_holdout)
```


## الانحدار الخطي



**تدريب نموذج الانحدار الخطي البسيط (المربعات الصغرى العادية).**


```python
linreg = LinearRegression()
linreg.fit(X_train_scaled, y_train);
```


**<font color='red'>السؤال 1:</font> ما المقصود بالأخطاء المربعة المتوسطة لتنبؤات النماذج في مجموعات القطارات والاحتجاز؟**


```python
print(
    "Mean squared error (train): %.3f"
    % mean_squared_error(y_train, linreg.predict(X_train_scaled))
)
print(
    "Mean squared error (test): %.3f"
    % mean_squared_error(y_holdout, linreg.predict(X_holdout_scaled))
)
```


**فرز الميزات حسب تأثيرها على الميزة المستهدفة (جودة النبيذ). احذر من أن المعاملات الإيجابية والسلبية الكبيرة تعني تأثيرًا كبيرًا على الهدف. من السهل استخدام `pandas.DataFrame` هنا.**
**<font color='red'>السؤال 2:</font> ما هي الميزة التي يتعامل معها نموذج الانحدار الخطي هذا باعتبارها الأكثر تأثيرًا على جودة النبيذ؟**


```python
linreg_coef = pd.DataFrame(
    {"coef": linreg.coef_, "coef_abs": np.abs(linreg.coef_)},
    index=data.columns.drop("quality"),
)
linreg_coef.sort_values(by="coef_abs", ascending=False)
```


## انحدار لاسو



** تدريب نموذج LASSO باستخدام $\alpha = 0.01$ (تنظيم ضعيف) وبيانات متدرجة. مرة أخرى، قم بتعيين Random_state=17.**


```python
lasso1 = Lasso(alpha=0.01, random_state=17)
lasso1.fit(X_train_scaled, y_train)
```


**ما هي الميزة الأقل إفادة في التنبؤ بجودة النبيذ، وفقًا لنموذج LASSO هذا؟**

```python
lasso1_coef = pd.DataFrame(
    {"coef": lasso1.coef_, "coef_abs": np.abs(lasso1.coef_)},
    index=data.columns.drop("quality"),
)
lasso1_coef.sort_values(by="coef_abs", ascending=False)
```


**قم بتدريب LassoCV باستخدام Random_state=17 لاختيار أفضل قيمة لـ $\alpha$ في التحقق المتبادل بخمسة أضعاف.**


```python
alphas = np.logspace(-6, 2, 200)
lasso_cv = LassoCV(random_state=17, cv=5, alphas=alphas)
lasso_cv.fit(X_train_scaled, y_train)
```

```python
lasso_cv.alpha_
```


**<font color='red'>السؤال 3:</font> ما الميزة الأقل إفادة في التنبؤ بجودة النبيذ، وفقًا لنموذج LASSO المضبوط؟**


```python
lasso_cv_coef = pd.DataFrame(
    {"coef": lasso_cv.coef_, "coef_abs": np.abs(lasso_cv.coef_)},
    index=data.columns.drop("quality"),
)
lasso_cv_coef.sort_values(by="coef_abs", ascending=False)
```


**<font color='red'>السؤال 4:</font> ما المقصود بالأخطاء المربعة المتوسطة لتنبؤات LASSO المضبوطة في مجموعات القطارات والحواجز؟**


```python
print(
    "Mean squared error (train): %.3f"
    % mean_squared_error(y_train, lasso_cv.predict(X_train_scaled))
)
print(
    "Mean squared error (test): %.3f"
    % mean_squared_error(y_holdout, lasso_cv.predict(X_holdout_scaled))
)
```


## غابة عشوائية



**قم بتدريب مجموعة عشوائية باستخدام معلمات مبتكرة، مع تعيين Random_state فقط على 17.**


```python
forest = RandomForestRegressor(random_state=17)
forest.fit(X_train_scaled, y_train)
```


**<font color='red'>السؤال 5:</font> ما هي الأخطاء التربيعية المتوسطة لنموذج RF في مجموعة التدريب، في التحقق المتبادل (cross_val_score مع التسجيل = 'neg_mean_squared_error' والوسائط الأخرى المتبقية مع القيم الافتراضية) وفي مجموعة الإيقاف؟**


```python
print(
    "Mean squared error (train): %.3f"
    % mean_squared_error(y_train, forest.predict(X_train_scaled))
)
print(
    "Mean squared error (cv): %.3f"
    % np.mean(
        np.abs(
            cross_val_score(
                forest, X_train_scaled, y_train, scoring="neg_mean_squared_error"
            )
        )
    )
)
print(
    "Mean squared error (test): %.3f"
    % mean_squared_error(y_holdout, forest.predict(X_holdout_scaled))
)
```


** قم بضبط المعلمات الفائقة `max_features` و`max_depth` باستخدام GridSearchCV وتحقق مرة أخرى من متوسط التحقق من الصحة المتبادل MSE وMSE في مجموعة الإيقاف.**


```python
forest_params = {"max_depth": list(range(10, 25)), "max_features": list(range(6, 12))}

locally_best_forest = GridSearchCV(
    RandomForestRegressor(n_jobs=-1, random_state=17),
    forest_params,
    scoring="neg_mean_squared_error",
    n_jobs=-1,
    cv=5,
    verbose=True,
)
locally_best_forest.fit(X_train_scaled, y_train)
```

```python
locally_best_forest.best_params_, locally_best_forest.best_score_
```


**<font color='red'>السؤال 6:</font> ما هي الأخطاء التربيعية المتوسطة لنموذج التردد اللاسلكي المضبوط في التحقق من الصحة (cross_val_score مع التسجيل='neg_mean_squared_error' والوسائط الأخرى المتبقية مع القيم الافتراضية) وفي مجموعة الإيقاف؟**


```python
print(
    "Mean squared error (cv): %.3f"
    % np.mean(
        np.abs(
            cross_val_score(
                locally_best_forest.best_estimator_,
                X_train_scaled,
                y_train,
                scoring="neg_mean_squared_error",
            )
        )
    )
)
print(
    "Mean squared error (test): %.3f"
    % mean_squared_error(y_holdout, locally_best_forest.predict(X_holdout_scaled))
)
```


** أهمية ميزة إخراج الترددات اللاسلكية. مرة أخرى، من الجيد تقديمه كإطار بيانات.**<br>
**<font color='red'>السؤال 7:</font> ما هي الميزة الأكثر أهمية وفقًا لنموذج Random Forest؟**


```python
rf_importance = pd.DataFrame(
    locally_best_forest.best_estimator_.feature_importances_,
    columns=["coef"],
    index=data.columns[:-1],
)
rf_importance.sort_values(by="coef", ascending=False)
```
