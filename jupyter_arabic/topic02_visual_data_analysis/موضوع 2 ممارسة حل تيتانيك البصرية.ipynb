```python
import numpy as np
import pandas as pd
import seaborn as sns

sns.set()
import matplotlib.pyplot as plt

# Graphics in SVG format are more sharp and legible
%config InlineBackend.figure_format = 'svg'
```


**قراءة البيانات**


```python
train_df = pd.read_csv("../../data/titanic_train.csv", index_col="PassengerId")
```

```python
train_df.head(2)
```

```python
train_df.describe(include="all")
```

```python
train_df.info()
```


**دعنا نسقط`Cabin`، ثم – جميع الصفوف ذات القيم المفقودة.**


```python
train_df = train_df.drop("Cabin", axis=1).dropna()
```

```python
train_df.shape
```


**1. أنشئ صورة لتصور جميع المخططات المبعثرة لكل زوج من الميزات `Age` و`Fare` و`SibSp` و`Parch` و`Survived`. (`scatter_matrix ` من Pandas أو `pairplot` من Seaborn)**


```python
sns.pairplot(train_df[["Survived", "Age", "Fare", "SibSp", "Parch"]]);
```


**2. كيف يعتمد سعر التذكرة (`Fare`) على `Pclass`؟ أنشئ مخططًا مربعًا.**


```python
sns.boxplot(x="Pclass", y="Fare", data=train_df);
```


**3. لنقم ببناء نفس المخطط ولكن مع تقييد قيم `Fare` لتكون أقل من 95% من المتجه الأولي (لإسقاط القيم المتطرفة التي تجعل المخطط أقل وضوحًا).**


```python
sns.boxplot(
    x="Pclass",
    y="Fare",
    data=train_df[train_df["Fare"] < train_df["Fare"].quantile(0.95)],
);
```


**4. كيف تعتمد نسبة الركاب الناجين على جنس الركاب؟ قم بتصويره باستخدام `Seaborn.countplot` باستخدام الوسيطة `hue`.**


```python
pd.crosstab(train_df["Sex"], train_df["Survived"])
```

```python
sns.countplot(x="Sex", hue="Survived", data=train_df);
```


لنفعل الشيء نفسه مع `Survived` و`Pclass`


```python
sns.countplot(x="Pclass", hue="Survived", data=train_df);
```


**5. كيف يختلف توزيع أسعار التذاكر لمن نجوا ومن لم ينجوا. قم بتصويره باستخدام `Seaborn.boxplot`**



الحقيقة المحزنة هي أن أولئك الذين نجوا، دفعوا عادةً مبلغًا أكبر بكثير مقابل تذاكرهم.


```python
sns.boxplot(x="Survived", y="Fare", data=train_df[train_df["Fare"] < 500]);
```

**6. كيف يعتمد البقاء على قيد الحياة على عمر الركاب؟  تحقق (رسميًا) من الافتراض القائل بأن الشباب (< 30 عامًا) نجوا بشكل متكرر أكثر من كبار السن (> 55 عامًا).**



دعونا أولا نبني هذا boxplot


```python
sns.boxplot(x="Survived", y="Age", data=train_df);
```


يمكننا تقسيم المزيد حسب الطبقات.


```python
sns.boxplot(x="Survived", hue="Pclass", y="Age", data=train_df);
```


حسنًا .. من الصعب استنتاج أي شيء. دعونا نفعل ذلك بطريقة أخرى.


```python
train_df["age_cat"] = train_df["Age"].apply(
    lambda age: 1 if age < 30 else 3 if age > 55 else 2
);
```

```python
pd.crosstab(train_df["age_cat"], train_df["Survived"])
```

```python
sns.countplot(x="age_cat", hue="Survived", data=train_df);
```


كان من الممكن أن نخمن أن نسبة الركاب الباقين على قيد الحياة أقل بين كبار السن. ومع ذلك، هناك عدد قليل جدا من الملاحظات لاستنتاجات جدية.
