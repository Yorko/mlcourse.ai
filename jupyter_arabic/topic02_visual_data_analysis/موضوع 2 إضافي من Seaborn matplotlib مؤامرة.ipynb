## 1. مجموعة البيانات
أولاً، سنقوم بتهيئة بيئتنا من خلال استيراد جميع المكتبات الضرورية. سنقوم أيضًا بتغيير إعدادات العرض لعرض المخططات بشكل أفضل.


```python
# Matplotlib forms basis for visualization in Python
import matplotlib.pyplot as plt
# We will use the Seaborn library
import seaborn as sns

sns.set()

# Graphics in SVG format are more sharp and legible
%config InlineBackend.figure_format = 'svg'

# Increase the default plot size and set the color scheme
plt.rcParams["figure.figsize"] = (8, 5)
plt.rcParams["image.cmap"] = "viridis"
import pandas as pd
```


الآن، لنقم بتحميل مجموعة البيانات التي سنستخدمها في `DataFrame`. لقد اخترت مجموعة بيانات حول مبيعات وتقييمات ألعاب الفيديو من [Kaggle Datasets](https://www.kaggle.com/rush4ratio/video-game-sales-with-ratings).
بعض الألعاب في مجموعة البيانات هذه تفتقر إلى التقييمات؛ لذلك، دعونا نقوم بالتصفية فقط لتلك الأمثلة التي تحتوي على جميع قيمها.


```python
df = pd.read_csv("../../data/video_games_sales.csv").dropna()
print(df.shape)
```


بعد ذلك، قم بطباعة ملخص `DataFrame` للتحقق من أنواع البيانات والتحقق من أن كل شيء غير فارغ.


```python
df.info()
```


نرى أن `pandas` قام بتحميل بعض الميزات الرقمية كنوع `object`. سنقوم بشكل صريح بتحويل هذه الأعمدة إلى `float` و`int`.


```python
df["User_Score"] = df["User_Score"].astype("float64")
df["Year_of_Release"] = df["Year_of_Release"].astype("int64")
df["User_Count"] = df["User_Count"].astype("int64")
df["Critic_Count"] = df["Critic_Count"].astype("int64")
```

يحتوي `DataFrame` الناتج على 6825 مثالًا و16 عمودًا. دعونا نلقي نظرة على الإدخالات القليلة الأولى باستخدام طريقة `head()` للتأكد من أن كل شيء قد تم تحليله بشكل صحيح. لجعل الأمر أكثر ملاءمة، قمت بإدراج المتغيرات التي سنستخدمها في هذا الدفتر فقط.


```python
useful_cols = [
    "Name",
    "Platform",
    "Year_of_Release",
    "Genre",
    "Global_Sales",
    "Critic_Score",
    "Critic_Count",
    "User_Score",
    "User_Count",
    "Rating",
]
df[useful_cols].head()
```


## 2. DataFrame.plot()
قبل أن ننتقل إلى Seaborn وPlotly، دعونا نناقش الطريقة الأبسط والأكثر ملائمة لتصور البيانات من `DataFrame`: باستخدام طريقة `plot()` الخاصة بها.
على سبيل المثال، سنقوم بإنشاء مخطط لمبيعات ألعاب الفيديو حسب البلد والسنة. أولاً، دعونا نحتفظ فقط بالأعمدة التي نحتاجها. بعد ذلك، سنحسب إجمالي المبيعات حسب السنة ونستدعي طريقة `plot()` على النتيجة `DataFrame`.


```python
df[[x for x in df.columns if "Sales" in x] + ["Year_of_Release"]].groupby(
    "Year_of_Release"
).sum().plot();
```


لاحظ أن تطبيق الأسلوب `plot()` في `pandas` يعتمد على `matplotlib`.



باستخدام المعلمة `kind`، يمكنك تغيير نوع قطعة الأرض إلى *مخطط شريطي* على سبيل المثال. يعد `matplotlib` مرنًا بشكل عام لتخصيص المخططات. يمكنك تغيير كل شيء تقريبًا في المخطط، ولكن قد تحتاج إلى البحث في [الوثائق](https://pandas.pydata.org/pandas-docs/stable/generated/pandas.DataFrame.plot.html) للعثور على المعلمات المقابلة. على سبيل المثال، المعلمة `rot` مسؤولة عن زاوية دوران علامات التجزئة على المحور x (للمخططات الرأسية):


```python
df[[x for x in df.columns if "Sales" in x] + ["Year_of_Release"]].groupby(
    "Year_of_Release"
).sum().plot(kind="bar", rot=45);
```


## 3. سيبورنالآن، دعنا ننتقل إلى مكتبة `Seaborn`. `seaborn` هي في الأساس واجهة برمجة تطبيقات عالية المستوى تعتمد على مكتبة `matplotlib`. من بين أمور أخرى، فهو يختلف عن الأخير لأنه يحتوي على إعدادات افتراضية أكثر ملاءمة للتخطيط. من خلال إضافة `import seaborn as sns; sns.set()` في التعليمات البرمجية الخاصة بك، ستصبح صور قطع الأراضي الخاصة بك أجمل بكثير. تحتوي هذه المكتبة أيضًا على مجموعة من الأدوات المعقدة للتصور والتي بخلاف ذلك (أي عند استخدام `matplotlib`) تتطلب قدرًا كبيرًا من التعليمات البرمجية.
#### زوج المؤامرة ()
دعونا نلقي نظرة على أول هذه المخططات المعقدة، *مخطط العلاقات الزوجية*، والذي يقوم بإنشاء مصفوفة من المخططات المبعثرة بشكل افتراضي. يساعدنا هذا النوع من الحبكة على تصور العلاقة بين المتغيرات المختلفة في مخرجات واحدة.


```python
# `pairplot()` may become very slow with the SVG format
%config InlineBackend.figure_format = 'png'
sns.pairplot(
    df[["Global_Sales", "Critic_Score", "Critic_Count", "User_Score", "User_Count"]]
);
```


كما ترون، تقع الرسوم البيانية للتوزيع على قطري المصفوفة. المخططات المتبقية عبارة عن مخططات مبعثرة لأزواج المعالم المقابلة.
#### ديستبلوت ()
من الممكن أيضًا رسم توزيع الملاحظات باستخدام `seaborn`'s `distplot()`. على سبيل المثال، دعونا نلقي نظرة على توزيع تقييمات النقاد: `Critic_Score`. افتراضيًا، يعرض المخطط رسمًا بيانيًا و[تقدير كثافة النواة](https://en.wikipedia.org/wiki/Kernel_density_estimation).


```python
%config InlineBackend.figure_format = 'svg'
sns.distplot(df["Critic_Score"]);
```


#### مؤامرة مشتركة ()
لإلقاء نظرة عن كثب على العلاقة بين متغيرين رقميين، يمكنك استخدام *مخطط مشترك*، وهو عبارة عن تقاطع بين المخطط المبعثر والرسم البياني. دعونا نرى كيف ترتبط الميزات `Critic_Score` و`User_Score`.


```python
sns.jointplot(x="Critic_Score", y="User_Score", data=df, kind="scatter");
```


#### بوكسبلوت()
نوع آخر مفيد من الحبكة هو *حبكة مربعة*. دعونا نقارن تقييمات النقاد لأفضل 5 منصات ألعاب.


```python
top_platforms = (
    df["Platform"].value_counts().sort_values(ascending=False).head(5).index.values
)
sns.boxplot(
    y="Platform",
    x="Critic_Score",
    data=df[df["Platform"].isin(top_platforms)],
    orient="h",
);
```

من المفيد قضاء المزيد من الوقت لمناقشة كيفية تفسير مخطط الصندوق. مكوناته هي *مربع* (من الواضح أن هذا هو سبب تسميته *مؤامرة مربعة*)، وما يسمى *الشعيرات*، وعدد من النقاط الفردية (*القيم المتطرفة*).
ويوضح المربع في حد ذاته الانتشار الربعي للتوزيع؛ يتم تحديد طوله بواسطة النسب المئوية $25\% \, (\text{Q1})$ و$75\% \, (\text{Q3})$. يشير الخط العمودي الموجود داخل المربع إلى متوسط ​​التوزيع ($50\%$). 
الشوارب هي الخطوط الممتدة من الصندوق. إنها تمثل نقاط البيانات المتناثرة بالكامل، وتحديدًا النقاط التي تقع ضمن الفاصل الزمني $(\text{Q1} - 1.5 \cdot \text{IQR}, \text{Q3} + 1.5 \cdot \text{IQR})$، حيث $\text{IQR} = \text{Q3} - \text{Q1}$ هو [النطاق الرباعي](https://en.wikipedia.org/wiki/Interquartile_range).
يتم رسم القيم المتطرفة التي تقع خارج النطاق المحدود بالشعيرات بشكل فردي.
#### خريطة الحرارة ()
النوع الأخير من المخططات الذي سنغطيه هنا هو *الخريطة الحرارية*. تتيح لك الخريطة الحرارية عرض توزيع متغير رقمي على متغيرين فئويين. دعونا نتصور إجمالي مبيعات الألعاب حسب النوع ومنصة الألعاب.


```python
platform_genre_sales = (
    df.pivot_table(
        index="Platform", columns="Genre", values="Global_Sales", aggfunc=sum
    )
    .fillna(0)
    .applymap(float)
)
sns.heatmap(platform_genre_sales, annot=True, fmt=".1f", linewidths=0.5);
```


## 4. مؤامرة
لقد قمنا بفحص بعض أدوات التصور بناءً على مكتبة `matplotlib`. ومع ذلك، ليس هذا هو الخيار الوحيد للتخطيط في `Python`. دعونا نلقي نظرة على مكتبة `plotly`. Plotly هي مكتبة مفتوحة المصدر تسمح بإنشاء مخططات تفاعلية داخل دفتر ملاحظات Jupyter دون الحاجة إلى استخدام Javascript.
الجمال الحقيقي للمخططات التفاعلية هو أنها توفر واجهة مستخدم لاستكشاف البيانات التفصيلية. على سبيل المثال، يمكنك رؤية القيم الرقمية الدقيقة عن طريق تمرير الماوس فوق النقاط، وإخفاء سلسلة غير مثيرة للاهتمام من التصور، وتكبير جزء معين من المؤامرة، وما إلى ذلك.قبل أن نبدأ، دعونا نستورد جميع الوحدات الضرورية ونقوم بتهيئة `plotly` عن طريق استدعاء الدالة `init_notebook_mode()`.


```python
import plotly
import plotly.graph_objs as go
from plotly.offline import download_plotlyjs, init_notebook_mode, iplot, plot

init_notebook_mode(connected=True)
```


#### مؤامرة الخط
أولاً، دعونا نبني *مخططًا خطيًا* يوضح عدد الألعاب التي تم إصدارها ومبيعاتها حسب العام.


```python
years_df = (
    df.groupby("Year_of_Release")[["Global_Sales"]]
    .sum()
    .join(df.groupby("Year_of_Release")[["Name"]].count())
)
years_df.columns = ["Global_Sales", "Number_of_Games"]
```


`Figure` هو الفصل الرئيسي وحصان العمل في `plotly`. وهو يتألف من البيانات (صفيف من الخطوط تسمى `traces` في هذه المكتبة) والنمط (الذي يمثله الكائن `layout`). في أبسط الحالات، يمكنك استدعاء الدالة `iplot` لإرجاع `traces` فقط.
تقوم المعلمة `show_link` بتبديل رؤية الروابط المؤدية إلى النظام الأساسي عبر الإنترنت `plot.ly` في مخططاتك. في أغلب الأحيان، لا تكون هذه الوظيفة مطلوبة، لذا قد ترغب في إيقاف تشغيلها عن طريق تمرير `show_link=False` لمنع النقرات غير المقصودة على تلك الروابط.


```python
# Create a line (trace) for the global sales
trace0 = go.Scatter(x=years_df.index, y=years_df["Global_Sales"], name="Global Sales")

# Create a line (trace) for the number of games released
trace1 = go.Scatter(
    x=years_df.index, y=years_df["Number_of_Games"], name="Number of games released"
)

# Define the data array
data = [trace0, trace1]

# Set the title
layout = {"title": "Statistics for video games"}

# Create a Figure and plot it
fig = go.Figure(data=data, layout=layout)
iplot(fig, show_link=False)
```


كخيار، يمكنك حفظ المؤامرة في ملف HTML:


```python
plotly.offline.plot(fig, filename="years_stats.html", show_link=False);
```


#### الرسم البياني الشريطي
دعونا نستخدم *مخططًا شريطيًا* لمقارنة الحصة السوقية لمنصات الألعاب المختلفة مقسمة حسب عدد الإصدارات الجديدة وإجمالي الإيرادات.


```python
# Do calculations and prepare the dataset
platforms_df = (
    df.groupby("Platform")[["Global_Sales"]]
    .sum()
    .join(df.groupby("Platform")[["Name"]].count())
)
platforms_df.columns = ["Global_Sales", "Number_of_Games"]
platforms_df.sort_values("Global_Sales", ascending=False, inplace=True)
```

```python
# Create a bar for the global sales
trace0 = go.Bar(
    x=platforms_df.index, y=platforms_df["Global_Sales"], name="Global Sales"
)

# Create a bar for the number of games released
trace1 = go.Bar(
    x=platforms_df.index,
    y=platforms_df["Number_of_Games"],
    name="Number of games released",
)

# Get together the data and style objects
data = [trace0, trace1]
layout = {"title": "Market share by gaming platform"}

# Create a `Figure` and plot it
fig = go.Figure(data=data, layout=layout)
iplot(fig, show_link=False)
```


#### مؤامرة مربعة
`plotly` يدعم أيضًا *مخططات الصندوق*. دعونا نفكر في توزيع تقييمات النقاد حسب نوع اللعبة.


```python
data = []

# Create a box trace for each genre in our dataset
for genre in df.Genre.unique():
    data.append(go.Box(y=df[df.Genre == genre].Critic_Score, name=genre))

# Visualize
iplot(data, show_link=False)
```


باستخدام `plotly`، يمكنك أيضًا إنشاء أنواع أخرى من التصورات. حتى مع الإعدادات الافتراضية، تبدو المؤامرات جميلة جدًا. بالإضافة إلى ذلك، تسهل المكتبة تعديل المعلمات المختلفة: الألوان، الخطوط، التسميات التوضيحية، التعليقات التوضيحية، وما إلى ذلك.
