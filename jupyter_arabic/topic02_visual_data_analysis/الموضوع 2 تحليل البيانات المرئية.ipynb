## 2. التصور أحادي المتغير
ينظر التحليل *أحادي المتغير* إلى ميزة واحدة في كل مرة. عندما نقوم بتحليل ميزة بشكل مستقل، فإننا عادةً ما نهتم في الغالب بـ *توزيع قيمها* ونتجاهل الميزات الأخرى في مجموعة البيانات.
أدناه، سننظر في أنواع إحصائية مختلفة من الميزات والأدوات المقابلة لتحليلها البصري الفردي.
#### 2.1 الميزات الكمية
*الميزات الكمية* تأخذ قيمًا عددية مرتبة. يمكن أن تكون هذه القيم *منفصلة*، مثل الأعداد الصحيحة، أو *مستمرة*، مثل الأعداد الحقيقية، وعادة ما تعبر عن عدد أو قياس.
##### الرسوم البيانية ومخططات الكثافة
أسهل طريقة لإلقاء نظرة على توزيع متغير رقمي هي رسم *الرسم البياني* الخاص به باستخدام طريقة `DataFrame` [`hist()`](https://pandas.pydata.org/pandas-docs/stable/generated/pandas.DataFrame.hist.html).


```python
features = ["Total day minutes", "Total intl calls"]
df[features].hist(figsize=(10, 4));
```


يقوم الرسم البياني بتجميع القيم في *صناديق* ذات نطاق قيمة متساوٍ. قد يحتوي شكل الرسم البياني على أدلة حول نوع التوزيع الأساسي: غاوسي، أسي، وما إلى ذلك. يمكنك أيضًا اكتشاف أي انحراف في شكله عندما يكون التوزيع منتظمًا تقريبًا ولكن به بعض الحالات الشاذة. تصبح معرفة توزيع قيم الميزات أمرًا مهمًا عند استخدام أساليب التعلم الآلي التي تفترض نوعًا معينًا (غالبًا غاوسي).
في المخطط أعلاه، نرى أن المتغير *إجمالي دقائق اليوم* يتم توزيعه بشكل طبيعي، في حين أن *إجمالي المكالمات الدولية* منحرف بشكل بارز إلى اليمين (ذيله أطول على اليمين).هناك أيضًا طريقة أخرى، غالبًا ما تكون أكثر وضوحًا، لفهم التوزيع: *مخططات الكثافة* أو، بشكل أكثر رسمية، *مخططات كثافة النواة*. يمكن اعتبارها نسخة [مُسهلة](https://en.wikipedia.org/wiki/Kernel_smoother) من الرسم البياني. ميزتها الرئيسية على الأخيرة هي أنها لا تعتمد على حجم الصناديق. لنقم بإنشاء مخططات كثافة لنفس المتغيرين:


```python
df[features].plot(
    kind="density", subplots=True, layout=(1, 2), sharex=False, figsize=(10, 4)
);
```


من الممكن أيضًا رسم توزيع الملاحظات باستخدام `seaborn`'s [`distplot()`](https://seaborn.pydata.org/generated/seaborn.distplot.html). على سبيل المثال، دعونا نلقي نظرة على توزيع *إجمالي دقائق اليوم*. افتراضيًا، يعرض المخطط الرسم البياني مع [تقدير كثافة النواة](https://en.wikipedia.org/wiki/Kernel_density_estimation) (KDE) في الأعلى.


```python
sns.distplot(df["Total intl calls"]);
```


يتم قياس ارتفاع أشرطة الرسم البياني هنا ويظهر الكثافة بدلاً من عدد الأمثلة في كل صندوق.
##### مؤامرة مربعة
نوع آخر مفيد من التصور هو *مؤامرة مربعة*. `seaborn` يقوم بعمل رائع هنا:


```python
sns.boxplot(x="Total intl calls", data=df);
```


دعونا نرى كيفية تفسير مؤامرة مربع. مكوناته هي *مربع* (من الواضح أن هذا هو سبب تسميته *مؤامرة مربعة*)، وما يسمى *الشعيرات*، وعدد من النقاط الفردية (*القيم المتطرفة*).
ويوضح المربع في حد ذاته الانتشار الربعي للتوزيع؛ يتم تحديد طوله بواسطة النسب المئوية $25th \, (\text{Q1})$ و$75th \, (\text{Q3})$. يمثل الخط العمودي داخل المربع الوسيط ($50\%$) للتوزيع. 
الشوارب هي الخطوط الممتدة من الصندوق. وهي تمثل نقاط البيانات المتناثرة بالكامل، وتحديدًا النقاط التي تقع ضمن الفاصل الزمني $(\text{Q1} - 1.5 \cdot \text{IQR}, \text{Q3} + 1.5 \cdot \text{IQR})$، حيث $\text{IQR} = \text{Q3} - \text{Q1}$ هو [النطاق الرباعي](https://en.wikipedia.org/wiki/Interquartile_range).
يتم رسم القيم المتطرفة التي تقع خارج النطاق الذي تحده الشعيرات بشكل فردي كنقاط سوداء على طول المحور المركزي.
يمكننا أن نرى أن عددًا كبيرًا من المكالمات الدولية نادر جدًا في بياناتنا.##### مؤامرة الكمان
النوع الأخير من قطع التوزيع التي سنأخذها بعين الاعتبار هو *مخطط الكمان*.
انظر إلى الأرقام أدناه. على اليسار، نرى مؤامرة مربعة مألوفة بالفعل. إلى اليمين، هناك *مؤامرة كمان* مع تقدير كثافة النواة على كلا الجانبين.


```python
_, axes = plt.subplots(1, 2, sharey=True, figsize=(6, 4))
sns.boxplot(data=df["Total intl calls"], ax=axes[0])
sns.violinplot(data=df["Total intl calls"], ax=axes[1]);
```


الفرق بين مخططات الصندوق والكمان هو أن الأول يوضح إحصائيات معينة تتعلق بأمثلة فردية في مجموعة بيانات بينما تركز مؤامرة الكمان أكثر على التوزيع السلس ككل.
في حالتنا، لا يساهم مخطط الكمان بأي معلومات إضافية حول البيانات حيث أن كل شيء واضح من المخطط الصندوقي وحده.
##### وصف ()
بالإضافة إلى الأدوات الرسومية، من أجل الحصول على الإحصائيات الرقمية الدقيقة للتوزيع، يمكننا استخدام الطريقة [`describe()`](https://pandas.pydata.org/pandas-docs/stable/generated/pandas.DataFrame.describe.html) لـ `DataFrame`:


```python
df[features].describe()
```


مخرجاتها في الغالب لا تحتاج إلى شرح. *25%* و*50%* و*75%* هي [النسب المئوية] المقابلة(https://en.wikipedia.org/wiki/Percentile).



#### 2.2 الميزات الفئوية والثنائية
*الميزات الفئوية* تأخذ عددًا ثابتًا من القيم. تقوم كل واحدة من هذه القيم بتعيين ملاحظة لمجموعة مقابلة، تُعرف باسم *الفئة*، والتي تعكس بعض الخصائص النوعية لهذا المثال. تعد المتغيرات *الثنائية* حالة خاصة مهمة من المتغيرات الفئوية عندما يكون عدد القيم المحتملة هو 2 بالضبط. إذا تم ترتيب قيم المتغير الفئوي، يطلق عليه *ترتيبي*.
##### جدول التكرار
دعونا نتحقق من توازن الفصل في مجموعة البيانات الخاصة بنا من خلال النظر في توزيع المتغير المستهدف: *معدل التراجع*. أولاً، سوف نحصل على جدول تكراري يوضح مدى تكرار كل قيمة للمتغير الفئوي. لهذا، سوف نستخدم الطريقة [`value_counts()`](https://pandas.pydata.org/pandas-docs/stable/generated/pandas.Series.value_counts.html):


```python
df["Churn"].value_counts()
```

افتراضيًا، يتم فرز الإدخالات في المخرجات من القيم الأكثر تكرارًا إلى الأقل تكرارًا.
في حالتنا، البيانات ليست *متوازنة*؛ أي أن الفئتين المستهدفتين لدينا، العملاء المخلصين وغير المخلصين، لا يتم تمثيلهم بالتساوي في مجموعة البيانات. قام جزء صغير فقط من العملاء بإلغاء اشتراكهم في خدمة الاتصالات. كما سنرى في المقالات التالية، قد تتضمن هذه الحقيقة بعض القيود على قياس أداء التصنيف، وفي المستقبل، قد نرغب في معاقبة أخطاء نموذجنا بشكل إضافي في التنبؤ بفئة الأقلية "Churn".



##### مؤامرة شريطية
الرسم البياني الشريطي هو تمثيل رسومي للجدول التكراري. أسهل طريقة لإنشائه هي استخدام وظيفة `seaborn` [`countplot()`](https://seaborn.pydata.org/generated/seaborn.countplot.html). هناك وظيفة أخرى في `seaborn` تسمى بشكل مربك إلى حد ما [`barplot()`](https://seaborn.pydata.org/generated/seaborn.barplot.html) وتستخدم في الغالب لتمثيل بعض الإحصائيات الأساسية لمتغير رقمي مجمع حسب ميزة فئوية.
دعونا نرسم التوزيعات لمتغيرين فئويين:


```python
_, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 4))

sns.countplot(x="Churn", data=df, ax=axes[0])
sns.countplot(x="Customer service calls", data=df, ax=axes[1]);
```

في حين أن الرسوم البيانية، التي تمت مناقشتها أعلاه، والمؤامرات الشريطية قد تبدو متشابهة، إلا أن هناك العديد من الاختلافات بينهما:
1. *الرسوم البيانية* هي الأنسب للنظر في توزيع المتغيرات الرقمية بينما يتم استخدام *المخططات الشريطية* للميزات الفئوية.
2. القيم الموجودة على المحور السيني في *الرسم البياني* هي قيم رقمية؛ يمكن أن تحتوي *المؤامرة الشريطية* على أي نوع من القيم على المحور السيني: الأرقام والسلاسل والقيم المنطقية.
3. المحور السيني *للرسم البياني* هو *محور إحداثي ديكارتي* لا يمكن تغيير القيم من خلاله؛ ترتيب *الأشرطة* غير محدد مسبقًا. ومع ذلك، من المفيد ملاحظة أن الأشرطة غالبًا ما يتم فرزها حسب الارتفاع، أي تكرار القيم. وأيضًا، عندما نأخذ في الاعتبار المتغيرات *الترتيبية* (مثل *مكالمات خدمة العملاء* في بياناتنا)، يتم ترتيب الأشرطة عادةً حسب قيمة المتغير.
يوضح الرسم البياني الأيسر أعلاه بوضوح عدم التوازن في المتغير المستهدف. يعطي المخطط الشريطي *مكالمات خدمة العملاء* الموجود على اليمين تلميحًا إلى أن غالبية العملاء يحلون مشاكلهم من خلال 2-3 مكالمات كحد أقصى. ولكن بما أننا نريد أن نكون قادرين على التنبؤ بطبقة الأقلية، فقد نكون أكثر اهتماماً بكيفية تصرف العملاء الأقل عدداً غير الراضين. من الممكن أن يكون ذيل تلك القطعة الشريطية يحتوي على معظم زبدنا. هذه مجرد فرضيات في الوقت الحالي، لذلك دعونا ننتقل إلى بعض التقنيات البصرية الأكثر إثارة للاهتمام والقوة.



## 3. التصور متعدد المتغيرات

## 4. تصورات مجموعة البيانات الكاملة



#### 4.1 نهج ساذج
لقد كنا ننظر إلى *جوانب* مختلفة من مجموعة البيانات الخاصة بنا عن طريق تخمين الميزات المثيرة للاهتمام واختيار عدد صغير منها في كل مرة للتصور. لقد تعاملنا فقط مع متغيرين أو ثلاثة متغيرات في وقت واحد وتمكنا بسهولة من ملاحظة البنية والعلاقات في البيانات. ولكن، ماذا لو أردنا عرض جميع الميزات وما زلنا قادرين على تفسير التصور الناتج؟
يمكننا استخدام `hist()` أو إنشاء مصفوفة مخطط تبعثر باستخدام `pairplot()` لمجموعة البيانات بأكملها للنظر في جميع ميزاتنا في وقت واحد. ولكن عندما يكون عدد الميزات مرتفعًا بدرجة كافية، فإن هذا النوع من التحليل البصري يصبح سريعًا بطيئًا وغير فعال. علاوة على ذلك، سنستمر في تحليل متغيراتنا بطريقة زوجية، وليس كلها مرة واحدة.



#### 4.2 تقليل الأبعادتحتوي معظم مجموعات البيانات الواقعية على العديد من الميزات، وأحيانًا عدة آلاف منها. ويمكن اعتبار كل واحد منهم بمثابة بُعد في مساحة نقاط البيانات. وبالتالي، فإننا في أغلب الأحيان نتعامل مع مجموعات بيانات عالية الأبعاد، حيث يكون التصور بأكمله صعبًا للغاية.
وللنظر إلى مجموعة البيانات ككل، نحتاج إلى تقليل عدد الأبعاد المستخدمة في التصور دون فقدان الكثير من المعلومات حول البيانات. تسمى هذه المهمة *تقليل الأبعاد* وهي مثال لمشكلة *التعلم غير الخاضع للإشراف* لأننا نحتاج إلى استخلاص ميزات جديدة منخفضة الأبعاد من البيانات نفسها، دون أي مدخلات خاضعة للإشراف.
إحدى طرق تقليل الأبعاد المعروفة هي *تحليل المكونات الرئيسية* (PCA)، والتي سندرسها لاحقًا في هذه الدورة. حدودها هي أنها خوارزمية *خطية* تتضمن قيودًا معينة على البيانات.
هناك أيضًا العديد من الأساليب غير الخطية، والتي تسمى مجتمعة *التعلم المتنوع*. ومن أشهرها *t-SNE*.



#### 4.3 طن-SNE
لنقم بإنشاء تمثيل [t-SNE](https://en.wikipedia.org/wiki/T-distributed_stochastic_neighbor_embedding) لنفس بيانات التغيير التي كنا نستخدمها. 
يبدو اسم الطريقة معقدًا ومخيفًا بعض الشيء: *تضمين الجوار العشوائي الموزع على شكل حرف T*. حساباتها مثيرة للإعجاب أيضًا (لن نتعمق فيها هنا، ولكن إذا كنت تشعر بالشجاعة، فإليك [المقالة الأصلية](http://www.jmlr.org/papers/volume9/vandermaaten08a/vandermaaten08a.pdf) بقلم Laurens van der Maaten وGeoffrey Hinton من [JMLR](http://www.jmlr.org/)). فكرتها الأساسية بسيطة: العثور على إسقاط لمساحة مميزة عالية الأبعاد على مستوى (أو طائرة مفرطة ثلاثية الأبعاد، لكنها دائمًا ما تكون ثنائية الأبعاد) بحيث تنتهي تلك النقاط التي كانت متباعدة في الفضاء الأولي ذو الأبعاد n متباعدة على المستوى. أولئك الذين كانوا قريبين في الأصل سيبقون قريبين من بعضهم البعض.بشكل أساسي، *تضمين الجوار* هو بحث عن تمثيل بيانات جديد وأقل أبعادًا يحافظ على جوار الأمثلة.
الآن، دعونا نقوم ببعض التدريبات. أولاً، نحتاج إلى استيراد بعض الفئات الإضافية:


```python
from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler
```


سنترك ميزات *State* و*Churn* ونحول القيم "Yes"/"No" للميزات الثنائية إلى قيم رقمية باستخدام [`pandas.Series.map()`](http://pandas.pydata.org/pandas-docs/stable/generated/pandas.Series.map.html):


```python
X = df.drop(["Churn", "State"], axis=1)
X["International plan"] = X["International plan"].map({"Yes": 1, "No": 0})
X["Voice mail plan"] = X["Voice mail plan"].map({"Yes": 1, "No": 0})
```


نحتاج أيضًا إلى تطبيع البيانات. ولهذا، سنطرح الوسط الحسابي من كل متغير ونقسمه على انحرافه المعياري. كل هذا يمكن القيام به باستخدام `StandardScaler`.


```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```


الآن، دعونا نبني تمثيل t-SNE:


```python
%%time
tsne = TSNE(random_state=17)
tsne_repr = tsne.fit_transform(X_scaled)
```


ورسمها:


```python
plt.scatter(tsne_repr[:, 0], tsne_repr[:, 1], alpha=0.5);
```


دعونا نلون تمثيل t-SNE هذا وفقًا للمخض (الأزرق للعملاء المخلصين، والبرتقالي لأولئك الذين مخضض).


```python
plt.scatter(
    tsne_repr[:, 0],
    tsne_repr[:, 1],
    c=df["Churn"].map({False: "blue", True: "orange"}),
    alpha=0.5,
);
```


يمكننا أن نرى أن العملاء الذين تخلوا عن العمل يتركزون في مناطق قليلة من مساحة الميزة ذات الأبعاد المنخفضة.
لفهم الصورة بشكل أفضل، يمكننا أيضًا تلوينها بالميزات الثنائية المتبقية: *الخطة الدولية* و *البريد الصوتي*. تشير النقاط البرتقالية هنا إلى المثيلات الإيجابية للميزة الثنائية المقابلة.


```python
_, axes = plt.subplots(1, 2, sharey=True, figsize=(12, 5))

for i, name in enumerate(["International plan", "Voice mail plan"]):
    axes[i].scatter(
        tsne_repr[:, 0],
        tsne_repr[:, 1],
        c=df[name].map({"Yes": "orange", "No": "blue"}),
        alpha=0.5,
    )
    axes[i].set_title(name);
```


من الواضح الآن، على سبيل المثال، أن العديد من العملاء غير الراضين الذين ألغوا اشتراكهم يتجمعون معًا في مجموعة واحدة تمثل الأشخاص الذين لديهم الخطة الدولية ولكن ليس لديهم بريد صوتي.وأخيرا، دعونا نلاحظ بعض عيوب t-SNE:
- التعقيد الحسابي العالي. من غير المرجح أن يكون [التنفيذ](http://scikit-learn.org/stable/modules/generated/sklearn.manifold.TSNE.html) في `scikit-learn` ممكنًا في مهمة حقيقية. إذا كان لديك عدد كبير من العينات، فيجب عليك تجربة [Multicore-TSNE](https://github.com/DmitryUlyanov/Multicore-TSNE) بدلاً من ذلك.
- يمكن أن تتغير الحبكة كثيرًا اعتمادًا على البذرة العشوائية، مما يعقد عملية التفسير. [هنا] (http://distill.pub/2016/misread-tsne/) هو برنامج تعليمي جيد حول t-SNE. بشكل عام، لا ينبغي عليك التوصل إلى أي استنتاجات بعيدة المدى بناءً على مثل هذه الرسوم البيانية لأنها يمكن أن تعادل التخمين البسيط. بالطبع، يمكن لبعض النتائج في صور t-SNE أن تلهم فكرة ويتم تأكيدها من خلال بحث أكثر شمولاً، لكن هذا لا يحدث كثيرًا.
في بعض الأحيان، باستخدام t-SNE، يمكنك الحصول على حدس جيد حقًا للبيانات. فيما يلي ورقة بحثية جيدة تعرض مثالاً على ذلك للأرقام المكتوبة بخط اليد: [Visualizing MNIST](https://colah.github.io/posts/2014-10-Visualizing-MNIST/).
