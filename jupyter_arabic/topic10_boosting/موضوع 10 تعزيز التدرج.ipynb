# 1. مقدمة وتاريخ التعزيز
لقد سمع الجميع تقريبًا في مجال التعلم الآلي عن تعزيز التدرج. يقوم العديد من علماء البيانات بتضمين هذه الخوارزمية في صندوق أدوات عالم البيانات الخاص بهم نظرًا للنتائج الجيدة التي تحققها في أي مشكلة معينة (غير معروفة).  
علاوة على ذلك، غالبًا ما يكون XGBoost هو الوصفة القياسية [للفوز](https://github.com/dmlc/xgboost/blob/master/demo/README.md#usecases) [مسابقات تعلم الآلة](http://blog.kaggle.com/tag/xgboost/). لقد أصبحت فكرة تكديس XGBoosts شائعة جدًا لدرجة أنها أصبحت مجرد ميم. علاوة على ذلك، يعد التعزيز عنصرًا مهمًا في [العديد من أنظمة التوصية](https://en.wikipedia.org/wiki/Learning_to_rank#Practical_usage_by_search_engines)؛ وفي بعض الأحيان، يتم اعتبارها [علامة تجارية](https://yandex.com/company/technologies/matrixnet/).
دعونا نلقي نظرة على تاريخ وتطور التعزيز.



نشأ التعزيز من [السؤال:](http://www.cis.upenn.edu/~mkearns/papers/boostnote.pdf) هل من الممكن الحصول على نموذج قوي واحد من عدد كبير من النماذج الضعيفة والبسيطة نسبيًا؟ عندما نقول "النماذج الضعيفة"، فإننا لا نعني النماذج الأساسية البسيطة مثل أشجار القرار، ولكن النماذج ذات الأداء الضعيف في الدقة، حيث يكون الرديء أفضل قليلاً من العشوائي.تم تحديد [إجابة رياضية إيجابية](http://www.cs.princeton.edu/~schapire/papers/strengthofweak.pdf) على هذا السؤال، ولكن الأمر استغرق بضع سنوات لتطوير خوارزميات تعمل بكامل طاقتها بناءً على هذا الحل، على سبيل المثال. adaBoost. تتخذ هذه الخوارزميات نهجًا جشعًا: أولاً، تقوم ببناء مجموعة خطية من النماذج البسيطة (الخوارزميات الأساسية) عن طريق إعادة وزن البيانات المدخلة. بعد ذلك، يتم بناء النموذج (عادة ما يكون شجرة قرارات) على كائنات تم التنبؤ بها بشكل غير صحيح في وقت سابق، والتي يتم إعطاؤها الآن أوزانًا أكبر.  
<spoiler title="More about AdaBoost">
تدرس العديد من دورات التعلم الآلي AdaBoost - سلف GBM (آلة تعزيز التدرج). ومع ذلك، منذ دمج AdaBoost مع GBM، أصبح من الواضح أن AdaBoost هو مجرد نسخة معينة من GBM.  
تتمتع الخوارزمية نفسها بتفسير مرئي واضح جدًا وحدس لتحديد الأوزان. دعونا نلقي نظرة على مشكلة تصنيف الألعاب التالية حيث سنقوم بتقسيم البيانات بين الأشجار ذات العمق 1 (المعروفة أيضًا باسم "جذوع الأشجار") في كل تكرار لـ AdaBoost. بالنسبة للتكرارين الأولين، لدينا الصورة التالية:

يتوافق حجم النقطة مع وزنها الذي تم تخصيصه للتنبؤ غير الصحيح.  وفي كل تكرار، يمكننا أن نرى أن هذه الأوزان تنمو - ولا تستطيع الجذوع التعامل مع هذه المشكلة. ومع ذلك، إذا أجرينا تصويتًا مرجحًا للجذوع، فسنحصل على التصنيفات الصحيحة:

الكود الزائف:
- تهيئة أوزان العينة $\Large w_i^{(0)} = \frac{1}{l}, i = 1, \dots, l$.
- للجميع $t = 1, \dots, T$
    * تدريب قاعدة الخوارزمية $\Large b_t$، دع $\epsilon_t$ يكون خطأً في التدريب.
    * $\Large \alpha_t = \frac{1}{2}ln\frac{1 - \epsilon_t}{\epsilon_t}$.
    * تحديث أوزان العينات: $\Large w_i^{(t)} = w_i^{(t-1)} e^{-\alpha_t y_i b_t(x_i)}, i = 1, \dots, l$.
    * تطبيع أوزان العينة: $\Large w_0^{(t)} = \sum_{j = 1}^l w_j^{(t)}, w_i^{(t)} = \frac{w_i^{(t)}}{w_0^{(t)}}, i = 1, \dots, l$.
- العودة $\sum_t^{T}\alpha_tb_t$[هنا](https://www.youtube.com/watch?v=k4G2VCuOMMg) هو مثال أكثر تفصيلاً لـ AdaBoost حيث، أثناء التكرار، يمكننا أن نرى زيادة الأوزان، خاصة على الحدود بين الفئات.
يعمل AdaBoost بشكل جيد، ولكن [الافتقار](https://www.cs.princeton.edu/courses/archive/spring07/cos424/papers/boosting-survey.pdf) إلى تفسير سبب نجاح الخوارزمية زرع بذور الشك. اعتبرها البعض خوارزمية فائقة، وحل سحري، لكن البعض الآخر كان متشككًا ويعتقد أن AdaBoost كان مجرد مبالغة في التجهيز. 
إن مشكلة التجهيز الزائد كانت موجودة بالفعل، خاصة عندما كانت البيانات تحتوي على قيم متطرفة قوية. ولذلك، في تلك الأنواع من المشاكل، كان AdaBoost غير مستقر. ولحسن الحظ، بدأ عدد قليل من الأساتذة في قسم الإحصاء في جامعة ستانفورد، الذين أنشأوا Lasso وElastic Net وRandom Forest، في البحث عن الخوارزمية. في عام 1999، توصل جيروم فريدمان إلى تعميم تعزيز تطوير الخوارزميات - Gradient Boosting (Machine)، المعروف أيضًا باسم GBM. من خلال هذا العمل، أنشأ فريدمان الأساس الإحصائي للعديد من الخوارزميات التي توفر النهج العام لتعزيز التحسين في الفضاء الوظيفي.
نشأت CART وbootstrap والعديد من الخوارزميات الأخرى من قسم الإحصاء في جامعة ستانفورد. وبذلك يكون القسم قد عزز أسمائهم في الكتب المدرسية المستقبلية. هذه الخوارزميات عملية للغاية، وبعض الأعمال الحديثة لم يتم اعتمادها على نطاق واسع بعد. على سبيل المثال، راجع [glinternet](https://arxiv.org/abs/1308.2719).  
لا تتوفر الكثير من تسجيلات الفيديو لفريدمان. على الرغم من وجود [مقابلة] (https://www.youtube.com/watch?v=8hupHmBVvb0) مثيرة جدًا للاهتمام معه حول إنشاء CART وكيفية حل المشكلات الإحصائية (التي تشبه تحليل البيانات وعلوم البيانات اليوم) منذ أكثر من 40 عامًا.
هناك أيضًا [محاضرة] رائعة (https://www.youtube.com/watch?v=zBk3PK3g-Fc) من Hastie، وهي عبارة عن عرض استرجاعي لتحليل البيانات من أحد مبتكري الأساليب التي نستخدمها كل يوم.بشكل عام، كان هناك انتقال من البحث الهندسي والخوارزمي إلى نهج متكامل لبناء ودراسة الخوارزميات. من منظور رياضي، لا يعد هذا تغييرًا كبيرًا - فنحن لا نزال نضيف (أو نعزز) خوارزميات ضعيفة ونوسع مجموعتنا من خلال تحسينات تدريجية لأجزاء من البيانات حيث كان النموذج غير دقيق. ولكن، هذه المرة، لم يتم بناء النموذج البسيط التالي على الأجسام المعاد وزنها فحسب، بل قام بتحسين تقريبه لتدرج الوظيفة الموضوعية الشاملة. يفتح هذا المفهوم خوارزمياتنا بشكل كبير للخيال والإضافات.

# 2. خوارزمية GBM
### بيان مشكلة تعلم الآلةسنقوم بحل مشكلة تقريب الوظيفة في بيئة تعليمية عامة تحت الإشراف. لدينا مجموعة من الميزات $ \large x $ والمتغيرات المستهدفة $\large y, \large \left\{ (x_i, y_i) \right\}_{i=1, \ldots,n}$ والتي نستخدمها لاستعادة التبعية $\large y = f(x) $. نحن نستعيد التبعية من خلال تقريب $ \large \hat{f}(x) $ ومن خلال فهم التقريب الأفضل عندما نستخدم دالة الخسارة $ \large L(y,f) $، والتي نريد تقليلها إلى الحد الأدنى: $ \large y \approx \hat{f}(x), \large \hat{f}(x) = \underset{f(x)}{\arg\min} \ L(y,f(x)) $.  
<img src='../../img/topic10_help_with_func.png'  align='center'>
في هذه اللحظة، لا نقوم بأي افتراضات فيما يتعلق بنوع الاعتماد $ \large f(x) $، أو نموذج التقريب لدينا $ \large \hat{f}(x) $، أو توزيع المتغير المستهدف $ \large y $. نتوقع فقط أن تكون الوظيفة $ \large L(y,f) $ قابلة للتمييز. نهجنا عام جدًا: نحدد $ \large \hat {f}(x) $ من خلال تقليل الخسارة:  
$$ \large  \hat{f}(x) = \underset{f(x)}{\arg\min} \ \mathbb {E} _{x,y}[L(y,f(x))]  $$
لسوء الحظ، فإن عدد الوظائف $ \large f(x) $ ليس كبيرًا فحسب، بل إن مساحتها الوظيفية ذات أبعاد لا نهائية. ولهذا السبب من المقبول بالنسبة لنا تحديد مساحة البحث ببعض عائلة الوظائف $ \large f(x, \theta), \theta \in \mathbb{R}^d $. يؤدي هذا إلى تبسيط الهدف كثيرًا لأنه لدينا الآن تحسين قابل للحل لقيم المعلمات:
$ \large \hat{f}(x) = f(x, \hat{\theta}),$
$$\large \hat{\theta} = \underset{\theta}{\arg\min} \ \mathbb {E} _{x,y}[L(y,f(x,\theta))] $$



غالبًا لا توجد حلول تحليلية بسيطة للعثور على المعلمات المثالية $ \large \hat{\theta} $، لذلك يتم عادةً تقريب المعلمات بشكل متكرر. للبدء، نكتب دالة الخسارة التجريبية $ \large L_{\theta}(\hat{\theta}) $ التي ستسمح لنا بتقييم معلماتنا باستخدام بياناتنا. بالإضافة إلى ذلك، دعنا نكتب التقريب $ \large \hat{\theta} $ لعدد $ \large M $ كمجموع:  
$ \كبير \hat{\theta} = \sum_{i = 1}^M \hat{\theta_i}, \\
\كبير L_{\theta}(\hat{\theta}) = \sum_{i = 1}^N L(y_i,f(x_i, \hat{\theta}))$بعد ذلك، الشيء الوحيد المتبقي هو العثور على خوارزمية تكرارية مناسبة لتقليل $\large L_{\theta}(\hat{\theta})$. النزول المتدرج هو الخيار الأبسط والأكثر استخدامًا. نحدد التدرج كـ $\large \nabla L_{\theta}(\hat{\theta})$ ونضيف إليه تقييماتنا التكرارية $\large \hat{\theta_i}$ (نظرًا لأننا نقوم بتقليل الخسارة، فإننا نضيف علامة الطرح). خطوتنا الأخيرة هي تهيئة التقريب الأول $\large \hat{\theta_0}$ واختيار عدد التكرارات $\large M$. دعونا نراجع خطوات هذه الخوارزمية غير الفعالة والساذجة لتقريب $\large \hat{\theta}$:
1. حدد التقريب الأولي للمعلمات $\large \hat{\theta} = \hat{\theta_0}$
2. لكل تكرار $\large t = 1, \dots, M$ كرر الخطوات من 3 إلى 7:
1. احسب تدرج دالة الخسارة $\large \nabla L_{\theta}(\hat{\theta})$ للتقريب الحالي $\large \hat{\theta}$
$\large \nabla L_{\theta}(\hat{\theta}) = \left[\frac{\partial L(y, f(x, \theta))}{\partial \theta}\right]_{\theta = \hat{\theta}}$
2. قم بتعيين التقريب التكراري الحالي $\large \hat{\theta_t}$ بناءً على التدرج المحسوب
$\large \hat{\theta_t} \leftarrow −\nabla L_{\theta}(\hat{\theta})$
3. قم بتحديث تقريب المعلمات $\large \hat{\theta}$:
$\large \hat{\theta} \leftarrow \hat{\theta} + \hat{\theta_t} = \sum_{i = 0}^t \hat{\theta_i} $
3. احفظ نتيجة التقريب $\large \hat{\theta}$:
$\large \hat{\theta} = \sum_{i = 0}^M \hat{\theta_i} $
4. استخدم الوظيفة التي تم العثور عليها $\large \hat{f}(x) = f(x, \hat{\theta})$

### نزول متدرج وظيفي
دعونا نتخيل للحظة أنه يمكننا إجراء التحسين في مساحة الوظيفة والبحث بشكل متكرر عن التقريبات $\large \hat{f}(x)$ كوظائف نفسها. سوف نعبر عن تقريبنا كمجموع من التحسينات الإضافية، كل منها عبارة عن وظيفة. من أجل التيسير، سنبدأ على الفور بمجموع التقريب الأولي $\large \hat{f_0}(x)$:
$$\large \hat{f}(x) = \sum_{i = 0}^M \hat{f_i}(x)$$
لم يحدث شيء بعد؛ لقد قررنا فقط أننا سنبحث عن التقريب $\large \hat{f}(x)$ ليس كنموذج كبير به الكثير من المعلمات (على سبيل المثال، الشبكة العصبية)، ولكن كمجموع من الوظائف، متظاهرين أننا نتحرك في الفضاء الوظيفي.من أجل إنجاز هذه المهمة، نحتاج إلى تقييد بحثنا ببعض عائلة الوظائف $\large \hat{f}(x) = h(x, \theta)$. هناك بعض المشكلات هنا - أولاً، يمكن أن يكون مجموع النماذج أكثر تعقيدًا من أي نموذج من هذه العائلة؛ ثانيا، الهدف العام لا يزال في الفضاء الوظيفي. دعونا نلاحظ أنه في كل خطوة، سنحتاج إلى تحديد المعامل الأمثل $\large \rho \in \mathbb{R}$. بالنسبة للخطوة $\large t$، المشكلة هي التالية:
$$\large \hat{f}(x) = \sum_{i = 0}^{t-1} \hat{f_i}(x), \\
\large (\rho_t,\theta_t) = \underset{\rho,\theta}{\arg\min} \ \mathbb {E} _{x,y}[L(y,\hat{f}(x) +  \rho \cdot h(x, \theta))], \\
\large \hat{f_t}(x) = \rho_t \cdot h(x, \theta_t)$$
هنا حيث يحدث السحر. لقد حددنا جميع أهدافنا بعبارات عامة، كما لو كان بإمكاننا تدريب أي نوع من النماذج $\large h(x, \theta)$ لأي نوع من وظائف الخسارة $\large L(y, f(x, \theta))$. من الناحية العملية، هذا أمر صعب للغاية، ولكن لحسن الحظ، هناك طريقة بسيطة لحل هذه المهمة.
من خلال معرفة التعبير عن تدرج دالة الخسارة، يمكننا حساب قيمتها على بياناتنا. لذلك، دعونا ندرب النماذج بحيث تكون توقعاتنا أكثر ارتباطًا بهذا التدرج (مع علامة الطرح). بمعنى آخر، سوف نستخدم المربعات الصغرى لتصحيح التوقعات باستخدام هذه البقايا. بالنسبة لمهام التصنيف والانحدار والتصنيف، سنقوم بتقليل الفرق المربع بين البقايا الزائفة $\large r$ وتوقعاتنا. بالنسبة للخطوة $\large t$، تبدو المشكلة النهائية كما يلي:
$$ \large \hat{f}(x) = \sum_{i = 0}^{t-1} \hat{f_i}(x), \\
\large r_{it} = -\left[\frac{\partial L(y_i, f(x_i))}{\partial f(x_i)}\right]_{f(x)=\hat{f}(x)}, \quad \mbox{for } i=1,\ldots,n ,\\
\large \theta_t = \underset{\theta}{\arg\min} \ \sum_{i = 1}^{n} (r_{it} - h(x_i, \theta))^2, \\
\large \rho_t = \underset{\rho}{\arg\min} \ \sum_{i = 1}^{n} L(y_i, \hat{f}(x_i) + \rho \cdot h(x_i, \theta_t))$$

### خوارزمية GBM الكلاسيكية لفريدمان
يمكننا الآن تعريف خوارزمية GBM الكلاسيكية التي اقترحها جيروم فريدمان في عام 1999. وهي خوارزمية خاضعة للإشراف وتحتوي على المكونات التالية:- مجموعة البيانات $\large \left\{ (x_i, y_i) \right\}_{i=1, \ldots,n}$؛
- عدد التكرارات $\large M$؛
- اختيار وظيفة الخسارة $\large L(y, f)$ مع تدرج محدد؛
- اختيار عائلة الوظائف من الخوارزميات الأساسية $\large h(x, \theta)$ مع إجراء التدريب؛
- معلمات تشعبية إضافية $\large h(x, \theta)$ (على سبيل المثال، في أشجار القرار، عمق الشجرة)؛
الشيء الوحيد المتبقي هو التقريب الأولي $\large f_0(x)$. للتبسيط، ولتقريب أولي، يتم استخدام قيمة ثابتة $\large \gamma$. يتم تحديد القيمة الثابتة، بالإضافة إلى المعامل الأمثل $\large \rho $، عبر البحث الثنائي أو خوارزمية بحث سطرية أخرى عبر دالة الخسارة الأولية (وليس التدرج). لذلك، لدينا خوارزمية GBM الخاصة بنا الموصوفة على النحو التالي:
1. قم بتهيئة GBM بقيمة ثابتة $\large \hat{f}(x) = \hat{f}_0, \hat{f}_0 = \gamma,  \gamma \in \mathbb{R}$
$\large \hat{f}_0 = \underset{\gamma}{\arg\min} \ \sum_{i = 1}^{n} L(y_i, \gamma)$
2. لكل تكرار $\large t = 1, \dots, M$، كرر:
1. احسب البقايا الزائفة $\large r_t$
$\large r_{it} = -\left[\frac{\partial L(y_i, f(x_i))}{\partial f(x_i)}\right]_{f(x)=\hat{f}(x)}, \quad \mbox{for } i=1,\ldots,n$
2. قم ببناء خوارزمية أساسية جديدة $\large h_t(x)$ كانحدار على البقايا الزائفة $\large \left\{ (x_i, r_{it}) \right\}_{i=1, \ldots,n}$
3. ابحث عن المعامل الأمثل $\large \rho_t $ في $\large h_t(x)$ فيما يتعلق بوظيفة الخسارة الأولية
$\large \rho_t = \underset{\rho}{\arg\min} \ \sum_{i = 1}^{n} L(y_i, \hat{f}(x_i) +  \rho \cdot h(x_i, \theta))$
4. احفظ $\large \hat{f_t}(x) = \rho_t \cdot h_t(x)$
5. تحديث التقريب الحالي $\large \hat{f}(x)$
$\large \hat{f}(x) \leftarrow \hat{f}(x) + \hat{f_t}(x) = \sum_{i = 0}^{t} \hat{f_i}(x)$
3. إنشاء نموذج GBM النهائي $\large \hat{f}(x)$
$\large \hat{f}(x) = \sum_{i = 0}^M \hat{f_i}(x) $
4. قهر Kaggle وبقية العالم

# 3. وظائف الخسارة
إذا أردنا حل مشكلة التصنيف بدلاً من الانحدار، فما الذي سيتغير؟ نحتاج فقط إلى اختيار دالة الخسارة المناسبة $\large L(y, f)$. هذه هي اللحظة الأكثر أهمية وعالية المستوى التي تحدد بالضبط كيفية تحسيننا والخصائص التي يمكن أن نتوقعها في النموذج النهائي.
كقاعدة عامة، لا نحتاج إلى اختراع هذا بأنفسنا، فقد قام الباحثون بذلك بالفعل من أجلنا. سنستكشف اليوم دوال الخسارة للهدفين الأكثر شيوعًا: الانحدار $\large y \in \mathbb{R}$ والتصنيف الثنائي $\large y \in \left\{-1, 1\right\}$. 
### وظائف خسارة الانحدار
لنبدأ بمشكلة الانحدار لـ $\large y \in \mathbb{R}$. من أجل اختيار دالة الخسارة المناسبة، نحتاج إلى النظر في خصائص التوزيع الشرطي $\large (y|x)$ التي نريد استعادتها. الخيارات الأكثر شيوعًا هي:- $\large L(y, f) = (y - f)^2$ ويعرف أيضًا باسم $\large L_2$ الخسارة أو الخسارة الغوسية. وهو الوسط الشرطي الكلاسيكي، وهو أبسط الحالات وأكثرها شيوعاً. إذا لم يكن لدينا أي معلومات أو متطلبات إضافية ليكون النموذج قويًا، فيمكننا استخدام الخسارة الغوسية.
- $\large L(y, f) = |y - f|$ ويعرف أيضًا باسم $\large L_1$ الخسارة أو خسارة لابلاس. للوهلة الأولى، لا تبدو هذه الدالة قابلة للاشتقاق، ولكنها في الواقع تحدد الوسيط الشرطي. الوسيط، كما نعلم، قوي بالنسبة للقيم المتطرفة، وهذا هو السبب في أن وظيفة الخسارة هذه أفضل في بعض الحالات. عقوبة الاختلافات الكبيرة ليست ثقيلة كما هي في $\large L_2$.
- $ \large \begin{equation} L(y, f) =\left\{ \begin{array}{@{}ll@{}} (1 - \alpha) \cdot |y - f|, & \text{if}\ y-f \leq 0 \\ \alpha \cdot |y - f|, & \text{if}\ y-f >0 \end{array}\right. \end{المعادلة}، \alpha \in (0,1)
$ a.k.a. $\large L_q$ loss or Quantile loss.  Instead of median, it uses quantiles. For example, $\large \alpha = 0.75$ يتوافق مع الكمية 75%. يمكننا أن نرى أن هذه الوظيفة غير متماثلة وتعاقب الملاحظات الموجودة على الجانب الأيمن من الكم المحدد.
<img src='@@KEEP_00039@@ width=60%>دعونا نستخدم وظيفة الخسارة $\large L_q$ في بياناتنا. الهدف هو استعادة الكمية الشرطية 75% لجيب التمام. دعونا نجمع كل شيء معًا من أجل GBM:
- بيانات اللعبة $\large \left\{ (x_i, y_i) \right\}_{i=1, \ldots,300}$ ✓
- عدد من التكرارات $\large M = 3$ ✓;
- دالة الخسارة للكميات $ \large \begin{equation} L_{0.75}(y, f) =\left\{
\begin{array}{@{}ll@{}} 0.25 \cdot |y - f|, & \text{if}\ y-f \leq 0 \\ 0.75 \cdot |y - f|, & \text{if}\ y-f >0 \end{array}\right. \end{المعادلة} $ ✓;
- التدرج $\large L_{0.75}(y, f)$ - الوظيفة المرجحة بواسطة $\large \alpha = 0.75$. سنقوم بتدريب النموذج القائم على الشجرة للتصنيف:
$\large r_{i} = -\left[\frac{\partial L(y_i, f(x_i))}{\partial f(x_i)}\right]_{f(x)=\hat{f}(x)} = $
$\large = \alpha I(y_i > \hat{f}(x_i) ) - (1 - \alpha)I(y_i \leq \hat{f}(x_i) ), \quad \mbox{for } i=1,\ldots,300$ ✓;
- شجرة القرار كخوارزمية أساسية $\large h(x)$ ✓;
- المعلمة الفائقة للأشجار: العمق = 2 ✓؛

### وظائف فقدان التصنيف
الآن، دعونا نلقي نظرة على مشكلة التصنيف الثنائي $\large y \in \left\{-1, 1\right\}$. لقد رأينا أن GBM يمكنها أيضًا تحسين وظائف الخسارة غير القابلة للتمييز. من الناحية الفنية، من الممكن حل هذه المشكلة بخسارة الانحدار $\large L_2$، لكن ذلك لن يكون صحيحًا.يتطلب توزيع المتغير الهدف منا استخدام تشابه السجل، لذلك نحتاج إلى وظائف خسارة مختلفة للأهداف مضروبة في توقعاتها: $\large y \cdot f$. الاختيارات الأكثر شيوعًا ستكون كما يلي:
- $\large L(y, f) = log(1 + exp(-2yf))$ ويعرف أيضًا باسم الخسارة اللوجستية أو خسارة برنولي. يتمتع هذا بخاصية مثيرة للاهتمام تعاقب حتى الفئات التي تم التنبؤ بها بشكل صحيح، مما يساعد ليس فقط على تحسين الخسارة ولكن أيضًا على إبعاد الفئات عن بعضها البعض، حتى لو تم التنبؤ بجميع الفئات بشكل صحيح.
- $\large L(y, f) = exp(-yf)$ ويعرف أيضًا باسم خسارة AdaBoost. إن AdaBoost الكلاسيكي يعادل GBM مع وظيفة الخسارة هذه. من الناحية النظرية، تشبه هذه الوظيفة إلى حد كبير الخسارة اللوجستية، ولكن لها عقوبة أسية أكبر إذا كان التنبؤ خاطئًا.

### الأوزان
في بعض الأحيان، يكون هناك موقف نريد فيه دالة خسارة أكثر تحديدًا لمشكلتنا. على سبيل المثال، في السلاسل الزمنية المالية، قد نرغب في إعطاء وزن أكبر للتحركات الكبيرة في السلاسل الزمنية؛ للتنبؤ بخسارة العملاء، من المفيد التنبؤ بخسارة العملاء ذوي القيمة الدائمة العالية (أو القيمة الدائمة: مقدار الأموال التي سيجلبها العميل في المستقبل).
<img src='@@KEEP_00049@@   align='center'>
سيخترع المحارب الإحصائي دالة الخسارة الخاصة به، ويكتب التدرج اللوني لها (للحصول على تدريب أكثر فعالية، قم بتضمين الهسي)، ويتحقق بعناية مما إذا كانت هذه الوظيفة تلبي الخصائص المطلوبة. ومع ذلك، هناك احتمال كبير لارتكاب خطأ ما في مكان ما، ومواجهة الصعوبات الحسابية، وقضاء قدر كبير من الوقت في البحث.وبدلاً من ذلك، تم اختراع أداة بسيطة جدًا (نادرًا ما يتم تذكرها عمليًا): وزن الملاحظات وتعيين وظائف الوزن. أبسط مثال على هذا الترجيح هو تحديد الأوزان لتوازن الطبقة. بشكل عام، إذا علمنا أن بعض المجموعات الفرعية من البيانات، سواء في متغيرات الإدخال $\large x$ أو في المتغير الهدف $\large y$، لها أهمية أكبر لنموذجنا، فإننا نخصص لها وزنًا أكبر $\large w(x,y)$. الهدف الرئيسي هو استيفاء المتطلبات العامة للأوزان:
$$ \large w_i \in \mathbb{R}, \\
\large w_i \geq 0 \quad \mbox{for } i=1,\ldots,n, \\
\large \sum_{i = 1}^n w_i > 0 $$
يمكن أن تقلل الأوزان بشكل كبير من الوقت المستغرق في ضبط دالة الخسارة للمهمة التي نقوم بحلها، كما تشجع أيضًا إجراء التجارب على خصائص النماذج المستهدفة. إن تعيين هذه الأوزان هو بالكامل وظيفة إبداعية. نحن ببساطة نضيف الأوزان العددية:
$$ \large L_{w}(y,f) = w \cdot L(y,f), \\
\large r_{it} =   - w_i \cdot \left[\frac{\partial L(y_i, f(x_i))}{\partial f(x_i)}\right]_{f(x)=\hat{f}(x)}, \quad \mbox{for } i=1,\ldots,n$$
من الواضح أننا، بالنسبة للأوزان التعسفية، لا نعرف الخصائص الإحصائية لنموذجنا. في كثير من الأحيان، قد يكون ربط الأوزان بالقيم $\large y$ معقدًا للغاية. على سبيل المثال، استخدام الأوزان المتناسبة مع $\large |y|$ في $\large L_1$ لا يعادل خسارة $\large L_2$ لأن التدرج لن يأخذ في الاعتبار قيم التنبؤات نفسها: $\large \hat{f}(x)$.
نذكر كل هذا حتى نتمكن من فهم إمكانياتنا بشكل أفضل. لنقم بإنشاء بعض الأوزان الغريبة جدًا لبيانات لعبتنا. سوف نحدد دالة الوزن غير المتماثلة بقوة على النحو التالي:
$$ \large \begin{equation} w(x) =\left\{   \begin{array}{@{}ll@{}}     0.1, & \text{if}\ x \leq 0 \\     0.1 + |cos(x)|, & \text{if}\ x >0 \end{array}\right. \end{equation} $$
<img src='@@KEEP_00050@@   align='center'>
باستخدام هذه الأوزان، نتوقع الحصول على خاصيتين: تفاصيل أقل للقيم السالبة $\large x$ وشكل الدالة، المشابه لجيب التمام الأولي. نحن نأخذ ضبط GBM الآخر من مثالنا السابق مع التصنيف بما في ذلك البحث الخطي عن المعاملات المثلى. دعونا ننظر إلى ما لدينا:<img src='@@KEEP_00051@@   align='center'>
لقد حققنا النتيجة التي توقعناها. أولاً، يمكننا أن نرى مدى قوة اختلاف المخلفات الزائفة؛ في التكرار الأولي، تبدو تقريبًا مثل جيب التمام الأصلي. ثانيًا، غالبًا ما يتم تجاهل الجزء الأيسر من الرسم البياني للدالة لصالح الجزء الأيمن، الذي له أوزان أكبر. ثالثًا، تلقت الوظيفة التي حصلنا عليها في التكرار الثالث اهتمامًا كافيًا وبدأت تبدو مشابهة لجيب التمام الأصلي (بدأت أيضًا في التناسب الزائد قليلاً).
تعتبر الأوزان أداة قوية ولكنها محفوفة بالمخاطر يمكننا استخدامها للتحكم في خصائص نموذجنا. إذا كنت ترغب في تحسين دالة الخسارة لديك، فمن المفيد أن تحاول حل مشكلة أكثر بساطة أولاً ولكن قم بإضافة أوزان إلى الملاحظات وفقًا لتقديرك.



# 4. الاستنتاج
اليوم، تعلمنا النظرية وراء تعزيز التدرج. إن GBM ليست مجرد خوارزمية محددة ولكنها منهجية شائعة لبناء مجموعات من النماذج. بالإضافة إلى ذلك، فإن هذه المنهجية مرنة وقابلة للتوسيع بما فيه الكفاية - من الممكن تدريب عدد كبير من النماذج، مع الأخذ في الاعتبار وظائف الخسارة المختلفة مع مجموعة متنوعة من وظائف الترجيح.
تُظهِر مسابقات التدريب والتعلم الآلي أنه في المشكلات القياسية (باستثناء الصور والصوت والبيانات المتفرقة للغاية)، غالبًا ما تكون GBM هي الخوارزمية الأكثر فعالية (ناهيك عن التراص والمجموعات عالية المستوى، حيث تكون GBM دائمًا جزءًا منها).  هناك أيضًا العديد من تعديلات GBM [للتعلم المعزز](https://arxiv.org/abs/1603.04119) (Minecraft, ICML 2016). بالمناسبة، خوارزمية Viola-Jones، التي لا تزال تستخدم في رؤية الكمبيوتر، [تعتمد على AdaBoost](https://en.wikipedia.org/wiki/Viola%E2%80%93Jones_object_detection_framework#Learning_algorithm).في هذه المقالة، قمنا عمدًا بحذف الأسئلة المتعلقة بتنظيم GBM، والعشوائية، والمعلمات المفرطة. لم يكن من قبيل الصدفة أننا استخدمنا عددًا صغيرًا من التكرارات $\large M = 3$ طوال الوقت. إذا استخدمنا 30 شجرة بدلاً من 3 وقمنا بتدريب GBM كما هو موضح، فلن تكون النتيجة متوقعة:
