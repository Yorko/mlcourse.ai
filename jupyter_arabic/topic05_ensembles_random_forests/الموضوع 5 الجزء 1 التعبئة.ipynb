## 1. الفرق
[نظرية هيئة المحلفين كوندورسيه](https://en.wikipedia.org/wiki/Condorcet%27s_jury_theorem) (1784) تدور حول مجموعة إلى حد ما. تنص على أنه إذا أصدر كل عضو في هيئة المحلفين حكمًا مستقلاً وكان احتمال القرار الصحيح من قبل كل محلف أكثر من 0.5، فإن احتمال القرار الصحيح من قبل هيئة المحلفين بأكملها يزيد مع العدد الإجمالي للمحلفين ويميل إلى واحد. من ناحية أخرى، إذا كان احتمال أن يكون على حق أقل من 0.5 لكل محلف، فإن احتمال اتخاذ القرار الصحيح من قبل هيئة المحلفين بأكملها يتناقص مع عدد المحلفين ويميل إلى الصفر. 
لنكتب تعبيرًا تحليليًا لهذه النظرية:
- $\large N$ هو إجمالي عدد المحلفين؛
- $\large m$ هو الحد الأدنى لعدد المحلفين الذي يمكن أن يشكل الأغلبية، وهو $\large m = floor(N/2) + 1$؛
- $\large {N \choose i}$ هو عدد $\large i$-المجموعات من مجموعة تحتوي على عناصر $\large N$.
- $\large p$ هو احتمال اتخاذ القرار الصحيح من قبل هيئة المحلفين؛
- $\large \mu$ هو احتمال اتخاذ القرار الصحيح من قبل هيئة المحلفين بأكملها.
ثم:
$$ \large \mu = \sum_{i=m}^{N}{N\choose i}p^i(1-p)^{N-i} $$
يمكن ملاحظة أنه إذا $\large p > 0.5$، ثم $\large \mu > p$. بالإضافة إلى ذلك، إذا $\large N \rightarrow \infty $، ثم $\large \mu \rightarrow 1$.دعونا نلقي نظرة على مثال آخر للمجموعات: الملاحظة المعروفة باسم [حكمة الجمهور](https://en.wikipedia.org/wiki/Wisdom_of_the_crowd). <img src="../../img/bull.png" align="right" width=15% height=15%> في عام 1906، قام [فرانسيس جالتون](https://en.wikipedia.org/wiki/Francis_Galton) بزيارة معرض ريفي في بليموث حيث رأى مسابقة تُقام للمزارعين.   حاول 800 مشارك تقدير وزن الثور المذبوح. وكان الوزن الحقيقي للثور 1198 جنيها. وعلى الرغم من عدم تمكن أي من المزارعين من تخمين الوزن الدقيق للحيوان، إلا أن متوسط ​​توقعاتهم كان 1197 رطلاً.
تم اعتماد فكرة مماثلة لتقليل الأخطاء في مجال التعلم الآلي.



## 2. التمهيد
*التجميع* (المعروف أيضًا باسم [تجميع Bootstrap](https://en.wikipedia.org/wiki/Bootstrap_aggregating)) هو أحد تقنيات التجميع الأولى والأكثر أساسية. تم اقتراحه من قبل [Leo Breiman](https://en.wikipedia.org/wiki/Leo_Breiman) في عام 1994. يعتمد التغليف على الطريقة الإحصائية لـ [bootstrapping](https://en.wikipedia.org/wiki/Bootstrapping_%28statistics%29)، مما يجعل تقييم العديد من إحصائيات النماذج المعقدة أمرًا ممكنًا.
تسير طريقة التمهيد على النحو التالي. يجب أن تكون هناك عينة $\large X$ بالحجم $\large N$. يمكننا عمل عينة جديدة من العينة الأصلية عن طريق سحب عناصر $\large N$ من الأخيرة بشكل عشوائي وموحد، مع الاستبدال. بمعنى آخر، نختار عنصرًا عشوائيًا من العينة الأصلية بالحجم $\large N$ ونقوم بذلك $\large N$ مرات. من المرجح أن يتم تحديد جميع العناصر بالتساوي، وبالتالي يتم رسم كل عنصر باحتمال متساو $\large \frac{1}{N}$.
لنفترض أننا نسحب الكرات من الكيس واحدة تلو الأخرى. في كل خطوة، يتم إعادة الكرة المحددة إلى الحقيبة بحيث يتم الاختيار التالي بشكل متساوٍ، أي من نفس عدد الكرات $\large N$. لاحظ أنه نظرًا لأننا أعدنا الكرات مرة أخرى، فقد تكون هناك نسخ مكررة في العينة الجديدة. دعنا نسمي هذه العينة الجديدة $\large X_1$.من خلال تكرار هذا الإجراء $\large M$ مرات، نقوم بإنشاء $\large M$ *عينات التمهيد* $\large X_1, \dots, X_M$. في النهاية، لدينا عدد كافٍ من العينات ويمكننا حساب إحصائيات مختلفة للتوزيع الأصلي.
![صورة](../../img/bootstrap_eng.png)
على سبيل المثال، سنستخدم مجموعة البيانات `telecom_churn` المألوفة. في السابق، عندما ناقشنا أهمية الميزة، رأينا أن أحد أهم الميزات في مجموعة البيانات هذه هو عدد المكالمات إلى خدمة العملاء. دعونا نتصور البيانات وننظر إلى توزيع هذه الميزة.


```python
import numpy as np
import pandas as pd
import seaborn as sns

sns.set()
%matplotlib inline
from matplotlib import pyplot as plt

telecom_data = pd.read_csv("../../data/telecom_churn.csv")

telecom_data.loc[telecom_data["Churn"] == False, "Customer service calls"].hist(
    label="Loyal"
)
telecom_data.loc[telecom_data["Churn"] == True, "Customer service calls"].hist(
    label="Churn"
)
plt.xlabel("Number of calls")
plt.ylabel("Density")
plt.legend();
```


يبدو أن العملاء المخلصين يقومون بإجراء مكالمات أقل لخدمة العملاء مقارنة بأولئك الذين يغادرون الخدمة في النهاية. الآن، قد يكون من الجيد تقدير متوسط ​​عدد مكالمات خدمة العملاء في كل مجموعة. وبما أن مجموعة البيانات لدينا صغيرة، فلن نحصل على تقدير جيد بمجرد حساب متوسط ​​العينة الأصلية. سيكون من الأفضل لنا تطبيق طريقة bootstrap. لنقم بإنشاء 1000 عينة تمهيدية جديدة من مجتمعنا الأصلي وننتج تقديرًا فاصلًا للمتوسط.


```python
def get_bootstrap_samples(data, n_samples):
    """Generate bootstrap samples using the bootstrap method."""
    indices = np.random.randint(0, len(data), (n_samples, len(data)))
    samples = data[indices]
    return samples


def stat_intervals(stat, alpha):
    """Produce an interval estimate."""
    boundaries = np.percentile(stat, [100 * alpha / 2.0, 100 * (1 - alpha / 2.0)])
    return boundaries


# Save the data about the loyal and former customers to split the dataset
loyal_calls = telecom_data.loc[
    telecom_data["Churn"] == False, "Customer service calls"
].values
churn_calls = telecom_data.loc[
    telecom_data["Churn"] == True, "Customer service calls"
].values

# Set the seed for reproducibility of the results
np.random.seed(0)

# Generate the samples using bootstrapping and calculate the mean for each of them
loyal_mean_scores = [
    np.mean(sample) for sample in get_bootstrap_samples(loyal_calls, 1000)
]
churn_mean_scores = [
    np.mean(sample) for sample in get_bootstrap_samples(churn_calls, 1000)
]

# Print the resulting interval estimates
print(
    "Service calls from loyal: mean interval", stat_intervals(loyal_mean_scores, 0.05)
)
print(
    "Service calls from churn: mean interval", stat_intervals(churn_mean_scores, 0.05)
)
```


لتفسير فترات الثقة، يمكنك معالجة [هذه](https://www.graphpad.com/guides/prism/7/statistics/stat_more_about_confidence_interval.htm?toc=0&printWindow) مذكرة موجزة أو أي دورة تدريبية حول الإحصائيات. ليس صحيحًا القول بأن فاصل الثقة يحتوي على 95% من القيم. لاحظ أن الفاصل الزمني للعملاء الأوفياء أضيق، وهو أمر معقول لأنهم يقومون بإجراء مكالمات أقل (0 أو 1 أو 2) مقارنة بالعملاء المختلين الذين يتصلون حتى يملوا ويقررون تبديل مقدمي الخدمة. 



## 3. التعبئة
الآن بعد أن استوعبت فكرة التمهيد، يمكننا الانتقال إلى *التعبئة*.لنفترض أن لدينا مجموعة تدريب $\large X$. باستخدام bootstrapping، نقوم بإنشاء عينات $\large X_1, \dots, X_M$. الآن، لكل عينة تمهيد، نقوم بتدريب المصنف الخاص بها $\large a_i(x)$. سيقوم المصنف النهائي بمتوسط ​​المخرجات من كل هذه المصنفات الفردية. وفي حالة التصنيف فإن هذه التقنية تقابل التصويت:
$$\large a(x) = \frac{1}{M}\sum_{i = 1}^M a_i(x).$$
الصورة أدناه توضح هذه الخوارزمية:
<img src="../../img/bagging.png" alt="image"/>



دعونا نفكر في مشكلة الانحدار مع الخوارزميات الأساسية $\large b_1(x), \dots , b_n(x)$. افترض أن هناك وظيفة هدف مثالية للإجابات الحقيقية $\large y(x)$ محددة لجميع المدخلات وأن التوزيع $\large p(x)$ محدد. يمكننا بعد ذلك التعبير عن الخطأ لكل دالة انحدار على النحو التالي:  
$$\large \varepsilon_i(x) = b_i(x) - y(x), \quad i = 1, \dots, n$$
والقيمة المتوقعة لمتوسط الخطأ التربيعي:  
$$\large \E_x\left[\left(b_i(x) - y(x)\right)^{2}\right] = \E_x\left[\varepsilon_i^{2}(x)\right].$$
بعد ذلك، سيبدو الخطأ المتوسط في جميع وظائف الانحدار كما يلي:  
$$ \large \E_1 = \frac{1}{n} \E_x\left[ \sum_i^n \varepsilon_i^{2}(x)\right]$$
سنفترض أن الأخطاء غير متحيزة وغير مترابطة، أي: 
$$\large \begin{array}{rcl} \E_x\left[\varepsilon_i(x)\right] &=& 0, \\
\E_x\left[\varepsilon_i(x)\varepsilon_j(x)\right] &=& 0, \quad i \neq j. \end{array}$$
الآن، لنقم بإنشاء دالة انحدار جديدة تعمل على حساب متوسط القيم من الوظائف الفردية: 
$$\large a(x) = \frac{1}{n}\sum_{i=1}^{n}b_i(x)$$
دعونا نجد خطأه التربيعي المتوسط:
$$\large \begin{array}{rcl}\E_n &=& \E_x\left[\frac{1}{n}\sum_{i=1}^{n}b_i(x)-y(x)\right]^2 \\
&=& \E_x\left[\frac{1}{n}\sum_{i=1}^{n}\varepsilon_i\right]^2 \\
&=& \frac{1}{n^2}\E_x\left[\sum_{i=1}^{n}\varepsilon_i^2(x) + \sum_{i \neq j}\varepsilon_i(x)\varepsilon_j(x)\right] \\
&=& \frac{1}{n}\E_1\end{array}$$
وبالتالي، من خلال حساب متوسط الإجابات الفردية، قمنا بتقليل متوسط الخطأ المربع بعامل $\large n$.
من درسنا السابق، دعونا نتذكر المكونات التي تشكل الخطأ الإجمالي خارج العينة:
$$\large \begin{array}{rcl} 
\Err\left(\vec{x}\right) &=& \E\left[\left(y - \hat{f}\left(\vec{x}\right)\right)^2\right] \\
&=& \sigma^2 + f^2 + \Var\left(\hat{f}\right) + \E\left[\hat{f}\right]^2 - 2f\E\left[\hat{f}\right] \\
&=& \left(f - \E\left[\hat{f}\right]\right)^2 + \Var\left(\hat{f}\right) + \sigma^2 \\
&=& \Bias\left(\hat{f}\right)^2 + \Var\left(\hat{f}\right) + \sigma^2
\end{array}$$


تعمل عملية التعبئة على تقليل تباين المصنف عن طريق تقليل الفرق في الخطأ عندما نقوم بتدريب النموذج على مجموعات بيانات مختلفة. وبعبارة أخرى، فإن التعبئة تمنع الإفراط في التجهيز. تأتي كفاءة التعبئة من حقيقة أن النماذج الفردية مختلفة تمامًا بسبب اختلاف بيانات التدريب وأخطاءها تلغي بعضها البعض أثناء التصويت. بالإضافة إلى ذلك، من المحتمل أن يتم حذف القيم المتطرفة في بعض عينات التدريب.
تدعم مكتبة `scikit-learn` التعبئة باستخدام مقدرات التعريف `BaggingRegressor` و`BaggingClassifier`. يمكنك استخدام معظم الخوارزميات كقاعدة.
دعونا نفحص كيفية عمل التعبئة عمليًا ونقارنها بشجرة القرارات. لهذا، سوف نستخدم مثالاً من [وثائق sklearn](http://scikit-learn.org/stable/auto_examples/ensemble/plot_bias_variance.html#sphx-glr-auto-examples-ensemble-plot-bias-variance-py).

كما ترون من الرسم البياني أعلاه، فإن التباين في الخطأ أقل بكثير بالنسبة للتعبئة. تذكر أننا أثبتنا ذلك من الناحية النظرية بالفعل.
التعبئة فعالة في مجموعات البيانات الصغيرة. يؤدي إسقاط جزء صغير من بيانات التدريب إلى إنشاء مصنفات أساسية مختلفة إلى حد كبير. إذا كانت لديك مجموعة بيانات كبيرة، فيمكنك إنشاء عينات تمهيدية بحجم أصغر بكثير.
من غير المرجح أن ينطبق المثال أعلاه على أي عمل حقيقي. وذلك لأننا افترضنا بقوة أن أخطائنا الفردية غير مترابطة. في أغلب الأحيان، يكون هذا أمرًا متفائلًا للغاية بالنسبة لتطبيقات العالم الحقيقي. عندما يكون هذا الافتراض خاطئا، فإن الانخفاض في الخطأ لن يكون كبيرا. في المحاضرات التالية، سنناقش بعض طرق التجميع الأكثر تطورًا، والتي تتيح تنبؤات أكثر دقة في مشاكل العالم الحقيقي.



## 4. خطأ خارج الحقيبةوبالنظر إلى المستقبل، في حالة Random Forest، ليست هناك حاجة لاستخدام التحقق المتبادل أو عينات الاحتجاز من أجل الحصول على تقدير خطأ غير متحيز. لماذا؟ لأنه في تقنيات التجميع، يتم تقدير الخطأ داخليًا.
 
يتم إنشاء أشجار عشوائية باستخدام عينات تمهيدية مختلفة من مجموعة البيانات الأصلية. يتم استبعاد ما يقرب من 37% من المدخلات من عينة تمهيد معينة ولا يتم استخدامها في إنشاء شجرة $\large k$-th.
وهذا من السهل إثباته. لنفترض أن هناك أمثلة $\large \ell$ في مجموعة البيانات الخاصة بنا. في كل خطوة، يكون لكل نقطة بيانات احتمالية متساوية للانتهاء في عينة تمهيد مع الاستبدال، الاحتمال $\large\frac{1}{\ell}.$ احتمال عدم وجود عينة تمهيد تحتوي على عنصر مجموعة بيانات معين (أي تم حذفها $\large \ell$ مرات) يساوي $\large (1 - \frac{1}{\ell})^\ell$. عندما $\large \ell \rightarrow +\infty$، يصبح مساويًا لـ [الحد الثاني الملحوظ](https://en.wikipedia.org/wiki/List_of_limits) $\large \frac{1}{e}$. ومن ثم، فإن احتمال اختيار مثال محدد هو $\large \approx  1 - \frac{1}{e} \approx 63\%$.
دعونا نتصور كيف يعمل تقدير الخطأ خارج الحقيبة (أو OOBE):
![صورة](../../img/oob.png)
يمثل الجزء العلوي من الشكل أعلاه مجموعة البيانات الأصلية لدينا. قمنا بتقسيمها إلى مجموعات التدريب (يسار) والاختبار (يمين). في الصورة اليسرى، نرسم شبكة تقسم مجموعة البيانات الخاصة بنا بشكل مثالي وفقًا للفئات. الآن، نستخدم نفس الشبكة لتقدير حصة الإجابات الصحيحة في مجموعة الاختبار لدينا. يمكننا أن نرى أن المصنف الخاص بنا أعطى إجابات غير صحيحة في تلك الحالات الأربع التي لم يتم استخدامها أثناء التدريب (على اليسار). ومن ثم، فإن دقة المصنف لدينا هي $\large \frac{11}{15}*100\% = 73.33\%$.لتلخيص ذلك، يتم تدريب كل خوارزمية أساسية على $\large \approx 63\%$ من الأمثلة الأصلية. يمكن التحقق من صحته على $\large \approx 37\%$ المتبقي. تقدير Out-of-Bag ليس أكثر من مجرد تقدير متوسط ​​للخوارزميات الأساسية على $\large \approx 37\%$ من المدخلات التي تم استبعادها من التدريب.
