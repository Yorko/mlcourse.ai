تمكن ليو بريمان من تطبيق عملية التمهيد ليس فقط في الإحصائيات ولكن أيضًا في التعلم الآلي. قام بالتعاون مع عادل كاتلر بتوسيع وتحسين خوارزمية الغابة العشوائية [اقترحها تين كام هو](http://ect.bell-labs.com/who/tkh/publications/papers/odt.pdf). لقد قاموا بدمج بناء الأشجار غير المترابطة باستخدام [CART](https://en.wikipedia.org/wiki/Predictive_analytics#Classification_and_regression_trees_.28CART.29)، والتعبئة، و[طريقة الفضاء الجزئي العشوائي](https://en.wikipedia.org/wiki/Random_subspace_method).  
تعتبر أشجار القرار خيارًا جيدًا للمصنف الأساسي في التعبئة لأنها متطورة جدًا ويمكن أن تحقق خطأ تصنيف صفر في أي عينة. تعمل طريقة الفضاء الجزئي العشوائي على تقليل الارتباط بين الأشجار وبالتالي تمنع التجهيز الزائد. من خلال التعبئة، يتم تدريب الخوارزميات الأساسية على مجموعات فرعية عشوائية مختلفة من مجموعة الميزات الأصلية.
تقوم الخوارزمية التالية ببناء مجموعة من النماذج باستخدام طريقة الفضاء الجزئي العشوائي:
1. اجعل عدد المثيلات يساوي $\large \ell$، وعدد الميزات يساوي $\large d$.
2. اختر $\large L$ كعدد النماذج الفردية في المجموعة. 
3. لكل طراز $\large l$، اختر عدد الميزات $\large dl < d$. كقاعدة عامة، يتم استخدام نفس قيمة $\large dl$ لجميع النماذج.
4. لكل نموذج $\large l$، أنشئ مجموعة تدريب عن طريق تحديد ميزات $\large dl$ بشكل عشوائي من المجموعة الكاملة لميزات $\large d$.
5. تدريب كل نموذج.
6. قم بتطبيق نموذج المجموعة الناتج على مثيل جديد من خلال دمج النتائج من كافة النماذج في $\large L$. يمكنك استخدام إما تصويت الأغلبية أو تجميع الاحتمالات الخلفية. 



## 1. الخوارزمية
خوارزمية إنشاء غابة عشوائية من الأشجار $\large N$ هي كما يلي:
* لكل $\large k = 1, \dots, N$:* قم بإنشاء نموذج التمهيد $\large X_k$.
    * أنشئ شجرة قرارات $\large b_k$ على العينة $\large X_k$:
        * اختر أفضل ميزة وفقًا للمعايير المحددة. قم بتقسيم العينة حسب هذه الميزة لإنشاء مستوى شجرة جديد. كرر هذا الإجراء حتى يتم استنفاد العينة.
        * بناء الشجرة حتى لا تحتوي أي ورقة من أوراقها على أكثر من $\large n_\text{min}$ مثلاً أو حتى الوصول إلى عمق معين.
        * لكل تقسيم، نقوم أولاً باختيار ميزات $\large m$ بشكل عشوائي من الميزات الأصلية $\large d$ ثم نبحث عن أفضل تقسيم تالي فقط بين المجموعة الفرعية.
         
يتم تعريف المصنف النهائي بواسطة:
$$\large a(x) = \frac{1}{N}\sum_{k = 1}^N b_k(x)$$
نحن نستخدم تصويت الأغلبية للتصنيف ومتوسط الانحدار.
بالنسبة لمشاكل التصنيف، من المستحسن تعيين $\large m = \sqrt{d}$. بالنسبة لمشكلات الانحدار، نستخدم عادةً $\large m = \frac{d}{3}$، حيث $\large d$ هو عدد الميزات. يوصى ببناء كل شجرة حتى تحتوي جميع أوراقها على $\large n_\text{min} = 1$ أمثلة للتصنيف و$\large n_\text{min} = 5$ أمثلة للانحدار. 
يمكنك رؤية الغابة العشوائية كتعبئة لأشجار القرار مع تعديل تحديد مجموعة فرعية عشوائية من الميزات عند كل تقسيم.
## 2. المقارنة مع أشجار القرار والتعبئة


```python
# Disable warnings in Anaconda
import warnings

import numpy as np

warnings.filterwarnings('ignore')

%matplotlib inline
from matplotlib import pyplot as plt

plt.style.use('ggplot')
plt.rcParams['figure.figsize'] = 10, 6
%config InlineBackend.figure_format = 'retina'

import seaborn as sns
from sklearn.datasets import make_circles
from sklearn.ensemble import (BaggingClassifier, BaggingRegressor,
                              RandomForestClassifier, RandomForestRegressor)
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
```

```python
n_train = 150        
n_test = 1000       
noise = 0.1

# Generate data
def f(x):
    x = x.ravel()
    return np.exp(-x ** 2) + 1.5 * np.exp(-(x - 2) ** 2)

def generate(n_samples, noise):
    X = np.random.rand(n_samples) * 10 - 5
    X = np.sort(X).ravel()
    y = np.exp(-X ** 2) + 1.5 * np.exp(-(X - 2) ** 2)\
        + np.random.normal(0.0, noise, n_samples)
    X = X.reshape((n_samples, 1))

    return X, y

X_train, y_train = generate(n_samples=n_train, noise=noise)
X_test, y_test = generate(n_samples=n_test, noise=noise)

# One decision tree regressor
dtree = DecisionTreeRegressor().fit(X_train, y_train)
d_predict = dtree.predict(X_test)

plt.figure(figsize=(10, 6))
plt.plot(X_test, f(X_test), "b")
plt.scatter(X_train, y_train, c="b", s=20)
plt.plot(X_test, d_predict, "g", lw=2)
plt.xlim([-5, 5])
plt.title("Decision tree, MSE = %.2f" 
          % np.sum((y_test - d_predict) ** 2))

# Bagging with a decision tree regressor
bdt = BaggingRegressor(DecisionTreeRegressor()).fit(X_train, y_train)
bdt_predict = bdt.predict(X_test)

plt.figure(figsize=(10, 6))
plt.plot(X_test, f(X_test), "b")
plt.scatter(X_train, y_train, c="b", s=20)
plt.plot(X_test, bdt_predict, "y", lw=2)
plt.xlim([-5, 5])
plt.title("Bagging for decision trees, MSE = %.2f" % np.sum((y_test - bdt_predict) ** 2));

# Random Forest
rf = RandomForestRegressor(n_estimators=10).fit(X_train, y_train)
rf_predict = rf.predict(X_test)

plt.figure(figsize=(10, 6))
plt.plot(X_test, f(X_test), "b")
plt.scatter(X_train, y_train, c="b", s=20)
plt.plot(X_test, rf_predict, "r", lw=2)
plt.xlim([-5, 5])
plt.title("Random forest, MSE = %.2f" % np.sum((y_test - rf_predict) ** 2));
```


كما يمكننا أن نرى من الرسوم البيانية لدينا وقيم MSE أعلاه، فإن الغابة العشوائية المكونة من 10 أشجار تحقق نتيجة أفضل من شجرة قرار واحدة ويمكن مقارنتها بالتعبئة باستخدام 10 أشجار. يتمثل الاختلاف الرئيسي بين الغابات العشوائية والتعبئة في أنه، في الغابة العشوائية، يتم تحديد أفضل ميزة للتقسيم من مجموعة فرعية عشوائية من الميزات المتاحة، بينما في التعبئة، يتم أخذ جميع الميزات في الاعتبار للتقسيم التالي الأفضل.
يمكننا أيضًا أن ننظر إلى مزايا الغابات العشوائية والتعبئة في مشاكل التصنيف:

```python
np.random.seed(42)
X, y = make_circles(n_samples=500, factor=0.1, noise=0.35, random_state=42)
X_train_circles, X_test_circles, y_train_circles, y_test_circles = \
    train_test_split(X, y, test_size=0.2)

dtree = DecisionTreeClassifier(random_state=42)
dtree.fit(X_train_circles, y_train_circles)

x_range = np.linspace(X.min(), X.max(), 100)
xx1, xx2 = np.meshgrid(x_range, x_range)
y_hat = dtree.predict(np.c_[xx1.ravel(), xx2.ravel()])
y_hat = y_hat.reshape(xx1.shape)
plt.contourf(xx1, xx2, y_hat, alpha=0.2)
plt.scatter(X[:,0], X[:,1], c=y, cmap='viridis', alpha=.7)
plt.title("Decision tree")
plt.show()

b_dtree = BaggingClassifier(DecisionTreeClassifier(), 
                            n_estimators=300, random_state=42)
b_dtree.fit(X_train_circles, y_train_circles)

x_range = np.linspace(X.min(), X.max(), 100)
xx1, xx2 = np.meshgrid(x_range, x_range)
y_hat = b_dtree.predict(np.c_[xx1.ravel(), xx2.ravel()])
y_hat = y_hat.reshape(xx1.shape)
plt.contourf(xx1, xx2, y_hat, alpha=0.2)
plt.scatter(X[:,0], X[:,1], c=y, cmap='viridis', alpha=.7)
plt.title("Bagging (decision trees)")
plt.show()

rf = RandomForestClassifier(n_estimators=300, random_state=42)
rf.fit(X_train_circles, y_train_circles)

x_range = np.linspace(X.min(), X.max(), 100)
xx1, xx2 = np.meshgrid(x_range, x_range)
y_hat = rf.predict(np.c_[xx1.ravel(), xx2.ravel()])
y_hat = y_hat.reshape(xx1.shape)
plt.contourf(xx1, xx2, y_hat, alpha=0.2)
plt.scatter(X[:,0], X[:,1], c=y, cmap='viridis', alpha=.7)
plt.title("Random forest")
plt.show()
```


توضح الأشكال أعلاه أن حدود القرار لشجرة القرار متعرجة تمامًا وتحتوي على الكثير من الزوايا الحادة التي تشير إلى فرط التجهيز وضعف القدرة على التعميم. سيكون لدينا مشكلة في وضع تنبؤات موثوقة بشأن بيانات الاختبار الجديدة. في المقابل، تتمتع خوارزمية التعبئة بحدود سلسة إلى حد ما وليس لديها علامات واضحة على التجهيز الزائد.

## 3. المعلمات
تقوم [مكتبة scikit-learn](http://scikit-learn.org/stable/) بتنفيذ غابات عشوائية من خلال توفير مُقدرين: `RandomForestClassifier` و`RandomForestRegressor`.
تظهر القائمة الكاملة لمعلمات الغابة العشوائية للانحدار أدناه:


- `n_estimators` — عدد الأشجار في الغابة (الافتراضي = 10)
   - `criterion` — الوظيفة المستخدمة لقياس جودة التقسيم. المعايير المدعومة هي "mse" لمتوسط الخطأ التربيعي، والذي يساوي تقليل التباين كمعيار لاختيار الميزة، و"mae" لمتوسط الخطأ المطلق (افتراضي = "mse")
   - `max_features` — عدد الميزات التي يجب مراعاتها عند البحث عن أفضل تقسيم. يمكنك تحديد عدد أو نسبة الميزات، أو الاختيار من بين القيم المتاحة: "تلقائي" (جميع الميزات)، "sqrt"، "log2". (افتراضي = "تلقائي")
   - `max_depth` — الحد الأقصى لعمق الشجرة (الافتراضي يعني أن العقد يتم توسيعها حتى تصبح جميع الأوراق نقية أو حتى تحتوي جميع الأوراق على أقل من min_samples_split عينات)
   - `min_samples_split` — الحد الأدنى لعدد العينات المطلوبة لتقسيم العقدة الداخلية. يمكن تحديده كرقم أو كنسبة مئوية من العدد الإجمالي للعينات (الافتراضي = 2)
   - `min_samples_leaf` — الحد الأدنى لعدد العينات المطلوبة في العقدة الطرفية (الافتراضي = 1)
   - `min_weight_fraction_leaf` — الحد الأدنى للكسر المرجح لمجموع الأوزان (لجميع عينات الإدخال) المطلوب وجوده عند العقدة الطرفية. يكون للعينات وزن متساوٍ عند عدم توفير Sample_weight (الافتراضي = 0)
   - `max_leaf_nodes` — الحد الأقصى لعدد الإجازات (الافتراضي = لا توجد قيود)
   - `min_impurity_split` — عتبة التوقف المبكر في نمو الأشجار. سوف تنقسم العقدة إذا كانت نجاستها أعلى من العتبة، وإلا فهي ورقة (الافتراضي = 1е-7)
   - `bootstrap` — ما إذا كان يتم استخدام عينات التمهيد عند بناء الأشجار (افتراضي = True)
   - `oob_score` — ما إذا كان سيتم استخدام عينات خارج الحقيبة لتقدير R^2 على البيانات غير المرئية (الافتراضي = False)
   - `n_jobs` - عدد المهام التي سيتم تشغيلها بالتوازي لكل من الملاءمة والتنبؤ. إذا كان -1، فإن عدد الوظائف iتم ضبطه على عدد النوى (الافتراضي = 1)
   - `random_state` — إذا كان int، فإن Random_state هي البذرة التي يستخدمها مولد الأرقام العشوائية؛ إذا كان مثيل RandomState، فإن Random_state هو منشئ الأرقام العشوائية؛ إذا لم يكن هناك شيء، فإن منشئ الأرقام العشوائية هو مثيل RandomState الذي يستخدمه np.random (الافتراضي = لا شيء)
   - `verbose` — يتحكم في إسهاب عملية بناء الشجرة (الافتراضي = 0)
   - `warm_start` - عند التعيين على True، أعد استخدام حل الاستدعاء السابق لملاءمة المجموعة وإضافة المزيد من المقدرات، وإلا، فما عليك سوى ملاءمة غابة جديدة بالكامل (افتراضي = False)



وفي حالة التصنيف، تكون المعلمات هي نفسها في الغالب. يختلف ما يلي فقط بالنسبة لـ `RandomForestClassifier` مقارنةً بـ `RandomForestRegressor`:
   - `criterion` — الوظيفة المستخدمة لقياس جودة التقسيم. المعايير المعتمدة هي "جيني" لشوائب جيني و"الانتروبيا" للحصول على المعلومات. ملاحظة: هذه المعلمة خاصة بالشجرة (افتراضي = "gini")
   - `class_weight` - وزن كل فئة (افتراضيًا، جميع الأوزان تساوي 1، ولكن يمكنك إنشاء قاموس بأوزان أو تحديده على أنه "متوازن" - يستخدم قيم الفئات لضبط الأوزان تلقائيًا بما يتناسب عكسيًا مع ترددات الفئة في بيانات الإدخال أو كـ "عينة فرعية متوازنة" - مثل "متوازن" باستثناء أنه يتم حساب الأوزان بناءً على عينة التمهيد لكل شجرة مزروعة)



فيما يلي المعلمات التي نحتاج إلى الاهتمام بها عند بناء نموذج جديد:- `n_estimators` - عدد الأشجار في الغابة؛
- `criterion` — الوظيفة المستخدمة لقياس جودة التقسيم؛
- `max_features` - عدد الميزات التي يجب مراعاتها عند البحث عن أفضل تقسيم؛
- `min_samples_leaf` — الحد الأدنى لعدد العينات المطلوبة في العقدة الطرفية؛
- `max_depth` — أقصى عمق للشجرة.
#### تدرب على الغابات العشوائية في مشكلة حقيقية
 
في هذا المثال، سننظر في توقع تراجع العملاء. هذه مشكلة تصنيف، لذا سنستخدم الدقة لتقييم النموذج.
أولاً، دعونا نبني مصنفًا بسيطًا سنستخدمه كخط أساس. ومن أجل التبسيط، سوف نستخدم الميزات الرقمية فقط.


```python
import pandas as pd
from sklearn.metrics import accuracy_score
from sklearn.model_selection import (GridSearchCV, StratifiedKFold,
                                     cross_val_score)

# Load data
df = pd.read_csv("../../data/telecom_churn.csv")

# Choose the numeric features
cols = []
for i in df.columns:
    if (df[i].dtype == "float64") or (df[i].dtype == 'int64'):
        cols.append(i)
        
# Divide the dataset into the input and target
X, y = df[cols].copy(), np.asarray(df["Churn"],dtype='int8')

# Initialize a stratified split of our dataset for the validation process
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Initialize the classifier with the default parameters 
rfc = RandomForestClassifier(random_state=42, n_jobs=-1)

# Train it on the training set
results = cross_val_score(rfc, X, y, cv=skf)

# Evaluate the accuracy on the test set
print("CV accuracy score: {:.2f}%".format(results.mean() * 100))
```


لدينا دقة تساوي 91.18%. الآن، دعونا نحاول تحسين هذه النتيجة، ونلقي نظرة على سلوك منحنيات التعلم عندما نقوم بتغيير المعلمات الأساسية.
لنبدأ بعدد الأشجار:


```python
# Initialize the validation
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Create lists to save the values of accuracy on training and test sets
train_acc = []
test_acc = []
temp_train_acc = []
temp_test_acc = []
trees_grid = [5, 10, 15, 20, 30, 50, 75, 100]

for ntrees in trees_grid:
    rfc = RandomForestClassifier(n_estimators=ntrees, random_state=42, n_jobs=-1)
    temp_train_acc = []
    temp_test_acc = []
    for train_index, test_index in skf.split(X, y):
        X_train, X_test = X.iloc[train_index], X.iloc[test_index]
        y_train, y_test = y[train_index], y[test_index]
        rfc.fit(X_train, y_train)
        temp_train_acc.append(rfc.score(X_train, y_train))
        temp_test_acc.append(rfc.score(X_test, y_test))
    train_acc.append(temp_train_acc)
    test_acc.append(temp_test_acc)
    
train_acc, test_acc = np.asarray(train_acc), np.asarray(test_acc)
print("Best CV accuracy is {:.2f}% with {} trees".format(max(test_acc.mean(axis=1))*100, 
                                                        trees_grid[np.argmax(test_acc.mean(axis=1))]))
```


## 4. التباين والانفصال
لنكتب تباين الغابة العشوائية كـ
$$\large \Var f(x) = \rho(x)\sigma^2(x)$$
$$\large \rho(x) = \Corr\left[T(x_1,\Theta_1(Z)),T(x_2,\Theta_2(Z))\right],$$
أين
* $\large \rho(x)$ هو نموذج معامل الارتباط بين أي شجرتين مستخدمتين في حساب المتوسط:
* $\large \Theta_1(Z)$ و$\large \Theta_2(Z)$ هما زوجان من الأشجار تم اختيارهما عشوائيًا على عناصر مختارة عشوائيًا من العينة $Z$؛
* $\large T(x,\Theta_i(Z))$ هو إخراج $\large i$-مصنف الشجرة على متجه الإدخال $\large x$؛
* $\large \sigma^2(x)$ هو تباين العينة لأي شجرة تم اختيارها عشوائيًا:
$$\large \sigma^2(x) = \Var\left[T(x,\Theta(X))\right]$$
من السهل الخلط بين $\large \rho(x)$ ومتوسط الارتباط بين الأشجار المدربة في غابة عشوائية معينة عندما نعتبر الأشجار بمثابة متجهات N ونحسب متوسط الارتباط الزوجي بينها. ولكن هذا ليس هو الحال.في الواقع، لا يرتبط هذا الارتباط الشرطي بشكل مباشر بعملية المتوسط، ويحذرنا اعتماد $\large \rho(x)$ على $\large x$ من هذا الاختلاف. $\large \rho(x)$ هو الارتباط النظري بين زوج من الأشجار العشوائية المقدرة على الإدخال $\large x$. وتأتي قيمتها من أخذ العينات المتكررة لمجموعة التدريب من السكان $\large Z$ والاختيار العشوائي اللاحق لزوج من الأشجار. في المصطلحات الإحصائية، هذا هو الارتباط الناتج عن توزيع العينات $\large Z$ و$\large \Theta$.
التباين الشرطي لأي زوج من الأشجار يساوي 0 لأن عملية التمهيد واختيار الميزة مستقلتان وموزعتان بشكل متطابق.
إذا أخذنا في الاعتبار تباين شجرة واحدة، فإنه بالكاد يعتمد على معلمات التقسيم ($\large m$). لكنها حاسمة بالنسبة للفرق. إن تباين الشجرة أعلى بكثير من تباين المجموعة. وكتاب *عناصر التعلم الإحصائي (تريفور هاستي، روبرت تيبشيراني، جيروم فريدمان)* له مثال رائع يوضح هذه الحقيقة:

## 5. التحيز
كما هو الحال في التعبئة، فإن انحياز الغابة العشوائية هو نفس انحياز شجرة واحدة $\large T(x,\Theta(Z))$:
$$ \large \begin{array}{rcl} \Bias &=& \mu(x) - \E_Z \, f_{rf}(x) \\
&=& \mu(x) - \E_Z \, \E_{\Theta | Z} \, T(x,\Theta(Z))\end{array}$$  
في القيمة المطلقة، يكون الانحياز عادة أعلى من الانحياز في الشجرة غير المقطوعة لأن التوزيع العشوائي وتقليل مساحة العينة يفرضان قيودهما الخاصة على النموذج. ولذلك، فإن التحسينات في دقة التنبؤ التي تم الحصول عليها عن طريق التعبئة والغابات العشوائية هي فقط نتيجة لتقليل التباين.



## 6. أشجار عشوائية للغايةتستخدم الأشجار العشوائية للغاية درجة أكبر من التوزيع العشوائي عند اختيار نقطة القطع عند تقسيم عقدة الشجرة. كما هو الحال في الغابات العشوائية، يتم استخدام مجموعة فرعية عشوائية من الميزات. ولكن بدلاً من البحث عن العتبات المثالية، يتم تحديد قيمها عشوائيًا لكل ميزة محتملة، ويتم استخدام أفضل واحدة من هذه العتبات التي تم إنشاؤها عشوائيًا كأفضل قاعدة لتقسيم العقدة. يؤدي هذا عادةً إلى حدوث انخفاض طفيف في تباين النموذج مع زيادة طفيفة في التحيز.
في مكتبة `scikit-learn`، يوجد تطبيقان للأشجار العشوائية للغاية: [ExtraTreesClassifier](http://scikit-learn.org/stable/modules/generated/sklearn.ensemble.ExtraTreesClassifier.html#sklearn.ensemble.ExtraTreesClassifier) و[ExtraTreesRegressor](http://scikit-learn.org/stable/modules/generated/sklearn.ensemble.ExtraTreesRegressor.html#sklearn.ensemble.ExtraTreesRegressor). 
يجب استخدام هذه الطريقة إذا كنت تعاني من فرط التناسب بشكل كبير مع الغابات العشوائية أو تعزيز التدرج.



## 7. أوجه التشابه بين الغابة العشوائية وأقرب الجيران 
تشبه طريقة الغابة العشوائية تقنية أقرب الجيران. تعتمد تنبؤات الغابات العشوائية على تسميات الأمثلة المتشابهة من مجموعة التدريب. وكلما ظهرت هذه الأمثلة في كثير من الأحيان في نفس ورقة الشجرة، كلما زاد تشابهها. دعونا نثبت هذا رسميا.
لنفكر في مشكلة الانحدار في دالة الخسارة التربيعية. دع $\large T_n(x)$ هو رقم ورقة الشجرة $\large n$- في غابة عشوائية مع الإدخال $\large x$.
استجابة الخوارزمية لمتجه الإدخال $\large x$ تساوي متوسط ​​الاستجابة لجميع أمثلة عينة التدريب التي تقع في الورقة $\large T_n(x)$. يمكن كتابة هذا كـ
$$\large b_n(x) = \sum_{i=1}^{l}w_n(x,x_i)y_i,$$
أين
$$ \large w_n(x, x_i) = \frac{\left[T_n(x) = T_n(x_i)\right]}{\sum_{j=1}^{l}\left[T_n(x) = T_n(x_j)\right]}$$  
ثم يكون رد التركيبة
$$ \large \begin{array}{rcl} a_n(x) &=& \frac{1}{N}\sum_{n=1}^{N}\sum_{i=1}^{l}w_n(x,x_i)y_i \\
&=& \sum_{i=1}^{l}\left(\frac{1}{N}\sum_{j=1}^{N}w_n(x,x_j)\right)y_i \end{array}$$
يمكنك أن ترى أن استجابة مجموعة التفرعات العشوائية عبارة عن مجموع مرجح للاستجابات على كافة أمثلة التدريب.ومن الجدير بالذكر أيضًا أن رقم الورقة $ \large T_n(x)$، حيث انتهى المثيل $\large x$، يعد ميزة قيمة في حد ذاته. على سبيل المثال، يعمل النهج التالي بشكل جيد: 1) يتم تدريب تكوين عدد صغير من الأشجار على عينة باستخدام غابة عشوائية أو تعزيز التدرج. 2) تتم إضافة الميزات الفئوية $\large T_1(x), \dots, T_n(x)$ إلى العينة.
هذه الميزات الجديدة هي نتيجة تقسيم المساحة غير الخطية، وهي توفر معلومات حول التشابه بين الأمثلة. وفي كتاب *عناصر التعلم الإحصائي* هناك مثال توضيحي جيد يوضح هذا التشابه بين الغابات العشوائية وتقنية الجيران الأقربين:

## 8. تحويل مجموعة البيانات إلى تمثيل عالي الأبعاد
تُستخدم الغابات العشوائية في الغالب في التعلم الخاضع للإشراف، ولكن هناك طريقة لتطبيقها في بيئة غير خاضعة للإشراف.
باستخدام طريقة `scikit-learn` [RandomTreesEmbedding](http://scikit-learn.org/stable/modules/generated/sklearn.ensemble.RandomTreesEmbedding.html#sklearn.ensemble.RandomTreesEmbedding)، يمكننا تحويل مجموعة البيانات الخاصة بنا إلى تمثيل عالي الأبعاد [متناثر](https://en.wikipedia.org/wiki/Sparse_matrix). نقوم أولاً ببناء أشجار عشوائية للغاية ثم نستخدم فهرس الورقة التي تحتوي على المثال كميزة جديدة.
على سبيل المثال، إذا ظهر الإدخال في الورقة الأولى، فإننا نقوم بتعيين $1$ كقيمة الميزة؛ إذا لم يكن الأمر كذلك، فإننا نعين $0$. وهذا ما يسمى *الترميز الثنائي*. يمكننا التحكم في عدد الميزات وتناثر البيانات عن طريق زيادة أو تقليل عدد الأشجار وعمقها. ونظرًا لأن نقاط البيانات القريبة من المحتمل أن تقع في نفس الورقة، فإن هذا التحويل يوفر تقديرًا ضمنيًا غير معلمي لكثافتها.



## 9. إيجابيات وسلبيات الغابات العشوائية

- فعال لمجموعات البيانات التي تحتوي على عدد كبير من الميزات والفئات.

- توجد طرق متطورة لتقدير أهمية الميزة.
- يعمل بشكل جيد مع البيانات المفقودة ويحافظ على دقة جيدة حتى في حالة فقدان جزء كبير من البيانات.
- يوفر وسيلة لوزن الفئات في مجموعة البيانات بأكملها وكذلك لكل عينة شجرة.
- تحت الغطاء، يحسب التقارب بين أزواج الأمثلة التي يمكن استخدامها لاحقًا في التجميع أو الكشف عن القيم المتطرفة أو تمثيلات البيانات المثيرة للاهتمام.
- يمكن توسيع الوظائف والخصائص المذكورة أعلاه لتشمل البيانات غير المسماة لتمكين التجميع غير الخاضع للرقابة، وتصور البيانات، واكتشاف البيانات الخارجية.
- متوازية بسهولة وقابلة للتطوير بدرجة كبيرة.
 
#### سلبيات:- بالمقارنة مع شجرة قرار واحدة، يكون تفسير مخرجات Random Forest أكثر صعوبة.
 - لا توجد قيم p رسمية لتقدير أهمية الميزة.
 - يؤدي أداءً أسوأ من الأساليب الخطية في حالة البيانات المتفرقة: إدخال النص، وحقيبة الكلمات، وما إلى ذلك.
 - على عكس الانحدار الخطي، فإن Random Forest غير قادر على الاستقراء. ولكن يمكن اعتبار هذا أيضًا ميزة لأن القيم المتطرفة لا تسبب قيمًا متطرفة في الغابات العشوائية.
 - عرضة للتجاوز في بعض المشاكل، خاصة عند التعامل مع البيانات المزعجة.
 - في حالة المتغيرات الفئوية ذات أرقام المستويات المتفاوتة، تفضل الغابات العشوائية المتغيرات ذات العدد الأكبر من المستويات. ستتناسب الشجرة بشكل أكبر مع الميزة ذات المستويات المتعددة لأن هذا يكتسب دقة أكبر.
 - إذا كانت مجموعة البيانات تحتوي على مجموعات من الميزات المترابطة، فقد يتم إعطاء الأفضلية للمجموعات ذات الحجم الأصغر. راجع [هذا العمل](http://rnowling.github.io/machine/learning/2015/08/11/random-forest-correlation-bias.html)
 - النموذج الناتج كبير ويتطلب الكثير من ذاكرة الوصول العشوائي.
